**3.4**
```python
class Held:
    punkte = 0

    def __init__(self, name):
        self.name = name

    def punkt_hinzufuegen(self):
        self.punkte += 1   # self statt Held!

a = Held('Aldric')
b = Held('Lyria')

a.punkt_hinzufuegen()
b.punkt_hinzufuegen()
a.punkt_hinzufuegen()

print(Held.punkte)
print(a.punkte)
print(b.punkte)
```

Im folgenden Code stecken drei Fehler. Finde und korrigiere alle drei.
Schreibe die korrigierte Version in die Code-Zelle und erklaere jeden Fehler als Kommentar.

```python
class Held:
    anzahl_helden = 0
    max_hp = 100

    def __init__(self, name, hp):
        self.name = name
        self.hp   = hp
        self.anzahl_helden += 1    # Fehler 1

    def heilen(self, menge):
        self.hp += menge
        if self.hp > self.max_hp:  # Fehler 2
            self.hp = max_hp       # Fehler 3

aldric = Held('Aldric', 80)
aldric.heilen(40)
print(aldric.hp)          # sollte 100 sein, nicht 120
print(Held.anzahl_helden) # sollte 1 sein
```
