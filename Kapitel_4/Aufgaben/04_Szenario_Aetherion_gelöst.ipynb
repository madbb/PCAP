Schreibe eine Klasse `Held` mit:

**Klassenvariablen:**
- `_registry = []` -- alle erstellten Helden
- `anzahl_helden = 0` -- Zaehler
- `MAX_HP = 500`
- `MIN_STAERKE = 1`

**Konstruktor:** `name`, `_hp` (max `MAX_HP`), `_staerke` (min `MIN_STAERKE`).
Erhoehe `anzahl_helden` und haenge `self` an `_registry`.

**Methoden:**
- `hp()`, `staerke()`: Getter
- `ist_am_leben()`: `True` wenn `_hp > 0`
- `erleidet_schaden(schaden)`: zieht Schaden ab, HP nie unter 0
- `heilen(menge)`: heilt, max `MAX_HP`, nur wenn noch am Leben
- `angreifen()`: gibt `_staerke * 2` zurueck
- `__str__`: `'Aldric (HP: 100/500, STR: 15)'`

**Klassenmethoden (als normale Methoden):**
- `alle_helden()`: gibt `_registry` zurueck
- `suche(name)`: gibt Held mit diesem Namen zurueck oder `None`


````python

class Held:
    _registry =  []
    anzahl_helden =  0
    MAX_HP = 500
    MIN_STAERKE = 1

    def __init__(self,name,hp,staerke):
        self.name = name
        self._hp = min(hp, Held.MAX_HP)
        self._staerke = max(staerke, Held.MIN_STAERKE)
        self.ausruestung = Ausruestung()
        self.inventar = Inventar()
        Held.anzahl_helden += 1
        Held._registry.append(self)

    def hp(self):
        return self._hp

    def staerke(self):
        return  self._staerke

    def ist_am_leben(self):
        return self._hp > 0

    def erleidet_schaden(self, schaden):
        reduziert = max(0, schaden - self.ausruestung.gesamt_schutz())
        self._hp = max(0,self._hp -reduziert)

    def heilen(self, menge):
        if self.ist_am_leben():
            self._hp = min(self._hp + menge, Held.MAX_HP)

    def angreifen(self):
        return self._staerke * 2 + self.ausruestung.gesamt_schaden()

    def alle_helden(self):
        return Held._registry

    def suche(self, name):
        for held in Held._registry:
            if held.name == name:
                return held
        return  None

    def __str__(self):
        return f'{self.name} (HP: {self._hp}/{Held.MAX_HP}, STR: {self._staerke}'
````

Schreibe drei Klassen die von `Held` erben:

**`Krieger(Held)`:**
- Zusatzattribut `ruestung` (Default `5`)
- `erleidet_schaden`: zieht zuerst `ruestung` vom Schaden ab, dann erst HP
- `angreifen()`: gibt `staerke * 2` zurueck (unveraendert)
- `__str__`: erweitert Held-String um `', RUE: X'`

**`Magierin(Held)`:**
- Zusatzattribut `_mana` (Default `80`, Min `0`)
- `angreifen()`: kostet 10 Mana, gibt `staerke * 3` zurueck.
  Ohne Mana: normaler Angriff `staerke * 2`, kein Manaverbrauch
- `mana()`: Getter
- `__str__`: erweitert um `', MANA: X'`

**`Waldlaeufer(Held)`:**
- Zaehlt Angriffe intern mit `_angriffe = 0`
- `angreifen()`: normaler Schaden, aber bei jedem zweiten Angriff Doppelschuss: `staerke * 4`
- `__str__`: unveraendert (erbt von `Held`)

Alle `__init__` rufen `super().__init__()` auf.


````python
class Krieger(Held):
    def __init__(self, name, hp, staerke, ruestung=5):
        super().__init__(name, hp, staerke)
        self.ruestung = ruestung

    def erleidet_schaden(self, schaden):
        reduziert = max(0, schaden - self.ruestung)
        super().erleidet_schaden(reduziert)

    def __str__(self):
        return super().__str__() + f', RUE: {self.ruestung}'


class Magierin(Held):
    def __init__(self, name, hp, staerke, mana=80):
        super().__init__(name, hp, staerke)
        self._mana = max(0, mana)

    def mana(self):
        return self._mana

    def angreifen(self):
        if self._mana >= 10:
            self._mana -= 10
            return self._staerke * 3
        return self._staerke * 2

    def __str__(self):
        return super().__str__() + f', MANA: {self._mana}'


class Waldlaeufer(Held):
    def __init__(self, name, hp, staerke):
        super().__init__(name, hp, staerke)
        self._angriffe = 0

    def angreifen(self):
        self._angriffe += 1
        if self._angriffe % 2 == 0:
            return self._staerke * 4
        return self._staerke * 2
````

````python
# Testblock
aldric = Krieger('Aldric', 100, 15)
lyria  = Magierin('Lyria', 80, 12)
thorn  = Waldlaeufer('Thorn', 90, 14)

aldric.erleidet_schaden(20)
assert aldric.hp() == 85   # 20 - 5 Ruestung = 15 Schaden

assert lyria.angreifen() == 12 * 3   # 36
assert lyria.mana() == 70
lyria._mana = 5
assert lyria.angreifen() == 12 * 2   # kein Mana: normaler Angriff
assert lyria.mana() == 5             # kein Verbrauch

# Waldlaeufer: Doppelschuss bei jedem zweiten Angriff
assert thorn.angreifen() == 14 * 2   # Angriff 1: normal
assert thorn.angreifen() == 14 * 4   # Angriff 2: Doppelschuss
assert thorn.angreifen() == 14 * 2   # Angriff 3: normal

# isinstance-Pruefungen
assert isinstance(aldric, Held)
assert isinstance(aldric, Krieger)
assert not isinstance(aldric, Magierin)

print('Abschnitt 2.1: bestanden')
````

Schreibe eine Klasse `Paladin(Krieger)` die:
- von `Krieger` erbt (bekommt also auch Ruestung)
- zusaetzlich `_mana = 40` hat
- `heilen_ziel(held, menge)`: heilt einen anderen Helden um `menge` HP,
  kostet 20 Mana. Gibt `False` zurueck wenn kein Mana reicht.
- `angreifen()`: ruft `super().angreifen()` auf und addiert `10` Bonusschaden
- `__str__`: erweitert Krieger-String um `', MANA: X'`


````python
class Paladin(Krieger):
    def __init__(self, name, hp, staerke, ruestung=5):
        super().__init__(name, hp, staerke, ruestung)
        self._mana = 40

    def heilen_ziel(self, held, menge):
        if self._mana <= 20:
            return False
        self._mana -=20
        held.heilen(menge)
        return True

    def angreifen(self):
        return super().angreifen() + 10

    def __str__(self):
        return super().__str__() + f', MANA: {self._mana}'
````

````python
# Testblock
sera   = Paladin('Sera', 90, 12)
aldric = Krieger('Aldric', 50, 15)

# Paladin erbt Ruestung vom Krieger
sera.erleidet_schaden(20)
assert sera.hp() == 75   # 20 - 5 Ruestung = 15 Schaden

# Angriff: Krieger-Basis + 10 Bonus
assert sera.angreifen() == 12 * 2 + 10   # 34

# Heilen
aldric.erleidet_schaden(30)
hp_vorher = aldric.hp()
ergebnis  = sera.heilen_ziel(aldric, 20)
assert ergebnis == True
assert aldric.hp() == hp_vorher + 20
assert sera._mana == 20

# Kein Mana mehr
assert sera.heilen_ziel(aldric, 20) == False

# Vererbungskette
assert isinstance(sera, Paladin)
assert isinstance(sera, Krieger)
assert isinstance(sera, Held)

print('Abschnitt 2.2: bestanden')
````

1. Schreibe eine Klasse `Gegenstand` mit `name`, `typ`, `schaden=0`, `schutz=0`, `wert=0`.
   `__str__`: `'Eisenschwert (Waffe, +25 Schaden)'`
   `__repr__`: `"Gegenstand('Eisenschwert', 'Waffe', schaden=25)"`

2. Schreibe eine Klasse `Ausruestung` mit vier Slots: `waffe`, `ruestung`, `helm`, `ring`
   (alle initial `None`).
   - `anlegen(gegenstand)`: weist anhand von `gegenstand.typ` den richtigen Slot zu
   - `ablegen(typ)`: leert den Slot, gibt den Gegenstand zurueck oder `None`
   - `gesamt_schaden()`: Summe aller `schaden`-Werte belegter Slots
   - `gesamt_schutz()`: Summe aller `schutz`-Werte belegter Slots
   - `__str__`: listet alle belegten Slots

3. Erweitere `Held.__init__` um `self.ausruestung = Ausruestung()`.
   Aendere `angreifen()`: gibt `_staerke * 2 + self.ausruestung.gesamt_schaden()` zurueck.
   Aendere `erleidet_schaden()`: zieht `self.ausruestung.gesamt_schutz()` vom Schaden ab.


````python
class Gegenstand:
    def __init__(self, name, typ, schaden=0, schutz=0, wert=0):
        self.name = name
        self.typ = typ
        self.schaden = schaden
        self.schutz = schutz
        self.wert = wert

    def __str__(self):
        teile = []
        if self.schaden:
            teile.append(f'+{self.schaden} Schaden')
        if self.schutz:
            teile.append(f'+{self.schutz} Schutz')
        details = ', '.join(teile) if teile else 'kein Bonus'
        return f'{self.name} ({self.typ}, {details})'

    def __repr__(self):
        teile = [f"'{self.name}'", f"'{self.typ}'"]
        if self.schaden:
            teile.append(f'schaden={self.schaden}')
        if self.schutz:
            teile.append(f'schutz={self.schutz}')
        if self.wert:
            teile.append(f'wert={self.wert}')
        return f"Gegenstand({', '.join(teile)})"


class Ausruestung:
    SLOT_TYPEN = {'Waffe': 'waffe', 'Ruestung': 'ruestung', 'Helm': 'helm', 'Ring': 'ring'}

    def __init__(self):
        self.waffe    = None
        self.ruestung = None
        self.helm     = None
        self.ring     = None

    def anlegen(self, gegenstand):
        slot = self.SLOT_TYPEN.get(gegenstand.typ)
        if slot:
            setattr(self, slot, gegenstand)

    def ablegen(self, typ):
        slot = self.SLOT_TYPEN.get(typ)
        if slot:
            g = getattr(self, slot)
            setattr(self, slot, None)
            return g
        return None

    def gesamt_schaden(self):
        return sum(
            g.schaden for g in [self.waffe, self.ruestung, self.helm, self.ring]
            if g is not None
        )

    def gesamt_schutz(self):
        return sum(
            g.schutz for g in [self.waffe, self.ruestung, self.helm, self.ring]
            if g is not None
        )

    def __str__(self):
        belegte = []
        for slot_name in ['waffe', 'ruestung', 'helm', 'ring']:
            g = getattr(self, slot_name)
            if g:
                belegte.append(f'{slot_name.capitalize()}: {g.name}')
        return ', '.join(belegte) if belegte else 'Keine Ausruestung'
````

````python
# Testblock
aldric  = Krieger('Aldric', 100, 15)
schwert = Gegenstand('Eisenschwert', 'Waffe', schaden=25)
helm    = Gegenstand('Eisenhelm', 'Helm', schutz=5)
ring    = Gegenstand('Kraftring', 'Ring', schaden=5)

aldric.ausruestung.anlegen(schwert)
aldric.ausruestung.anlegen(helm)
aldric.ausruestung.anlegen(ring)

# Angriff: STR*2 + Ausruestungsschaden
assert aldric.angreifen() == 15 * 2 + 25 + 5   # 60

# Schaden: Ruestung des Kriegers + Helm
aldric.erleidet_schaden(20)
# Krieger-Ruestung 5 + Helm 5 = 10 Schutz -> 10 Schaden
assert aldric.hp() == 90

abgelegt = aldric.ausruestung.ablegen('Waffe')
assert abgelegt is schwert
assert aldric.angreifen() == 15 * 2 + 5   # ohne Schwert

print('Abschnitt 3.1: bestanden')
print(aldric.ausruestung)
````

Schreibe eine Klasse `Inventar` mit `self._items = []`.

Implementiere:
- `__len__`: Anzahl Gegenstaende
- `__contains__`: `gegenstand in inventar` (prueft per `is`)
- `__iter__`: Iteration ueber alle Gegenstaende
- `__add__`: zwei Inventare zusammenfuehren
- `__str__`: alle Gegenstaende zeilenweise
- `aufnehmen(gegenstand)`: fuegt hinzu
- `ablegen(name)`: entfernt nach Name, gibt `True`/`False` zurueck
- `wert_gesamt()`: Summe aller `wert`-Attribute

Erweitere `Held.__init__` um `self.inventar = Inventar()`.


````python
class Inventar:
    def __init__(self):
        self._items = []

    def aufnehmen(self, item):
        self._items.append(item)

    def ablegen(self, name):
        for i in self._items:
            if i.name == name:
                self._items.remove(i)
                return True
        return False

    def wert_gesamt(self):
        return sum(i.wert for i in self._items)

    def __len__(self):
        return len(self._items)

    def __contains__(self, item):
        return item in self._items   # is / ==

    def __iter__(self):
        return iter(self._items)

    def __add__(self, other):
        neu = Inventar()
        neu._items = self._items + other._items
        return neu

    def __str__(self):
        if not self._items:
            return 'Inventar leer'
        return '\n'.join(str(i) for i in self._items)
````

````python
# Testblock
aldric    = Krieger('Aldric', 100, 15)
heiltrank = Gegenstand('Heiltrank', 'Trank', wert=10)
schluessel = Gegenstand('Kelchschluessel', 'Schluessel', wert=5)
goldmuenze = Gegenstand('Goldmuenze', 'Muenze', wert=1)

aldric.inventar.aufnehmen(heiltrank)
aldric.inventar.aufnehmen(schluessel)
aldric.inventar.aufnehmen(goldmuenze)

assert len(aldric.inventar) == 3
assert heiltrank in aldric.inventar
assert aldric.inventar.wert_gesamt() == 16

assert aldric.inventar.ablegen('Heiltrank') == True
assert len(aldric.inventar) == 2
assert aldric.inventar.ablegen('Zauberstab') == False

# Inventare kombinieren
lyria = Magierin('Lyria', 80, 12)
lyria.inventar.aufnehmen(Gegenstand('Manastein', 'Stein', wert=20))

gemeinsam = aldric.inventar + lyria.inventar
assert len(gemeinsam) == 3   # 2 von Aldric + 1 von Lyria

print('Abschnitt 3.2: bestanden')
````

1. Definiere eine Liste `BEKANNTE_FAEHIGKEITEN = ['angreifen', 'heilen_ziel', 'zaubern', 'schleichen', 'tanzen']`.
2. Schreibe eine Funktion `faehigkeiten_von(charakter)`,
   die alle Methoden aus `BEKANNTE_FAEHIGKEITEN` zurueckgibt, die der Charakter besitzt.
   Nutze `hasattr()`.
3. Schreibe eine Funktion `kann(charakter, faehigkeit)`,
   die `True` zurueckgibt wenn der Charakter die Faehigkeit hat.
4. Schreibe eine Funktion `ausfuehren(charakter, faehigkeit, *args)`,
   die die Faehigkeit aufruft wenn vorhanden, sonst eine Meldung ausgibt.
   Nutze `getattr()`.
5. Testblock nicht veraendern:


````python
BEKANNTE_FAEHIGKEITEN = ['angreifen', 'heilen_ziel', 'zaubern', 'schleichen', 'tanzen']

def faehigkeiten_von(charakter):
    return [f for f in BEKANNTE_FAEHIGKEITEN if hasattr(charakter, f)]

def kann(charakter, faehigkeit):
    return hasattr(charakter, faehigkeit)

def ausfuehren(charakter, faehigkeit, *args):
    if hasattr(charakter, faehigkeit):
        methode = getattr(charakter, faehigkeit)
        return methode(*args)
    print(f'{charakter.name} kann nicht {faehigkeit}')
    return None


````

````python
# Testblock
aldric = Krieger('Aldric', 100, 15)
sera   = Paladin('Sera', 90, 12)
lyria  = Magierin('Lyria', 80, 12)

# Krieger hat nur angreifen
assert 'angreifen' in faehigkeiten_von(aldric)
assert 'heilen_ziel' not in faehigkeiten_von(aldric)

# Paladin hat angreifen und heilen_ziel
assert 'angreifen' in faehigkeiten_von(sera)
assert 'heilen_ziel' in faehigkeiten_von(sera)

assert kann(sera, 'heilen_ziel') == True
assert kann(aldric, 'zaubern') == False

# ausfuehren: Paladin heilt Aldric
aldric.erleidet_schaden(40)
hp_vorher = aldric.hp()
ausfuehren(sera, 'heilen_ziel', aldric, 20)
assert aldric.hp() == hp_vorher + 20

# ausfuehren: Aldric kann nicht zaubern
ausfuehren(aldric, 'zaubern')   # gibt Meldung aus, kein Fehler

print('Abschnitt 4.1: bestanden')
````

1. Schreibe eine Funktion `charakter_report(charakter)`,

```

```
   Nutze `type(charakter).__name__` und `type(charakter).__bases__`.
2. Schreibe eine Funktion `truppe_analysieren(charaktere)`,
   die ein Dictionary zurueckgibt mit:
   - `'gesamt'`: Gesamtzahl
   - `'am_leben'`: Anzahl lebender Charaktere
   - `'typen'`: Dictionary `{Klassenname: Anzahl}`
   - `'kann_heilen'`: Liste der Charaktere mit `heilen_ziel`
3. Testblock nicht veraendern:


````python
def charakter_report(charakter):

    klasse = type(charakter).__name__  # Krieger

    basisklassen = [b.__name__ for b in type(charakter).__bases__]

    faehigkeiten = faehigkeiten_von(charakter)

    """
    === Aldric ===
    Klasse    : Krieger
    Erbt von  : Held
    Lebt      : True
    Faehigkeiten: angreifen"""

    lines = [
        f'=== {charakter.name} ===',
        f'Klasse        : {klasse}',
        f'Erbt von      : {", ".join(basisklassen)}',
        f'Lebt          : {charakter.ist_am_leben()}',
        f'Faehigkeiten  : {', '.join(faehigkeiten) if faehigkeiten else "keine"}'
    ]

    return '\n'.join(lines)


def truppe_analysieren(charaktere):

    typen       = {}
    kann_heilen = []
    am_leben    = 0

    for c in charaktere:

        name = type(c).__name__
        typen[name] = typen.get(name, 0) + 1

        if c.ist_am_leben():
            am_leben += 1

        if hasattr(c, "heilen_ziel"):
            kann_heilen.append(c)

    return {
        'gesamt':       len(charaktere),
        'am_leben':     am_leben,
        'typen':        typen,
        'kann_heilen':  kann_heilen
    }


````
