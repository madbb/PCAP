Schritt 1 ist entscheidend: **ein Modul wird nur einmal geladen**, egal wie oft du es importierst.

````python
import sys

# Erstes Importieren
import math
id_erstes_mal = id(sys.modules['math'])

# Zweites Importieren -- kein erneutes Laden
import math
id_zweites_mal = id(sys.modules['math'])

print('Selbes Objekt?', id_erstes_mal == id_zweites_mal)  # True

# Das bedeutet auch: Code auf Modulebene laeuft nur einmal
# (wichtig fuer Seiteneffekte beim Import)
````

`sys.path` ist eine Liste von Verzeichnissen, die Python der Reihe nach durchsucht. Das erste Verzeichnis mit einer passenden Datei gewinnt.

Die Reihenfolge ist festgelegt:

1. Das Verzeichnis des ausgefuehrten Skripts (oder `''` fuer das aktuelle Verzeichnis im interaktiven Modus)
2. Verzeichnisse aus der Umgebungsvariablen `PYTHONPATH`
3. Installationsabhaengige Standardverzeichnisse (stdlib, site-packages)

````python
import sys

print('sys.path Eintraege:')
for i, pfad in enumerate(sys.path):
    print(f'  [{i}] {pfad if pfad else "(aktuelles Verzeichnis)"}')
````

````python
import sys
import os

# sys.path zur Laufzeit erweitern
# Nuetzlich wenn eigene Module in einem anderen Verzeichnis liegen
eigenes_verzeichnis = os.path.abspath('./meine_module')

if eigenes_verzeichnis not in sys.path:
    sys.path.insert(0, eigenes_verzeichnis)  # vorne einfuegen = hoechste Prioritaet

print('Erster Eintrag:', sys.path[0])

# Hinweis: sys.path.append() fuegt hinten an -- niedrigere Prioritaet
# sys.path.insert(0, ...) fuegt vorne an -- hoehere Prioritaet als stdlib
# Das kann stdlib-Module verdecken -- mit Vorsicht einsetzen
````

Python bietet mehrere Wege, etwas zu importieren. Sie unterscheiden sich darin, was in den aktuellen Namespace eingetragen wird.

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;"><span style="font-size:0.6rem;letter-spacing:0.1em;text-transform:uppercase;color:#80cbc4;background:rgba(128,203,196,0.1);border:1px solid rgba(128,203,196,0.25);border-radius:3px;padding:0.15em 0.5em;margin-right:0.65em;vertical-align:middle;white-space:nowrap;">PCAP 1.1</span>Form 1: import modul</span></div>

Importiert das Modul und bindet seinen Namen im aktuellen Namespace. Zugriff auf Inhalte ueber den Modulnamen.

````python
import math

print(math.pi)           # 3.141592...
print(math.sqrt(16))     # 4.0

# Im Namespace ist nur 'math' eingetragen, nicht 'pi' oder 'sqrt'
# Das ist die sicherste Form: klar, woher jede Funktion stammt
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;"><span style="font-size:0.6rem;letter-spacing:0.1em;text-transform:uppercase;color:#80cbc4;background:rgba(128,203,196,0.1);border:1px solid rgba(128,203,196,0.25);border-radius:3px;padding:0.15em 0.5em;margin-right:0.65em;vertical-align:middle;white-space:nowrap;">PCAP 1.1</span>Form 2: import modul as alias</span></div>

Wie Form 1, aber der Name im Namespace wird umbenannt. Nützlich bei langen Modulnamen oder Namenskonflikten.

````python
import math as m
import datetime as dt

print(m.pi)
print(dt.date.today())

# Konventionelle Aliases (de-facto Standard in der Python-Welt):
# import numpy as np
# import pandas as pd
# import matplotlib.pyplot as plt
````

Importiert einen bestimmten Namen direkt in den aktuellen Namespace. Der Modulname selbst wird nicht eingetragen.

````python
from math import sqrt, pi

print(sqrt(25))   # 5.0  -- kein 'math.' noetig
print(pi)         # 3.141592...

# 'math' ist NICHT im Namespace
try:
    print(math.e)
except NameError as e:
    print(f'Fehler: {e}')

# Vorteil: kuerzerer Code
# Nachteil: nicht sofort erkennbar, woher sqrt kommt
````

````python
from math import sqrt as wurzel
from datetime import datetime as Zeitpunkt

print(wurzel(9))           # 3.0
print(Zeitpunkt.now())     # aktueller Zeitstempel

# Nuetzlich bei Namenskonflikten zwischen zwei Modulen
# z.B. from modul_a import connect as connect_a
#      from modul_b import connect as connect_b
````

Importiert alle oeffentlichen Namen eines Moduls in den aktuellen Namespace.

````python
from math import *

print(sin(pi / 2))   # 1.0
print(cos(0))        # 1.0
print(e)             # 2.718...

# Was wurde alles importiert?
import math
print('\nAnzahl importierter Namen:', len([n for n in dir(math) if not n.startswith('_')]))
````

> **`import *` -- wann ja, wann nein?**
>
> In Produktionscode gilt `import *` als schlechte Praxis: Es ist nicht nachvollziehbar, woher ein Name stammt, und es koennen unbemerkt Namen im aktuellen Namespace ueberschrieben werden.
>
> Einzige Ausnahme: interaktive Shell oder Jupyter-Notebooks fuer schnelle Erkundung.
>
> Ein Modul kann ueber `__all__` steuern, welche Namen bei `import *` exportiert werden:

````python
# Beispiel: Modul mit __all__
# Angenommen, das waere der Inhalt von 'mein_modul.py':

# __all__ = ['oeffentlich']   # nur 'oeffentlich' wird bei import * exportiert
#
# def oeffentlich():
#     return 'ich bin oeffentlich'
#
# def _intern():
#     return 'ich bin intern (Konvention: Unterstrich)'
#
# def auch_intern():
#     return 'ich bin NICHT in __all__, also auch nicht bei import *'

# Schauen wir uns __all__ eines echten Moduls an:
import os.path
print('os.path.__all__:', os.path.__all__ if hasattr(os.path, '__all__') else 'nicht definiert')

import json
print('json.__all__:', json.__all__ if hasattr(json, '__all__') else 'nicht definiert')
````

Jedes Python-Modul hat eine Variable `__name__`. Ihr Wert haengt davon ab, wie das Modul verwendet wird:

- Wird die Datei **direkt ausgefuehrt** (`python datei.py`), ist `__name__` gleich `'__main__'`
- Wird die Datei **importiert**, ist `__name__` gleich dem Modulnamen (z.B. `'datei'`)

Das erlaubt es, Code zu schreiben, der nur beim direkten Ausfuehren laeuft -- nicht beim Import.

````python
# In einem Notebook ist __name__ immer '__main__'
print('__name__ in diesem Notebook:', __name__)

# Simulieren wir, wie das in einer .py-Datei aussehen wuerde:
# ------------------------------------------------------------------
# # berechnung.py
#
# def quadrat(n):
#     return n * n
#
# def wuerfel(n):
#     return n * n * n
#
# if __name__ == '__main__':
#     # Dieser Block laeuft NUR wenn berechnung.py direkt gestartet wird
#     # Beim Import durch ein anderes Modul wird er uebersprungen
#     print('Teste quadrat:', quadrat(5))   # 25
#     print('Teste wuerfel:', wuerfel(3))   # 27
# ------------------------------------------------------------------
````

````python
# Warum ist das wichtig?
# Ohne if __name__ == '__main__' wuerde dieser Code beim Import ausgefuehrt:

# -- schlechtes Beispiel --
# def berechne():
#     return 42
#
# print('Ergebnis:', berechne())   # laeuft IMMER, auch beim Import
#                                   # nervt jeden, der das Modul importiert

# -- gutes Beispiel --
# def berechne():
#     return 42
#
# if __name__ == '__main__':
#     print('Ergebnis:', berechne())   # laeuft nur bei direktem Aufruf

# Wir koennen __name__ verschiedener importierter Module pruefen:
import math
import os
print('math.__name__:', math.__name__)
print('os.__name__:  ', os.__name__)
````

Ein zirkulaerer Import entsteht, wenn Modul A Modul B importiert und Modul B gleichzeitig Modul A importiert.

```
modul_a.py: import modul_b
modul_b.py: import modul_a   # Zirkel!
```

Python erkennt das und bricht nicht in eine Endlosschleife -- aber das Modul ist moeglicherweise noch nicht vollstaendig geladen, wenn es das zweite Mal importiert wird. Das fuehrt zu schwer nachvollziehbaren `ImportError` oder `AttributeError`.

````python
import os

# Wir simulieren zirkulaere Imports mit temporaeren Dateien
os.makedirs('zirkel_demo', exist_ok=True)

with open('zirkel_demo/modul_a.py', 'w') as f:
    f.write('from zirkel_demo import modul_b\n')
    f.write('def funktion_a():\n    return "A"\n')

with open('zirkel_demo/modul_b.py', 'w') as f:
    f.write('from zirkel_demo import modul_a\n')
    f.write('def funktion_b():\n    return modul_a.funktion_a() + "B"\n')

with open('zirkel_demo/__init__.py', 'w') as f:
    f.write('')

import sys
try:
    from zirkel_demo import modul_a
    print(modul_a.funktion_a())
except Exception as e:
    print(f'{type(e).__name__}: {e}')
finally:
    # Aufraeumen
    import shutil
    shutil.rmtree('zirkel_demo', ignore_errors=True)
    for key in list(sys.modules.keys()):
        if 'zirkel_demo' in key:
            del sys.modules[key]
````

Es gibt drei gaengige Loesungen:

**1. Struktur ueberdenken:** Meistens ist ein zirkulaerer Import ein Zeichen, dass zwei Module zu eng gekoppelt sind. Die gemeinsam genutzten Teile in ein drittes Modul auslagern.

**2. Import innerhalb der Funktion:** Den Import erst dort ausfuehren, wo er benoetigt wird -- dann ist das Modul bereits vollstaendig geladen.

```python
# modul_a.py
def funktion_a():
    from modul_b import funktion_b   # Import erst bei Aufruf
    return funktion_b()
```

**3. Typannotationen mit `TYPE_CHECKING`:** Wenn der Zirkel nur wegen Typhinweisen entsteht:

```python
from __future__ import annotations
from typing import TYPE_CHECKING
if TYPE_CHECKING:
    from modul_b import KlasseB   # wird zur Laufzeit nicht ausgefuehrt
```

Python-Module sind Objekte -- sie haben Attribute, die Informationen ueber das Modul liefern.

````python
import math
import os

print('Name:       ', math.__name__)
print('Datei:      ', math.__file__ if hasattr(math, '__file__') else 'eingebaut (C-Extension)')
print('Paket:      ', math.__package__)
print('Dokumentation (Auszug):') 
print(math.__doc__[:200] if math.__doc__ else 'keine')
print()
print('Oeffentliche Namen:', [n for n in dir(math) if not n.startswith('_')])
````

````python
# Wo liegt ein Modul auf der Festplatte?
import json
import csv
import pathlib

for modul in [json, csv, pathlib]:
    if hasattr(modul, '__file__'):
        print(f'{modul.__name__:12} --> {modul.__file__}')
````

| Thema | Kernaussage |
|-------|-------------|
| Import-Ablauf | Suche in `sys.modules` --> `sys.path` --> Laden --> Ausfuehren --> Cachen |
| Einmaliges Laden | `sys.modules` verhindert mehrfaches Laden desselben Moduls |
| `sys.path` | Liste der Suchpfade; erweiterbar zur Laufzeit |
| Import-Formen | `import m`, `import m as a`, `from m import x`, `from m import x as a`, `from m import *` |
| `__name__` | `'__main__'` bei direktem Aufruf, Modulname bei Import |
| Absolut vs. relativ | Absolut bevorzugt; relativ nuetzlich in grossen Paketen |
| Zirkulaere Imports | Strukturproblem -- durch Auslagerung oder lokalen Import loesen |
| `reload()` | Erzwingt erneutes Laden -- noetig bei Aenderungen im laufenden Interpreter |

**Merksatz:** `import` ist keine passive Deklaration. Python fuehrt Modulcode aus, cacht das Ergebnis, und bindet Namen. Wer das versteht, versteht auch, warum Import-Reihenfolge und Seiteneffekte auf Modulebene wichtig sind.
