Python liest zunächst den Quelltext als reinen Text ein. Der **Lexer** zerlegt diesen Text in sogenannte *Tokens* -- die kleinsten bedeutungstragenden Einheiten der Sprache.

Aus der Zeile `x = 42 + 1` werden z.B. diese Tokens:

| Token | Typ |
|-------|-----|
| `x` | NAME |
| `=` | EQUAL |
| `42` | NUMBER |
| `+` | PLUS |
| `1` | NUMBER |

Der **Parser** prüft danach, ob die Reihenfolge der Tokens der Python-Grammatik entspricht. Fehlt eine Klammer oder ein Doppelpunkt, bricht Python hier ab -- das ist ein `SyntaxError`.

````python
# Wir koennen uns die Tokens einer Code-Zeile direkt anschauen
import tokenize
import io

quellcode = 'x = 42 + 1'

tokens = tokenize.generate_tokens(io.StringIO(quellcode).readline)

for tok in tokens:
    print(tok)
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;"><span style="font-size:0.6rem;letter-spacing:0.1em;text-transform:uppercase;color:#80cbc4;background:rgba(128,203,196,0.1);border:1px solid rgba(128,203,196,0.25);border-radius:3px;padding:0.15em 0.5em;margin-right:0.65em;vertical-align:middle;white-space:nowrap;">Kursinhalt</span>Schritt 2: Abstract Syntax Tree (AST)</span></div>

Aus den Tokens baut Python einen **Abstract Syntax Tree** -- eine Baumstruktur, die die logische Bedeutung des Codes darstellt. Rein syntaktische Details wie Leerzeichen oder Klammern um Ausdrücke fallen dabei weg.

Den AST kann man sich in Python direkt anzeigen lassen:

````python
import ast

quellcode = """
def greet(name):
    return 'Hallo, ' + name
"""

baum = ast.parse(quellcode)
print(ast.dump(baum, indent=2))
````

Die Ausgabe wirkt auf den ersten Blick komplex. Wichtig ist das Prinzip: Python arbeitet nicht mit dem Originaltext, sondern mit dieser strukturierten Baumdarstellung. Das ermöglicht Werkzeuge wie Linter, Formatter und statische Typprüfung.

> **Praxisbezug:** Tools wie `pylint`, `mypy` oder `black` analysieren denselben AST, den Python intern erzeugt.

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;"><span style="font-size:0.6rem;letter-spacing:0.1em;text-transform:uppercase;color:#80cbc4;background:rgba(128,203,196,0.1);border:1px solid rgba(128,203,196,0.25);border-radius:3px;padding:0.15em 0.5em;margin-right:0.65em;vertical-align:middle;white-space:nowrap;">Kursinhalt</span>Schritt 3: Compiler und Bytecode</span></div>

Aus dem AST erzeugt Pythons interner Compiler **Bytecode** -- einen plattformunabhängigen Zwischencode, der effizienter zu verarbeiten ist als der Originaltext.

Bytecode ist kein Maschinencode (den dein Prozessor direkt versteht), sondern Instruktionen für die Python Virtual Machine.

Mit dem Modul `dis` (Disassembler) kann man den Bytecode einer Funktion lesbar darstellen:

````python
import dis

def addiere(a, b):
    ergebnis = a + b
    return ergebnis

dis.dis(addiere)
````

Die Spalten bedeuten:

| Spalte | Bedeutung |
|--------|-----------|
| Zeilennummer | Zeile im Quellcode |
| Byte-Offset | Position im Bytecode |
| Instruktion | Was die PVM tun soll |
| Argument | Operand der Instruktion |

`LOAD_FAST` lädt eine lokale Variable. `BINARY_OP` führt die Addition aus. `RETURN_VALUE` gibt den Wert zurück.

Du musst Bytecode nicht auswendig kennen -- aber wenn du verstehst, dass Python intern auf dieser Ebene arbeitet, erklären sich viele Verhaltensweisen von selbst.

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;"><span style="font-size:0.6rem;letter-spacing:0.1em;text-transform:uppercase;color:#80cbc4;background:rgba(128,203,196,0.1);border:1px solid rgba(128,203,196,0.25);border-radius:3px;padding:0.15em 0.5em;margin-right:0.65em;vertical-align:middle;white-space:nowrap;">PCAP 1.5</span>.pyc-Dateien: Bytecode auf der Festplatte</span></div>

Python speichert den kompilierten Bytecode in `.pyc`-Dateien, damit er beim nächsten Programmstart nicht neu erzeugt werden muss. Das beschleunigt den Start.

Sie liegen im Unterordner `__pycache__` neben dem Quellcode:

```
mein_projekt/
    mein_modul.py
    __pycache__/
        mein_modul.cpython-311.pyc
```

Der Dateiname enthält die Python-Version (`cpython-311` = CPython 3.11). Wenn du den Quellcode änderst oder eine andere Python-Version verwendest, erstellt Python die `.pyc`-Datei automatisch neu.

> **Wichtig:** `.pyc`-Dateien gehören nicht ins Git-Repository. In einer guten `.gitignore` steht deshalb immer `__pycache__/` und `*.pyc`.

````python
# Wir erzeugen eine .pyc-Datei manuell und schauen, wo sie landet
import py_compile
import os

# Erstelle eine kleine Testdatei
with open('beispiel_modul.py', 'w') as f:
    f.write('def greet():\n    return "Hallo"\n')

# Kompiliere sie
py_compile.compile('beispiel_modul.py')

# Zeige den Inhalt von __pycache__
if os.path.exists('__pycache__'):
    print('Inhalt von __pycache__:')
    for datei in os.listdir('__pycache__'):
        print(' ', datei)
else:
    print('__pycache__ noch nicht vorhanden')
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;"><span style="font-size:0.6rem;letter-spacing:0.1em;text-transform:uppercase;color:#80cbc4;background:rgba(128,203,196,0.1);border:1px solid rgba(128,203,196,0.25);border-radius:3px;padding:0.15em 0.5em;margin-right:0.65em;vertical-align:middle;white-space:nowrap;">Kursinhalt</span>Schritt 4: Die Python Virtual Machine (PVM)</span></div>

Die **PVM** ist das Herzstück von Python. Sie liest den Bytecode Instruktion für Instruktion und führt ihn aus. Die PVM ist eine Laufzeitumgebung -- sie verwaltet:

- den **Call Stack** (welche Funktionen gerade aktiv sind)
- den **Heap** (dynamisch allozierter Speicher für Objekte)
- den **Garbage Collector** (automatische Speicherfreigabe)

Die bekannteste PVM-Implementierung ist **CPython** (in C geschrieben). Daneben gibt es PyPy (JIT-Compiler, deutlich schneller für numerischen Code), Jython (läuft auf der JVM) und weitere.

Wenn Menschen von "Python" sprechen, meinen sie fast immer CPython.

````python
import sys
import platform

print('Python-Version:   ', sys.version)
print('Implementierung:  ', platform.python_implementation())
print('Compiler:         ', platform.python_compiler())
print('Plattform:        ', sys.platform)
````

<div style="background:#0b1929;border-left:4px solid #4fc3f7;border-radius:4px;padding:0.75em 1.2em;margin:2rem 0 1.2rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(1rem,2.2vw,1.25rem);font-weight:600;color:#e8eaf6;letter-spacing:-0.01em;"><span style="font-size:0.65rem;letter-spacing:0.1em;text-transform:uppercase;color:#4fc3f7;background:rgba(79,195,247,0.12);border:1px solid rgba(79,195,247,0.3);border-radius:3px;padding:0.2em 0.55em;margin-right:0.75em;vertical-align:middle;font-family:'Segoe UI',sans-serif;white-space:nowrap;">Kursinhalt</span>Interpretiert oder kompiliert?</span></div>

<span style="font-size:clamp(1rem,2.2vw,1.35rem);font-weight:600;color:#d0d0e0;letter-spacing:-0.01em;"><span style="font-size:0.7rem;letter-spacing:0.12em;text-transform:uppercase;color:#4fc3f7;background:rgba(79,195,247,0.08);border:1px solid rgba(79,195,247,0.2);border-radius:3px;padding:0.15em 0.5em;margin-right:0.7em;vertical-align:middle;font-family:'Segoe UI',sans-serif;">Kursinhalt</span>Interpretiert oder kompiliert?</span></div>


Python wird oft als "interpretierte Sprache" bezeichnet -- das stimmt, ist aber vereinfacht.

| Eigenschaft | C / C++ | Java | Python |
|-------------|---------|------|--------|
| Kompilierung | zu Maschinencode | zu JVM-Bytecode | zu PVM-Bytecode |
| Ausführung | direkt durch CPU | durch JVM | durch PVM |
| Typen geprüft | zur Compilezeit | zur Compilezeit | zur Laufzeit |
| Geschwindigkeit | sehr hoch | hoch | mittel |

Genauer gesagt ist Python **erst kompiliert (zu Bytecode), dann interpretiert (durch die PVM)**. Der Kompilierungsschritt ist nur deutlich schneller und weniger sichtbar als bei C.

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Dynamische Typisierung</span></div>

Ein wesentlicher Unterschied zu C oder Java: Python prüft Typen **zur Laufzeit**, nicht beim Kompilieren. Das hat Konsequenzen:

````python
# In Python kann dieselbe Variable verschiedene Typen annehmen
x = 42
print(type(x))  # int

x = 'jetzt ein String'
print(type(x))  # str

x = [1, 2, 3]
print(type(x))  # list

# Kein Fehler bisher -- Python prüft Typen erst, wenn eine Operation ausgefuehrt wird
# Dieser Code ist syntaktisch korrekt, faellt aber zur Laufzeit auf:
# ergebnis = 'Text' + 42  # TypeError: can only concatenate str (not "int") to str
````

<div style="background:#0b1929;border-left:4px solid #4fc3f7;border-radius:4px;padding:0.75em 1.2em;margin:2rem 0 1.2rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(1rem,2.2vw,1.25rem);font-weight:600;color:#e8eaf6;letter-spacing:-0.01em;"><span style="font-size:0.65rem;letter-spacing:0.1em;text-transform:uppercase;color:#4fc3f7;background:rgba(79,195,247,0.12);border:1px solid rgba(79,195,247,0.3);border-radius:3px;padding:0.2em 0.55em;margin-right:0.75em;vertical-align:middle;font-family:'Segoe UI',sans-serif;white-space:nowrap;">Kursinhalt</span>Alles ist ein Objekt</span></div>

<span style="font-size:clamp(1rem,2.2vw,1.35rem);font-weight:600;color:#d0d0e0;letter-spacing:-0.01em;"><span style="font-size:0.7rem;letter-spacing:0.12em;text-transform:uppercase;color:#4fc3f7;background:rgba(79,195,247,0.08);border:1px solid rgba(79,195,247,0.2);border-radius:3px;padding:0.15em 0.5em;margin-right:0.7em;vertical-align:middle;font-family:'Segoe UI',sans-serif;">Kursinhalt</span>Alles ist ein Objekt</span></div>


In Python ist buchstäblich alles ein Objekt -- Zahlen, Strings, Funktionen, Klassen, Module. Jedes Objekt hat:

- eine **Identität** (Speicheradresse, abfragbar mit `id()`)
- einen **Typ** (abfragbar mit `type()`)
- einen **Wert**

Das hat praktische Konsequenzen, die viele Anfänger überraschen:

````python
# Variablen sind Referenzen auf Objekte, keine Behaelter
a = [1, 2, 3]
b = a           # b zeigt auf dasselbe Objekt wie a

b.append(4)

print('a:', a)  # [1, 2, 3, 4] -- auch a ist veraendert!
print('b:', b)  # [1, 2, 3, 4]

print('Selbes Objekt?', a is b)   # True
print('id(a):', id(a))
print('id(b):', id(b))
````

````python
# Bei unveraenderlichen (immutable) Typen sieht es anders aus
x = 10
y = x

y = 20  # y zeigt jetzt auf ein neues Objekt

print('x:', x)  # 10 -- unveraendert
print('y:', y)  # 20
print('id(x):', id(x))
print('id(y):', id(y))  # andere Adresse als id(x)
````

Wenn du `import math` schreibst, führt Python intern mehrere Schritte aus:

1. **Prüfung in `sys.modules`:** Wurde das Modul schon importiert? Dann wird die gecachte Version zurückgegeben -- kein erneutes Laden.
2. **Suche nach der Datei:** Python durchsucht `sys.path` nach einer passenden `.py`-Datei oder einem Paketverzeichnis.
3. **Kompilierung:** Der gefundene Quellcode wird zu Bytecode kompiliert (und als `.pyc` gecacht).
4. **Ausführung:** Der Modulcode wird einmalig von oben nach unten ausgeführt. Funktionen und Klassen werden dabei definiert, aber noch nicht aufgerufen.
5. **Eintrag in `sys.modules`:** Das Modul-Objekt wird registriert.

Das bedeutet: **Importieren ist keine passive Operation.** Der Code auf Modulebene wird ausgeführt.





````python
import sys

# sys.modules ist ein Dictionary aller bisher importierten Module
print('math schon importiert?', 'math' in sys.modules)

import math

print('math jetzt importiert?', 'math' in sys.modules)
print('Typ des Objekts:       ', type(sys.modules['math']))
````

````python
# sys.path zeigt, wo Python nach Modulen sucht
print('Python sucht Module in:')
for pfad in sys.path:
    print(' ', pfad if pfad else '(aktuelles Verzeichnis)')
````

Python verwaltet Speicher automatisch. Du musst Objekte nicht explizit freigeben wie in C. Den Mechanismus dahinter nennt man **Garbage Collection**.

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Reference Counting</span></div>

CPython zählt für jedes Objekt, wie viele Referenzen darauf zeigen. Sinkt dieser Zähler auf null, wird das Objekt sofort freigegeben.

````python
import sys

a = [1, 2, 3]
print('Referenzen auf a:', sys.getrefcount(a))  # mindestens 2 (a + Argument in getrefcount)

b = a
print('Nach b = a:      ', sys.getrefcount(a))  # jetzt eine mehr

del b
print('Nach del b:      ', sys.getrefcount(a))  # wieder eine weniger
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Zyklische Referenzen</span></div>

Reference Counting allein reicht nicht, wenn sich zwei Objekte gegenseitig referenzieren. Dafür hat Python einen zusätzlichen **Cycle Garbage Collector**, der solche Zyklen erkennt und auflöst.

In der Praxis musst du dich darum selten kümmern -- es erklärt aber, warum Python gelegentlich kurze GC-Pausen hat.

| Schritt | Was passiert |
|--------|---------------|
| Lexer / Parser | Quelltext wird in Tokens zerlegt, Syntax geprüft |
| AST | Logische Baumstruktur des Codes |
| Compiler | AST wird zu Bytecode umgewandelt |
| `.pyc`-Datei | Bytecode wird in `__pycache__` gecacht |
| PVM | Bytecode wird Instruktion für Instruktion ausgeführt |
| Import | Suche in `sys.modules`, dann `sys.path`, dann Ausführung des Modulcodes |
| Speicher | Reference Counting + Cycle GC |

**Merksatz:** Python kompiliert deinen Code immer -- aber der Compiler ist so schnell und so gut versteckt, dass du ihn normalerweise nicht bemerkst.
