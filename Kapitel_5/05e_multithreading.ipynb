1. Konzept: Nebenläufigkeit
</span>
</div>

Nebenläufigkeit bedeutet, dass mehrere Aufgaben scheinbar gleichzeitig ablaufen. Python bietet dafuer zwei Ansaetze:

| | `threading` | `multiprocessing` |
|-|-------------|------------------|
| **Einheit** | Thread | Prozess |
| **Speicher** | Geteilt | Getrennt |
| **GIL** | Begrenzt durch GIL | Kein GIL |
| **Gut fuer** | I/O-Aufgaben (Netzwerk, Dateien) | CPU-intensive Berechnungen |
| **Overhead** | Gering | Hoeher |

**Der GIL (Global Interpreter Lock):** CPython erlaubt nur einem Thread gleichzeitig Python-Bytecode auszufuehren. Threads sind deshalb bei reinen Python-Berechnungen nicht schneller als ein einzelner Thread -- aber bei Wartezeiten (Netzwerk, Dateizugriff) laeuft die CPU weiter, waehrend gewartet wird.

2. Thread erstellen und starten
</span>
</div>

```python
import threading
import time

def lade_daten(quelle, dauer):
    """Simuliert das Laden von Daten aus einer externen Quelle."""
    print(f'Starte Laden von {quelle}...')
    time.sleep(dauer)   # simuliert Netzwerklatenz
    print(f'Fertig: {quelle} ({dauer}s)')

# Sequenziell: 2 + 3 = 5 Sekunden
print('--- Sequenziell ---')
start = time.time()
lade_daten('Datenbank', 1)
lade_daten('API', 1)
print(f'Gesamt: {time.time() - start:.1f}s')

print()

# Mit Threads: ca. max(2, 3) = 3 Sekunden
print('--- Mit Threads ---')
start = time.time()

t1 = threading.Thread(target=lade_daten, args=('Datenbank', 1))
t2 = threading.Thread(target=lade_daten, args=('API', 1))

t1.start()
t2.start()

t1.join()   # Warten bis t1 fertig ist
t2.join()   # Warten bis t2 fertig ist

print(f'Gesamt: {time.time() - start:.1f}s')
```

```python
import threading, time

def aufgabe(name, dauer):
    time.sleep(dauer)
    print(f'{name} fertig')

t = threading.Thread(target=aufgabe, args=('Hintergrund', 0.5))
t.start()

# Ohne join() laeuft der Hauptthread weiter
print('Hauptthread laeuft weiter...')

t.join()  # erst hier wird gewartet
print('Nach join() -- Thread ist fertig')
```

3. Thread als Klasse
</span>
</div>

```python
import threading

class BerechnungsThread(threading.Thread):
    """Thread der eine Berechnung durchfuehrt und das Ergebnis speichert."""

    def __init__(self, name, werte):
        super().__init__()
        self.name    = name
        self.werte   = werte
        self.summe   = None
        self.maximum = None

    def run(self):   # wird von start() aufgerufen -- hier nie direkt aufrufen!
        self.summe   = sum(self.werte)
        self.maximum = max(self.werte)
        print(f'{self.name}: Summe={self.summe}, Max={self.maximum}')


# Drei Threads parallel
t1 = BerechnungsThread('Block-1', range(1,    1001))
t2 = BerechnungsThread('Block-2', range(1001, 2001))
t3 = BerechnungsThread('Block-3', range(2001, 3001))

for t in [t1, t2, t3]:
    t.start()

for t in [t1, t2, t3]:
    t.join()

print(f'Gesamtsumme: {t1.summe + t2.summe + t3.summe}')
```

4. Race Conditions und Locks
</span>
</div>

Eine **Race Condition** entsteht, wenn zwei Threads gleichzeitig eine geteilte Variable lesen und schreiben. Das Ergebnis haengt dann von der zufaelligen Ausfuehrungsreihenfolge ab -- nicht deterministisch.

**Warum passiert das?** `zaehler += 1` ist keine atomare Operation -- intern sind es drei Schritte: lesen, addieren, schreiben. Zwischen diesen Schritten kann ein anderer Thread unterbrochen werden.

```python
import threading

# --- OHNE Lock: Race Condition ---
zaehler = 0

def erhoehe(n):
    global zaehler
    for _ in range(n):
        zaehler += 1  # nicht atomar!

threads = [threading.Thread(target=erhoehe, args=(10000,)) for _ in range(5)]
for t in threads: t.start()
for t in threads: t.join()

print(f'Ohne Lock: {zaehler} (erwartet 50000)')
# Ergebnis ist oft kleiner als 50000 -- Datenverlust durch Race Condition
```

```python
import threading

# --- MIT Lock: sicher ---
zaehler = 0
lock = threading.Lock()

def erhoehe_sicher(n):
    global zaehler
    for _ in range(n):
        with lock:       # acquire() -- nur ein Thread gleichzeitig
            zaehler += 1
        # release() automatisch am Ende des with-Blocks

threads = [threading.Thread(target=erhoehe_sicher, args=(10000,)) for _ in range(5)]
for t in threads: t.start()
for t in threads: t.join()

print(f'Mit Lock:   {zaehler} (erwartet 50000)')
```

5. Thread-Kommunikation mit Event
</span>
</div>

Ein `threading.Event` erlaubt es Threads, auf ein Signal zu warten. Ein Thread signalisiert mit `set()`, ein anderer wartet mit `wait()`.

```python
import threading
import time

daten_bereit = threading.Event()

def produzent():
    print('Produzent: Daten werden vorbereitet...')
    time.sleep(1)
    print('Produzent: Daten bereit!')
    daten_bereit.set()   # Signal geben

def konsument():
    print('Konsument: Warte auf Daten...')
    daten_bereit.wait()  # Blockiert bis Event gesetzt wird
    print('Konsument: Daten empfangen und verarbeite...')

t1 = threading.Thread(target=produzent)
t2 = threading.Thread(target=konsument)

t2.start()   # Konsument startet und wartet
t1.start()   # Produzent startet und bereitet Daten vor

t1.join()
t2.join()
```

6. Daemon-Threads
</span>
</div>

Ein **Daemon-Thread** laeuft im Hintergrund und wird automatisch beendet, wenn der Hauptthread endet -- unabhaengig davon ob er fertig ist. Normal-Threads halten das Programm am Laufen bis sie abgeschlossen sind.

```python
import threading
import time

def heartbeat():
    """Sendet regelmaessig ein Lebenszeichen -- laeuft endlos."""
    i = 0
    while True:
        print(f'Heartbeat {i}')
        i += 1
        time.sleep(0.3)

# Daemon-Thread: endet wenn Hauptthread endet
t = threading.Thread(target=heartbeat, daemon=True)
t.start()

# Hauptthread laeuft noch kurz
time.sleep(1)
print('Hauptthread endet -- Daemon-Thread wird automatisch gestoppt')
# kein t.join() noetig
```

**Zusammenfassung**

| Konzept | Kernaussage |
|---------|-------------|
| Thread | Ausfuehrungsstrang innerhalb eines Prozesses -- teilt Speicher |
| GIL | CPython: ein Thread gleichzeitig -- Threads gut fuer I/O, nicht fuer CPU |
| `Thread(target, args)` | Thread erstellen |
| `.start()` | Thread starten -- ruft intern `run()` auf |
| `.join()` | Hauptthread wartet bis dieser Thread fertig ist |
| Race Condition | Nicht-deterministisch bei geteiltem Zugriff ohne Schutz |
| `Lock` | Gegenseitiger Ausschluss -- mit `with lock:` verwenden |
| `Event` | Kommunikation zwischen Threads -- `set()` und `wait()` |
| Daemon-Thread | Beendet sich automatisch mit dem Hauptthread |
