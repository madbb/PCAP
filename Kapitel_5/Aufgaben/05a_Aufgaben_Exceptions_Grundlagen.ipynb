Im folgenden Code stecken vier Fehler. Finde und korrigiere alle vier.

```python
def berechne(werte, index):
    try:
        ergebnis = werte[index] / werte[0]
    except IndexError, ZeroDivisionError:   # Fehler 1
        return None
    except Exception:                       # Fehler 2: steht vor spezifischeren Bloecken
        return -1
    except ValueError:
        return -2
    else:
        print('Kein Fehler')
        return ergebnis
    finally:
        return 0                            # Fehler 3: ueberschreibt alle Rueckgaben

werte = [4, 8, 'x']
raise berechne(werte, 0)                    # Fehler 4
```
