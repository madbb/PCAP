**5.2**
```python
class MeinFehler(Exception):
    def __init__(self, msg):
        super().__init__(msg + '!')
        self.args = (msg,)

try:
    raise MeinFehler('hallo')
except MeinFehler as e:
    print(e)
    print(e.args)
```

**5.3**
```python
class MeinFehler(Exception):
    def __init__(self, msg):
        Exception.__init__(self, msg, msg)

try:
    raise MeinFehler('ab')
except MeinFehler as e:
    print(e)
    print(e.args)
    print(len(e.args))
```

**5.4**
```python
class MeinFehler(Exception):
    def __init__(self, wert):
        self.wert = wert
        # kein super().__init__() Aufruf!

try:
    raise MeinFehler('test')
except MeinFehler as e:
    print(e)
    print(e.args)
    print(e.wert)
```

**9.3**
```python
class MeinFehler(ValueError):
    def __init__(self, msg):
        super().__init__(msg.upper())
        self.original = msg

try:
    raise MeinFehler('hallo')
except ValueError as e:
    print(str(e))
    print(e.original)
    print(isinstance(e, ValueError))
    print(isinstance(e, MeinFehler))
```
