1. Das Problem: Tests die niemand ausfuehrt
</span>
</div>

Wir haben jetzt 30+ Tests. Sie laufen lokal auf unserem Rechner. Aber was ist mit dem Kollegen der eine Aenderung pusht und vergisst die Tests auszufuehren? Oder der Entwickler der auf einem anderen Betriebssystem arbeitet und dessen Umgebung sich leicht unterscheidet?

Tests die niemand ausfuehrt nuetzen nichts. Die Loesung: Tests laufen **automatisch** -- bei jedem Push in das Repository, auf einem neutralen Server, unabhaengig davon wer den Code eingecheckt hat.

Das nennt sich **Continuous Integration** (CI) -- und GitHub bietet dafuer einen integrierten Dienst: **GitHub Actions**.

2. Was ist CI/CD?
</span>
</div>

**CI -- Continuous Integration:** Jede Code-Aenderung wird sofort automatisch gebaut und getestet. Fehler werden frueh entdeckt -- bevor sie sich anhaeufen.

**CD -- Continuous Delivery/Deployment:** Nach erfolgreichem Test wird die Anwendung automatisch in eine Test- oder Produktionsumgebung eingespielt.

In diesem Notebook konzentrieren wir uns auf CI -- den Test-Teil.

Der Ablauf mit GitHub Actions:

```
Entwickler pusht Code
        |
        v
GitHub Actions startet automatisch
        |
        v
Frische Umgebung wird aufgebaut (Ubuntu, Python 3.11)
        |
        v
Abhaengigkeiten installiert (pip install pytest)
        |
        v
pytest wird ausgefuehrt
        |
     /     \
Gruен       Rot
Alles OK    Entwickler bekommt Email + Badge wird rot
```

GitHub Actions werden in **YAML**-Dateien konfiguriert. YAML ist ein einfaches Format fuer strukturierte Daten -- aehnlich wie JSON, aber lesbarer. Es funktioniert mit Einrueckungen statt Klammern.

Die wichtigsten Regeln:
- Einrueckung mit **Leerzeichen** (nie Tabs)
- Schluessel-Wert-Paare: `schluessel: wert`
- Listen: Zeilen beginnen mit `-`
- Strings mit Sonderzeichen in Anfuehrungszeichen

````python
# YAML vs. Python -- Vergleich
yaml_beispiel = """
# YAML
name: Aldric
klasse: Krieger
hp: 100
faehigkeiten:
  - Schwertangriff
  - Schildblock
  - Kampfschrei
"""

python_aequivalent = {
    'name': 'Aldric',
    'klasse': 'Krieger',
    'hp': 100,
    'faehigkeiten': ['Schwertangriff', 'Schildblock', 'Kampfschrei']
}

print('YAML und Python beschreiben dieselbe Struktur:')
print(python_aequivalent)
````

4. Die erste Pipeline -- Schritt fuer Schritt
</span>
</div>

Eine GitHub Actions Workflow-Datei lebt immer in diesem Pfad im Repository:

```
.github/
    workflows/
        tests.yml     <-- hier liegt unsere Pipeline
```

Schauen wir uns die minimale Pipeline Zeile fuer Zeile an:

````python
pipeline_erklaert = """
# Name des Workflows -- erscheint in GitHub unter "Actions"
name: Tests

# Wann soll der Workflow ausgefuehrt werden?
on:
  push:                    # Bei jedem Push...
    branches: [ main ]     # ...auf den main-Branch
  pull_request:            # Und bei jedem Pull Request
    branches: [ main ]

# Was soll ausgefuehrt werden?
jobs:
  test:                              # Name des Jobs
    runs-on: ubuntu-latest           # Betriebssystem: aktuelles Ubuntu

    steps:                           # Einzelne Schritte in Reihenfolge

      - name: Code auschecken        # Schritt 1: Repository herunterladen
        uses: actions/checkout@v4    # vorgefertigte Action von GitHub

      - name: Python einrichten      # Schritt 2: Python installieren
        uses: actions/setup-python@v5
        with:
          python-version: '3.11'

      - name: Abhaengigkeiten        # Schritt 3: Pakete installieren
        run: pip install pytest

      - name: Tests ausfuehren       # Schritt 4: pytest starten
        run: pytest -v
"""

print(pipeline_erklaert)
````

````python
import os

# Verzeichnisstruktur anlegen
os.makedirs('.github/workflows', exist_ok=True)

# Die minimale Pipeline
pipeline_minimal = """name: Tests

on:
  push:
    branches: [ main ]
  pull_request:
    branches: [ main ]

jobs:
  test:
    runs-on: ubuntu-latest

    steps:
      - name: Code auschecken
        uses: actions/checkout@v4

      - name: Python einrichten
        uses: actions/setup-python@v5
        with:
          python-version: '3.11'

      - name: Abhaengigkeiten installieren
        run: pip install pytest

      - name: Tests ausfuehren
        run: pytest -v
"""

with open('.github/workflows/tests.yml', 'w') as f:
    f.write(pipeline_minimal)

print('Datei erstellt: .github/workflows/tests.yml')
print()
print('Naechste Schritte:')
print('  git add .github/workflows/tests.yml')
print('  git commit -m "CI: Tests mit GitHub Actions"')
print('  git push')
print()
print('Danach: GitHub -> Repository -> Actions -> Tests')
````

5. Erweiterte Pipeline -- mit Coverage und mehreren Python-Versionen
</span>
</div>

Die minimale Pipeline ist gut fuer den Einstieg. In der Praxis will man mehr: Coverage messen, auf mehreren Python-Versionen testen, einen Coverage-Report als Artefakt speichern.

GitHub Actions unterstuetzt **Matrix-Builds** -- denselben Job gleichzeitig mit verschiedenen Konfigurationen ausfuehren:

````python
pipeline_erweitert = """name: Tests

on:
  push:
    branches: [ main ]
  pull_request:
    branches: [ main ]

jobs:
  test:
    runs-on: ubuntu-latest

    # Matrix: testet auf mehreren Python-Versionen gleichzeitig
    strategy:
      matrix:
        python-version: ['3.10', '3.11', '3.12']

    steps:
      - name: Code auschecken
        uses: actions/checkout@v4

      - name: Python ${{ matrix.python-version }} einrichten
        uses: actions/setup-python@v5
        with:
          python-version: ${{ matrix.python-version }}

      - name: Abhaengigkeiten installieren
        run: pip install pytest pytest-cov

      - name: Tests mit Coverage
        run: pytest -v --cov=combat --cov=held --cov-report=xml

      - name: Coverage-Report speichern
        uses: actions/upload-artifact@v4
        with:
          name: coverage-py${{ matrix.python-version }}
          path: coverage.xml
"""

with open('.github/workflows/tests_erweitert.yml', 'w') as f:
    f.write(pipeline_erweitert)

print('Erweiterte Pipeline erstellt: .github/workflows/tests_erweitert.yml')
````

7. Was passiert wenn ein Test fehlschlaegt?
</span>
</div>

Wenn ein Test in der Pipeline fehlschlaegt:

1. GitHub markiert den Commit mit einem roten X
2. Der Badge im README wird rot
3. Wer den Pull Request gestellt hat, bekommt eine E-Mail
4. Der Merge in `main` kann blockiert werden -- bis alle Tests bestehen

Das ist der entscheidende Unterschied zu lokalem Testen: Kein Code der Tests bricht kommt in den Hauptbranch -- unabhaengig davon ob jemand vergessen hat zu testen.

Schauen wir uns an wie man einen fehlerhaften Commit simuliert und was passiert:

````python
# Simulation: jemand aendert combat.py und bricht einen Test
combat_kaputt = '''
def berechne_schaden(angriff, verteidigung):
    if angriff < 0 or verteidigung < 0:
        raise ValueError("Angriff und Verteidigung muessen >= 0 sein")
    schaden = angriff - verteidigung
    return max(0, schaden)   # Bug wieder eingebaut: max(0, ...) statt max(1, ...)

def trefferchance(angreifer_geschwindigkeit, ziel_geschwindigkeit):
    if angreifer_geschwindigkeit <= 0 or ziel_geschwindigkeit <= 0:
        raise ValueError("Geschwindigkeit muss > 0 sein")
    chance = 50 + (angreifer_geschwindigkeit - ziel_geschwindigkeit) * 5
    return max(10, min(95, chance))

def kritischer_treffer(schaden, krit_faktor=1.5):
    if krit_faktor < 1.0:
        raise ValueError("krit_faktor muss >= 1.0 sein")
    if schaden < 0:
        raise ValueError("Schaden muss >= 0 sein")
    return int(schaden * krit_faktor)

def heilen(aktuelle_hp, max_hp, heilmenge):
    if max_hp <= 0:
        raise ValueError("max_hp muss > 0 sein")
    if heilmenge <= 0:
        raise ValueError("heilmenge muss > 0 sein")
    if aktuelle_hp <= 0:
        return aktuelle_hp
    return min(max_hp, aktuelle_hp + heilmenge)

def ist_besiegt(hp):
    return hp <= 0

def statuseffekt_anwenden(schaden, effekt):
    if schaden < 0:
        raise ValueError("Schaden muss >= 0 sein")
    faktoren = {
        "normal": 1.0, "vergiftet": 1.5,
        "geschwaeht": 0.75, "verstaerkt": 2.0,
    }
    if effekt not in faktoren:
        raise ValueError(f"Unbekannter Statuseffekt: {effekt!r}")
    return int(schaden * faktoren[effekt])
'''

with open('../combat/combat.py', 'w') as f:
    f.write(combat_kaputt)

print('combat.py mit Bug gespeichert -- was sagt die Pipeline?')
````

````python
import subprocess

# Lokal testen -- genau das wuerde die Pipeline auf GitHub machen
result = subprocess.run(
    ['python', '-m', 'pytest', 'test_combat.py', '-v', '--tb=short'],
    capture_output=True, text=True
)
print(result.stdout[-3000:])  # letzte 3000 Zeichen -- relevante Fehler
````

Genau diese Ausgabe wuerde die GitHub Actions Pipeline produzieren -- und den Commit als fehlerhaft markieren. Niemand muss manuell pruefen. Die Pipeline macht es automatisch.

Jetzt stellen wir `combat.py` wieder her:

````python
combat_korrekt = '''
"""combat.py -- Kampfsystem Aetherion, Version 2 (alle Bugs behoben)"""


def berechne_schaden(angriff, verteidigung):
    if angriff < 0 or verteidigung < 0:
        raise ValueError("Angriff und Verteidigung muessen >= 0 sein")
    return max(1, angriff - verteidigung)


def trefferchance(angreifer_geschwindigkeit, ziel_geschwindigkeit):
    if angreifer_geschwindigkeit <= 0 or ziel_geschwindigkeit <= 0:
        raise ValueError("Geschwindigkeit muss > 0 sein")
    chance = 50 + (angreifer_geschwindigkeit - ziel_geschwindigkeit) * 5
    return max(10, min(95, chance))


def kritischer_treffer(schaden, krit_faktor=1.5):
    if krit_faktor < 1.0:
        raise ValueError("krit_faktor muss >= 1.0 sein")
    if schaden < 0:
        raise ValueError("Schaden muss >= 0 sein")
    return int(schaden * krit_faktor)


def heilen(aktuelle_hp, max_hp, heilmenge):
    if max_hp <= 0:
        raise ValueError("max_hp muss > 0 sein")
    if heilmenge <= 0:
        raise ValueError("heilmenge muss > 0 sein")
    if aktuelle_hp <= 0:
        return aktuelle_hp
    return min(max_hp, aktuelle_hp + heilmenge)


def ist_besiegt(hp):
    return hp <= 0


def statuseffekt_anwenden(schaden, effekt):
    if schaden < 0:
        raise ValueError("Schaden muss >= 0 sein")
    faktoren = {
        "normal": 1.0, "vergiftet": 1.5,
        "geschwaeht": 0.75, "verstaerkt": 2.0,
    }
    if effekt not in faktoren:
        raise ValueError(f"Unbekannter Statuseffekt: {effekt!r}")
    return int(schaden * faktoren[effekt])
'''

with open('../combat/combat.py', 'w') as f:
    f.write(combat_korrekt)

# Alle Tests laufen -- alles gruen
result = subprocess.run(
    ['python', '-m', 'pytest', 'test_combat.py', '-v'],
    capture_output=True, text=True
)
# Nur die Zusammenfassung zeigen
for zeile in result.stdout.split('\n'):
    if 'passed' in zeile or 'failed' in zeile or 'error' in zeile:
        print(zeile)
````
