1. Warum Listbox und Treeview?
</span>
</div>

In 11a und 11b haben wir Fenster, Widgets, Layout und Ereignisse kennengelernt. Alle Beispiele zeigten dabei einzelne Werte -- ein Label, ein Eingabefeld, ein Button. Das reicht für einfache Formulare, aber nicht für Anwendungen die mit Daten arbeiten.

Das MP3-Projekt zum Beispiel braucht eine Ansicht die viele Titel gleichzeitig anzeigt -- mit Titelname, Interpret, Album und Dauer nebeneinander. Dafür gibt es zwei Widgets:

**Listbox** -- zeigt eine Liste von Einträgen, einer pro Zeile. Einfach, aber ohne Spalten. Gut für einfache Auswahllisten: eine Liste von Dateinamen, eine Liste von Optionen.

**Treeview** -- zeigt Daten in Spalten, ähnlich einer Tabelle. Jede Zeile kann Werte in mehreren Spalten haben. Der Name "Tree" kommt daher, dass es auch Baumstrukturen darstellen kann (Zeilen mit Unterzeilen) -- aber für unsere Zwecke benutzen wir es als einfache Tabelle.

Wir lernen zuerst die Listbox, weil sie einfacher ist und die Grundprinzipien zeigt. Dann kommt der Treeview der für das Projekt gebraucht wird.

2. Listbox -- eine einfache Auswahlliste
</span>
</div>

Eine `Listbox` zeigt eine scrollbare Liste von Texteinträgen. Der Nutzer kann einen oder mehrere Einträge auswählen -- ähnlich wie in einem Dateiauswahldialog.

**Erstellen:** `tk.Listbox(parent, height=10)` -- `height` ist die Anzahl sichtbarer Zeilen, nicht Pixel. Ohne `height` zeigt die Listbox fünf Zeilen.

**Einträge hinzufügen:** `.insert(position, wert)` -- fügt einen Eintrag an der angegebenen Position ein. `tk.END` fügt am Ende an. `0` fügt ganz vorne ein.

**Einträge entfernen:** `.delete(start, end)` -- löscht Einträge von `start` bis `end`. `.delete(0, tk.END)` leert die gesamte Liste.

**Ausgewählten Eintrag lesen:** `.curselection()` gibt ein Tupel mit den Indizes der ausgewählten Einträge zurück -- zum Beispiel `(2,)` wenn der dritte Eintrag markiert ist. Das Tupel ist leer wenn nichts ausgewählt ist. Mit `.get(index)` holt man den Text an einem bestimmten Index.

````python
# fenster_15_listbox.py
import tkinter as tk

def hinzufuegen():
    text = eingabe.get().strip()
    if text:                          # nicht hinzufuegen wenn leer
        liste.insert(tk.END, text)    # ans Ende der Liste anhaengen
        eingabe.delete(0, tk.END)     # Eingabefeld leeren

def loeschen():
    auswahl = liste.curselection()    # Tupel mit Indizes der markierten Eintraege
    if not auswahl:
        return                        # nichts tun wenn nichts ausgewaehlt
    # Rueckwaerts loeschen damit Indizes nicht verrutschen
    for index in reversed(auswahl):
        liste.delete(index)

def ausgeben():
    auswahl = liste.curselection()
    if auswahl:
        index = auswahl[0]            # ersten ausgewaehlten Index nehmen
        text = liste.get(index)       # Text an diesem Index lesen
        info_label.config(text=f'Ausgewaehlt: {text}')
    else:
        info_label.config(text='Nichts ausgewaehlt')

root = tk.Tk()
root.title('Listbox')
root.geometry('320x300')

# Eingabebereich oben
eingabe_frame = tk.Frame(root)
eingabe_frame.pack(fill='x', padx=10, pady=8)

eingabe = tk.Entry(eingabe_frame, width=20)
eingabe.pack(side='left')

tk.Button(eingabe_frame, text='Hinzufuegen', command=hinzufuegen).pack(side='left', padx=5)

# Listbox in der Mitte
# height=8: acht Zeilen sichtbar
liste = tk.Listbox(root, height=8, width=35)
liste.pack(padx=10, pady=5)

# Vorab ein paar Eintraege befuellen
for eintrag in ['Apfel', 'Banane', 'Kirsche', 'Dattel', 'Erdbeere']:
    liste.insert(tk.END, eintrag)

# Buttons unten
button_frame = tk.Frame(root)
button_frame.pack(pady=5)

tk.Button(button_frame, text='Ausgewaehlt anzeigen', command=ausgeben).pack(side='left', padx=4)
tk.Button(button_frame, text='Loeschen',             command=loeschen).pack(side='left', padx=4)

# Infozeile unten
info_label = tk.Label(root, text='')
info_label.pack(pady=5)

root.mainloop()
````

3. Scrollbar -- wenn die Liste länger als sichtbar ist
</span>
</div>

Eine Listbox hat keinen eingebauten Scrollbalken. Wenn die Liste mehr Einträge hat als Zeilen sichtbar sind, kann man zwar mit der Maus scrollen -- aber ohne sichtbare Scrollbar ist das für den Nutzer nicht erkennbar.

Eine `Scrollbar` ist ein eigenständiges Widget das man mit einer Listbox oder einem anderen scrollbaren Widget **verbindet**. Die Verbindung funktioniert in beide Richtungen:

- Die Scrollbar steuert die Listbox: `scrollbar.config(command=liste.yview)` -- wenn der Nutzer die Scrollbar bewegt, ruft sie `liste.yview()` auf
- Die Listbox informiert die Scrollbar: `liste.config(yscrollcommand=scrollbar.set)` -- wenn sich die Listbox scrollt, aktualisiert sie die Scrollbar-Position

Das klingt kompliziert, ist aber immer dasselbe Muster -- man kopiert es einmal und versteht es dann.

**Platzierung:** Listbox und Scrollbar müssen nebeneinander platziert werden. Dafür eignet sich ein Frame mit `pack(side='left')` und `pack(side='right')`.

````python
# fenster_16_scrollbar.py
import tkinter as tk

root = tk.Tk()
root.title('Listbox mit Scrollbar')
root.geometry('280x250')

# Frame der Listbox und Scrollbar zusammenhaelt
liste_frame = tk.Frame(root)
liste_frame.pack(padx=10, pady=10, fill='both', expand=True)

# Scrollbar erstellen -- noch nicht verbunden
scrollbar = tk.Scrollbar(liste_frame)
scrollbar.pack(side='right', fill='y')     # rechts, volle Hoehe

# Listbox erstellen und mit Scrollbar verbinden
# yscrollcommand: wenn Listbox scrollt, aktualisiert sie die Scrollbar
liste = tk.Listbox(liste_frame, yscrollcommand=scrollbar.set)
liste.pack(side='left', fill='both', expand=True)

# Scrollbar mit Listbox verbinden
# command: wenn Scrollbar bewegt wird, scrollt sie die Listbox
scrollbar.config(command=liste.yview)

# Viele Eintraege hinzufuegen damit die Scrollbar sichtbar wird
for i in range(1, 31):
    liste.insert(tk.END, f'Eintrag {i:02d}')

root.mainloop()
````

4. Treeview -- Daten in Spalten anzeigen
</span>
</div>

Der `ttk.Treeview` ist das leistungsfähigste Listenelement in tkinter. Er kommt aus dem `ttk`-Submodul -- "ttk" steht für "themed tk" und bezeichnet modernere Widgets die besser aussehen als die klassischen tk-Widgets.

**Der Import:** Der Treeview kommt aus `tkinter.ttk`, nicht aus `tkinter` direkt:

```python
import tkinter as tk
from tkinter import ttk

baum = ttk.Treeview(root)
```

Der Treeview hat zwei Konzepte die man verstehen muss bevor man ihn benutzt:

**Spalten** -- man definiert welche Spalten es gibt, mit welchen Namen. Die erste Spalte (`#0`) ist immer vorhanden und heißt intern die "Tree-Spalte" -- sie zeigt das Baumstruktur-Symbol. Wenn man den Treeview als reine Tabelle nutzt, blendet man diese Spalte aus.

**Zeilen** -- man fügt Zeilen mit `.insert()` ein. Anders als bei der Listbox übergibt man hier einen Eltern-Knoten (für reine Tabellen immer `''`), eine Position (`tk.END`) und die Werte als Tupel.

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;">
<span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Treeview Schritt für Schritt aufbauen</span>
</div>

Ein Treeview entsteht in vier Schritten:

**Schritt 1: Spalten definieren** -- beim Erstellen gibt man eine Liste von Spalten-IDs an. Das sind interne Namen, keine Überschriften:

```python
baum = ttk.Treeview(root, columns=('titel', 'interpret', 'dauer'), show='headings')
```

`show='headings'` blendet die erste Baumspalte (`#0`) aus -- wir wollen eine reine Tabelle ohne Baumstruktur.

**Schritt 2: Überschriften setzen** -- für jede Spalte wird eine sichtbare Überschrift definiert:

```python
baum.heading('titel',    text='Titel')
baum.heading('interpret', text='Interpret')
baum.heading('dauer',    text='Dauer')
```

**Schritt 3: Spaltenbreiten setzen** -- optional aber empfohlen:

```python
baum.column('titel',    width=200)
baum.column('interpret', width=150)
baum.column('dauer',    width=60)
```

**Schritt 4: Zeilen einfügen** -- `.insert()` mit dem Elternknoten `''` für Toplevel-Zeilen:

```python
baum.insert('', tk.END, values=('Bohemian Rhapsody', 'Queen', '5:55'))
```

````python
# fenster_17_treeview.py
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title('Treeview als Tabelle')
root.geometry('550x300')

# Schritt 1: Treeview erstellen
# columns: interne Namen der Spalten -- keine Ueberschriften!
# show='headings': Baumspalte #0 ausblenden, nur echte Spalten zeigen
baum = ttk.Treeview(root, columns=('titel', 'interpret', 'album', 'dauer'), show='headings')

# Schritt 2: Ueberschriften setzen
baum.heading('titel',     text='Titel')
baum.heading('interpret', text='Interpret')
baum.heading('album',     text='Album')
baum.heading('dauer',     text='Dauer')

# Schritt 3: Spaltenbreiten und Ausrichtung
# anchor: Textausrichtung innerhalb der Spalte
# 'w' = west = linksbuendig
# 'center' = zentriert
baum.column('titel',     width=200, anchor='w')
baum.column('interpret', width=150, anchor='w')
baum.column('album',     width=130, anchor='w')
baum.column('dauer',     width=60,  anchor='center')

baum.pack(fill='both', expand=True, padx=10, pady=10)

# Schritt 4: Zeilen einfuegen
# '' = kein Elternelement (Toplevel-Zeile)
# values: Tupel mit den Spaltenwerten in derselben Reihenfolge wie columns
songs = [
    ('Bohemian Rhapsody', 'Queen',        'A Night at the Opera', '5:55'),
    ('Hotel California',  'Eagles',       'Hotel California',     '6:30'),
    ('Stairway to Heaven','Led Zeppelin', 'Led Zeppelin IV',      '8:02'),
    ('Smells Like Teen Spirit', 'Nirvana','Nevermind',            '5:01'),
    ('Purple Haze',       'Jimi Hendrix', 'Are You Experienced',  '2:52'),
]

for song in songs:
    baum.insert('', tk.END, values=song)

root.mainloop()
````

5. Treeview -- Auswahl, Einfügen, Löschen
</span>
</div>

Ein statischer Treeview ist schön -- aber für das MP3-Projekt muss man Einträge hinzufügen, auswählen und entfernen können.

**Ausgewählte Zeile lesen:** `.selection()` gibt ein Tupel von Item-IDs zurück. Eine Item-ID ist ein interner Bezeichner den tkinter beim `.insert()` automatisch vergibt -- man kann aber auch selbst einen vergeben. Mit `.item(id, 'values')` holt man die Spaltenwerte einer Zeile.

```python
auswahl = baum.selection()       # Tupel von Item-IDs
if auswahl:
    item_id = auswahl[0]         # erste ausgewaehlte Zeile
    werte = baum.item(item_id, 'values')  # Werte als Tupel
    titel = werte[0]             # erste Spalte
```

**Zeile löschen:** `.delete(item_id)` löscht eine Zeile anhand ihrer ID. `.delete(*baum.get_children())` löscht alle Zeilen auf einmal.

**Auf Klick reagieren:** Das Ereignis `'<<TreeviewSelect>>'` wird ausgelöst wenn der Nutzer eine Zeile anklickt. Man bindet es mit `.bind()`:

```python
baum.bind('<<TreeviewSelect>>', bei_auswahl)
```

Die Callback-Funktion bekommt ein Event-Objekt -- aber man braucht es oft nicht, weil man die Auswahl einfach mit `.selection()` direkt aus dem Treeview liest.

````python
# fenster_18_treeview_interaktion.py
import tkinter as tk
from tkinter import ttk

def bei_auswahl(event):
    # event wird nicht benoetigt -- Auswahl direkt aus Treeview lesen
    auswahl = baum.selection()
    if not auswahl:
        return
    item_id = auswahl[0]
    werte = baum.item(item_id, 'values')
    # werte ist ein Tupel: (titel, interpret, dauer)
    info_label.config(text=f'{werte[0]} von {werte[1]}')

def zeile_hinzufuegen():
    titel    = titel_var.get().strip()
    interpret = interpret_var.get().strip()
    if not titel or not interpret:
        return
    baum.insert('', tk.END, values=(titel, interpret, '?:??'))
    titel_var.set('')
    interpret_var.set('')

def zeile_loeschen():
    auswahl = baum.selection()
    for item_id in auswahl:
        baum.delete(item_id)           # jede ausgewaehlte Zeile loeschen
    info_label.config(text='')

root = tk.Tk()
root.title('Treeview Interaktion')
root.geometry('480x380')

# --- Eingabebereich oben ---
eingabe_frame = tk.Frame(root)
eingabe_frame.pack(fill='x', padx=10, pady=8)

titel_var    = tk.StringVar()
interpret_var = tk.StringVar()

tk.Label(eingabe_frame, text='Titel:').grid(    row=0, column=0, sticky='w')
tk.Entry(eingabe_frame, textvariable=titel_var, width=20).grid(row=0, column=1, padx=5)

tk.Label(eingabe_frame, text='Interpret:').grid(row=0, column=2, sticky='w', padx=(10,0))
tk.Entry(eingabe_frame, textvariable=interpret_var, width=20).grid(row=0, column=3, padx=5)

tk.Button(eingabe_frame, text='Hinzufuegen', command=zeile_hinzufuegen).grid(row=0, column=4, padx=5)

# --- Treeview mit Scrollbar ---
baum_frame = tk.Frame(root)
baum_frame.pack(fill='both', expand=True, padx=10, pady=5)

scrollbar = ttk.Scrollbar(baum_frame, orient='vertical')
scrollbar.pack(side='right', fill='y')

baum = ttk.Treeview(
    baum_frame,
    columns=('titel', 'interpret', 'dauer'),
    show='headings',
    yscrollcommand=scrollbar.set    # Treeview informiert Scrollbar
)
scrollbar.config(command=baum.yview)  # Scrollbar steuert Treeview

baum.heading('titel',     text='Titel')
baum.heading('interpret', text='Interpret')
baum.heading('dauer',     text='Dauer')

baum.column('titel',     width=200)
baum.column('interpret', width=160)
baum.column('dauer',     width=70, anchor='center')

baum.pack(side='left', fill='both', expand=True)

# Auf Auswahl reagieren
baum.bind('<<TreeviewSelect>>', bei_auswahl)

# Startdaten
for song in [
    ('Bohemian Rhapsody', 'Queen',   '5:55'),
    ('Hotel California',  'Eagles',  '6:30'),
    ('Smells Like Teen Spirit', 'Nirvana', '5:01'),
]:
    baum.insert('', tk.END, values=song)

# --- Buttons und Info unten ---
unten_frame = tk.Frame(root)
unten_frame.pack(fill='x', padx=10, pady=5)

tk.Button(unten_frame, text='Ausgewaehlte Zeile loeschen', command=zeile_loeschen).pack(side='left')

info_label = tk.Label(unten_frame, text='')
info_label.pack(side='left', padx=15)

root.mainloop()
````

6. Treeview -- Sortierung durch Klick auf Spaltenüberschrift
</span>
</div>

Professionelle Tabellen erlauben dem Nutzer durch Klick auf eine Überschrift nach dieser Spalte zu sortieren. Der Treeview bietet das nicht eingebaut -- man implementiert es selbst.

Das Prinzip:

1. Alle Zeilen aus dem Treeview auslesen
2. Die Daten nach der gewünschten Spalte sortieren
3. Alle Zeilen löschen
4. Die sortierten Daten neu einfügen

Um auf Klicks auf die Überschrift zu reagieren, übergibt man der `heading()`-Methode ein `command`-Argument:

```python
baum.heading('titel', text='Titel', command=lambda: sortiere('titel'))
```

Hier sieht man warum `lambda` praktisch ist -- man braucht eine Funktion ohne Argumente für `command`, will aber trotzdem übergeben nach welcher Spalte sortiert wird.

````python
# fenster_19_sortieren.py
import tkinter as tk
from tkinter import ttk

# Sortierrichtung pro Spalte merken: True = aufsteigend, False = absteigend
sortierung = {'titel': True, 'interpret': True, 'dauer': True}

def sortiere(spalte):
    # Alle Item-IDs aus dem Treeview holen
    alle_ids = baum.get_children('')

    # Daten als Liste von Tupeln sammeln: (spaltenwert, item_id)
    # Damit wir nach dem Wert sortieren koennen, die ID aber noch kennen
    daten = [(baum.item(item_id, 'values'), item_id) for item_id in alle_ids]

    # Spaltenindex bestimmen
    spalten = ('titel', 'interpret', 'dauer')
    idx = spalten.index(spalte)

    # Sortieren nach dem Wert an Position idx
    aufsteigend = sortierung[spalte]
    daten.sort(key=lambda x: x[0][idx], reverse=not aufsteigend)

    # Richtung fuer naechsten Klick umkehren
    sortierung[spalte] = not aufsteigend

    # Zeilen in neuer Reihenfolge verschieben
    for reihe, (werte, item_id) in enumerate(daten):
        baum.move(item_id, '', reihe)   # .move() verschiebt ohne neu einzufuegen

root = tk.Tk()
root.title('Treeview sortierbar')
root.geometry('480x280')

baum = ttk.Treeview(root, columns=('titel', 'interpret', 'dauer'), show='headings')

# command in heading: Klick auf Ueberschrift loest Sortierung aus
baum.heading('titel',     text='Titel',     command=lambda: sortiere('titel'))
baum.heading('interpret', text='Interpret', command=lambda: sortiere('interpret'))
baum.heading('dauer',     text='Dauer',     command=lambda: sortiere('dauer'))

baum.column('titel',     width=220)
baum.column('interpret', width=170)
baum.column('dauer',     width=70, anchor='center')

baum.pack(fill='both', expand=True, padx=10, pady=10)

songs = [
    ('Bohemian Rhapsody',      'Queen',        '5:55'),
    ('Hotel California',       'Eagles',       '6:30'),
    ('Stairway to Heaven',     'Led Zeppelin', '8:02'),
    ('Smells Like Teen Spirit','Nirvana',       '5:01'),
    ('Purple Haze',            'Jimi Hendrix', '2:52'),
    ('Johnny B. Goode',        'Chuck Berry',  '2:41'),
]
for song in songs:
    baum.insert('', tk.END, values=song)

tk.Label(root, text='Auf eine Spaltenüberschrift klicken um zu sortieren.',
         fg='gray').pack(pady=3)

root.mainloop()
````

Alle Konzepte aus diesem Notebook in einem lauffähigen Beispiel: Treeview mit Scrollbar, Einträge hinzufügen und löschen, Auswahl anzeigen, sortierbare Spalten. Das ist die Basis für das spätere MP3-Projekt.

````python
# fenster_20_musikverwaltung.py -- Abschlussbeispiel
import tkinter as tk
from tkinter import ttk

SPALTEN = ('titel', 'interpret', 'album', 'dauer')
sortierung = {s: True for s in SPALTEN}

def sortiere(spalte):
    alle_ids = baum.get_children('')
    daten = [(baum.item(i, 'values'), i) for i in alle_ids]
    idx = SPALTEN.index(spalte)
    daten.sort(key=lambda x: x[0][idx], reverse=not sortierung[spalte])
    sortierung[spalte] = not sortierung[spalte]
    for pos, (_, item_id) in enumerate(daten):
        baum.move(item_id, '', pos)

def hinzufuegen():
    t = titel_var.get().strip()
    i = interpret_var.get().strip()
    a = album_var.get().strip() or 'Unbekannt'
    if t and i:
        baum.insert('', tk.END, values=(t, i, a, '?:??'))
        for var in (titel_var, interpret_var, album_var):
            var.set('')

def loeschen():
    for item_id in baum.selection():
        baum.delete(item_id)
    status_var.set('Eintrag gelöscht.')

def bei_auswahl(event):
    auswahl = baum.selection()
    if auswahl:
        werte = baum.item(auswahl[0], 'values')
        status_var.set(f'Ausgewaehlt: {werte[0]} -- {werte[1]}')

# ------- Fenster aufbauen -------
root = tk.Tk()
root.title('Mini-Musikverwaltung')
root.geometry('620x420')

# --- Eingabebereich ---
eingabe_frame = tk.Frame(root, pady=6)
eingabe_frame.pack(fill='x', padx=10)
eingabe_frame.columnconfigure(1, weight=1)
eingabe_frame.columnconfigure(3, weight=1)
eingabe_frame.columnconfigure(5, weight=1)

titel_var    = tk.StringVar()
interpret_var = tk.StringVar()
album_var    = tk.StringVar()

tk.Label(eingabe_frame, text='Titel:').grid(    row=0, column=0, sticky='w')
tk.Entry(eingabe_frame, textvariable=titel_var).grid(row=0, column=1, sticky='ew', padx=(3,8))
tk.Label(eingabe_frame, text='Interpret:').grid(row=0, column=2, sticky='w')
tk.Entry(eingabe_frame, textvariable=interpret_var).grid(row=0, column=3, sticky='ew', padx=(3,8))
tk.Label(eingabe_frame, text='Album:').grid(    row=0, column=4, sticky='w')
tk.Entry(eingabe_frame, textvariable=album_var).grid(row=0, column=5, sticky='ew', padx=(3,8))
tk.Button(eingabe_frame, text='Hinzufuegen', command=hinzufuegen).grid(row=0, column=6)

# --- Treeview mit Scrollbar ---
baum_frame = tk.Frame(root)
baum_frame.pack(fill='both', expand=True, padx=10, pady=5)

sb = ttk.Scrollbar(baum_frame, orient='vertical')
sb.pack(side='right', fill='y')

baum = ttk.Treeview(baum_frame, columns=SPALTEN, show='headings', yscrollcommand=sb.set)
sb.config(command=baum.yview)

for spalte, breite, titel in [
    ('titel',    220, 'Titel'),
    ('interpret',150, 'Interpret'),
    ('album',    140, 'Album'),
    ('dauer',     70, 'Dauer'),
]:
    baum.heading(spalte, text=titel, command=lambda s=spalte: sortiere(s))
    baum.column( spalte, width=breite, anchor='w' if spalte != 'dauer' else 'center')

baum.pack(side='left', fill='both', expand=True)
baum.bind('<<TreeviewSelect>>', bei_auswahl)

# Startdaten
for song in [
    ('Bohemian Rhapsody',      'Queen',        'A Night at the Opera', '5:55'),
    ('Hotel California',       'Eagles',       'Hotel California',     '6:30'),
    ('Stairway to Heaven',     'Led Zeppelin', 'Led Zeppelin IV',      '8:02'),
    ('Smells Like Teen Spirit','Nirvana',       'Nevermind',           '5:01'),
    ('Purple Haze',            'Jimi Hendrix', 'Are You Experienced',  '2:52'),
]:
    baum.insert('', tk.END, values=song)

# --- Buttons und Statuszeile ---
unten_frame = tk.Frame(root)
unten_frame.pack(fill='x', padx=10, pady=5)

tk.Button(unten_frame, text='Ausgewaehlten Titel loeschen', command=loeschen).pack(side='left')

status_var = tk.StringVar(value='Bereit.')
tk.Label(unten_frame, textvariable=status_var, fg='gray').pack(side='left', padx=12)

root.mainloop()
````

**Zusammenfassung**

| Konzept | Erklärung |
|---------|------------|
| `tk.Listbox(root, height=8)` | Einfache Liste, `height` in Zeilen |
| `.insert(tk.END, wert)` | Eintrag ans Ende anhängen |
| `.delete(index)` | Eintrag löschen |
| `.curselection()` | Tupel mit Indizes der ausgewählten Einträge |
| `.get(index)` | Text an Position `index` lesen |
| `selectmode='extended'` | Mehrfachauswahl mit Shift/Strg |
| `tk.Scrollbar(frame)` | Scrollbar -- muss mit Widget verbunden werden |
| `yscrollcommand=sb.set` | Widget informiert Scrollbar beim Scrollen |
| `sb.config(command=widget.yview)` | Scrollbar steuert Widget |
| `from tkinter import ttk` | Moderner Widget-Satz (Themed Tk) |
| `ttk.Treeview(root, columns=..., show='headings')` | Tabelle ohne Baumspalte |
| `.heading(spalte, text='...')` | Sichtbare Überschrift setzen |
| `.column(spalte, width=...)` | Spaltenbreite und Ausrichtung |
| `.insert('', tk.END, values=(...))` | Zeile einfügen, `''` = kein Elternelement |
| `.selection()` | Tupel mit Item-IDs der ausgewählten Zeilen |
| `.item(id, 'values')` | Spaltenwerte einer Zeile als Tupel |
| `.delete(item_id)` | Zeile löschen |
| `.get_children('')` | Alle Item-IDs der Toplevel-Zeilen |
| `.move(id, '', position)` | Zeile verschieben ohne neu einzufügen |
| `baum.bind('<<TreeviewSelect>>', fn)` | Auf Zeilenauswahl reagieren |
| `.heading(spalte, command=fn)` | Klick auf Überschrift löst Funktion aus |
