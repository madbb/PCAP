1. Das Problem mit grossen Listen
</span>
</div>

List Comprehensions sind elegant -- aber sie haben einen Nachteil: Die gesamte Liste wird sofort in den Speicher geladen.

Bei kleinen Listen ist das kein Problem. Aber was wenn man eine Million Spieler-Statistiken verarbeiten will? Oder die Zeilen einer sehr grossen Logdatei? Dann will man nicht alles auf einmal im Speicher haben -- man will die Werte **einen nach dem anderen** verarbeiten.

```python
import sys

# List Comprehension: alle Werte sofort im Speicher
liste = [x**2 for x in range(1_000_000)]
print(f'Liste: {sys.getsizeof(liste):,} Bytes')   # ca. 8 MB

# Wenn man nur die Summe braucht, ist das verschwenderisch
summe = sum(liste)
print(f'Summe: {summe:,}')
```

Die Liste braucht Megabytes an Speicher -- obwohl wir nur die Summe brauchen und jeden Einzelwert sofort nach der Addition vergessen koennten. Die Loesung heisst **Generator**: Werte werden nicht alle auf einmal berechnet, sondern **lazy** -- jeweils nur wenn der naechste benoetigt wird.

2. Generator Expressions -- wie List Comprehension, aber lazy
</span>
</div>

Eine **Generator Expression** sieht aus wie eine List Comprehension -- nur mit runden statt eckigen Klammern. Sie erzeugt aber keine Liste, sondern ein **Generator-Objekt**. Das berechnet Werte erst wenn sie abgerufen werden.

```python
import sys

# List Comprehension -- eckige Klammern, alles sofort
liste = [x**2 for x in range(1_000_000)]
print(f'Liste:     {sys.getsizeof(liste):,} Bytes')

# Generator Expression -- runde Klammern, lazy
generator = (x**2 for x in range(1_000_000))
print(f'Generator: {sys.getsizeof(generator):,} Bytes')  # nur ~200 Bytes!

print(type(generator))   # <class 'generator'>
```

Der Generator ist winzig im Speicher -- er speichert nicht die Werte, sondern nur die Anweisung wie die Werte berechnet werden. Werte entstehen erst beim Abruf.

```python
# Generator direkt in Funktionen -- ohne extra Klammern
# sum(), min(), max(), any(), all() akzeptieren direkt Generators

summe = sum(x**2 for x in range(1_000_000))   # kein Speicherproblem
print(f'Summe: {summe:,}')

maximum = max(x**2 for x in range(100))
print(f'Maximum: {maximum}')

# any() / all() mit Generator -- bricht ab sobald Ergebnis feststeht
helden_hp = [100, 45, 80, 12, 60, 95]
print(any(hp < 20 for hp in helden_hp))    # True -- sobald 12 gefunden, fertig
print(all(hp > 0  for hp in helden_hp))    # True -- alle groesser als 0
```

3. Wie ein Generator funktioniert -- next() und StopIteration
</span>
</div>

Ein Generator ist ein **Iterator** -- ein Objekt das weiss wo es gerade ist und auf Anfrage den naechsten Wert liefert. Mit `next()` kann man Werte einzeln abrufen. Wenn keine Werte mehr kommen, wirft der Generator eine `StopIteration`-Exception.

Das ist der Mechanismus der hinter jeder `for`-Schleife steckt -- Python ruft intern `next()` auf, bis `StopIteration` kommt.

```python
gen = (x**2 for x in range(4))

# Werte einzeln abrufen
print(next(gen))   # 0
print(next(gen))   # 1
print(next(gen))   # 4
print(next(gen))   # 9

try:
    print(next(gen))   # StopIteration!
except StopIteration:
    print('Generator erschoepft')

# Ein erschoepfter Generator liefert keine Werte mehr
print(list(gen))   # []  -- leer
```

Generators sind einmalig -- sie koennen nicht zurueckgespult werden
</span>
</div>

```python
gen = (x for x in range(5))

# Erste Iteration: alle Werte
print(list(gen))    # [0, 1, 2, 3, 4]

# Zweite Iteration: leer! Generator ist erschoepft
print(list(gen))    # []  -- nichts mehr

# Loesung: neuen Generator erstellen
gen = (x for x in range(5))  # neues Objekt
print(list(gen))    # [0, 1, 2, 3, 4]
```

4. Generator-Funktionen mit yield
</span>
</div>

Generator Expressions sind praktisch fuer einfache Faelle. Fuer komplexere Logik gibt es **Generator-Funktionen** -- normale Funktionen die statt `return` das Schluesselwort `yield` verwenden.

Der Unterschied zu `return`: `return` beendet die Funktion. `yield` **pausiert** sie und liefert einen Wert. Beim naechsten Aufruf von `next()` laeuft die Funktion weiter -- genau da wo sie aufgehoert hat.

```python
# Eine normale Funktion
def gib_zahlen_liste():
    return [1, 2, 3]    # alle auf einmal

# Eine Generator-Funktion
def gib_zahlen_generator():
    yield 1   # pausiert hier, liefert 1
    yield 2   # beim naechsten next(): weiter, liefert 2
    yield 3   # beim naechsten next(): weiter, liefert 3
              # danach: StopIteration

# Aufruf erzeugt ein Generator-Objekt -- noch keine Berechnung!
gen = gib_zahlen_generator()
print(type(gen))    # <class 'generator'>

print(next(gen))    # 1
print(next(gen))    # 2
print(next(gen))    # 3

# Oder einfach mit for-Schleife
for zahl in gib_zahlen_generator():
    print(zahl, end=' ')
```

```python
# Praxisbeispiel: Spieler-Logs zeilenweise verarbeiten
# Simuliert das Lesen einer sehr grossen Datei

def lese_kampf_logs(logs):
    """Gibt nur Kampfereignisse zurueck -- lazy, ohne alles im Speicher."""
    for zeile in logs:
        if 'KAMPF' in zeile:
            yield zeile.strip()

alle_logs = [
    'INFO: Spiel gestartet',
    'KAMPF: Aldric greift Drachen an -- 45 Schaden',
    'INFO: Held bewegt sich',
    'KAMPF: Drache trifft Aldric -- 30 Schaden',
    'INFO: Gegenstand gefunden',
    'KAMPF: Aldric besiegt Drachen',
]

# Nur Kampfereignisse -- lazy
for ereignis in lese_kampf_logs(alle_logs):
    print(ereignis)
```

5. Verbindung zu map() und filter()
</span>
</div>

In Notebook 03f haben wir `map()` und `filter()` kennengelernt. Jetzt macht ihr Verhalten vollstaendig Sinn: Beide geben **keinen List** zurueck, sondern einen **lazy Iterator** -- also im Prinzip einen Generator.

Das erklaert warum man `list()` drumherum braucht wenn man eine Liste will -- und warum sie speichereffizient sind.

```python
zahlen = [1, 4, 9, 16, 25]

# map() gibt einen lazy Iterator zurueck -- kein list!
ergebnis = map(lambda x: x**2, zahlen)
print(type(ergebnis))   # <class 'map'> -- ein Iterator
print(next(ergebnis))   # 1  -- wie bei einem Generator
print(next(ergebnis))   # 16

# filter() genau so
positiv = filter(lambda x: x > 0, [-2, -1, 0, 1, 2])
print(type(positiv))    # <class 'filter'>

# Alle drei Varianten sind aequivalent und lazy:
data = range(1_000_000)

v1 = map(lambda x: x*2, data)          # lazy map
v2 = (x*2 for x in data)              # lazy generator expression

# Beide verbrauchen kaum Speicher -- Werte entstehen erst beim Abruf
import sys
print(sys.getsizeof(v1), sys.getsizeof(v2))  # beide klein
```

**Zusammenfassung**

| Konzept | Erklaerung |
|---------|------------|
| Lazy Evaluation | Werte werden erst berechnet wenn sie benoetigt werden |
| Generator Expression | `(ausdruck for x in it)` -- wie List Comprehension, aber lazy |
| Generator-Funktion | Funktion mit `yield` -- pausiert und liefert Werte einzeln |
| `next()` | Naechsten Wert abrufen -- `StopIteration` wenn erschoepft |
| Einmalig | Generator kann nicht zurueckgespult werden |
| `map()`, `filter()` | Geben lazy Iteratoren zurueck -- wie Generators |

**Wann Generator statt Liste:**
- Grosse Datenmengen die nicht alle gleichzeitig im Speicher sein muessen
- Wenn man nur einmal durchiteriert
- Als Argument fuer `sum()`, `min()`, `max()`, `any()`, `all()`

**Wann Liste:**
- Wenn man mehrfach iterieren will
- Wenn man Indexzugriff braucht
- Wenn man die Laenge braucht (`len()` funktioniert nicht bei Generators)
