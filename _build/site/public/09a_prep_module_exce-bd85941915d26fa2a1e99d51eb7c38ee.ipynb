```python
import math

# Alle PCAP-relevanten math-Funktionen
print(math.ceil(4.1))       # 5   -- kleinste ganze Zahl >= x
print(math.ceil(-4.1))      # -4  -- Achtung: Richtung!
print(math.floor(4.9))      # 4   -- groesste ganze Zahl <= x
print(math.floor(-4.9))     # -5  -- Achtung!
print(math.trunc(4.9))      # 4   -- Nachkommastellen weg, Richtung Null
print(math.trunc(-4.9))     # -4  -- NICHT -5 (Unterschied zu floor!)
print(math.factorial(5))    # 120 -- 5*4*3*2*1
print(math.hypot(3, 4))     # 5.0 -- Hypotenuse: sqrt(3^2 + 4^2)
print(math.sqrt(16))        # 4.0 -- immer float!
```

**Checkliste -- das muss sitzen:**

- [ ] except-Reihenfolge: spezifisch vor allgemein
- [ ] `except (E1, E2)`: mehrere in einem Block
- [ ] `else`: laeuft nur wenn kein Fehler
- [ ] `finally`: laeuft immer -- auch nach return
- [ ] `raise` vs. `raise` ohne Argument
- [ ] `assert` -- wirft AssertionError
- [ ] Hierarchie: BaseException > Exception > ArithmeticError > ZeroDivisionError usw.
- [ ] `args` -- was `print(e)` anzeigt (ein Arg: direkt; mehrere: Tupel)
- [ ] `Exception.__init__` setzt `self.args`
- [ ] `self.args` manuell ueberschreiben veraendert die Ausgabe

```python
# Reihenfolge: spezifisch zuerst
try:
    x = 1 / 0
except ZeroDivisionError:   # spezifisch -- greift
    print('ZeroDivisionError')
except ArithmeticError:     # allgemeiner -- wuerde auch greifen, aber zu spaet
    print('ArithmeticError')
except Exception:           # noch allgemeiner
    print('Exception')
```

```python
# else und finally
def teste(wert):
    try:
        ergebnis = 10 / wert
    except ZeroDivisionError:
        print('Fehler!')
    else:
        print(f'Kein Fehler: {ergebnis}')  # nur wenn kein Fehler
    finally:
        print('Immer!')                     # immer

teste(2)   # Kein Fehler: 5.0 / Immer!
print()
teste(0)   # Fehler! / Immer!
```

```python
# args und was print(e) ausgibt
e1 = ValueError('eine Meldung')
e2 = ValueError('erstes', 'zweites')
e3 = ValueError()

print(str(e1))   # 'eine Meldung'          -- ein Arg: direkt
print(str(e2))   # "('erstes', 'zweites')" -- mehrere: Tupel
print(str(e3))   # ''                       -- leer
```

```python
# Eigene Exception -- self.args bestimmt print(e)
class MeinFehler(Exception):
    def __init__(self, msg):
        Exception.__init__(self, msg + msg)  # args = (msg+msg,)
        self.args = (msg,)                   # args = (msg,) -- ueberschrieben!

try:
    raise MeinFehler('hallo')
except MeinFehler as e:
    print(e)        # 'hallo' -- nicht 'hallohallo'!
    print(e.args)   # ('hallo',)
```
