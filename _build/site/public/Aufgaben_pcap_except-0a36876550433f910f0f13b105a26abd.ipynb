# Lösungen
---



1.  **A) `Ein Fehler ist aufgetreten`**  
    Ein `except:` ohne Angabe einer Exception (ein "nacktes" except) fängt alle Exceptions ab, die von `BaseException` erben.



2.  **C) `<class 'ValueError'>`**  
    `except ValueError as e:` fängt die `ValueError`-Exception und weist die Exception-Instanz der Variablen `e` zu. `type(e)` gibt daher die Klasse der Exception zurück.



3.  **B) `Fehler gefangen`**  
    Der Zugriff `my_list[0]` auf eine leere Liste löst einen `IndexError` aus. Da `IndexError` im Tupel `(IndexError, KeyError)` enthalten ist, wird dieser `except`-Block ausgeführt.



4.  **D)**

<pre>
    Schritt 1
    Schritt 2
    Kein Fehler
</pre> 
    Im `try`-Block tritt keine Exception auf. Daher wird der `except`-Block übersprungen und der `else`-Block am Ende ausgeführt.



5.  **B) `Exception gefangen`**  
    Python prüft die `except`-Blöcke von oben nach unten. Da `ZeroDivisionError` eine Unterklasse von `Exception` ist, fängt der erste allgemeinere Block (`except Exception:`) den Fehler bereits ab.



6.  **C) `Zu jung!`**  
    Die Funktion löst eine `ValueError`-Exception mit der Nachricht 'Zu jung!' aus. Diese wird gefangen, und `print(e)` gibt die String-Repräsentation der Exception aus, was in diesem Fall die Nachricht ist.



7.  **C) Der Code bricht mit einem `AssertionError` ab.**  
    Die `assert`-Bedingung `x > 0` ist `False`. Daher wird ein `AssertionError` mit der optionalen Nachricht 'x muss positiv sein' ausgelöst und das Programm beendet (da die Exception nicht gefangen wird).



8.  **B) `Ein spezieller Fehler`**  
    Eine eigene Exception-Klasse wird definiert, mit einer Nachricht ausgelöst und wieder gefangen. `print(e)` gibt die Nachricht aus, mit der die Exception instanziiert wurde.



9.  **A) `('Fehler', 404, 'Nicht gefunden')`**  
    Die Eigenschaft `.args` einer Exception ist ein Tupel, das alle Argumente enthält, die bei der Erstellung der Exception übergeben wurden.



10. **B) `Gefangen` wird ausgegeben, gefolgt von einem `ZeroDivisionError` Traceback.**  
    Der `ZeroDivisionError` wird gefangen und `Gefangen` wird gedruckt. `raise e` löst dieselbe Exception erneut aus (re-raising), was dazu führt, dass das Programm mit einem `ZeroDivisionError` abbricht.



11. **A) `Error`**  
    `SubError` ist eine Unterklasse von `Error`. Da der `except`-Block für die Basisklasse `Error` zuerst kommt, fängt er die `SubError`-Instanz ab.



12. **B)**
    <pre>Start
    Division durch Null
</pre>
    Im `try`-Block tritt ein `ZeroDivisionError` auf. Die Ausführung des Blocks wird sofort abgebrochen (`Ende` wird nicht gedruckt). Der passende `except`-Block wird ausgeführt. Der `else`-Block wird übersprungen.



13. **C) `Wert ist Null`**  
    `assert wert` ist `False`, da `wert` 0 ist. Ein `AssertionError` wird mit der angegebenen Nachricht ausgelöst. Der `except`-Block fängt diesen Fehler und `print(e)` gibt die Nachricht aus.



14. **B) `1`**  
    Die Operation `1 + "2"` löst einen `TypeError` aus. Standardmäßig wird die Fehlermeldung als ein einziges String-Argument an die Exception übergeben. `e.args` ist daher ein Tupel mit einem Element, und `len()` davon ist 1.



15. **C) `Fehlercode: 404`**  
    Die `__str__`-Methode der benutzerdefinierten Exception wird aufgerufen, wenn die Instanz `e` gedruckt wird. Diese Methode formatiert die Ausgabe wie definiert.



16. **A)** 
    <pre>Wird immer ausgeführt
    Gefangen
    </pre>  
    Wenn `funktion()` aufgerufen wird, wird ein `ValueError` ausgelöst. Bevor die Exception die Funktion verlässt, wird der `finally`-Block ausgeführt. Danach wird die Exception vom äußeren `try...except`-Block gefangen.



17. **D)**
    <pre>   
    Division durch Null!  
    Fertig.  
    Ergebnis: 5.0  
    Fertig.
    </pre>

Beim ersten Funktionsaufruf `divide(10, 2)` wird keine Ausnahme ausgelöst → `else`-Block wird ausgeführt → `Ergebnis: 5.0`.

Danach wird immer `finally` ausgeführt → Fertig.

Beim zweiten Aufruf `divide(5, 0)` löst `x / y` eine `ZeroDivisionError` aus → `except`-Block wird ausgeführt → `Division durch Null!`

Danach wird wieder `finally` ausgeführt → Fertig.



18. **A)** 
<pre> 
    1
    4 
</pre> 
    `int("a")` löst einen `ValueError` aus. Der erste `except`-Block passt und druckt `1`. Der `else`-Block wird übersprungen, da ein Fehler auftrat. Der `finally`-Block wird immer ausgeführt und druckt `4`.



19. **A) `Erwischt!`**  
    Ein `raise` ohne Argument in einem `except`-Block löst die gerade aktive Exception erneut aus. `f()` fängt den `ZeroDivisionError` und löst ihn sofort wieder aus. Der äußere `except`-Block fängt ihn dann.



20. **B) `Hallo Welt`**  
    Wenn Python im optimierten Modus (mit dem Flag `-O`) ausgeführt wird, werden alle `assert`-Anweisungen ignoriert. Die `assert`-Zeile wird übersprungen und der Code läuft normal weiter.



21. **B) `Falscher Schlüssel`**  
    Der Zugriff auf einen nicht existierenden Schlüssel in einem Dictionary löst einen `KeyError` aus. Der erste `except`-Block fängt diesen nicht ab, aber der zweite (`except KeyError:`) passt.



22. **C) Der Code bricht mit einem `TypeError` ab, der besagt, dass eine Exception-Instanz erwartet wurde.**  
    `raise` erwartet eine *Instanz* einer Exception (z.B. `IndexError()`), nicht die Klasse selbst. Das Auslösen einer Klasse führt zu einem `TypeError`.



23. **A) `()`**  
    Wenn eine Exception ohne Argumente instanziiert wird (`Exception()`), ist ihre `.args`-Eigenschaft ein leeres Tupel.



24. **B) `10.0 T 3.33... 2.5 `**  
    Die Schleife teilt 10 durch jedes Element. Bei `1`, `3` und `4` funktioniert dies. Bei `'2'` schlägt die Division fehl, da `n` ein String ist (`TypeError`). Der entsprechende `except`-Block wird ausgeführt und druckt `T`.



25. **A) `Lookup Fehler`**  
    Der Zugriff auf `my_dict[0]` löst einen `KeyError` aus. Da `KeyError` eine Unterklasse von `LookupError` ist und der `except`-Block für `LookupError` zuerst steht, fängt dieser die Exception ab.
