Inhalt der Datei:
```python
import platform

MODUL_INFO = "System Tools v1.0"

def kurz_info():
    '''Gibt System, Architektur und Python-Version als String zurueck.'''
    return f"{platform.system()} | {platform.machine()} | Python {platform.python_version()}"

def ist_windows():
    '''Gibt True zurueck, wenn das Programm auf Windows laeuft.'''
    return platform.system() == 'Windows'

def python_major():
    '''Gibt die Major-Versionsnummer als Integer zurueck.'''
    return int(platform.python_version_tuple()[0])
```
