## Lösungen



**1. B) `Basis True Unabhängig`**
Die Instanz `obj1` der Klasse `Abgeleitet` hat keine eigene `__str__`-Methode, erbt sie aber von `Basis`. Daher gibt `print(obj1)` "Basis" aus. `isinstance(obj1, Basis)` ist `True`, da `Abgeleitet` von `Basis` erbt. `obj2` hat seine eigene `__str__`-Methode.



**2. A) `True False True`**
`kopie_ref = original` bewirkt, dass beide Variablen auf dasselbe Objekt im Speicher zeigen (`is` ist `True`). `kopie_wert = original[:]` erstellt eine neue Liste mit demselben Inhalt. Es ist ein anderes Objekt (`is` ist `False`), aber der Inhalt ist identisch (`==` ist `True`).



**3. C) `flatter`\
   `flatter`**
Die Schleife prüft jedes Objekt mit `hasattr(tier, 'fliegen')`. Nur die Instanzen von `Vogel` haben diese Methode. Der Code ruft sie also für das erste und das dritte Objekt auf und ignoriert das `Fisch`-Objekt, ohne einen Fehler zu verursachen.



**4. A)**
1.  **`"dynamic" in dir(a)` gibt `True` zurück:** Die Zeile `a.dynamic = 123` fügt der *Instanz* `a` ein neues Attribut hinzu. Die Funktion `dir(a)` gibt eine Liste aller gültigen Attribute für das Objekt `a` zurück, was sowohl Klassen- als auch Instanzattribute umfasst. Da `dynamic` ein Attribut der Instanz `a` ist, ist es in der Liste von `dir(a)` enthalten.
2.  **`hasattr(A, "dynamic")` gibt `False` zurück:** `hasattr()` prüft hier, ob die *Klasse* `A` ein Attribut namens `dynamic` hat. Das Attribut wurde jedoch nur der *Instanz* `a` hinzugefügt, nicht der Klasse `A` selbst. Daher hat die Klasse `A` dieses Attribut nicht.


**5. B) `HTTPRequest True`**
`obj.__class__` gibt das Klassenobjekt zurück, von dem die Instanz `obj` erstellt wurde (in diesem Fall `HTTPRequest`). `__name__` dieses Klassenobjekts ist der String `"HTTPRequest"`. Dieser String ist eine Instanz der Klasse `str`.



**6. C) `True False`**
Python speichert häufig verwendete kleine ganze Zahlen (-5 bis 256) in einem Cache, um Speicher zu sparen. `a` und `b` (Wert 100) verweisen auf dasselbe Objekt im Cache (`a is b` ist `True`). `c` und `d` (Wert 300) liegen außerhalb dieses Bereichs, daher werden zwei separate Objekte erstellt (`c is d` ist `False`).



**7. C) `[10, 3.14, 20]`**
`isinstance(d, (int, float))` prüft, ob `d` eine Instanz von `int` ODER `float` ist. Die List Comprehension sammelt alle Elemente aus `daten`, für die diese Bedingung zutrifft.



**8. C) `Fehler`**
`issubclass()` erwartet als erstes Argument eine Klasse, nicht eine Instanz. Der Versuch, `issubclass()` mit der Instanz `Auto()` aufzurufen, führt zu einem `TypeError`. Die korrekte Prüfung für eine Instanz wäre `isinstance(Auto(), Motor)`.



**9. D) `collections __main__`**
Das `__module__`-Attribut gibt an, in welchem Modul eine Klasse definiert wurde. `Counter` wurde im Modul `collections` definiert. `LokaleKlasse` wurde im Hauptskript definiert, dessen Modulname immer `__main__` ist.



**10. C) `2 True`**
Das `__bases__`-Attribut einer Klasse gibt ein Tupel ihrer direkten Elternklassen zurück. `FileHandler` hat zwei direkte Eltern (`Reader` und `Writer`), also ist die Länge 2. Das Attribut selbst ist immer ein Tupel.



**11. B) `Config None`**
Der erste Aufruf `get_property(c, "port")` ist erfolgreich, da das `Config`-Objekt das Attribut `port` hat. Die Funktion gibt jedoch `obj.__class__.__name__` zurück, was `"Config"` ist. Der zweite Aufruf `get_property(c, "user")` schlägt fehl, da das Attribut nicht existiert. Daher wird der `default`-Wert `None` zurückgegeben.



**12. B) `False True`**
Obwohl `t1` und `t2` identischen Inhalt haben (`t1 == t2` ist `True`), sind es zwei separate Tupel-Objekte im Speicher. Der `is`-Operator prüft die Speicheradresse und gibt daher `False` zurück. (Anders als bei kleinen ganzen Zahlen werden Tupel nicht standardmäßig zwischengespeichert).



**13. C) `3`**
Die List Comprehension iteriert durch den Zoo. Für jedes `tier` wird dessen Klasse (`tier.__class__`) genommen und mit `issubclass()` geprüft, ob sie von `Tier` erbt. Dies trifft auf die Klassen `Hund` und `Katze` zu. Drei der vier Objekte im Zoo erfüllen diese Bedingung.



**14. C) `Typfehler!`**
Die `__str__`-Methode muss per Definition immer ein String-Objekt zurückgeben. Die Rückgabe von `None` (oder einem anderen Typ) verstößt gegen diese Regel und löst zur Laufzeit einen `TypeError` aus.



**15. B) `Aufrufbares Objekt: meine_funktion`**
Die Funktion `beschreibe` prüft ihre Eingabe. `meine_funktion` ist keine Liste. Die nächste Prüfung `hasattr(obj, '__call__')` ist `True`, da Funktionen aufrufbar sind. Daher wird der zweite `if`-Block ausgeführt, der `obj.__name__` (`"meine_funktion"`) verwendet, um den String zu formatieren.
