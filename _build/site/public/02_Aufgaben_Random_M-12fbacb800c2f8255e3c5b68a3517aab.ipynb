**Schritt 1:** Erstelle eine neue Datei `wuerfel_tools.py` im selben Verzeichnis wie dieses Notebook.
Inhalt der Datei:
```python
import random

MODUL_INFO = "Wuerfel Tools v1.0"

def wuerfel(seiten=6):
    '''Wirft einen Wuerfel mit der angegebenen Seitenanzahl.'''
    return random.randint(1, seiten)

def mehrfach_wuerfeln(anzahl, seiten=6):
    '''Wirft einen Wuerfel mehrfach und gibt alle Ergebnisse als Liste zurueck.'''
    return [random.randint(1, seiten) for _ in range(anzahl)]
```
