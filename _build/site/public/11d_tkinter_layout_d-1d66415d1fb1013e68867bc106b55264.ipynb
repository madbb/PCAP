1. Schriften -- font-Parameter
</span>

Den `font`-Parameter kennen wir schon aus 11a -- er steuert die Schrift eines Widgets. Bisher haben wir ihn kaum genutzt. Hier lernen wir was man alles damit machen kann.

**Kurzform als Tupel:** Die einfachste Art eine Schrift anzugeben ist ein Tupel:

```python
font=('Arial', 12)             # Schriftart und Größe
font=('Arial', 12, 'bold')     # fett
font=('Arial', 12, 'italic')   # kursiv
font=('Arial', 12, 'bold italic')  # beides
```

**Schriftarten die immer verfügbar sind** -- plattformunabhängig:
- `'TkDefaultFont'` -- die Standardschrift von tkinter
- `'TkFixedFont'` -- Monospace-Schrift (gut für Code)
- `'TkHeadingFont'` -- Überschriften-Schrift

**Schriftarten die auf den meisten Systemen vorhanden sind:**
- Windows: `'Arial'`, `'Courier New'`, `'Times New Roman'`, `'Segoe UI'`
- macOS/Linux: `'Helvetica'`, `'Courier'`, `'Times'`

**Vorsicht:** Wenn man eine Schrift angibt die auf dem System nicht vorhanden ist, fällt tkinter auf eine Standardschrift zurück -- ohne Fehlermeldung. Wenn Portabilität wichtig ist, besser `'TkDefaultFont'` benutzen.

**Font-Objekt aus `tkinter.font`:** Für mehr Kontrolle -- zum Beispiel wenn man dieselbe Schrift an vielen Stellen benutzen will und sie zentral ändern möchte -- gibt es `tkinter.font.Font`:

```python
from tkinter import font

ueberschrift_font = font.Font(family='Arial', size=16, weight='bold')
tk.Label(root, text='Titel', font=ueberschrift_font).pack()
```

Wenn man dann `ueberschrift_font.config(size=20)` aufruft, aktualisieren sich alle Widgets die dieses Font-Objekt benutzen automatisch.

````python
# fenster_21_schriften.py
import tkinter as tk

root = tk.Tk()
root.title('Schriften')
root.geometry('380x280')

# Verschiedene Schriften im Vergleich
tk.Label(root, text='Standard (kein font-Parameter)').pack(pady=4)

tk.Label(root, text='Arial 12 fett',
         font=('Arial', 12, 'bold')).pack(pady=4)

tk.Label(root, text='Courier New 11 kursiv',
         font=('Courier New', 11, 'italic')).pack(pady=4)

tk.Label(root, text='TkFixedFont -- immer Monospace',
         font='TkFixedFont').pack(pady=4)

tk.Label(root, text='Groß: Arial 18',
         font=('Arial', 18)).pack(pady=4)

# Button mit eigener Schrift
tk.Button(root, text='Button mit groesser Schrift',
          font=('Arial', 13, 'bold')).pack(pady=8)

root.mainloop()
````

2. ttk.Style -- Treeview und andere ttk-Widgets gestalten
</span>
</div>

Bei klassischen tk-Widgets funktionieren `bg`, `fg` und `font` direkt. Bei ttk-Widgets -- also `ttk.Treeview`, `ttk.Button`, `ttk.Scrollbar` -- funktioniert das nicht so. ttk-Widgets verwenden ein Style-System das von den Betriebssystem-Themes getrennt ist.

Um zum Beispiel die Schriftgröße im Treeview zu ändern, benutzt man `ttk.Style`:

```python
from tkinter import ttk

style = ttk.Style()
style.configure('Treeview', font=('Arial', 11), rowheight=25)
style.configure('Treeview.Heading', font=('Arial', 11, 'bold'))
```

**`rowheight`** -- die Zeilenhöhe in Pixeln. Standard ist oft zu gering. `25` ist ein guter Wert für normale Schriftgrößen.

**`Treeview.Heading`** -- der Style-Name für die Überschriftenzeile. Die Überschriften sind ein eigener Style-Bereich getrennt von den Datenzeilen.

**Abwechselnde Zeilenfarben (Striping):** Der Treeview unterstützt Tags -- damit kann man einzelnen Zeilen Farben zuweisen:

```python
baum.tag_configure('gerade',   background='#f0f0f0')
baum.tag_configure('ungerade', background='white')

# Beim Einfügen Tag mitgeben:
baum.insert('', tk.END, values=(...), tags=('gerade',))
```

````python
# fenster_22_treeview_style.py
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title('Treeview gestaltet')
root.geometry('500x280')

# ttk.Style fuer Treeview konfigurieren
style = ttk.Style()
style.configure('Treeview',
                font=('Segoe UI', 10),
                rowheight=26)           # Zeilenhoehe in Pixeln
style.configure('Treeview.Heading',
                font=('Segoe UI', 10, 'bold'))

baum = ttk.Treeview(root, columns=('titel', 'interpret', 'dauer'), show='headings')

baum.heading('titel',     text='Titel')
baum.heading('interpret', text='Interpret')
baum.heading('dauer',     text='Dauer')
baum.column('titel',     width=220)
baum.column('interpret', width=170)
baum.column('dauer',     width=70, anchor='center')

# Abwechselnde Zeilenfarben definieren
baum.tag_configure('gerade',   background='#f5f5f5')
baum.tag_configure('ungerade', background='#ffffff')

baum.pack(fill='both', expand=True, padx=10, pady=10)

songs = [
    ('Bohemian Rhapsody',       'Queen',        '5:55'),
    ('Hotel California',        'Eagles',       '6:30'),
    ('Stairway to Heaven',      'Led Zeppelin', '8:02'),
    ('Smells Like Teen Spirit', 'Nirvana',      '5:01'),
    ('Purple Haze',             'Jimi Hendrix', '2:52'),
]

for index, song in enumerate(songs):
    tag = 'gerade' if index % 2 == 0 else 'ungerade'
    baum.insert('', tk.END, values=song, tags=(tag,))

root.mainloop()
````

</span>

Eine Menüleiste ("Datei", "Bearbeiten", "Hilfe" etc.) ist in fast jeder Desktopanwendung vorhanden. In tkinter baut man sie so auf:

**Schritt 1:** Eine `Menu`-Instanz als Hauptmenüleiste erstellen und mit dem Fenster verbinden:

```python
menubar = tk.Menu(root)
root.config(menu=menubar)   # Fenster bekommt diese Menüleiste
```

**Schritt 2:** Untermenüs erstellen und in die Leiste einhängen:

```python
datei_menu = tk.Menu(menubar, tearoff=0)  # tearoff=0 verhindert abreiÃbares Menü
menubar.add_cascade(label='Datei', menu=datei_menu)
```

`tearoff=0` ist wichtig -- ohne diesen Parameter hat jedes Menü oben eine gestrichelte Linie die es "abreißbar" macht. Das sieht altmodisch aus und verwirrt Nutzer. Immer auf `0` setzen.

**Schritt 3:** Einträge zum Untermenü hinzufügen:

```python
datei_menu.add_command(label='Öffnen',    command=oeffnen)
datei_menu.add_command(label='Speichern', command=speichern)
datei_menu.add_separator()               # Trennlinie
datei_menu.add_command(label='Beenden',  command=root.quit)
```

**`add_separator()`** fügt eine horizontale Trennlinie ein -- gut um zusammengehörige Einträge zu gruppieren.

````python
# fenster_23_menu.py
import tkinter as tk

def datei_neu():
    status.config(text='Neu')

def datei_oeffnen():
    status.config(text='Oeffnen...')

def datei_speichern():
    status.config(text='Gespeichert.')

def info_anzeigen():
    status.config(text='Meine Anwendung v1.0')

root = tk.Tk()
root.title('Menüleiste')
root.geometry('350x200')

# --- Menüleiste aufbauen ---
menubar = tk.Menu(root)
root.config(menu=menubar)   # Fenster bekommt diese Menüleiste

# Datei-Menü
datei_menu = tk.Menu(menubar, tearoff=0)   # tearoff=0: kein abreißbares Menü
menubar.add_cascade(label='Datei', menu=datei_menu)

datei_menu.add_command(label='Neu',        command=datei_neu)
datei_menu.add_command(label='Öffnen ...',  command=datei_oeffnen)
datei_menu.add_command(label='Speichern',  command=datei_speichern)
datei_menu.add_separator()                 # Trennlinie
datei_menu.add_command(label='Beenden',    command=root.quit)

# Hilfe-Menü
hilfe_menu = tk.Menu(menubar, tearoff=0)
menubar.add_cascade(label='Hilfe', menu=hilfe_menu)
hilfe_menu.add_command(label='Info', command=info_anzeigen)

# --- Fensterinhalt ---
tk.Label(root, text='Menüleiste ausprobieren',
         font=('Arial', 12)).pack(expand=True)

status = tk.Label(root, text='Bereit.', fg='gray', anchor='w')
status.pack(side='bottom', fill='x', padx=5, pady=3)

root.mainloop()
````

4. filedialog -- Datei- und Ordnerauswahl
</span>
</div>

Für das MP3-Projekt müssen Nutzer Musikdateien vom Dateisystem auswählen. Dafür gibt es `tkinter.filedialog` -- ein Submodul das die nativen Dateidialoge des Betriebssystems öffnet.

**Import:**

```python
from tkinter import filedialog
```

**Die wichtigsten Funktionen:**

`filedialog.askopenfilename()` -- öffnet einen Dialog zur Auswahl einer einzelnen Datei. Gibt den Pfad als String zurück. Gibt einen leeren String zurück wenn der Nutzer abbricht.

`filedialog.askopenfilenames()` -- wie oben, aber für mehrere Dateien gleichzeitig. Gibt ein Tupel von Pfaden zurück.

`filedialog.askdirectory()` -- Ordnerauswahl statt Dateiauswahl. Gibt den Ordnerpfad zurück.

`filedialog.asksaveasfilename()` -- Speicherdialog. Gibt den gewählten Pfad zurück, erstellt die Datei aber nicht automatisch -- das muss man selbst tun.

**Parameter die alle teilen:**

- `title='...'` -- Titel des Dialogfensters
- `initialdir='...'` -- Startverzeichnis des Dialogs
- `filetypes=[('Beschreibung', '*.endung')]` -- Filter für Dateitypen

**Wichtig:** Der Rückgabewert muss immer auf leer geprüft werden. Wenn der Nutzer auf "Abbrechen" klickt, gibt die Funktion `''` zurück -- nicht `None`.

````python
# fenster_24_filedialog.py
import tkinter as tk
from tkinter import filedialog

def datei_auswaehlen():
    pfad = filedialog.askopenfilename(
        title='Datei auswaehlen',
        initialdir='/',                              # Startverzeichnis
        filetypes=[
            ('Textdateien',  '*.txt'),
            ('Python-Dateien', '*.py'),
            ('Alle Dateien', '*.*'),                 # immer als letzte Option
        ]
    )
    if pfad:                                         # leer wenn Abbruch
        pfad_label.config(text=pfad)
    else:
        pfad_label.config(text='Abgebrochen.')

def mehrere_auswaehlen():
    pfade = filedialog.askopenfilenames(
        title='Mehrere Dateien auswaehlen',
        filetypes=[('Alle Dateien', '*.*')]
    )
    if pfade:
        pfad_label.config(text=f'{len(pfade)} Dateien ausgewaehlt')
    else:
        pfad_label.config(text='Abgebrochen.')

def ordner_auswaehlen():
    ordner = filedialog.askdirectory(
        title='Ordner auswaehlen',
        initialdir='/'
    )
    if ordner:
        pfad_label.config(text=ordner)
    else:
        pfad_label.config(text='Abgebrochen.')

def speichern_als():
    pfad = filedialog.asksaveasfilename(
        title='Speichern unter',
        defaultextension='.txt',                     # wird angehaengt wenn Nutzer keine gibt
        filetypes=[('Textdateien', '*.txt')]
    )
    if pfad:
        pfad_label.config(text=f'Speichern nach: {pfad}')
    else:
        pfad_label.config(text='Abgebrochen.')

root = tk.Tk()
root.title('filedialog')
root.geometry('420x230')

button_frame = tk.Frame(root)
button_frame.pack(pady=15)

tk.Button(button_frame, text='Eine Datei',       command=datei_auswaehlen,   width=18).grid(row=0, column=0, padx=5, pady=4)
tk.Button(button_frame, text='Mehrere Dateien',  command=mehrere_auswaehlen, width=18).grid(row=0, column=1, padx=5, pady=4)
tk.Button(button_frame, text='Ordner',           command=ordner_auswaehlen,  width=18).grid(row=1, column=0, padx=5, pady=4)
tk.Button(button_frame, text='Speichern unter',  command=speichern_als,      width=18).grid(row=1, column=1, padx=5, pady=4)

tk.Label(root, text='Ergebnis:').pack()
pfad_label = tk.Label(root, text='', fg='#1a6ea8', wraplength=380)
pfad_label.pack(pady=5)

root.mainloop()
````

5. messagebox -- Meldungen und Bestätigungen
</span>
</div>

`tkinter.messagebox` zeigt kleine Dialogfenster an -- für Fehlermeldungen, Warnungen, Informationen und Bestätigungen. Auch diese Dialoge sehen nativ aus und passen sich dem Betriebssystem an.

**Import:**

```python
from tkinter import messagebox
```

**Die wichtigsten Funktionen:**

| Funktion | Symbol | Verwendung |
|----------|--------|------------|
| `messagebox.showinfo(titel, text)` | ℹ | Reine Information, nur OK-Button |
| `messagebox.showwarning(titel, text)` | ⚠ | Warnung, nur OK-Button |
| `messagebox.showerror(titel, text)` | ✗ | Fehlermeldung, nur OK-Button |
| `messagebox.askyesno(titel, text)` | ? | Ja/Nein-Frage, gibt `True`/`False` zurück |
| `messagebox.askokcancel(titel, text)` | ? | OK/Abbrechen, gibt `True`/`False` zurück |

**Rückgabewerte bei Frage-Dialogen:** `askyesno` und `askokcancel` geben `True` zurück wenn der Nutzer zustimmt, `False` wenn er abbricht. Das macht die Abfrage sehr einfach:

```python
if messagebox.askyesno('Bestätigung', 'Wirklich löschen?'):
    datei_loeschen()
```

````python
# fenster_25_messagebox.py
import tkinter as tk
from tkinter import messagebox

def info():
    messagebox.showinfo('Information', 'Die Datei wurde gespeichert.')
    status.config(text='Info angezeigt.')

def warnung():
    messagebox.showwarning('Warnung', 'Die Datei existiert bereits.')
    status.config(text='Warnung angezeigt.')

def fehler():
    messagebox.showerror('Fehler', 'Datei konnte nicht geöffnet werden.')
    status.config(text='Fehler angezeigt.')

def ja_nein():
    antwort = messagebox.askyesno('Bestätigung', 'Wirklich alle Eintraege loeschen?')
    # antwort ist True (Ja) oder False (Nein)
    status.config(text=f'Antwort: {"Ja" if antwort else "Nein"}')

def ok_abbrechen():
    antwort = messagebox.askokcancel('Beenden', 'Anwendung wirklich beenden?')
    if antwort:
        root.quit()      # nur beenden wenn bestaetigt
    else:
        status.config(text='Beenden abgebrochen.')

root = tk.Tk()
root.title('messagebox')
root.geometry('320x240')

for text, fn in [
    ('showinfo',      info),
    ('showwarning',   warnung),
    ('showerror',     fehler),
    ('askyesno',      ja_nein),
    ('askokcancel',   ok_abbrechen),
]:
    tk.Button(root, text=text, command=fn, width=20).pack(pady=4)

status = tk.Label(root, text='')
status.pack(pady=5)

root.mainloop()
````

Alle Elemente aus Kapitel 11 in einem vollständigen Anwendungsgerüst: Menüleiste mit Dateidialog, Treeview für die Titelliste, Statuszeile, Bestätigungsdialoge, Schriften und Styles. Der Code hat noch keine echte Musikwiedergabe -- das kommt im MP3-Projekt -- aber er zeigt wie die GUI-Schicht aufgebaut ist.

````python
# fenster_26_mp3_geruest.py -- GUI-Grundgeruest fuer das MP3-Projekt
import tkinter as tk
from tkinter import ttk, filedialog, messagebox
import os

# -------------------------------------------------------
# Hilfsfunktionen (Platzhalter -- im Projekt ersetzt)
# -------------------------------------------------------

def titel_hinzufuegen():
    """Öffnet Dateidialog und fügt ausgewaehlte MP3s zur Liste hinzu."""
    pfade = filedialog.askopenfilenames(
        title='MP3-Dateien auswaehlen',
        filetypes=[('MP3-Dateien', '*.mp3'), ('Alle Audiodateien', '*.mp3 *.wav *.ogg')]
    )
    for pfad in pfade:
        dateiname = os.path.basename(pfad)    # nur den Dateinamen ohne Pfad
        name_ohne_endung = os.path.splitext(dateiname)[0]
        # Im echten Projekt wuerde man hier ID3-Tags auslesen
        baum.insert('', tk.END, values=(name_ohne_endung, 'Unbekannt', 'Unbekannt', '?:??'), tags=('ungerade' if len(baum.get_children()) % 2 else 'gerade',))
    status_var.set(f'{len(baum.get_children())} Titel in der Liste.')

def ausgewaehlt_loeschen():
    auswahl = baum.selection()
    if not auswahl:
        return
    if messagebox.askyesno('Löschen', f'{len(auswahl)} Eintrag/Eintraege aus der Liste entfernen?'):
        for item_id in auswahl:
            baum.delete(item_id)
        status_var.set(f'{len(baum.get_children())} Titel in der Liste.')

def alle_loeschen():
    if not baum.get_children():
        return
    if messagebox.askyesno('Alle loeschen', 'Alle Eintraege entfernen?'):
        baum.delete(*baum.get_children())
        status_var.set('Liste geleert.')

def abspielen_platzhalter():
    auswahl = baum.selection()
    if not auswahl:
        messagebox.showinfo('Hinweis', 'Bitte einen Titel auswaehlen.')
        return
    werte = baum.item(auswahl[0], 'values')
    status_var.set(f'Spiele: {werte[0]}')

def beenden():
    if messagebox.askokcancel('Beenden', 'Anwendung beenden?'):
        root.quit()

# -------------------------------------------------------
# Fenster
# -------------------------------------------------------
root = tk.Tk()
root.title('MP3-Player')
root.geometry('620x440')
root.minsize(480, 320)

# Style
style = ttk.Style()
style.configure('Treeview', font=('Segoe UI', 10), rowheight=24)
style.configure('Treeview.Heading', font=('Segoe UI', 10, 'bold'))

# -------------------------------------------------------
# Menüleiste
# -------------------------------------------------------
menubar = tk.Menu(root)
root.config(menu=menubar)

datei_menu = tk.Menu(menubar, tearoff=0)
menubar.add_cascade(label='Datei', menu=datei_menu)
datei_menu.add_command(label='Titel hinzufügen ...', command=titel_hinzufuegen)
datei_menu.add_separator()
datei_menu.add_command(label='Beenden', command=beenden)

bearbeiten_menu = tk.Menu(menubar, tearoff=0)
menubar.add_cascade(label='Bearbeiten', menu=bearbeiten_menu)
bearbeiten_menu.add_command(label='Ausgewaehlte loeschen', command=ausgewaehlt_loeschen)
bearbeiten_menu.add_command(label='Alle loeschen',         command=alle_loeschen)

# -------------------------------------------------------
# Toolbar
# -------------------------------------------------------
toolbar = tk.Frame(root, bg='#e8e8e8', height=38)
toolbar.pack(side='top', fill='x')
toolbar.pack_propagate(False)

tk.Button(toolbar, text='+ Hinzufügen', command=titel_hinzufuegen,
          relief='flat', bg='#e8e8e8').pack(side='left', padx=6, pady=5)
tk.Button(toolbar, text='▶ Abspielen',  command=abspielen_platzhalter,
          relief='flat', bg='#e8e8e8').pack(side='left', padx=4, pady=5)
tk.Button(toolbar, text='✕ Loeschen',   command=ausgewaehlt_loeschen,
          relief='flat', bg='#e8e8e8').pack(side='left', padx=4, pady=5)

# -------------------------------------------------------
# Treeview mit Scrollbar
# -------------------------------------------------------
inhalt = tk.Frame(root)
inhalt.pack(fill='both', expand=True, padx=8, pady=(6, 4))

sb = ttk.Scrollbar(inhalt, orient='vertical')
sb.pack(side='right', fill='y')

SPALTEN = ('titel', 'interpret', 'album', 'dauer')
baum = ttk.Treeview(inhalt, columns=SPALTEN, show='headings', yscrollcommand=sb.set)
sb.config(command=baum.yview)

for spalte, breite, text in [
    ('titel',    230, 'Titel'),
    ('interpret',150, 'Interpret'),
    ('album',    130, 'Album'),
    ('dauer',     70, 'Dauer'),
]:
    baum.heading(spalte, text=text)
    baum.column(spalte, width=breite, anchor='w' if spalte != 'dauer' else 'center')

baum.tag_configure('gerade',   background='#f5f5f5')
baum.tag_configure('ungerade', background='#ffffff')

baum.pack(side='left', fill='both', expand=True)

# Doppelklick = Abspielen
baum.bind('<Double-Button-1>', lambda e: abspielen_platzhalter())

# -------------------------------------------------------
# Statusleiste
# -------------------------------------------------------
statusleiste = tk.Frame(root, bg='#e0e0e0', height=22)
statusleiste.pack(side='bottom', fill='x')
statusleiste.pack_propagate(False)

status_var = tk.StringVar(value='Bereit. Titel über Datei > Titel hinzufügen laden.')
tk.Label(statusleiste, textvariable=status_var, bg='#e0e0e0',
         anchor='w', font=('Segoe UI', 9)).pack(side='left', padx=6)

root.mainloop()
````

**Zusammenfassung**

| Konzept | Erklärung |
|---------|------------|
| `font=('Arial', 12, 'bold')` | Schrift als Tupel: Name, Größe, Stil |
| `'TkDefaultFont'`, `'TkFixedFont'` | Plattformunabhängige Systemschriften |
| `ttk.Style()` | Style-Objekt für ttk-Widgets |
| `style.configure('Treeview', font=..., rowheight=...)` | Treeview-Schrift und Zeilenhöhe setzen |
| `style.configure('Treeview.Heading', ...)` | Überschriften-Zeile des Treeview |
| `baum.tag_configure('name', background='...')` | Tag mit Farbe definieren |
| `baum.insert(..., tags=('name',))` | Tag beim Einfügen zuweisen |
| `tk.Menu(root)` + `root.config(menu=...)` | Menüleiste erstellen und anhängen |
| `tk.Menu(menubar, tearoff=0)` | Untermenü -- `tearoff=0` immer setzen |
| `menubar.add_cascade(label='...', menu=...)` | Untermenü in Leiste einhängen |
| `menu.add_command(label='...', command=...)` | Menüeintrag hinzufügen |
| `menu.add_separator()` | Trennlinie im Menü |
| `from tkinter import filedialog` | Dateidialog-Modul |
| `filedialog.askopenfilename(filetypes=[...])` | Einzelne Datei auswählen |
| `filedialog.askopenfilenames(...)` | Mehrere Dateien auswählen |
| `filedialog.askdirectory(...)` | Ordner auswählen |
| `filedialog.asksaveasfilename(...)` | Speicherpfad auswählen |
| Rückgabe bei Abbruch | Immer `''` -- nicht `None` |
| `from tkinter import messagebox` | Meldungsdialog-Modul |
| `messagebox.showinfo/showwarning/showerror(titel, text)` | Reine Meldung |
| `messagebox.askyesno(titel, text)` | Ja/Nein, gibt `True`/`False` |
| `messagebox.askokcancel(titel, text)` | OK/Abbrechen, gibt `True`/`False` |
