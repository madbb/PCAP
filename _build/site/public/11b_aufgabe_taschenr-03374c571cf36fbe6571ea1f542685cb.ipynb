Jetzt kommt die eigentliche Logik. Schreibe eine Funktion `berechne(operator)` die aufgerufen wird wenn der Nutzer einen Operator-Button klickt.

**Was in diesem Schritt entsteht:** Der Taschenrechner rechnet. Klickt man auf `+`, werden die zwei Zahlen addiert und das Ergebnis erscheint im Fenster.

**Aufgaben:**

1. Erstelle ein Label das das Ergebnis anzeigt. Platziere es unterhalb der Buttons. Am Anfang ist es leer.
2. Schreibe die Funktion `berechne(operator)`. Sie soll:
   - Die Werte aus beiden Entry-Feldern mit `.get()` auslesen
   - Die Werte von String in Zahl umwandeln
   - Je nach `operator` die passende Rechenoperation ausführen
   - Das Ergebnis als vollständige Rechnung im Label anzeigen: `12 + 4 = 16`
3. Verbinde die Buttons mit der Funktion: Ersetze `lambda: None` durch den richtigen lambda-Aufruf.

**Hilfestellungen:**

> Aus 11b: `entry.get()` gibt immer einen String zurück. `float('12')` wandelt ihn in eine Zahl um. Warum `float` statt `int`? Weil der Nutzer auch Dezimalzahlen eingeben können soll.

> Aus 11b: `label.config(text='...')` aktualisiert den angezeigten Text eines Labels.

> Aus 11b: `command=lambda: berechne('+')` -- so übergibt man den Operator als Argument. Der `lambda f=farbe`-Trick aus dem Schleifenbeispiel ist hier nicht nötig, weil die Werte nicht aus einer Schleife kommen.

> Für die vier Operationen brauchst du keine externe Bibliothek. `a + b`, `a - b`, `a * b`, `a / b` reichen vollständig.

> Tipp für die Anzeige: f-Strings machen das einfach. `f'{a} {operator} {b} = {ergebnis}'` ergibt zum Beispiel `'12.0 + 4.0 = 16.0'`.

````python
def berechne(operator):
    try:
        a = float(eingabe1.get())
        b = float(eingabe2.get())
    except ValueError:
        ergebnis_label.config(
            text='Fehler bitte zwei gültige Zahlen eingeben',
            fg='red'
        )
        return

    if operator == ' / '  and b == 0:
        ergebnis_label.config(
            text='Fehler: Division durch null',
            fg='red'
        )
        return

    if operator == ' + ':
        ergebnis = a + b
    if operator == ' - ':
        ergebnis = a - b
    if operator == ' * ':
        ergebnis = a * b
    if operator == ' / ':
        ergebnis = a / b

    def fint(n):
        return int(n) if n == int(n) else n

    ergebnis_label.config(
        text=f'{fint(a)} {operator} {fint(b)} = {fint(ergebnis)}',
        fg='black'

    )

#---Ergebnis---

ergebnis_label = tk.Label(root, text ='',font=('Arial', 11))
ergebnis_label.pack(pady=6)


````
