**Frage 40**

Was ist die Ausgabe?
```python
import tempfile, os

with tempfile.NamedTemporaryFile(mode='w', suffix='.txt', delete=False) as f:
    f.write('hello')
    name = f.name

with open(name, 'r') as f:
    content = f.read()

os.unlink(name)
print(content)
```
A) `hello`

B) `None`

C) `''`

D) `FileNotFoundError`
