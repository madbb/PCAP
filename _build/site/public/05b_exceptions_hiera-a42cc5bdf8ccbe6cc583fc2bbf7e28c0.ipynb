1. Exceptions sind Klassen

In 05a haben wir Exceptions verwendet, ohne genau zu wissen was sie sind. Jetzt schauen wir genauer hin.

`ValueError`, `TypeError`, `ZeroDivisionError` -- das sind keine Sonderkonstrukte in Python. Das sind ganz normale **Klassen**, die von einer gemeinsamen Basisklasse erben. Und wenn eine Exception ausgeloest wird, wird ein Objekt dieser Klasse erstellt.

Das kann man direkt pruefen:

````python
# Exceptions sind Klassen
print(type(ValueError))       # <class 'type'> -- es ist eine Klasse
print(ValueError.__bases__)   # (<class 'Exception'>,) -- erbt von Exception

# Wenn eine Exception ausgeloest wird, entsteht ein Objekt
try:
    int('keine Zahl')
except ValueError as e:
    print(type(e))             # <class 'ValueError'> -- ein Objekt
    print(isinstance(e, ValueError))   # True
    print(isinstance(e, Exception))    # True -- Exception ist Elternklasse
    print(isinstance(e, BaseException)) # True -- noch hoeher
````

Das erklaert warum `except Exception` auch `ValueError` faengt -- `ValueError` ist eine Unterklasse von `Exception`, und `isinstance` gibt True zurueck. Python prueft bei jedem `except`-Block: ist das Exception-Objekt eine Instanz dieser Klasse oder einer Unterklasse?

2. Die Hierarchie -- wer erbt von wem
</span>

Die wichtigsten Klassen und ihre Beziehungen:

```
BaseException
  +-- SystemExit           (sys.exit() -- kein normaler Fehler)
  +-- KeyboardInterrupt    (Strg+C)
  +-- Exception            (alle normalen Fehler)
        +-- ArithmeticError
        |     +-- ZeroDivisionError
        |     +-- OverflowError
        +-- LookupError
        |     +-- IndexError
        |     +-- KeyError
        +-- ValueError
        +-- TypeError
        +-- AttributeError
        +-- NameError
        +-- OSError
              +-- FileNotFoundError
              +-- PermissionError
```

**BaseException vs Exception:** `BaseException` ist die Wurzel aller Exceptions. Aber `SystemExit` und `KeyboardInterrupt` sollen normalerweise nicht abgefangen werden -- sie sind kein Programmierfehler. Deshalb gilt die Faustregel: In eigenen `except`-Bloecken immer `Exception` oder spezifischere Klassen verwenden, nie `BaseException`.

````python
# Vererbungsstruktur live untersuchen
print(ZeroDivisionError.__mro__)
# ZeroDivisionError -> ArithmeticError -> Exception -> BaseException -> object

print(FileNotFoundError.__mro__)
# FileNotFoundError -> OSError -> Exception -> BaseException -> object

# Ein except fuer ArithmeticError faengt auch ZeroDivisionError und OverflowError
for fehler in [ZeroDivisionError, OverflowError, ValueError]:
    try:
        raise fehler('Test')
    except ArithmeticError:
        print(f'{fehler.__name__} -- von ArithmeticError gefangen')
    except Exception:
        print(f'{fehler.__name__} -- nicht ArithmeticError, faellt durch zu Exception')
````

3. except E as e -- das Exception-Objekt
</span>

Mit `except ValueError as e` bekommt man Zugriff auf das Exception-Objekt selbst. Weil Exceptions Klassen sind, hat das Objekt Attribute und Methoden -- genau wie jedes andere Objekt.

````python
try:
    int('keine Zahl')
except ValueError as e:
    print(f'Typ:         {type(e).__name__}')
    print(f'Meldung:     {e}')          # str(e) -- die lesbare Fehlermeldung
    print(f'repr:        {repr(e)}')    # repr(e) -- technische Darstellung
    print(f'args:        {e.args}')     # Tupel der Argumente
````

4. args -- das Herzstuck des Exception-Objekts
</span>
</div>

Jede Exception hat ein Attribut `args` -- ein **Tupel** der Argumente die beim Erstellen uebergeben wurden. Das ist der Schluessel zum Verstaendnis wie `print(e)` und `str(e)` funktionieren.

Intern macht Python bei `str(e)` folgendes:
- Wenn `args` ein Tupel mit **einem** Element ist: gibt dieses Element als String zurueck
- Wenn `args` mehrere Elemente hat: gibt das gesamte Tupel als String zurueck
- Wenn `args` leer ist: gibt einen leeren String zurueck

````python
# args verstehen
e1 = ValueError('eine Meldung')
e2 = ValueError('erstes', 'zweites')   # zwei Argumente!
e3 = ValueError()                       # kein Argument

print(e1.args)   # ('eine Meldung',)   -- Tupel mit einem Element
print(e2.args)   # ('erstes', 'zweites') -- Tupel mit zwei Elementen
print(e3.args)   # ()                   -- leeres Tupel

print()

# Was macht str() / print() damit?
print(str(e1))   # 'eine Meldung'       -- ein Element: direkt
print(str(e2))   # "('erstes', 'zweites')" -- mehrere: Tupel als String
print(str(e3))   # ''                    -- leer: leerer String
````

````python
# args ist direkt erreichbar -- einzelne Werte auslesbar
try:
    raise ValueError('Ungueltige HP', -50, 'Feld: hp')
except ValueError as e:
    print(f'Meldung:  {e.args[0]}')
    print(f'Wert:     {e.args[1]}')
    print(f'Feld:     {e.args[2]}')
    print(f'Alle:     {e.args}')
    print(f'print(e): {e}')       # gibt das gesamte Tupel aus, weil 3 Elemente
````

Das Wissen ueber `args` ist wichtig fuer eigene Exception-Klassen -- denn in `__init__` entscheidet man, was in `args` landet, und damit was `print(e)` anzeigt. Das schauen wir uns in Notebook 05c genau an.

5. Exception-Klassen als Event-Klassen
</span>
</div>

Der PCAP-Syllabus nennt den Begriff **"event classes"** -- damit sind Exception-Klassen gemeint. Dieser Begriff betont dass Exceptions nicht nur Fehler signalisieren koennen, sondern allgemein **Ereignisse** -- auch solche die kein Programmfehler sind.

Ein Beispiel: In manchen Architekturen nutzt man Exceptions um Kontrollfluss zu signalisieren -- z.B. "Spieler hat das Spiel beendet" oder "Kein weiteres Element vorhanden":

````python
# Exception als Ereignis-Signal -- nicht unbedingt ein Fehler
class SpielEnde(Exception):
    """Wird ausgeloest wenn der Spieler das Spiel beenden moechte."""
    pass

class HeldGestorben(Exception):
    """Wird ausgeloest wenn ein Held stirbt."""
    def __init__(self, heldenname):
        self.heldenname = heldenname
        super().__init__(f'{heldenname} ist gefallen')


def kampfrunde(held_hp, gegner_schaden):
    held_hp -= gegner_schaden
    if held_hp <= 0:
        raise HeldGestorben('Aldric')   # Ereignis signalisieren
    return held_hp


try:
    hp = 100
    hp = kampfrunde(hp, 40)
    print(f'HP nach Runde 1: {hp}')
    hp = kampfrunde(hp, 40)
    print(f'HP nach Runde 2: {hp}')
    hp = kampfrunde(hp, 40)   # toedlich!
except HeldGestorben as e:
    print(f'Ereignis: {e}')   # 'Aldric ist gefallen'
    print(f'Held:     {e.heldenname}')
````

**Zusammenfassung**

| Konzept | Kernaussage |
|---------|-------------|
| Exceptions sind Klassen | `ValueError`, `TypeError` usw. sind normale Python-Klassen |
| `except` prueft `isinstance` | Ein `except ArithmeticError` faengt auch Unterklassen wie `ZeroDivisionError` |
| `BaseException` | Wurzel aller Exceptions -- in `except` fast nie verwenden |
| `Exception` | Basis aller normalen Fehler -- haeufigster Fallback |
| `args` | Tupel der Konstruktor-Argumente -- bestimmt was `str(e)` anzeigt |
| `str(e)` bei einem arg | Gibt das einzelne Element direkt zurueck |
| `str(e)` bei mehreren args | Gibt das gesamte Tupel als String zurueck |
| Event-Klassen | Exceptions koennen Ereignisse signalisieren -- nicht nur Fehler |
