2. isxxx() -- Pruefmethoden
</span>
</div>

Stell dir vor, du verarbeitest Benutzereingaben. Bevor du eine Eingabe als Zahl weiterverarbeitest, willst du sicherstellen dass sie wirklich eine Zahl ist. Bevor du einen Namen speicherst, pruefst du ob er nur Buchstaben enthaelt.

Dafuer gibt es die `is`-Methoden -- sie geben immer `True` oder `False` zurueck:

| Methode | Gibt True wenn... |
|---------|------------------|
| `.isalpha()` | alle Zeichen Buchstaben sind |
| `.isdigit()` | alle Zeichen Dezimalziffern sind |
| `.isalnum()` | alle Zeichen Buchstaben oder Ziffern sind |
| `.islower()` | alle Buchstaben Kleinbuchstaben sind |
| `.isupper()` | alle Buchstaben Grossbuchstaben sind |
| `.isspace()` | alle Zeichen Leerzeichen, Tab oder Zeilenumbruch sind |
| `.istitle()` | jedes Wort mit Grossbuchstaben beginnt |

Wichtig: Ein **leerer String** gibt bei allen `is`-Methoden `False` zurueck.

````python
# Eingabe validieren -- Praxisbeispiel
def verarbeite_eingabe(eingabe):
    if not eingabe:            # leerer String
        return 'Leer'
    if eingabe.isdigit():
        return f'Zahl: {int(eingabe)}'
    if eingabe.isalpha():
        return f'Wort: {eingabe}'
    return f'Gemischt: {eingabe}'

print(verarbeite_eingabe('42'))      # Zahl: 42
print(verarbeite_eingabe('Hallo'))   # Wort: Hallo
print(verarbeite_eingabe('H3llo'))   # Gemischt: H3llo
print(verarbeite_eingabe(''))        # Leer

# Alle is-Methoden im Ueberblick
print('abc'.isalpha())        # True
print('abc1'.isalpha())       # False -- Ziffer enthalten
print('123'.isdigit())        # True
print('abc123'.isalnum())     # True
print('hello'.islower())      # True
print('Hello World'.istitle())# True
print('   \t\n'.isspace())    # True
````

````python
# isdigit() prueft ob JEDES Zeichen eine Dezimalziffer ist
print('123'.isdigit())    # True
print('-5'.isdigit())     # False -- Minuszeichen ist keine Ziffer
print('3.14'.isdigit())   # False -- Punkt ist keine Ziffer
print(''.isdigit())       # False -- leerer String

# Hoechgestellte Zahlen sind technisch Ziffern
print('²'.isdigit())      # True  -- Unicode-Hochzahl
print('½'.isdigit())      # False -- Bruch ist keine Ziffer
print('½'.isnumeric())    # True  -- isnumeric() ist weiter gefasst

# Wenn man wirklich eine gueltige ganze Zahl pruefen will:
def ist_ganze_zahl(text):
    try:
        int(text)
        return True
    except ValueError:
        return False

print(ist_ganze_zahl('-5'))    # True
print(ist_ganze_zahl('3.14'))  # False
````

3. split() und join() -- String und Liste ineinander umwandeln
</span>
</div>

Viele Datenformate kommen als Text an -- CSV-Zeilen, Pfadangaben, Konfigurationsstrings. Um damit zu arbeiten, muss man den Text aufteilen. Und umgekehrt: wenn man Daten als Text ausgeben will, muss man eine Liste wieder zusammensetzen.

`split()` und `join()` sind ein Paar -- sie sind Umkehroperationen.

````python
# split() -- String in Liste zerlegen

# Ohne Argument: trennt an Leerzeichen (inkl. Tab und Zeilenumbruch)
satz = 'Python ist eine tolle Sprache'
print(satz.split())           # ['Python', 'ist', 'eine', 'tolle', 'Sprache']

# Mit Trennzeichen
csv = 'Aldric,Krieger,100,15'
print(csv.split(','))         # ['Aldric', 'Krieger', '100', '15']

# Mit maxsplit: maximale Anzahl Splits
print('a:b:c:d'.split(':', 2))  # ['a', 'b', 'c:d'] -- nur 2 Splits

# Achtung: mehrere Trennzeichen hintereinander erzeugen leere Strings
print('a,,b'.split(','))         # ['a', '', 'b']
````

````python
# join() -- Liste zu String zusammenfuegen
# Syntax: 'trennzeichen'.join(liste)

helden = ['Aldric', 'Lyria', 'Thorn']
print(', '.join(helden))     # Aldric, Lyria, Thorn
print(' und '.join(helden))  # Aldric und Lyria und Thorn
print(''.join(helden))       # AldricLyriaThorn

# split und join sind Umkehroperationen
original  = 'Python ist toll'
geteilt   = original.split()
zusammen  = ' '.join(geteilt)
print(original == zusammen)   # True
````

4. sort() und sorted()
</span>
</div>

`sort()` und `sorted()` stehen im PCAP-Syllabus explizit im String-Abschnitt -- weil sie haeufig auf Listen von Strings angewendet werden. Der zentrale Unterschied:

- `list.sort()` -- sortiert die Liste **direkt** (in-place) und gibt `None` zurueck
- `sorted(iterable)` -- gibt eine **neue** sortierte Liste zurueck, Original unveraendert

````python
helden = ['Thorn', 'Aldric', 'Lyria', 'Mira']

# sorted() -- neue Liste, Original bleibt
sortiert = sorted(helden)
print('sortiert: ', sortiert)   # ['Aldric', 'Lyria', 'Mira', 'Thorn']
print('original: ', helden)     # unveraendert

# sort() -- aendert die Liste selbst
helden.sort()
print('nach sort():', helden)   # jetzt sortiert

# Absteigend
print(sorted(helden, reverse=True))

# Gross-/Kleinschreibung ignorieren
gemischt = ['thorn', 'Aldric', 'LYRIA']
print(sorted(gemischt, key=str.lower))  # ['Aldric', 'LYRIA', 'thorn']
````

5. find(), rfind() und index() -- Suchen in Strings
</span>
</div>

Alle drei suchen einen Teilstring und geben seine Position zurueck. Der Unterschied liegt darin, was passiert wenn nichts gefunden wird -- und das ist in der Pruefung der entscheidende Punkt:

| Methode | Nicht gefunden | Sucht von |
|---------|---------------|----------|
| `.find(sub)` | gibt `-1` zurueck | Links |
| `.rfind(sub)` | gibt `-1` zurueck | Rechts (letztes Vorkommen) |
| `.index(sub)` | wirft `ValueError` | Links |

````python
text = 'Python ist toll und Python macht Spass'

# find() -- erstes Vorkommen, -1 wenn nicht gefunden
print(text.find('Python'))    # 0  -- Position des ersten 'Python'
print(text.find('Java'))      # -1 -- nicht gefunden

# rfind() -- letztes Vorkommen
print(text.rfind('Python'))   # 19 -- zweites 'Python'

# Mit Start- und Endposition suchen
print(text.find('Python', 5))      # 19 -- sucht ab Position 5
print(text.find('Python', 5, 15))  # -1 -- nur in Bereich 5-15 suchen
````

````python
text = 'Python ist toll'

# index() -- wie find(), aber Exception statt -1
print(text.index('ist'))    # 7

try:
    text.index('Java')      # ValueError!
except ValueError as e:
    print(f'ValueError: {e}')

# Wann welche verwenden?
# find(): wenn unsicher ob der String vorkommt -- pruefen mit != -1
# index(): wenn der String definitiv vorhanden sein muss -- Exception zeigt den Fehler

email = 'benutzer@example.com'
if email.find('@') != -1:           # sicherer Check
    print('Gueltige E-Mail')

protokoll = 'GET /index.html HTTP/1.1'
pos = protokoll.index('HTTP')       # muss vorhanden sein
print(protokoll[pos:])
````

6. Weitere nuetzliche Methoden
</span>
</div>

````python
# Gross-/Kleinschreibung
print('python'.upper())            # PYTHON
print('PYTHON'.lower())            # python
print('hallo welt'.capitalize())   # Hallo welt  -- nur erstes Zeichen
print('hallo welt'.title())        # Hallo Welt  -- jedes Wort
print('Hallo'.swapcase())          # hALLO

# Leerzeichen entfernen
text = '  Hallo Welt  '
print(repr(text.strip()))    # 'Hallo Welt'   -- beidseitig
print(repr(text.lstrip()))   # 'Hallo Welt  ' -- nur links
print(repr(text.rstrip()))   # '  Hallo Welt' -- nur rechts

# Ersetzen und pruefen
print('Hallo Welt'.replace('Welt', 'Python'))  # Hallo Python
print('Hallo Welt'.replace('l', 'L'))          # HaLLo WeLt -- alle Vorkommen
print('Hallo'.startswith('Ha'))  # True
print('Hallo'.endswith('lo'))    # True
print('Hallo'.count('l'))        # 2  -- Anzahl Vorkommen

# Ausrichten und Auffuellen
print('42'.zfill(5))             # 00042  -- mit Nullen auffuellen
print('Hi'.center(10, '-'))      # ----Hi----
print('Hi'.ljust(10, '.'))       # Hi........
print('Hi'.rjust(10, '.'))       # ........Hi
````

**Zusammenfassung**

| Methode | PCAP | Kernaussage |
|---------|------|-------------|
| `.isalpha()`, `.isdigit()` usw. | 3.3 | Geben `True`/`False` -- leerer String immer `False` |
| `.isdigit()` | 3.3 | Erkennt keine negativen Zahlen oder Kommazahlen |
| `.split(sep)` | 3.3 | String → Liste; Standard-Trenner: Leerzeichen |
| `sep.join(liste)` | 3.3 | Liste → String; Aufruf auf dem Trennzeichen |
| `list.sort()` | 3.3 | In-place, gibt `None` zurueck |
| `sorted(iterable)` | 3.3 | Neue Liste, Original unveraendert |
| `.find(sub)` | 3.3 | Position oder `-1` |
| `.rfind(sub)` | 3.3 | Letztes Vorkommen oder `-1` |
| `.index(sub)` | 3.3 | Position oder `ValueError` |
