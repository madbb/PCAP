**4.1**
```python
def erstelle_zaehler():
    n = 0
    def naechster():
        nonlocal n
        n += 1
        return n
    return naechster

z = erstelle_zaehler()
print(z())
print(z())
print(z())
```

**4.2**
```python
def erstelle_zaehler():
    n = 0
    def naechster():
        nonlocal n
        n += 1
        return n
    return naechster

a = erstelle_zaehler()
b = erstelle_zaehler()
print(a())
print(a())
print(b())
print(a())
```

**4.3**
```python
def f():
    x = 10
    def g():
        x = 20   # kein nonlocal!
        return x
    return x, g()

print(f())
```
