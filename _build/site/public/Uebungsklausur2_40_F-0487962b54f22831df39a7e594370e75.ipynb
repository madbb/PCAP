**Frage 25**

```python
class Animal:
    def __init__(self, name):
        self.name = name

class Cat(Animal):
    def __init__(self, name, indoor):
        super().__init__(name)
        self.indoor = indoor

c = Cat('Mimi', True)
print(c.name, c.indoor)
```

**Frage 26**

Was ist die Ausgabe?
```python
class Foo:
    def method(self):
        return 'Foo'

class Bar(Foo):
    def method(self):
        return super().method() + 'Bar'

class Baz(Bar):
    pass

print(Baz().method())
```
A) `Baz`

B) `FooBar`

C) `BarFoo`

D) `Bar`

**Frage 27**

Was gibt der Code aus? (MRO / Mehrfachvererbung)
```python
class Left:
    def who(self):
        return 'Left'

class Right:
    def who(self):
        return 'Right'

class Child(Left, Right):
    pass

print(Child().who())
```
A) `Left`

B) `Right`

C) `Left Right`

D) `AttributeError`

**Frage 28**

Was ist die Ausgabe?
```python
class Counter:
    _count = 0

    def __init__(self):
        Counter._count += 1

    @classmethod
    def get_count(cls):
        return cls._count

Counter()
Counter()
print(Counter.get_count())
```
A) `0`

B) `1`

C) `2`

D) `3`
