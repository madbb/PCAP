Einleitung: Warum Module und Pakete?</span></div>

Stellt euch vor, eure Python-Projekte wachsen und werden immer komplexer. Irgendwann wird es echt unübersichtlich, den gesamten Code in einer einzigen Datei zu haben. Das führt meistens zu ein paar typischen Problemen:

- **Lesbarkeitsprobleme**: Lange Dateien sind wie ein Labyrinth – schwer zu durchblicken und zu verstehen.
- **Wartungsprobleme**: Wenn ihr an einer Stelle etwas ändert, kann das ungewollt andere Teile eures Codes durcheinanderbringen.
- **Mangelnde Wiederverwendbarkeit**: Code, den ihr für ein Projekt geschrieben habt, lässt sich nur mühsam in einem anderen Projekt wiederverwenden.

Pythons Antwort auf diese Herausforderungen sind **Module** und **Pakete**.

**Modul:** Eine einzelne `.py`-Datei. Darin könnt ihr Python-Definitionen (Funktionen, Klassen, Variablen) und Anweisungen sammeln. Module helfen dabei, euren Code logisch zu gliedern -- wie Kapitel in einem Buch.

**Paket (Package):** Ein Ordner, der eine Sammlung von Modulen enthält, die thematisch zusammengehören. Diese Ordner können weitere Unterordner (Unterpakete) enthalten. Die einzelnen Teile sprecht ihr über Punktnotation an: `paket.modul.funktion`.

Die Verwendung von Modulen und Paketen bietet folgende Vorteile: Code-Organisation, Wiederverwendbarkeit, Namensraumverwaltung (Namenskonflikte vermeiden) und bessere Zusammenarbeit im Team.

Erstellen und Verwenden eines einfachen Moduls</span></div>

Jede Python-Datei, die ihr erstellt, kann im Prinzip schon als Modul dienen. Lasst uns das direkt ausprobieren und ein eigenes kleines Modul basteln, das uns ein paar nützliche Funktionen bereitstellt.

Stellt euch vor, wir brauchen ein paar einfache mathematische Helferlein und eine Begrüßungsfunktion. Erstellt eine Datei namens `meine_helfer.py` mit folgendem Code:

```python
# meine_helfer.py
print("Modul meine_helfer wird importiert!")

PI_NAHE = 3.14159
EULER_ZAHL_NAHE = 2.71828

def addieren(x, y):
    """Gibt die Summe von x und y zurück."""
    print(f"  [meine_helfer] Addiere: {x} + {y}")
    return x + y

def begruesse(name):
    """Gibt eine persönliche Begrüßung zurück."""
    return f"Hallo {name}, willkommen im Modul!"

# Dieser Teil wird nur ausgeführt, wenn ihr die Datei direkt startet,
# nicht, wenn sie als Modul importiert wird.
if __name__ == "__main__":
    print()
    print("meine_helfer.py wird direkt als Skript ausgeführt.")
    
    test_summe = addieren(5, 3)
    print(f"Testergebnis für addieren(5, 3): {test_summe}")
    
    test_begruessung = begruesse("Python-Lerner")
    print(test_begruessung)
    print(f"Der Wert von PI_NAHE ist: {PI_NAHE}")
```

Nun wollen wir unser Modul selber aufrufen:

````python
print("Starte hauptanwendung.py...")
import meine_helfer # Achtung: Die print-Anweisung oben im Modul wird jetzt ausgeführt!

print("Zugriff auf Elemente aus meine_helfer")
print(f"Der Näherungswert für Pi ist: {meine_helfer.PI_NAHE}")
print(f"Die Euler'sche Zahl ist etwa: {meine_helfer.EULER_ZAHL_NAHE}")

summe1 = meine_helfer.addieren(10, 5)
print(f"Das Ergebnis von 10 + 5 ist: {summe1}")

summe2 = meine_helfer.addieren(meine_helfer.PI_NAHE, meine_helfer.EULER_ZAHL_NAHE)
print(f"Pi + e ist ungefähr: {summe2}")

willkommen = meine_helfer.begruesse("Tim")
print(willkommen)

print("... hauptanwendung.py beendet.")
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Was macht <code style="background:rgba(79,195,247,0.1);color:#4fc3f7;padding:0.1em 0.3em;border-radius:3px;">if __name__ == "__main__"</code>?</span></div>

Diese Zeile sieht erstmal kompliziert aus, ist aber sehr nützlich! Sie hilft dabei, zu unterscheiden, ob eine Datei direkt gestartet oder nur als Modul importiert wurde.

`__name__` ist eine eingebaute Variable in jeder `.py`-Datei. Der Wert hängt davon ab, wie die Datei genutzt wird:

**1. Direktes Ausführen** (`python meine_helfer.py` oder als Notebook in Jupyter) → `__name__` ist automatisch `"__main__"`. Bedeutet: Die Datei wird gerade direkt gestartet.

**2. Importieren als Modul** (`import meine_datei`) → `__name__` ist der Modulname, also z. B. `"meine_helfer"`. Bedeutet: Die Datei ist nur ein Hilfsmodul im Hintergrund.

```python
if __name__ == "__main__":
    # Wird die Datei direkt gestartet? → Dann führe diesen Code aus.
    # Wird sie nur importiert? → Dann überspringe diesen Teil.
    test_summe = addieren(5, 3)
    print(f"Testergebnis für addieren(5, 3): {test_summe}")
```

Aliase -- kurze Namen fuer lange Pfade</span></div>

Wir haben `import math as m` schon gesehen. Aliase können aber noch mehr: Ihr könnt auch einzelne Funktionen beim Import umbenennen.

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Warum Aliase?</span></div>

- Lange Modulnamen würden den Code unlesbar machen (z. B. `numpy`, `pandas`, `matplotlib.pyplot`)
- Bestimmte Aliase sind Konventionen, die alle Python-Entwicklerinnen und -Entwickler kennen
- Ihr könnt Namenskollisionen vermeiden, indem ihr eine importierte Funktion umbenennt

````python
# Modul-Alias: der Originalname ist danach NICHT mehr verfuegbar
import math as m

print(m.sqrt(9))   # funktioniert
# print(math.sqrt(9))  # NameError! math existiert nicht mehr im Namespace
````

````python
# Funktions-Alias: sehr nuetzlich wenn ihr eine eigene Funktion gleichen Namens habt
from math import sqrt as math_sqrt

def sqrt(x):
    """Eigene Version: prueft auf negative Eingabe."""
    if x < 0:
        raise ValueError(f"Negative Wurzel nicht definiert: {x}")
    return math_sqrt(x)

print(sqrt(25))       # nutzt unsere Funktion
print(math_sqrt(25))  # nutzt math.sqrt direkt
````

Wenn ihr `import meine_helfer` schreibt, muss Python wissen, wo es suchen soll. Die Antwort steckt in `sys.path` -- einer Liste von Verzeichnissen, die Python der Reihe nach durchsucht.

````python
import sys

# sys.path zeigt, wo Python nach Modulen sucht -- in dieser Reihenfolge!
for pfad in sys.path:
    print(pfad if pfad else '(aktuelles Verzeichnis)')
````

Das eigene Modul wirklich importieren</span></div>

Oben haben wir `meine_helfer.py` als Markdown-Codeblock gezeigt. Jetzt erstellen wir die Datei tatsächlich und importieren sie -- damit ihr den ganzen Ablauf live seht.

````python
# Schritt 1: Datei meine_helfer.py im aktuellen Verzeichnis anlegen

modul_code = '''
# meine_helfer.py
print("Modul meine_helfer wird importiert!")

PI_NAHE = 3.14159
EULER_ZAHL_NAHE = 2.71828

def addieren(x, y):
    """Gibt die Summe von x und y zurueck."""
    print(f"  [meine_helfer] Addiere: {x} + {y}")
    return x + y

def begruesse(name):
    """Gibt eine persoenliche Begrueszung zurueck."""
    return f"Hallo {name}, willkommen im Modul!"

if __name__ == "__main__":
    print()
    print("meine_helfer.py wird direkt als Skript ausgefuehrt.")
    test_summe = addieren(5, 3)
    print(f"Testergebnis fuer addieren(5, 3): {test_summe}")
    test_begruessung = begruesse("Python-Lerner")
    print(test_begruessung)
    print(f"Der Wert von PI_NAHE ist: {PI_NAHE}")
'''

with open('meine_helfer.py', 'w', encoding='utf-8') as f:
    f.write(modul_code)

print("meine_helfer.py wurde erstellt!")
````

````python
# Schritt 2: Modul importieren -- die print-Anweisung im Modul laeuft jetzt!
import meine_helfer

print("\nZugriff auf Elemente aus meine_helfer:")
print(f"PI_NAHE:       {meine_helfer.PI_NAHE}")
print(f"EULER_ZAHL:    {meine_helfer.EULER_ZAHL_NAHE}")
print(f"addieren:      {meine_helfer.addieren(10, 5)}")
print(f"begruesse:     {meine_helfer.begruesse('Tim')}")
````

Beachtet: Das `print("Modul meine_helfer wird importiert!")` läuft **nur einmal**, egal wie oft ihr `import meine_helfer` schreibt. Python speichert importierte Module in einem internen Cache (`sys.modules`) und lädt sie nicht ein zweites Mal.

````python
# Zweiter Import -- kein erneutes Ausfuehren des Modulcodes!
import meine_helfer
print("Zweiter Import: kein Print-Aufruf aus dem Modul")

import sys
print("Im Cache?", 'meine_helfer' in sys.modules)  # True
````

Pakete -- Module in Ordnern</span></div>

Wenn ein Projekt größer wird, reicht eine einzelne Moduldatei irgendwann nicht mehr. Pakete lösen das: Ein Paket ist einfach ein **Ordner mit einer `__init__.py`-Datei** darin. Diese Datei -- die auch leer sein darf -- signalisiert Python: *Dieser Ordner ist ein Paket.*

Eine typische Paketstruktur sieht so aus:

```
mein_paket/
    __init__.py       <-- macht den Ordner zum Paket
    rechner.py
    texte.py
```

````python
import os

# Ordnerstruktur anlegen
os.makedirs('mein_paket', exist_ok=True)

# __init__.py -- kann leer sein oder Paket-Infos enthalten
with open('mein_paket/__init__.py', 'w', encoding='utf-8') as f:
    f.write('# Paket mein_paket\n')
    f.write('VERSION = "1.0"\n')

# rechner.py
with open('mein_paket/rechner.py', 'w', encoding='utf-8') as f:
    f.write('def addiere(a, b):\n    return a + b\n\n')
    f.write('def subtrahiere(a, b):\n    return a - b\n')

# texte.py
with open('mein_paket/texte.py', 'w', encoding='utf-8') as f:
    f.write('def begruesse(name):\n    return f"Hallo {name}!"\n')

print("Paket erstellt:")
for root, dirs, files in os.walk('mein_paket'):
    for datei in files:
        print(f"  {root}/{datei}")
````

````python
# Paket importieren -- greift auf __init__.py zu
import mein_paket
print(mein_paket.VERSION)

# Untermodul importieren
from mein_paket import rechner
print(rechner.addiere(4, 6))

# Direkt aus einem Untermodul importieren
from mein_paket.texte import begruesse
print(begruesse("Kurs"))
````

Die einzelnen Teile sprecht ihr über **Punktnotation** an: `paket.modul.funktion`.

Ohne `__init__.py` erkennt Python den Ordner nicht als Paket -- der Import schlägt dann mit einem `ModuleNotFoundError` fehl.

Was ist <code style="color:#4fc3f7;font-size:0.9em;">__pycache__</code>?</span></div>

Nach dem ersten Import eines eigenen Moduls werdet ihr in eurem Verzeichnis einen neuen Ordner namens `__pycache__` sehen. Der wurde nicht von euch angelegt -- Python hat das automatisch gemacht.

````python
import os

if os.path.exists('__pycache__'):
    print("__pycache__ existiert. Inhalt:")
    for datei in os.listdir('__pycache__'):
        print(f"  {datei}")
else:
    print("Noch kein __pycache__ vorhanden")
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Was steckt darin?</span></div>

Python übersetzt euren Quellcode (`.py`) beim ersten Import in sogenannten **Bytecode** (`.pyc`). Der Bytecode wird in `__pycache__` gespeichert. Beim nächsten Import prüft Python:

- Hat sich die `.py`-Datei verändert? → Dann neu übersetzen.
- Unverändert? → Bytecode direkt laden. Das ist schneller.

Ihr müsst `__pycache__` **nicht in euer Git-Repository** aufnehmen -- es wird automatisch neu erzeugt. Tragt es einfach in eure `.gitignore` ein:

```
__pycache__/
*.pyc
```
