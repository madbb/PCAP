In Notebook 03a haben wir gesehen: Funktionen sind Objekte, sie koennen Variablen zugewiesen und in Datenstrukturen gespeichert werden.

Am Ende haben wir eine Funktion `wende_an()` geschrieben, die eine andere Funktion als Argument entgegennimmt:

```python
def wende_an(funktion, werte):
    return [funktion(wert) for wert in werte]
```

Eine Funktion, die **eine andere Funktion als Argument annimmt oder eine Funktion zurueckgibt**, nennt man **hoeherwertige Funktion** (Higher-Order Function).

Dieses Notebook zeigt, warum das so nuetzlich ist -- und welche hoeherwertigen Funktionen Python direkt mitbringt.

2. sorted() -- ein bekanntes Beispiel neu betrachtet

Du kennst `sorted()` bereits. Aber hast du dich gefragt, was der `key`-Parameter eigentlich ist?

`key` erwartet eine **Funktion**. Vor dem Vergleich zweier Elemente ruft `sorted()` diese Funktion auf jedem Element auf und vergleicht dann die Ergebnisse -- nicht die Originale.

Ohne `key` wird direkt verglichen. Mit `key` wird erst transformiert, dann verglichen.

````python
woerter = ['Banane', 'apfel', 'Kirsche', 'dattel']

# Ohne key: alphabetisch nach Unicode -- Grossbuchstaben kommen vor Kleinbuchstaben
print(sorted(woerter))
# ['Banane', 'Kirsche', 'apfel', 'dattel']  <-- B und K vor a und d

# Mit key=str.lower: erst alle in Kleinbuchstaben umwandeln, dann vergleichen
print(sorted(woerter, key=str.lower))
# ['apfel', 'Banane', 'dattel', 'Kirsche']  <-- jetzt intuitiv alphabetisch
````

`str.lower` ist hier eine Funktion -- ohne Klammern, denn wir wollen nicht `str.lower()` aufrufen, sondern `sorted()` die Funktion uebergeben, damit es sie selbst aufrufen kann.

Man kann jede Funktion als `key` uebergeben -- auch eigene:

````python
# Eigene Funktion als key
def letzter_buchstabe(wort):
    return wort[-1]

woerter = ['Banane', 'Apfel', 'Kirsche', 'Dattel']
print(sorted(woerter, key=letzter_buchstabe))
# Sortiert nach dem letzten Buchstaben jedes Wortes
````

````python
# Auch bei komplexen Daten sehr nuetzlich
personen = [
    {'name': 'Alice', 'alter': 30},
    {'name': 'Bob',   'alter': 25},
    {'name': 'Carol', 'alter': 35},
]

# Nach Alter sortieren -- key bekommt ein Dictionary, gibt das Alter zurueck
def nach_alter(person):
    return person['alter']

print(sorted(personen, key=nach_alter))

# min() und max() funktionieren genauso
juengste = min(personen, key=nach_alter)
print(f"Juengste Person: {juengste['name']}, {juengste['alter']} Jahre")
````

Nicht nur Python-Builtins koennen hoeherwertig sein -- wir koennen selbst solche Funktionen schreiben. Der Syllabus nennt das explizit: *"self-defined functions taking lambdas as arguments"*.

Wir bauen eine Verarbeitungspipeline: eine Funktion, die eine Liste von Daten entgegennimmt und eine Funktion, die bestimmt was mit jedem Element passiert.

````python
def verarbeite_liste(daten, transformation):
    """
    Wendet eine Transformationsfunktion auf jedes Element an.
    Die aufrufende Stelle bestimmt, was die Transformation tut.
    """
    return [transformation(element) for element in daten]

temperaturen_celsius = [0, 20, 37, 100]

# Umrechnung in Fahrenheit
def celsius_zu_fahrenheit(c):
    return c * 9/5 + 32

print(verarbeite_liste(temperaturen_celsius, celsius_zu_fahrenheit))
# [32.0, 68.0, 98.6, 212.0]

# Dieselbe Funktion, andere Transformation
print(verarbeite_liste(temperaturen_celsius, lambda c: round(c * 9/5 + 32, 1)))
print(verarbeite_liste(temperaturen_celsius, str))
````

````python
# Noch ein Beispiel: Funktion die filtert und transformiert
def filtere_und_wandle(daten, bedingung, transformation):
    """
    Behaelt nur Elemente die die Bedingung erfuellen
    und wendet dann die Transformation an.
    """
    return [transformation(x) for x in daten if bedingung(x)]

zahlen = [-4, -2, 0, 1, 3, 5, 8]

# Nur positive Zahlen, quadriert
ergebnis = filtere_und_wandle(
    zahlen,
    bedingung=lambda x: x > 0,
    transformation=lambda x: x ** 2
)
print(ergebnis)  # [1, 9, 25, 64]

# Nur gerade Zahlen, als String
ergebnis2 = filtere_und_wandle(
    zahlen,
    bedingung=lambda x: x % 2 == 0,
    transformation=str
)
print(ergebnis2)  # ['-4', '-2', '0', '8']
````

</span>

Hoeherwertige Funktionen koennen auch Funktionen **zurueckgeben**. Das hoert sich zunachst seltsam an -- warum sollte man so etwas brauchen?

Ein konkretes Szenario: Du moechtest mehrere Validierungsfunktionen erstellen, die alle nach demselben Muster funktionieren -- pruefe ob ein Wert in einem bestimmten Bereich liegt. Statt jede einzeln zu schreiben, schreibt man eine Funktion die Validatoren **erzeugt**:

````python
def erstelle_bereichspruefer(minimum, maximum):
    """
    Gibt eine Funktion zurueck, die prueft ob ein Wert
    zwischen minimum und maximum liegt.
    """
    def pruefer(wert):
        return minimum <= wert <= maximum
    return pruefer

# Drei verschiedene Pruefer erstellen
ist_schulnote    = erstelle_bereichspruefer(1, 6)
ist_prozentwert  = erstelle_bereichspruefer(0, 100)
ist_temperatur   = erstelle_bereichspruefer(-273, 1000)

# Jeder Pruefer ist eine eigene Funktion
print(ist_schulnote(3))      # True
print(ist_schulnote(0))      # False
print(ist_prozentwert(75))   # True
print(ist_prozentwert(150))  # False

# Nuetzlich: Validierung in einem Schritt
eingaben = [1, 3, 7, 2, 9, 4]
gueltige_noten = [n for n in eingaben if ist_schulnote(n)]
print(gueltige_noten)  # [1, 3, 2, 4]
````

Was passiert hier technisch? `erstelle_bereichspruefer(1, 6)` ruft die aeussere Funktion auf -- `minimum` und `maximum` werden als `1` und `6` gesetzt. Die innere Funktion `pruefer` wird erstellt und **zurueckgegeben**.

`ist_schulnote` zeigt jetzt auf diese innere Funktion. Wenn man `ist_schulnote(3)` aufruft, weiss `pruefer` noch immer dass `minimum=1` und `maximum=6` -- diese Werte wurden "eingeschlossen".

Das nennt man **Closure** -- das ist das Thema von Notebook 03c.

---

**Zusammenfassung**

Eine **hoeherwertige Funktion** nimmt eine Funktion als Argument oder gibt eine Funktion zurueck.

| Beispiel | Was sie tut |
|----------|-------------|
| `sorted(daten, key=fn)` | Sortiert nach dem Ergebnis von `fn` |
| `min(daten, key=fn)` | Findet Minimum nach `fn` |
| `max(daten, key=fn)` | Findet Maximum nach `fn` |
| Eigene HOF | Nimmt Funktion entgegen, wendet sie flexibel an |
| Funktion als Rueckgabe | Erzeugt spezialisierte Funktionen nach Bedarf |

Der Kern: Hoeherwertige Funktionen trennen **was** getan werden soll (die uebergebene Funktion) von **wie** es organisiert wird (die hoeherwertige Funktion selbst). Das macht Code wiederverwendbar und flexibel.
