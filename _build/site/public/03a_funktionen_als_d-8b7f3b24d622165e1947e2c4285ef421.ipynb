Funktionen kennst du bereits: Du definierst sie mit `def`, gibst ihnen Argumente und rufst sie auf.

```python
def begruessung(name):
    return f'Hallo, {name}!'

begruessung('Alice')  # Aufruf
```

Soweit so bekannt. Jetzt kommt eine Idee, die sich auf den ersten Blick merkwuerdig anfuehlt:

**In Python ist eine Funktion ein Objekt -- genau wie eine Zahl, ein String oder eine Liste.**

Was das bedeutet, schauen wir uns Schritt fuer Schritt an.

2. Eine Funktion hat einen Typ
</span>
</div>

Jedes Objekt in Python hat einen Typ. `type()` verrät ihn. Schauen wir, was `type()` bei einer Funktion zurueckgibt:

````python
def begruessung(name):
    return f'Hallo, {name}!'

# Was ist der Typ einer Funktion?
print(type(42))           # <class 'int'>
print(type('Hallo'))      # <class 'str'>
print(type([1, 2, 3]))    # <class 'list'>
print(type(begruessung))  # <class 'function'>
````

Python antwortet mit `<class 'function'>`. Eine Funktion ist also ein Objekt der Klasse `function` -- genau wie `42` ein Objekt der Klasse `int` ist.

Und genau wie jedes andere Objekt hat eine Funktion:
- eine **Identitaet** (Speicheradresse, abfragbar mit `id()`)
- einen **Typ** (`function`)
- **Attribute** -- zum Beispiel ihren Namen

````python
# Funktionen haben Attribute
print(begruessung.__name__)  # 'begruessung' -- der Name der Funktion
print(id(begruessung))       # Speicheradresse -- eine eindeutige Zahl
````

3. Eine Funktion einer Variable zuweisen
</span>
</div>

Weil eine Funktion ein Objekt ist, kann man sie einer Variable zuweisen -- genau wie eine Zahl oder einen String.

Was passiert also bei `def begruessung(name):`?

Python erstellt ein Funktionsobjekt im Speicher und bindet es an den Namen `begruessung`. Das ist nicht anders als `x = 42` -- Python erstellt ein Int-Objekt und bindet es an den Namen `x`.

Und weil `begruessung` nur ein Name fuer das Objekt ist, kann man das Objekt auch an einen anderen Namen binden:

````python
def begruessung(name):
    return f'Hallo, {name}!'

# greet ist jetzt ein zweiter Name fuer dasselbe Objekt
greet = begruessung

# Beide Namen rufen dieselbe Funktion auf
print(begruessung('Alice'))  # Hallo, Alice!
print(greet('Bob'))          # Hallo, Bob!

# Sie zeigen auf dasselbe Objekt
print(greet is begruessung)  # True
print(id(greet) == id(begruessung))  # True -- gleiche Speicheradresse
````

Wichtig: `greet = begruessung` ruft die Funktion **nicht** auf. Es erstellt nur eine zweite Referenz auf dasselbe Objekt. Das ist derselbe Mechanismus wie bei:

```python
a = [1, 2, 3]
b = a           # b zeigt auf dieselbe Liste
```

4. Der Unterschied: Referenz vs. Aufruf
</span>
</div>

Das ist einer der haeufigsten Anfaengerfehler -- und einer der wichtigsten Unterschiede:

- `begruessung` -- **Referenz** auf das Funktionsobjekt. Nichts wird ausgefuehrt.
- `begruessung('Alice')` -- **Aufruf**. Die Funktion wird ausgefuehrt und gibt einen Wert zurueck.

Der einzige aeusserliche Unterschied: die Klammern `()`. Aber der Effekt ist fundamental verschieden.

````python
def verdopple(x):
    return x * 2

# Ohne Klammern: Referenz -- wir speichern die Funktion
a = verdopple
print(type(a))   # <class 'function'>
print(a(5))      # Erst jetzt wird sie ausgefuehrt: 10

# Mit Klammern: Aufruf -- wir speichern das Ergebnis
b = verdopple(5)
print(type(b))   # <class 'int'>
print(b)         # 10 -- die Zahl, nicht die Funktion
````

Dieser Fehler passiert haeufig, wenn man eine Funktion an eine andere uebergeben moechte -- aber versehentlich schon aufruft:

````python
woerter = ['Banane', 'apfel', 'Kirsche']

# RICHTIG: sorted bekommt die Funktion str.lower als key
print(sorted(woerter, key=str.lower))   # ['apfel', 'Banane', 'Kirsche']

# FALSCH: str.lower() wird aufgerufen -- aber auf was? Es gibt kein Argument
try:
    print(sorted(woerter, key=str.lower()))  # TypeError!
except TypeError as e:
    print(f'Fehler: {e}')
````

5. Funktionen in Datenstrukturen speichern
</span>
</div>

Weil Funktionen Objekte sind, koennen sie auch in Listen, Dictionaries oder Tupeln gespeichert werden -- genau wie Zahlen oder Strings.

Das klingt abstrakt. Hier ein konkretes Szenario: Ein einfacher Taschenrechner, der die gewuenschte Operation als Name entgegennimmt.

````python
# Vier Grundrechenarten als Funktionen
def addiere(a, b):     return a + b
def subtrahiere(a, b): return a - b
def multipliziere(a, b): return a * b
def dividiere(a, b):   return a / b if b != 0 else None

# Dictionary: Name -> Funktion
operationen = {
    '+': addiere,
    '-': subtrahiere,
    '*': multipliziere,
    '/': dividiere,
}

# Jetzt kann der Nutzer eine Operation als String waehlen
a, b, op = 10, 3, '*'

if op in operationen:
    ergebnis = operationen[op](a, b)  # Funktion nachschlagen und aufrufen
    print(f'{a} {op} {b} = {ergebnis}')
````

Der entscheidende Vorteil: Der Code `operationen[op](a, b)` muss nicht wissen, was `op` konkret ist. Kein `if op == '+': ...elif op == '-': ...`. Die Tabelle erledigt das -- und neue Operationen koennen einfach eingetragen werden.

6. Eine Funktion als Argument uebergeben
</span>
</div>

Weil Funktionen Objekte sind, koennen sie auch als Argument an eine andere Funktion uebergeben werden.

Das kennen wir eigentlich schon -- `sorted()` macht genau das mit dem `key`-Parameter. Wir schreiben jetzt selbst eine solche Funktion.

Das Ziel: Eine Funktion `wende_an()`, die eine beliebige Transformation auf jedes Element einer Liste anwendet.

````python
def wende_an(funktion, werte):
    """Wendet eine Funktion auf jedes Element einer Liste an."""
    ergebnis = []
    for wert in werte:
        ergebnis.append(funktion(wert))  # funktion wird mit jedem Element aufgerufen
    return ergebnis

zahlen = [1, 4, 9, 16, 25]

# Dieselbe Funktion wende_an -- aber mit verschiedenen Transformationen
import math
print(wende_an(math.sqrt, zahlen))  # Wurzel jeder Zahl
print(wende_an(str, zahlen))        # Jede Zahl als String
print(wende_an(lambda x: x * 2, zahlen))  # Jede Zahl verdoppelt
````

`wende_an` weiss nichts darueber, was `funktion` konkret tut. Es koennte `math.sqrt` sein, `str`, eine selbst definierte Funktion oder ein Lambda. Die Funktion ist austauschbar -- der Rest des Codes bleibt gleich.

Eine Funktion die eine andere als Argument entgegennimmt (oder zurueckgibt) nennt man **hoeherwertige Funktion** (Higher-Order Function). Dazu mehr in Notebook 03b.

---

**Zusammenfassung**

Funktionen sind in Python **First-Class-Objekte**. Das bedeutet konkret:

| Was man tun kann | Beispiel |
|-----------------|----------|
| Typ abfragen | `type(begruessung)` → `<class 'function'>` |
| Variable zuweisen | `greet = begruessung` |
| In Liste/Dict speichern | `ops = {'+': addiere, '-': subtrahiere}` |
| Als Argument uebergeben | `wende_an(math.sqrt, zahlen)` |

Der wichtigste Unterschied: `funktion` ist eine Referenz -- `funktion()` ist ein Aufruf. Die Klammern machen den Unterschied.
