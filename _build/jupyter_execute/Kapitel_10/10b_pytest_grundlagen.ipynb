1. Warum pytest statt unittest?
</span>
</div>

In Kapitel 5 haben wir `unittest` kennengelernt. Es funktioniert -- aber es hat Boilerplate. Jeder Test braucht eine Klasse, jede Klasse muss von `TestCase` erben, und statt `assert a == b` schreibt man `self.assertEqual(a, b)`.

Vergleich derselben Tests in beiden Frameworks:

```python
# unittest
import unittest
class TestSchaden(unittest.TestCase):
    def test_normaler_schaden(self):
        self.assertEqual(berechne_schaden(50, 20), 30)
    def test_mindestschaden(self):
        self.assertEqual(berechne_schaden(10, 50), 1)

# pytest
def test_normaler_schaden():
    assert berechne_schaden(50, 20) == 30
def test_mindestschaden():
    assert berechne_schaden(10, 50) == 1
```

pytest-Tests sind einfache Funktionen -- kein `self`, keine Klasse, normales `assert`. pytest findet und fuehrt sie automatisch aus. Und wenn ein Test fehlschlaegt, zeigt pytest genau was der tatsaechliche und was der erwartete Wert war -- ohne dass man das selbst formulieren muss.

In der Praxis ist pytest der Standard.

2. Installation und Projektstruktur
</span>

```bash
pip install pytest
```

pytest findet Tests automatisch -- aber nur wenn man sich an die Konventionen haelt:

| Was | Konvention |
|-----|------------|
| Testdateien | `test_*.py` oder `*_test.py` |
| Testfunktionen | beginnen mit `test_` |
| Testklassen (optional) | beginnen mit `Test` |

Unsere Projektstruktur fuer dieses Kapitel:

```
projekt/
    combat.py           <-- der Code der getestet wird
    test_combat.py      <-- unsere Tests
```

Tests ausfuehren im Terminal:

```bash
pytest              # alle Tests im aktuellen Verzeichnis
pytest -v           # verbose -- zeigt jeden einzelnen Test
pytest -v test_combat.py   # nur diese Datei
```

3. Der erste Test -- und was pytest daraus macht
</span>
</div>

Wir schreiben unsere erste Testdatei. Jede Testfunktion prueft genau **eine** Sache -- das ist wichtig. Wenn ein Test fehlschlaegt, soll sofort klar sein was kaputt ist.

````python
# Testdatei erstellen -- so wuerde test_combat.py beginnen
test_code = '''
from combat import berechne_schaden


def test_normaler_schaden():
    assert berechne_schaden(50, 20) == 30


def test_gleiche_werte():
    # Angriff == Verteidigung -- Mindestschaden 1
    assert berechne_schaden(20, 20) == 1


def test_verteidigung_groesser():
    # Verteidigung > Angriff -- trotzdem Mindestschaden 1
    assert berechne_schaden(10, 50) == 1


def test_kein_angriff():
    # Angriff 0, Verteidigung 0 -- Mindestschaden 1
    assert berechne_schaden(0, 0) == 1
'''

with open('test_combat_schritt1.py', 'w') as f:
    f.write(test_code)

print('test_combat_schritt1.py erstellt')
````

````python
# Tests ausfuehren -- direkt aus dem Notebook
import subprocess
result = subprocess.run(
    ['python', '-m', 'pytest', 'test_combat_schritt1.py', '-v'],
    capture_output=True, text=True
)
print(result.stdout)
if result.stderr:
    print(result.stderr)
````

Die Ausgabe von pytest lesen:

```
PASSED  test_combat_schritt1.py::test_normaler_schaden
FAILED  test_combat_schritt1.py::test_gleiche_werte
```

Beim fehlgeschlagenen Test zeigt pytest:

```
assert berechne_schaden(20, 20) == 1
AssertionError: assert 0 == 1
```

Man sieht sofort: die Funktion gibt `0` zurueck, erwartet war `1`. Das ist der Bug aus 10a -- `max(0, schaden)` statt `max(1, schaden)`. pytest hat ihn gefunden.

4. Aequivalenzklassen und Grenzwerte als Tests
</span>
</div>

In 10a haben wir Aequivalenzklassen und Grenzwerte als Konzept kennengelernt. Jetzt uebersetzen wir sie direkt in Tests.

Fuer `heilen()` hatten wir diese Klassen identifiziert:

| Klasse | Repraesentant | Test-Name |
|--------|--------------|----------|
| Normalfall | `heilen(60, 100, 20)` = 80 | `test_heilen_normalfall` |
| HP bereits voll | `heilen(100, 100, 10)` = 100 | `test_heilen_bereits_voll` |
| Heilung uebersteigt Maximum | `heilen(90, 100, 50)` = 100 | `test_heilen_cap_bei_max_hp` |
| Grenze: genau max_hp erreicht | `heilen(90, 100, 10)` = 100 | `test_heilen_genau_max_hp` |
| Toter Held | `heilen(0, 100, 50)` = 0 | `test_heilen_toter_held` |
| HP negativ | `heilen(-10, 100, 50)` = -10 | `test_heilen_negative_hp` |
| Ungueltige Heilmenge | `heilen(60, 100, 0)` -> ValueError | `test_heilen_ungueltige_menge` |

````python
test_heilen = '''
import pytest
from combat import heilen


# --- Aequivalenzklasse: Normalfall ---
def test_heilen_normalfall():
    assert heilen(60, 100, 20) == 80


# --- Aequivalenzklasse: HP bereits voll ---
def test_heilen_bereits_voll():
    assert heilen(100, 100, 10) == 100


# --- Grenzwert: Heilung uebersteigt Maximum ---
def test_heilen_knapp_unter_max():
    # 90 + 9 = 99 -- noch unter max_hp
    assert heilen(90, 100, 9) == 99

def test_heilen_genau_max_hp():
    # 90 + 10 = 100 -- genau max_hp
    assert heilen(90, 100, 10) == 100

def test_heilen_cap_bei_max_hp():
    # 90 + 11 = 101 -- wird auf 100 gekappt
    assert heilen(90, 100, 11) == 100


# --- Grenzwert: tot/lebendig ---
def test_heilen_hp_negativ():
    # Negativer HP-Wert -- kein Effekt
    assert heilen(-10, 100, 50) == -10

def test_heilen_hp_null():
    # HP genau 0 -- Held ist tot, kein Effekt
    assert heilen(0, 100, 50) == 0

def test_heilen_hp_eins():
    # HP = 1 -- Held lebt noch, Heilung wirkt
    assert heilen(1, 100, 20) == 21


# --- Aequivalenzklasse: Ungueltige Eingaben ---
def test_heilen_heilmenge_null():
    with pytest.raises(ValueError):
        heilen(60, 100, 0)

def test_heilen_heilmenge_negativ():
    with pytest.raises(ValueError):
        heilen(60, 100, -10)

def test_heilen_max_hp_null():
    with pytest.raises(ValueError):
        heilen(60, 0, 10)
'''

with open('test_heilen.py', 'w') as f:
    f.write(test_heilen)
print('test_heilen.py erstellt')
````

````python
import subprocess
result = subprocess.run(
    ['python', '-m', 'pytest', 'test_heilen.py', '-v'],
    capture_output=True, text=True
)
print(result.stdout)
````

pytest zeigt genau welche Tests bestehen und welche fehlschlagen. Die fehlgeschlagenen Tests -- `test_heilen_cap_bei_max_hp`, `test_heilen_genau_max_hp` -- bestaetigen den Bug: der max_hp-Cap fehlt.

Das ist der Wert systematischen Testens: Nicht nur *irgendwas* testen, sondern gezielt die Grenzen und Klassen -- und damit genau die Stellen wo Bugs lauern.

````python
alle_tests = '''
import pytest
from combat import (
    berechne_schaden,
    trefferchance,
    kritischer_treffer,
    heilen,
    ist_besiegt,
    statuseffekt_anwenden,
)


# ============================================================
# berechne_schaden
# ============================================================

def test_schaden_normalfall():
    assert berechne_schaden(50, 20) == 30

def test_schaden_gleiche_werte_mindestschaden():
    # Grenzwert: angriff == verteidigung -> Mindestschaden 1
    assert berechne_schaden(20, 20) == 1

def test_schaden_verteidigung_groesser_mindestschaden():
    # Aequivalenzklasse: verteidigung > angriff -> immer 1
    assert berechne_schaden(10, 50) == 1

def test_schaden_angriff_null():
    # Grenzwert: minimaler Angriff
    assert berechne_schaden(0, 0) == 1

def test_schaden_negative_werte_werfen_fehler():
    with pytest.raises(ValueError):
        berechne_schaden(-1, 0)


# ============================================================
# trefferchance
# ============================================================

def test_trefferchance_gleich_schnell():
    # 50 + (10-10)*5 = 50
    assert trefferchance(10, 10) == 50

def test_trefferchance_minimum_bei_10():
    # Grenzwert: sehr langsamer Angreifer -> Minimum 10
    assert trefferchance(1, 20) == 10

def test_trefferchance_maximum_bei_95():
    # Grenzwert: sehr schneller Angreifer -> Maximum 95
    assert trefferchance(20, 1) == 95

def test_trefferchance_nullwert_wirft_fehler():
    with pytest.raises(ValueError):
        trefferchance(0, 10)


# ============================================================
# kritischer_treffer
# ============================================================

def test_krit_standard_faktor():
    assert kritischer_treffer(30) == 45

def test_krit_gibt_int_zurueck():
    # Rueckgabetyp muss int sein
    assert isinstance(kritischer_treffer(30), int)

def test_krit_benutzerdefinierter_faktor():
    assert kritischer_treffer(30, krit_faktor=2.0) == 60

def test_krit_abrunden():
    # 10 * 1.5 = 15.0 -> int 15
    assert kritischer_treffer(10) == 15
    assert isinstance(kritischer_treffer(10), int)

def test_krit_faktor_unter_eins_wirft_fehler():
    with pytest.raises(ValueError):
        kritischer_treffer(30, krit_faktor=0.5)


# ============================================================
# heilen
# ============================================================

def test_heilen_normalfall():
    assert heilen(60, 100, 20) == 80

def test_heilen_cap_bei_max_hp():
    assert heilen(90, 100, 50) == 100

def test_heilen_genau_max_hp():
    assert heilen(90, 100, 10) == 100

def test_heilen_knapp_unter_max():
    assert heilen(90, 100, 9) == 99

def test_heilen_toter_held_hp_null():
    assert heilen(0, 100, 50) == 0

def test_heilen_toter_held_hp_negativ():
    assert heilen(-10, 100, 50) == -10

def test_heilen_hp_eins_lebt_noch():
    assert heilen(1, 100, 20) == 21

def test_heilen_heilmenge_null_wirft_fehler():
    with pytest.raises(ValueError):
        heilen(60, 100, 0)


# ============================================================
# ist_besiegt
# ============================================================

def test_ist_besiegt_lebendig():
    assert ist_besiegt(100) == False

def test_ist_besiegt_grenzwert_eins():
    # hp=1: noch lebendig
    assert ist_besiegt(1) == False

def test_ist_besiegt_grenzwert_null():
    # hp=0: genau besiegt
    assert ist_besiegt(0) == True

def test_ist_besiegt_negative_hp():
    # hp=-10: durch starken Treffer -- muss auch True sein
    assert ist_besiegt(-10) == True


# ============================================================
# statuseffekt_anwenden
# ============================================================

def test_statuseffekt_normal():
    assert statuseffekt_anwenden(20, 'normal') == 20

def test_statuseffekt_vergiftet():
    assert statuseffekt_anwenden(20, 'vergiftet') == 30

def test_statuseffekt_geschwaeht():
    assert statuseffekt_anwenden(20, 'geschwaeht') == 15

def test_statuseffekt_verstaerkt():
    assert statuseffekt_anwenden(20, 'verstaerkt') == 40

def test_statuseffekt_unbekannt_wirft_fehler():
    # Unbekannter Effekt muss ValueError werfen
    with pytest.raises(ValueError):
        statuseffekt_anwenden(20, 'eingefroren')

def test_statuseffekt_tippfehler_wirft_fehler():
    # Gross-/Kleinschreibung oder Tippfehler -> ValueError
    with pytest.raises(ValueError):
        statuseffekt_anwenden(20, 'Vergiftet')
'''

with open('test_combat.py', 'w') as f:
    f.write(alle_tests)
print('test_combat.py erstellt')
````

````python
import subprocess
result = subprocess.run(
    ['python', '-m', 'pytest', 'test_combat.py', '-v'],
    capture_output=True, text=True
)
print(result.stdout)
````

Jetzt beheben wir alle sechs Bugs. Die Tests bleiben unveraendert -- sie sind die Spezifikation. Der Code wird so angepasst, dass die Tests bestehen.

````python
combat_fixed = '''
"""
combat.py -- Kampfsystem fuer das Rollenspiel "Aetherion"
Version 2 -- alle Bugs behoben
"""


def berechne_schaden(angriff, verteidigung):
    if angriff < 0 or verteidigung < 0:
        raise ValueError("Angriff und Verteidigung muessen >= 0 sein")
    schaden = angriff - verteidigung
    return max(1, schaden)   # FIX: max(1, ...) statt max(0, ...)


def trefferchance(angreifer_geschwindigkeit, ziel_geschwindigkeit):
    if angreifer_geschwindigkeit <= 0 or ziel_geschwindigkeit <= 0:
        raise ValueError("Geschwindigkeit muss > 0 sein")
    chance = 50 + (angreifer_geschwindigkeit - ziel_geschwindigkeit) * 5
    return max(10, min(95, chance))   # FIX: min(95, ...) ergaenzt


def kritischer_treffer(schaden, krit_faktor=1.5):
    if krit_faktor < 1.0:
        raise ValueError("krit_faktor muss >= 1.0 sein")
    if schaden < 0:
        raise ValueError("Schaden muss >= 0 sein")
    return int(schaden * krit_faktor)   # FIX: int() ergaenzt


def heilen(aktuelle_hp, max_hp, heilmenge):
    if max_hp <= 0:
        raise ValueError("max_hp muss > 0 sein")
    if heilmenge <= 0:
        raise ValueError("heilmenge muss > 0 sein")
    if aktuelle_hp <= 0:
        return aktuelle_hp
    return min(max_hp, aktuelle_hp + heilmenge)   # FIX: min(max_hp, ...) ergaenzt


def ist_besiegt(hp):
    return hp <= 0   # FIX: <= statt ==


def statuseffekt_anwenden(schaden, effekt):
    if schaden < 0:
        raise ValueError("Schaden muss >= 0 sein")
    faktoren = {
        'normal':    1.0,
        'vergiftet': 1.5,
        'geschwaeht': 0.75,
        'verstaerkt': 2.0,
    }
    if effekt not in faktoren:
        raise ValueError(f"Unbekannter Statuseffekt: {effekt!r}")   # FIX
    return int(schaden * faktoren[effekt])
'''

with open('../combat/combat.py', 'w') as f:
    f.write(combat_fixed)
print('combat.py (Version 2) gespeichert')
````

````python
# Alle Tests erneut ausfuehren -- jetzt sollten alle gruen sein
import subprocess
result = subprocess.run(
    ['python', '-m', 'pytest', 'test_combat.py', '-v'],
    capture_output=True, text=True
)
print(result.stdout)
````

**Zusammenfassung**

| Konzept | Erklaerung |
|---------|------------|
| pytest vs. unittest | Keine Klassen, kein self, normales assert |
| Testdatei | `test_*.py` -- pytest findet sie automatisch |
| Testfunktion | Beginnt mit `test_` -- eine Funktion, eine Sache |
| `assert a == b` | pytest zeigt bei Fehler was tatsaechlich zurueckgegeben wurde |
| `pytest.raises(E)` | Prueft ob eine Exception ausgeloest wird |
| `pytest -v` | Zeigt jeden Test einzeln |
| Aequivalenzklassen | Gruppen gleichartiger Eingaben -- ein Test pro Gruppe |
| Grenzwertanalyse | Darunter, die Grenze, darueber -- dort lauern Bugs |
| Tests als Spezifikation | Tests beschreiben was der Code tun soll -- Code wird angepasst bis Tests bestehen |
