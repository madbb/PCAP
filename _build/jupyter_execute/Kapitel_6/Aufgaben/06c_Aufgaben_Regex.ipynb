**5.1**
```python
import re
print(re.findall(r'\d+', 'abc123def456'))
print(re.findall(r'\d', 'abc123def456'))
print(len(re.findall(r'\d+', 'abc123def456')))
```

**5.2**
```python
import re
m = re.search(r'(\d+)', 'HP: 100, Mana: 80')
print(m.group())
print(m.group(0))
print(m.group(1))
print(m.start())
```

**5.3**
```python
import re
print(bool(re.match(r'\d+', '123abc')))
print(bool(re.fullmatch(r'\d+', '123abc')))
print(bool(re.search(r'\d+', 'abcdef')))
```

**5.4**
