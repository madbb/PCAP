**Checkliste -- das muss sitzen:**

- [ ] List Comprehension: `if` am Ende filtert, `if-else` im Ausdruck transformiert
- [ ] Reihenfolge verschachtelter `for`-Klauseln bestimmt das Ergebnis
- [ ] Lambda: nur ein Ausdruck, kein `return`, kein `if`-Block
- [ ] `map()` und `filter()` geben lazy Iteratoren -- nicht Listen
- [ ] Closures: Referenz auf Variable -- nicht Wert zum Definitionszeitpunkt
- [ ] Schleifenfalle: alle Closures sehen dasselbe `i` nach der Schleife
- [ ] `nonlocal` noetig zum Veraendern von Enclosing-Variablen
- [ ] Textmodus gibt `str`, Binaermodus gibt `bytes`
- [ ] Modus `'w'` loescht Inhalt sofort beim Oeffnen
- [ ] `readline()` gibt `''` am Dateiende
- [ ] Cursor nach `read()` am Ende -- `seek(0)` zum Zuruecksetzen
- [ ] `bytearray` veraenderlich, `bytes` unveraenderlich
- [ ] Indexzugriff auf `bytes`/`bytearray` gibt `int` -- nicht `bytes`

```python
zahlen = [-3, -1, 0, 2, 4, 7]

# if AM ENDE -- filtert
positiv = [x for x in zahlen if x > 0]
print(positiv)   # [2, 4, 7]  -- nur 3 Elemente

# if-else IM AUSDRUCK -- alle bleiben, zwei Ausgaben
absolut = [x if x >= 0 else -x for x in zahlen]
print(absolut)   # [3, 1, 0, 2, 4, 7]  -- 6 Elemente

# Reihenfolge bei verschachtelten for-Klauseln
a = [(x, y) for x in range(3) for y in range(3) if x != y]
b = [(x, y) for y in range(3) for x in range(3) if x != y]
print('Gleich?', a == b)   # False -- andere Reihenfolge!
```

```python
# Lambda -- nur ein Ausdruck
quadrat = lambda x: x**2
print(quadrat(5))   # 25

# Ternary-Ausdruck ist erlaubt (ein Ausdruck)
absolut = lambda x: x if x >= 0 else -x
print(absolut(-3))   # 3

# map() und filter() geben Iteratoren
m = map(lambda x: x**2, [1, 2, 3])
print(type(m))       # <class 'map'>  -- kein list!
print(list(m))       # [1, 4, 9]
print(list(m))       # []  -- erschoepft!

f = filter(lambda x: x > 0, [-1, 0, 1, 2])
print(type(f))       # <class 'filter'>
print(list(f))       # [1, 2]
```

```python
# Closure-Schleifenfalle
funktionen = []
for i in range(3):
    funktionen.append(lambda: i)   # Referenz auf i!

print([f() for f in funktionen])  # [2, 2, 2] -- alle sehen i=2!

# Fix: Default-Parameter friert Wert ein
funktionen = []
for i in range(3):
    funktionen.append(lambda i=i: i)  # i=i: Wert wird kopiert

print([f() for f in funktionen])  # [0, 1, 2]

# nonlocal
def zaehler():
    n = 0
    def erhoehe():
        nonlocal n   # ohne nonlocal: UnboundLocalError
        n += 1
        return n
    return erhoehe

z = zaehler()
print(z(), z(), z())   # 1 2 3
```

```python
# Text vs. Binaer
with open('test.txt', 'w') as f:
    f.write('Hallo')

with open('test.txt', 'r') as f:   # Textmodus
    print(type(f.read()))           # <class 'str'>

with open('test.txt', 'rb') as f:  # Binaermodus
    print(type(f.read()))           # <class 'bytes'>

# Modus 'w' loescht sofort
with open('test.txt', 'w') as f:
    pass   # nichts schreiben -- aber alter Inhalt ist weg!
with open('test.txt', 'r') as f:
    print(repr(f.read()))   # ''  -- leer!

# Cursor-Falle
with open('test.txt', 'w') as f:
    f.write('abc')
with open('test.txt', 'r') as f:
    print(f.read())    # 'abc'
    print(f.read())    # ''  -- Cursor am Ende!
```

```python
# bytes vs. bytearray
b  = bytes([65, 66, 67])      # unveraenderlich
ba = bytearray([65, 66, 67])  # veraenderlich

# Indexzugriff gibt int!
print(b[0])     # 65  -- nicht b'A'
print(ba[0])    # 65
print(b[0:1])   # b'A' -- Slicing gibt bytes

# bytearray ist veraenderlich
ba[0] = 97   # 'a' statt 'A'
print(ba)    # bytearray(b'aBC')

try:
    b[0] = 97   # TypeError!
except TypeError as e:
    print(f'TypeError: {e}')
```
