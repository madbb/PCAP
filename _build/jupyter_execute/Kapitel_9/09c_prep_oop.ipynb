**Checkliste -- das muss sitzen:**

- [ ] `self` ist Konvention -- kein Schluesselwort
- [ ] `self` vergessen in Methode -- TypeError
- [ ] Klassenvariable per Instanz setzen -- erzeugt Instanzvariable, verdeckt Klassenvariable
- [ ] `__dict__` -- was steht wo
- [ ] Name Mangling: `__x` wird zu `_Klassenname__x`
- [ ] `__name__` existiert an der **Klasse**, nicht an der Instanz
- [ ] `isinstance()` prueft Vererbungskette -- `type()` nur direkte Klasse
- [ ] `issubclass(X, X)` gibt True
- [ ] `super()` folgt MRO -- nicht direkt die Elternklasse
- [ ] MRO-Reihenfolge: von links nach rechts, dann aufwaerts
- [ ] `is` prueft Identitaet -- `==` prueft Gleichheit
- [ ] Overriding: Kindklasse gewinnt immer

```python
class Held:
    def __init__(self, name):
        self.name = name

    def info(self):        # self korrekt
        return self.name

    def kaputt():          # self vergessen
        return 'kaputt'

h = Held('Aldric')
print(h.info())    # Aldric

try:
    h.kaputt()     # TypeError: kaputt() takes 0 positional arguments but 1 was given
except TypeError as e:
    print(f'TypeError: {e}')

# self ist nur Konvention
class Test:
    def __init__(this, wert):   # 'this' statt 'self' -- gueltig
        this.wert = wert

print(Test(42).wert)   # 42
```

```python
class Held:
    gilde = 'Silberne Klinge'   # Klassenvariable

    def __init__(self, name):
        self.name = name

a = Held('Aldric')
b = Held('Lyria')

# Per Klasse setzen -- alle sehen die Aenderung
Held.gilde = 'Goldene Flamme'
print(a.gilde, b.gilde)   # Goldene Flamme  Goldene Flamme

# Per Instanz setzen -- erzeugt Instanzvariable, verdeckt Klassenvariable
a.gilde = 'Freie Klingen'
print(a.gilde)             # Freie Klingen  -- Instanzvariable
print(b.gilde)             # Goldene Flamme -- Klassenvariable unveraendert
print(Held.gilde)          # Goldene Flamme -- Klasse unveraendert
print(a.__dict__)          # {'name': 'Aldric', 'gilde': 'Freie Klingen'}
print(b.__dict__)          # {'name': 'Lyria'}  -- kein gilde-Eintrag
```

```python
class Tier:
    pass

class Held(Tier):
    pass

class Krieger(Held):
    pass

k = Krieger()

# __name__ nur an der Klasse
print(Krieger.__name__)      # 'Krieger'
try:
    print(k.__name__)        # AttributeError!
except AttributeError as e:
    print(f'AttributeError: {e}')
print(type(k).__name__)      # 'Krieger' -- Umweg

# isinstance vs. type
print(type(k) == Krieger)    # True
print(type(k) == Held)       # False -- type prueft nur direkte Klasse
print(isinstance(k, Krieger)) # True
print(isinstance(k, Held))    # True  -- Vererbungskette!
print(isinstance(k, Tier))    # True

# issubclass
print(issubclass(Krieger, Held))     # True
print(issubclass(Krieger, Krieger))  # True -- sich selbst!
```

```python
class A:
    def methode(self): return 'A'

class B(A):
    def methode(self): return 'B'

class C(A):
    def methode(self): return 'C'

class D(B, C):
    pass

# MRO: D -> B -> C -> A -> object
print([k.__name__ for k in D.__mro__])

d = D()
print(d.methode())   # 'B' -- B kommt in MRO zuerst

# is vs. ==
a1 = Held('x') if False else A()
a2 = A()
a3 = a1

a1 = A(); a2 = A(); a3 = a1
print(a1 is a2)   # False -- verschiedene Objekte
print(a1 is a3)   # True  -- dieselbe Referenz
print(a1 == a2)   # False -- kein __eq__ definiert, faellt auf 'is' zurueck
```

```python
class Held:
    def __init__(self):
        self.name    = 'Aldric'   # oeffentlich
        self._rang   = 'Rekrut'   # geschuetzt (Konvention)
        self.__code  = 'geheim'   # privat -- Name Mangling!

h = Held()
print(h.name)            # Aldric
print(h._rang)           # Rekrut  -- geht, aber bitte nicht

try:
    print(h.__code)      # AttributeError!
except AttributeError:
    print('__code nicht direkt erreichbar')

# Echter Name nach Mangling:
print(h._Held__code)     # geheim -- so heisst es wirklich
print(h.__dict__)        # zeigt _Held__code
```
