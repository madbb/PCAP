## Block 5: Vererbung

**Frage 26**

Was gibt der Code aus?
```python
class A:
    def greet(self):
        return 'A'

class B(A):
    def greet(self):
        return super().greet() + 'B'

print(B().greet())
```
A) `B`

B) `AB`

C) `BA`

D) `A`

**Frage 27**

Was ist die Ausgabe?
```python
class X:
    def __init__(self, n):
        self.n = n

    def __str__(self):
        return f'X({self.n})'

print(str(X(5)))
```
A) `X`

B) `5`

C) `X(5)`

D) `X object`

C) `Q`
