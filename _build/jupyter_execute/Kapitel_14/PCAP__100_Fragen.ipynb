**Frage 90**

Was gibt dieser Code aus?
```python
def counter():
    n = 0
    def inc():
        nonlocal n
        n += 1
        return n
    return inc

a = counter()
b = counter()
a(); a()
b()
print(a(), b())
```
A) `2 1`  
B) `3 2`  
C) `3 1`  
D) `2 2`

**Frage 96**

Was gibt dieser Code aus?
```python
import tempfile, os
with tempfile.NamedTemporaryFile(mode='w', suffix='.txt', delete=False) as f:
    f.write('line1\nline2\n')
    name = f.name
with open(name) as f:
    lines = f.readlines()
os.unlink(name)
print(len(lines), lines[0].strip())
```

C) `2 line1`  
D) `2 line1\n`

**Frage 100**

Was gibt dieser Code aus?
```python
result = []
for i in range(3):
    def f(x, i=i):
        return x + i
    result.append(f)
print([fn(10) for fn in result])
```
A) `[12, 12, 12]`  
B) `[10, 10, 10]`  
C) `[10, 11, 12]`
