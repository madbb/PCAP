2. Schreibe einen Generator `top_spieler(daten, n)`, der die `n` Helden mit

   Hinweis: Hier muss man leider alle Daten lesen um zu sortieren.
3. Schreibe einen Generator `batches(daten, groesse)`, der die Daten in
   Gruppen zu je `groesse` Elementen yieldet.
   Teste: `list(batches([1,2,3,4,5], 2))` ergibt `[[1,2], [3,4], [5]]`.
4. Vergleiche den Speicherbedarf zwischen:
   - `list(verwundete(datenstrom))` (alles auf einmal)
   - dem Generator `verwundete(datenstrom)` selbst

Dokumentation: https://docs.python.org/3/glossary.html#term-generator


````python
import random
import sys

random.seed(42)

datenstrom = [
    {'id': i,
     'hp': random.randint(1, 100),
     'punkte': random.randint(0, 2000)}
    for i in range(100_000)
    ]

def verwundete(daten):
    for held in daten:
        if held['hp'] < 20:
            yield held
````

````python
gen_obj_verwundete = verwundete(datenstrom)
for held in gen_obj_verwundete:
    print(held)
````
