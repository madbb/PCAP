1. Das Muster das immer wieder auftaucht
</span>
</div>

Beim Arbeiten mit Listen taucht ein Muster staendig auf: Eine Liste durchlaufen, jeden Wert transformieren oder filtern, Ergebnis in neue Liste sammeln.

Hier drei typische Beispiele aus einem Spielkontext:

````python
helden_hp = [100, 45, 80, 12, 60, 95]

# Aufgabe 1: Alle HP verdoppeln (Buff)
gebufft = []
for hp in helden_hp:
    gebufft.append(hp * 2)

# Aufgabe 2: Nur verletzte Helden (HP < 50)
verletzt = []
for hp in helden_hp:
    if hp < 50:
        verletzt.append(hp)

# Aufgabe 3: HP als Prozentwert (max 100)
prozent = []
for hp in helden_hp:
    prozent.append(f'{hp}%')

print(gebufft)
print(verletzt)
print(prozent)
````

Alle drei folgen demselben Schema: leere Liste anlegen, Schleife, `append()`. Das ist viel Boilerplate fuer eine einfache Idee. Python hat dafuer eine kompaktere Schreibweise: die **List Comprehension**.

<div style="background:#0b1929;border-left:4px solid #4fc3f7;border-radius:4px;padding:0.75em 1.2em;margin:2rem 0 1.2rem 0;font-family:'Segoe UI',sans-serif;">
<span style="font-size:clamp(1rem,2.2vw,1.25rem);font-weight:600;color:#e8eaf6;letter-spacing:-0.01em;">

2. Das Grundprinzip
</span>
</div>

Eine List Comprehension fasst das Schleife-und-append-Muster in eine einzige Zeile:

```
[ausdruck for element in iterable]
```

Lies es so: *"Erstelle eine Liste mit `ausdruck` fuer jedes `element` in `iterable`."*

````python
helden_hp = [100, 45, 80, 12, 60, 95]

# Vorher: 3 Zeilen
gebufft = []
for hp in helden_hp:
    gebufft.append(hp * 2)

# Nachher: 1 Zeile -- identisches Ergebnis
gebufft = [hp * 2 for hp in helden_hp]

print(gebufft)   # [200, 90, 160, 24, 120, 190]

# Weitere Beispiele
prozent  = [f'{hp}%' for hp in helden_hp]
als_str  = [str(hp) for hp in helden_hp]
quadrate = [x**2 for x in range(1, 6)]

print(prozent)
print(quadrate)
````

3. if zum Filtern
</span>
</div>

Mit einem `if` am Ende der Comprehension kommen nur Elemente in die Liste, die die Bedingung erfuellen. Elemente die nicht passen, werden uebersprungen -- die Ergebnisliste ist kuerzer als die Eingabe.

```
[ausdruck for element in iterable if bedingung]
```

````python
helden_hp = [100, 45, 80, 12, 60, 95]

# Nur verletzte Helden (HP unter 50)
verletzt = [hp for hp in helden_hp if hp < 50]
print(verletzt)   # [45, 12]

# Nur gerade Zahlen
gerade = [x for x in range(1, 11) if x % 2 == 0]
print(gerade)     # [2, 4, 6, 8, 10]

# Filtern und gleichzeitig transformieren
# Verletzte Helden heilen: nur HP < 50, und auf 50 aufstocken
geheilt = [50 for hp in helden_hp if hp < 50]
print(geheilt)    # [50, 50]  -- zwei Helden wurden geheilt

# Oder: HP auf 50 aufstocken und in die Liste schreiben
geheilt2 = [max(hp, 50) for hp in helden_hp if hp < 50]
print(geheilt2)   # [50, 50]
````

4. if-else im Ausdruck -- transformieren statt filtern
</span>
</div>

Das `if` am Ende filtert -- nicht alle Elemente kommen in die Liste. Manchmal will man aber alle Elemente behalten und sie nur unterschiedlich behandeln. Dafuer gibt es `if-else` im **Ausdruck** -- ganz vorne, vor dem `for`.

Die beiden Varianten sehen aehnlich aus, tun aber fundamental Verschiedenes:

````python
helden_hp = [100, 45, 80, 12, 60, 95]

# if AM ENDE -- filtert: nicht alle Elemente kommen rein
nur_verletzte = [hp for hp in helden_hp if hp < 50]
print(f'Gefiltert ({len(nur_verletzte)} Eintraege): {nur_verletzte}')

# if-else IM AUSDRUCK -- transformiert: alle Elemente kommen rein
# Verletzte bekommen Mindest-HP 50, Gesunde bleiben unveraendert
behandelt = [max(hp, 50) if hp < 50 else hp for hp in helden_hp]
print(f'Behandelt ({len(behandelt)} Eintraege): {behandelt}')

# Merkhilfe: Wo steht das if?
# [ausdruck_a if bed else ausdruck_b  for x in it]  --> alle Elemente, zwei Ausgaben
# [ausdruck                           for x in it if bed]  --> nur passende Elemente

# Noch ein Beispiel: Status als Text
status = ['kritisch' if hp < 50 else 'ok' for hp in helden_hp]
print(status)
````

5. Verschachtelte Comprehensions
</span>
</div>

Eine List Comprehension kann mehrere `for`-Klauseln enthalten. Die Reihenfolge entspricht der Reihenfolge verschachtelter Schleifen -- aussen nach innen, links nach rechts.

Ein Anwendungsfall: Alle moeglichen Held-Gegner-Paarungen in einem Turnier:

````python
helden  = ['Aldric', 'Lyria', 'Thorn']
gegner  = ['Drache', 'Riese']

# Alle Paarungen -- zwei for-Klauseln
paarungen = [(held, g) for held in helden for g in gegner]
print(paarungen)

# Aequivalente verschachtelte Schleife
paarungen_schleife = []
for held in helden:
    for g in gegner:
        paarungen_schleife.append((held, g))

print(paarungen == paarungen_schleife)  # True
````

````python
# Klassisches Beispiel: Matrix abflachen
schadenswerte = [[10, 20, 30],
                 [40, 50, 60],
                 [70, 80, 90]]

alle_werte = [wert for zeile in schadenswerte for wert in zeile]
print(alle_werte)   # [10, 20, 30, 40, 50, 60, 70, 80, 90]

# Nur bestimmte Werte: Schaden ueber 40
hoher_schaden = [w for zeile in schadenswerte for w in zeile if w > 40]
print(hoher_schaden)
````

Reihenfolge der for-Klauseln bestimmt das Ergebnis
</span>

````python
# Die Reihenfolge der for-Klauseln aendert das Ergebnis grundlegend
# Genau wie bei verschachtelten Schleifen

# Variante A: erst x, dann y
a = [(x, y) for x in range(3) for y in range(3) if x != y]
print('A:', a)

# Variante B: erst y, dann x -- andere Reihenfolge im Ergebnis!
b = [(x, y) for y in range(3) for x in range(3) if x != y]
print('B:', b)

print('Gleich?', a == b)   # False -- verschiedene Reihenfolge
````

6. Dict- und Set-Comprehensions
</span>
</div>

Das Comprehension-Prinzip funktioniert nicht nur fuer Listen -- auch fuer Dictionaries und Sets. Der Syntax ist identisch, nur die Klammern aendern sich.

````python
helden = ['Aldric', 'Lyria', 'Thorn']

# Dict Comprehension -- {schluessel: wert for ...}
namenlaengen = {name: len(name) for name in helden}
print(namenlaengen)   # {'Aldric': 6, 'Lyria': 5, 'Thorn': 5}

# Set Comprehension -- {ausdruck for ...} -- keine Duplikate
laengen = {len(name) for name in helden}
print(laengen)        # {5, 6} -- Thorn und Lyria beide Laenge 5

# Praxisbeispiel: HP-Tabelle aus zwei Listen
namen  = ['Aldric', 'Lyria', 'Thorn']
hp_werte = [100, 80, 90]
hp_tabelle = {name: hp for name, hp in zip(namen, hp_werte)}
print(hp_tabelle)     # {'Aldric': 100, 'Lyria': 80, 'Thorn': 90}
````

---

**Zusammenfassung**

| Konzept | PCAP | Kernaussage |
|---------|------|-------------|
| `[ausdruck for x in it]` | 5.1 | Grundform -- ersetzt Schleife mit append |
| `[ausdruck for x in it if bed]` | 5.1 | Filtert -- Ergebnisliste kuerzer als Eingabe |
| `[a if bed else b for x in it]` | 5.1 | Transformiert -- alle Elemente bleiben |
| Zwei for-Klauseln | 5.1 | Wie verschachtelte Schleifen -- Reihenfolge beachten |
| Dict Comprehension | Kursinhalt | `{k: v for x in it}` |
| Set Comprehension | Kursinhalt | `{ausdruck for x in it}` -- keine Duplikate |
