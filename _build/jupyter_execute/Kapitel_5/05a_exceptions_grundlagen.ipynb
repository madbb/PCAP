1. Warum ueberhaupt Exceptions?
</span>
</div>

Programme laufen nicht immer wie geplant. Eine Datei ist nicht vorhanden. Der Nutzer gibt keine Zahl ein sondern Text. Ein Netzwerk bricht ab. Eine Division durch Null passiert.

Ohne Fehlerbehandlung bricht Python einfach ab und gibt eine Fehlermeldung aus -- das Programm ist tot. Das ist fuer Produktionscode keine Option.

Schauen wir uns das Problem an:

```python
# Ohne Fehlerbehandlung: Absturz
def berechne_schaden(angriff, verteidigung):
    return angriff / verteidigung   # Was wenn verteidigung = 0?

# Das hier funktioniert
print(berechne_schaden(100, 5))    # 20.0

# Das hier bricht das Programm ab
print(berechne_schaden(100, 0))    # ZeroDivisionError!
```

Ein erster Gedanke waere: pruefen wir das vorher mit `if`:

```python
def berechne_schaden(angriff, verteidigung):
    if verteidigung == 0:
        return 0
    return angriff / verteidigung

print(berechne_schaden(100, 0))   # 0 -- kein Absturz
```

Das funktioniert hier -- aber was wenn man nicht im Voraus weiss was alles schief gehen kann? Beim Lesen einer Datei koennte die Datei fehlen, die Berechtigungen koennten fehlen, die Datei koennte leer sein, die Kodierung koennte falsch sein. Man kann nicht alle Faelle vorher pruefen.

Exceptions erlauben einen anderen Ansatz: **Versuch es -- und reagiere wenn etwas schief geht.** Das ist das `try/except`-Muster.

2. Das Grundprinzip: try / except
</span>
</div>

Der `try`-Block enthaelt Code der fehlschlagen koennte. Der `except`-Block wird nur ausgefuehrt wenn ein Fehler des angegebenen Typs aufgetreten ist. Laeuft der `try`-Block ohne Fehler durch, wird `except` komplett uebersprungen.

```python
def berechne_schaden(angriff, verteidigung):
    try:
        return angriff / verteidigung
    except ZeroDivisionError:
        print('Verteidigung ist 0 -- kein Schaden berechnet')
        return 0

print(berechne_schaden(100, 5))   # 20.0  -- try laeuft durch, except wird uebersprungen
print(berechne_schaden(100, 0))   # except greift, gibt 0 zurueck
```

Sobald der Fehler im `try`-Block auftritt, bricht Python die Ausfuehrung des `try`-Blocks sofort ab und springt in den passenden `except`-Block. Alles was nach dem Fehler im `try`-Block stehen wuerde, wird nicht mehr ausgefuehrt:

```python
try:
    print('Vor dem Fehler')    # wird ausgefuehrt
    x = 1 / 0                  # Fehler!
    print('Nach dem Fehler')   # wird NICHT ausgefuehrt
except ZeroDivisionError:
    print('Im except-Block')   # wird ausgefuehrt

print('Nach dem try/except')   # laeuft weiter
```

3. Alle Varianten von try/except
</span>
</div>

```python
# Variante 1: Mehrere except-Bloecke fuer verschiedene Fehlertypen
def lese_zahl(text):
    try:
        zahl = int(text)
        ergebnis = 100 / zahl
        return ergebnis
    except ValueError:
        print(f'{text!r} ist keine gueltige Zahl')
    except ZeroDivisionError:
        print('Division durch Null')

lese_zahl('10')     # funktioniert
lese_zahl('abc')    # ValueError
lese_zahl('0')      # ZeroDivisionError
```

```python
# Variante 2: Mehrere Exceptions in einem except-Block (als Tupel)
def verarbeite(text):
    try:
        return 100 / int(text)
    except (ValueError, ZeroDivisionError):
        print('Ungueltige Eingabe -- Zahl oder Null')
        return None

verarbeite('abc')
verarbeite('0')
```

```python
# Variante 3: else -- laeuft nur wenn KEIN Fehler aufgetreten ist
def lade_held(name):
    helden = {'Aldric': 100, 'Lyria': 80}
    try:
        hp = helden[name]
    except KeyError:
        print(f'Held {name!r} nicht gefunden')
    else:
        # Nur ausgefuehrt wenn kein Fehler aufgetreten ist
        print(f'{name} geladen mit {hp} HP')

lade_held('Aldric')   # kein Fehler -> else
lade_held('Mira')     # KeyError -> except, kein else
```

```python
# Variante 4: finally -- laeuft IMMER, egal ob Fehler oder nicht
def lade_datei(pfad):
    datei = None
    try:
        datei = open(pfad, 'r')
        inhalt = datei.read()
        return inhalt
    except FileNotFoundError:
        print(f'Datei nicht gefunden: {pfad}')
        return None
    finally:
        # Datei IMMER schliessen -- auch wenn ein Fehler auftritt
        if datei:
            datei.close()
            print('Datei geschlossen')

lade_datei('nicht_vorhanden.txt')
```

`finally` ist besonders nuetzlich fuer Aufraeumarbeiten -- Dateien schliessen, Datenbankverbindungen trennen -- weil es garantiert ausgefuehrt wird, selbst wenn ein unerwarteter Fehler auftritt oder die Funktion mit `return` verlassen wird.

4. Die Reihenfolge der except-Bloecke
</span>
</div>

Python prueft die `except`-Bloecke von oben nach unten und nimmt den ersten der passt. Das hat eine wichtige Konsequenz: Spezifischere Exceptions muessen **vor** allgemeineren stehen.

Das liegt an der Vererbungshierarchie -- `ZeroDivisionError` erbt von `ArithmeticError` erbt von `Exception`. Ein `except Exception` faengt also auch `ZeroDivisionError` ab.

```python
# FALSCH -- Exception ist zu weit oben, ZeroDivisionError wird nie erreicht
try:
    x = 1 / 0
except Exception:           # faengt ALLES -- auch ZeroDivisionError
    print('Irgendein Fehler')
except ZeroDivisionError:   # wird NIE erreicht!
    print('Division durch Null')

print()

# RICHTIG -- Spezifisches zuerst, Allgemeines danach
try:
    x = 1 / 0
except ZeroDivisionError:   # spezifisch zuerst
    print('Division durch Null')
except Exception:            # allgemeiner Fallback
    print('Unbekannter Fehler')
```

5. raise -- Exceptions selbst ausloesen
</span>
</div>

Manchmal stellt man fest, dass eine Eingabe ungueltig ist -- und moechte das dem Aufrufer mitteilen. Statt `return None` oder `return -1` ist es sauberer, eine Exception auszuloesen. Der Aufrufer kann dann selbst entscheiden wie er damit umgeht.

```python
def setze_hp(wert):
    if not isinstance(wert, int):
        raise TypeError(f'HP muss ein int sein, nicht {type(wert).__name__}')
    if wert < 0:
        raise ValueError(f'HP kann nicht negativ sein: {wert}')
    return wert

print(setze_hp(100))    # 100

try:
    setze_hp(-5)
except ValueError as e:
    print(f'Fehler: {e}')

try:
    setze_hp('hundert')
except TypeError as e:
    print(f'Fehler: {e}')
```

```python
# raise ohne Argument -- faengt und wirft die aktuelle Exception weiter
# Nuetzlich wenn man etwas loggen will, aber die Exception nicht schlucken

def verarbeite_heldendaten(daten):
    try:
        hp = int(daten['hp'])
        return hp
    except (KeyError, ValueError) as e:
        print(f'LOG: Fehler beim Verarbeiten der Heldendaten: {e}')
        raise   # Exception weiterwerfen -- Aufrufer soll sie auch sehen

try:
    verarbeite_heldendaten({'hp': 'ungueltig'})
except ValueError:
    print('Aufrufer hat den Fehler empfangen')
```

6. assert -- Vorbedingungen pruefen
</span>
</div>

`assert bedingung, meldung` ist eine kompakte Form um Vorbedingungen zu pruefen. Wenn die Bedingung `False` ist, wirft Python einen `AssertionError` mit der angegebenen Meldung.

Es ist konzeptionell verschieden von `raise`: `assert` sagt *"das darf hier niemals passieren -- wenn doch, ist der Code kaputt"*. `raise` sagt *"das ist ein erwartbarer Fehlerfall den der Aufrufer behandeln soll"*.

```python
def teile_party(helden, gruppen):
    assert len(helden) > 0, 'Keine Helden vorhanden'
    assert gruppen > 0, 'Anzahl Gruppen muss positiv sein'
    assert len(helden) >= gruppen, 'Zu wenige Helden fuer so viele Gruppen'

    groesse = len(helden) // gruppen
    return [helden[i:i+groesse] for i in range(0, len(helden), groesse)]

helden = ['Aldric', 'Lyria', 'Thorn', 'Mira']
print(teile_party(helden, 2))   # [['Aldric', 'Lyria'], ['Thorn', 'Mira']]

try:
    teile_party([], 2)
except AssertionError as e:
    print(f'AssertionError: {e}')
```

**Zusammenfassung**

| Konzept | Kernaussage |
|---------|-------------|
| `try/except` | Versuch es -- wenn ein bestimmter Fehler auftritt, handle ihn |
| Mehrere `except` | Von spezifisch nach allgemein -- Python nimmt den ersten passenden |
| `except (E1, E2)` | Mehrere Typen in einem Block |
| `else` | Laeuft nur wenn kein Fehler aufgetreten ist |
| `finally` | Laeuft immer -- fuer Aufraeumarbeiten |
| `raise` | Exception selbst ausloesen |
| `raise` (ohne Argument) | Aktuelle Exception weiterwerfen |
| `assert` | Vorbedingung pruefen -- fuer interne Konsistenz |
