1. Warum automatisiert testen?
</span>
</div>

Manuelles Testen -- Code ausfuehren und Ausgabe visuell pruefen -- hat ein zentrales Problem: Es skaliert nicht. Bei jeder Aenderung muesste man alle bisherigen Szenarien erneut manuell durchgehen. Automatisierte Tests tun das in Sekunden.

**Das AAA-Muster** beschreibt den Aufbau eines guten Tests:
- **Arrange** -- Vorbedingungen herstellen
- **Act** -- Die zu testende Aktion ausfuehren
- **Assert** -- Ergebnis pruefen

**Arten von Tests:**

| Typ | Testet | Geschwindigkeit |
|-----|--------|----------------|
| **Unittest** | Eine Funktion/Klasse isoliert | Sehr schnell |
| **Integrationstest** | Zusammenspiel mehrerer Komponenten | Mittel |
| **Systemtest** | Das gesamte System | Langsam |

Wir konzentrieren uns auf Unittests -- die Basis jeder Teststrategie.

2. Grundstruktur
</span>

Jede Testklasse erbt von `unittest.TestCase`. Testmethoden muessen mit `test_` beginnen -- nur dann werden sie automatisch ausgefuehrt.

````python
import unittest

# --- Zu testender Code ---
def addiere(a, b):
    return a + b

def dividiere(a, b):
    if b == 0:
        raise ValueError('Division durch Null nicht erlaubt')
    return a / b

def ist_palindrom(text):
    bereinigt = text.lower().replace(' ', '')
    return bereinigt == bereinigt[::-1]

# --- Testklasse ---
class TestRechner(unittest.TestCase):

    def test_addiere_positiv(self):
        # Arrange
        a, b = 3, 5
        # Act
        ergebnis = addiere(a, b)
        # Assert
        self.assertEqual(ergebnis, 8)

    def test_addiere_negativ(self):
        self.assertEqual(addiere(-1, -2), -3)

    def test_addiere_float(self):
        self.assertAlmostEqual(addiere(0.1, 0.2), 0.3, places=10)

    def test_dividiere_ganzzahl(self):
        self.assertEqual(dividiere(10, 2), 5.0)

    def test_dividiere_durch_null_wirft_exception(self):
        with self.assertRaises(ValueError):
            dividiere(5, 0)


class TestPalindrom(unittest.TestCase):

    def test_einfaches_palindrom(self):
        self.assertTrue(ist_palindrom('racecar'))

    def test_kein_palindrom(self):
        self.assertFalse(ist_palindrom('python'))

    def test_grosskleinschreibung_ignoriert(self):
        self.assertTrue(ist_palindrom('Anna'))

    def test_mit_leerzeichen(self):
        self.assertTrue(ist_palindrom('never odd or even'))


# In Notebooks so ausfuehren:
unittest.main(argv=[''], exit=False, verbosity=2)
````

3. Assert-Methoden im Ueberblick
</span>
</div>

| Methode | Prueft | Typischer Einsatz |
|---------|--------|------------------|
| `assertEqual(a, b)` | `a == b` | Rueckgabewert pruefen |
| `assertNotEqual(a, b)` | `a != b` | Unveraendertheit pruefen |
| `assertTrue(x)` | `bool(x) is True` | Bedingung pruefen |
| `assertFalse(x)` | `bool(x) is False` | Verneinung pruefen |
| `assertIsNone(x)` | `x is None` | Kein Rueckgabewert |
| `assertIsNotNone(x)` | `x is not None` | Objekt wurde erstellt |
| `assertIn(a, b)` | `a in b` | Listeninhalt pruefen |
| `assertNotIn(a, b)` | `a not in b` | Ausschluss pruefen |
| `assertRaises(E)` | Exception ausgeloest | Fehlerverhalten testen |
| `assertAlmostEqual(a, b)` | `a ≈ b` | Kommazahlen vergleichen |
| `assertIsInstance(a, T)` | `isinstance(a, T)` | Typ pruefen |

````python
import unittest

class TestAssertionen(unittest.TestCase):

    def test_alle_gaengigen_assertions(self):
        # Gleichheit
        self.assertEqual('hallo'.upper(), 'HALLO')
        self.assertNotEqual(1, 2)

        # Wahrheitswert
        self.assertTrue('python' in 'ich lerne python')
        self.assertFalse([] == [1])

        # None-Pruefung
        self.assertIsNone(None)
        self.assertIsNotNone(42)

        # Enthaltensein
        self.assertIn(3, [1, 2, 3, 4])
        self.assertNotIn('java', ['python', 'rust'])

        # Typ
        self.assertIsInstance(3.14, float)
        self.assertIsInstance([1, 2], list)

        # Kommazahlen
        self.assertAlmostEqual(0.1 + 0.2, 0.3, places=10)

    def test_raises_als_context_manager(self):
        # assertRaises als Context Manager -- empfohlene Form
        with self.assertRaises(ValueError):
            int('kein int')

        with self.assertRaises(TypeError):
            'text' + 42

        with self.assertRaises(KeyError):
            {}['nicht_vorhanden']

unittest.main(argv=[''], exit=False, verbosity=2)
````

4. setUp() und tearDown()
</span>
</div>

`setUp()` wird vor **jedem** Testfall aufgerufen und stellt einen sauberen Ausgangszustand her. `tearDown()` laeuft nach jedem Testfall und raumt auf -- z.B. Testdateien loeschen oder Datenbankverbindungen schliessen.

Warum das wichtig ist: Tests muessen voneinander unabhaengig sein. Ein fehlschlagender Test darf den naechsten nicht beeinflussen.

````python
import unittest
import os

class Notizblock:
    def __init__(self):
        self._notizen = []

    def hinzufuegen(self, text):
        if not text.strip():
            raise ValueError('Notiz darf nicht leer sein')
        self._notizen.append(text.strip())

    def loeschen(self, index):
        if index < 0 or index >= len(self._notizen):
            raise IndexError('Ungueltigter Index')
        self._notizen.pop(index)

    def alle(self):
        return list(self._notizen)

    def anzahl(self):
        return len(self._notizen)


class TestNotizblock(unittest.TestCase):

    def setUp(self):
        # Vor jedem Test: frischer Notizblock mit zwei Eintraegen
        self.block = Notizblock()
        self.block.hinzufuegen('Erste Notiz')
        self.block.hinzufuegen('Zweite Notiz')

    def tearDown(self):
        # Nach jedem Test: hier nichts noetig
        # Bei Datenbankverbindungen oder Dateien wuerde man hier aufraeumen
        pass

    def test_anfangszustand(self):
        self.assertEqual(self.block.anzahl(), 2)

    def test_hinzufuegen(self):
        self.block.hinzufuegen('Dritte Notiz')
        self.assertEqual(self.block.anzahl(), 3)
        self.assertIn('Dritte Notiz', self.block.alle())

    def test_leere_notiz_verboten(self):
        with self.assertRaises(ValueError):
            self.block.hinzufuegen('   ')  # nur Leerzeichen

    def test_loeschen(self):
        self.block.loeschen(0)
        self.assertEqual(self.block.anzahl(), 1)
        self.assertNotIn('Erste Notiz', self.block.alle())

    def test_loeschen_ungueltig(self):
        with self.assertRaises(IndexError):
            self.block.loeschen(99)

    def test_alle_gibt_kopie(self):
        # alle() soll eine Kopie zurueckgeben, nicht die interne Liste
        liste = self.block.alle()
        liste.append('manipuliert')
        self.assertEqual(self.block.anzahl(), 2)  # intern unveraendert


unittest.main(argv=[''], exit=False, verbosity=2)
````

5. Tests mit Dateien -- setUp und tearDown in der Praxis
</span>
</div>

````python
import unittest
import os

def zeilen_zaehlen(pfad):
    """Zaehlt die Zeilen in einer Textdatei."""
    with open(pfad, 'r') as f:
        return len(f.readlines())

def woerter_zaehlen(pfad):
    """Zaehlt alle Woerter in einer Textdatei."""
    with open(pfad, 'r') as f:
        return sum(len(zeile.split()) for zeile in f)


class TestDateiAnalyse(unittest.TestCase):

    def setUp(self):
        # Testdatei anlegen
        self.testdatei = 'test_temp.txt'
        with open(self.testdatei, 'w') as f:
            f.write('Hallo Welt\n')
            f.write('Python ist toll\n')
            f.write('Testen macht Spass\n')

    def tearDown(self):
        # Testdatei nach jedem Test loeschen
        if os.path.exists(self.testdatei):
            os.remove(self.testdatei)

    def test_zeilen(self):
        self.assertEqual(zeilen_zaehlen(self.testdatei), 3)

    def test_woerter(self):
        self.assertEqual(woerter_zaehlen(self.testdatei), 9)

    def test_datei_nicht_vorhanden(self):
        with self.assertRaises(FileNotFoundError):
            zeilen_zaehlen('existiert_nicht.txt')


unittest.main(argv=[''], exit=False, verbosity=2)
````

6. Teststruktur in einem Projekt
</span>
</div>

In einem echten Projekt liegen Tests in separaten Dateien. Konventionen:

```
mein_projekt/
    rechner.py          # Produktionscode
    tests/
        __init__.py
        test_rechner.py # Tests
```

Dateiname beginnt mit `test_` -- dann findet `unittest` sie automatisch.

Tests ausfuehren im Terminal:

```bash
# Einzelne Datei
python -m unittest tests/test_rechner.py

# Alle Tests im Projekt finden und ausfuehren
python -m unittest discover

# Ausfuehrlich mit Einzelergebnissen
python -m unittest discover -v
```

**Gute Testnamen** erklaeren was getestet wird:

| Schlecht | Gut |
|----------|-----|
| `test_1` | `test_addiere_zwei_positive_zahlen` |
| `test_fehler` | `test_dividiere_durch_null_wirft_value_error` |
| `test_ok` | `test_leere_liste_gibt_none_zurueck` |

---

**Zusammenfassung**

| Konzept | Kernaussage |
|---------|-------------|
| `unittest.TestCase` | Basisklasse -- alle Testklassen erben davon |
| `test_*` | Nur Methoden mit diesem Praefix werden automatisch ausgefuehrt |
| `setUp()` | Vor jedem Testfall -- sauberer Ausgangszustand |
| `tearDown()` | Nach jedem Testfall -- aufraeumen |
| `assertEqual` | Haeufigste Assert-Methode |
| `assertRaises` | Exception-Verhalten als Context Manager testen |
| `assertAlmostEqual` | Fuer Kommazahlen -- wegen Gleitkommafehlern |
| Testisolation | Jeder Test muss unabhaengig von anderen funktionieren |
