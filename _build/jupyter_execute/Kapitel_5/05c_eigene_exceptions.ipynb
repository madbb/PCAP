1. Warum eigene Exception-Klassen?
</span>
</div>

Pythons eingebaute Exceptions sind allgemein: `ValueError`, `TypeError`, `RuntimeError`. Sie sagen was technisch falsch gelaufen ist. Aber manchmal will man sagen was **inhaltlich** falsch gelaufen ist -- bezogen auf die eigene Anwendungslogik.

Stell dir vor, du baust ein Rollenspielsystem. Ein Held versucht einen Zauber zu wirken, hat aber zu wenig Mana. Das ist kein `ValueError` -- es ist ein `ZuWenigManaFehler`. Dieser Fehler traegt eine eigene Bedeutung im Kontext des Spiels.

Weil Exceptions Klassen sind, kann man sie auch selbst schreiben.

2. Die einfachste Form
</span>
</div>

Die minimale eigene Exception-Klasse erbt einfach von `Exception` und hat keinen eigenen Code. `pass` genuegt -- alles andere kommt von der Elternklasse.

```python
class ZuWenigManaFehler(Exception):
    pass

class HeldNichtGefundenFehler(Exception):
    pass


def wirke_zauber(held_mana, zaubermana):
    if held_mana < zaubermana:
        raise ZuWenigManaFehler(
            f'Zu wenig Mana: {held_mana} vorhanden, {zaubermana} benoetigt'
        )
    return held_mana - zaubermana


try:
    mana = wirke_zauber(30, 50)
except ZuWenigManaFehler as e:
    print(f'Fehler: {e}')   # Zu wenig Mana: 30 vorhanden, 50 benoetigt

# Pruefe Vererbung
try:
    raise ZuWenigManaFehler('Test')
except Exception as e:                       # greift auch!
    print(f'Auch Exception: {type(e).__name__}')
```

3. Mit __init__ und eigenen Attributen
</span>
</div>

Haeufig will man mehr als nur eine Meldung transportieren -- auch konkrete Daten. Dafuer schreibt man einen eigenen `__init__`. Der Trick: Man muss `super().__init__()` aufrufen, damit die Basisklasse ihre Initialisierung durchfuehren kann -- insbesondere `self.args` setzen.

```python
class ZuWenigManaFehler(Exception):
    def __init__(self, vorhanden, benoetigt):
        self.vorhanden = vorhanden
        self.benoetigt = benoetigt
        # super().__init__() setzt self.args und ermoeglicht str(e)
        super().__init__(
            f'Zu wenig Mana: {vorhanden} vorhanden, {benoetigt} benoetigt'
        )


try:
    raise ZuWenigManaFehler(30, 50)
except ZuWenigManaFehler as e:
    print(f'Meldung:          {e}')             # str(e) -- die Meldung
    print(f'Vorhanden:        {e.vorhanden}')   # eigenes Attribut
    print(f'Benoetigt:        {e.benoetigt}')   # eigenes Attribut
    print(f'Differenz:        {e.benoetigt - e.vorhanden}')
    print(f'args:             {e.args}')        # was super().__init__ gesetzt hat
```

4. super().__init__() vs. Exception.__init__() -- und was self.args wirklich ist
</span>
</div>

Hier kommt das Herzstuck dieses Notebooks -- und der Schluessel um PCAP-Aufgaben zu loesen, die mit eigenen Exception-Klassen und `args` arbeiten.

`super().__init__(meldung)` und `Exception.__init__(self, meldung)` sind in einfacher Vererbung funktional identisch. Der Unterschied wird sichtbar bei Mehrfachvererbung -- aber die PCAP-Pruefung setzt bewusst beide Formen ein.

Was `Exception.__init__` intern tut:
- Es setzt `self.args = (meldung,)` -- ein Tupel mit den uebergebenen Argumenten
- Und das bestimmt was `str(e)` und `print(e)` anzeigen

Man kann `self.args` auch **danach manuell ueberschreiben**. Das ist selten aber in der PCAP-Pruefung ein wichtiges Konzept.

```python
# super().__init__ und Exception.__init__ -- Vergleich
class FehlerA(Exception):
    def __init__(self, msg):
        super().__init__(msg)          # empfohlener Weg

class FehlerB(Exception):
    def __init__(self, msg):
        Exception.__init__(self, msg)  # direkter Aufruf -- aequivalent

a = FehlerA('Meldung A')
b = FehlerB('Meldung B')

print(str(a), a.args)   # Meldung A  ('Meldung A',)
print(str(b), b.args)   # Meldung B  ('Meldung B',)
```

```python
# Was passiert wenn man Exception.__init__ mit anderen Argumenten aufruft
# als spaeter in self.args steht?

class TrickyException(Exception):
    def __init__(self, msg):
        # Schritt 1: Exception.__init__ bekommt msg + msg
        # --> self.args wird auf (msg + msg,) gesetzt
        Exception.__init__(self, msg + msg)

        # Schritt 2: self.args wird danach manuell ueberschrieben
        # --> self.args ist jetzt (msg,)
        self.args = (msg,)

e = TrickyException('hallo')
print(e.args)   # ('hallo',)  -- Schritt 2 hat Schritt 1 ueberschrieben
print(str(e))   # 'hallo'     -- ein Element: direkt als String

try:
    raise TrickyException('ex')
except TrickyException as e:
    print(e)    # 'ex'  -- nicht 'exex'!
```

Das ist der entscheidende Mechanismus: `Exception.__init__(self, msg + msg)` setzt `self.args = ('exex',)`. Die naechste Zeile `self.args = (msg,)` setzt `self.args` auf `('ex',)` zurueck. `print(e)` zeigt dann `'ex'` -- nicht `'exex'`.

**Merke:** `self.args` ist ein ganz normales Attribut. Es kann nach dem `__init__`-Aufruf der Elternklasse noch veraendert werden. Was am Ende in `self.args` steht, bestimmt die Ausgabe von `str(e)`.

```python
# Die verschiedenen Auswirkungen von args auf str(e) -- Ueberblick

class BeispielException(Exception):
    pass

# Ein Argument -> str(e) gibt den Wert direkt
e1 = BeispielException('hallo')
print(f'args={e1.args}  str={str(e1)}')

# Zwei Argumente -> str(e) gibt das Tupel
e2 = BeispielException('hallo', 'welt')
print(f'args={e2.args}  str={str(e2)}')

# Kein Argument -> str(e) ist leer
e3 = BeispielException()
print(f'args={e3.args}  str={str(e3)!r}')

# Manuelles Ueberschreiben von args
e4 = BeispielException('urspruenglich')
e4.args = ('ueberschrieben',)
print(f'args={e4.args}  str={str(e4)}')
```

5. Eigene Exception-Hierarchien bauen
</span>
</div>

Genau wie Pythons eingebaute Exceptions eine Hierarchie haben, kann man eigene Hierarchien aufbauen. Das erlaubt einem Aufrufer, entweder spezifisch oder allgemein zu fangen.

```python
# Eigene Hierarchie fuer ein Rollenspielsystem
class SpielFehler(Exception):
    """Basisklasse fuer alle Spielfehler."""
    pass

class HeldFehler(SpielFehler):
    """Fehler die mit einem Helden zusammenhaengen."""
    def __init__(self, held, meldung):
        self.held = held
        super().__init__(f'[{held}] {meldung}')

class ZuWenigManaFehler(HeldFehler):
    def __init__(self, held, vorhanden, benoetigt):
        self.vorhanden = vorhanden
        self.benoetigt = benoetigt
        super().__init__(held,
            f'Zu wenig Mana ({vorhanden}/{benoetigt})')

class HeldBesiegtFehler(HeldFehler):
    def __init__(self, held):
        super().__init__(held, 'Dieser Held wurde bereits besiegt')

class KarteFehler(SpielFehler):
    """Fehler die mit der Spielkarte zusammenhaengen."""
    pass


def wirke_zauber(held, mana_vorhanden, mana_benoetigt):
    raise ZuWenigManaFehler(held, mana_vorhanden, mana_benoetigt)


# Spezifisch fangen
try:
    wirke_zauber('Lyria', 20, 50)
except ZuWenigManaFehler as e:
    print(f'Mana-Problem: {e}')
    print(f'Fehlend: {e.benoetigt - e.vorhanden} Mana')

print()

# Allgemein fangen -- faengt alle HeldFehler
try:
    wirke_zauber('Lyria', 20, 50)
except HeldFehler as e:
    print(f'Held-Problem: {e} (Held: {e.held})')

print()

# Noch allgemeiner -- faengt alle Spielfehler
try:
    wirke_zauber('Lyria', 20, 50)
except SpielFehler as e:
    print(f'Spielfehler: {e}')
```

Wann von Exception, wann von spezifischeren Klassen erben?
</span>
</div>

| Erbt von | Wann sinnvoll |
|----------|---------------|
| `Exception` | Allgemeiner Anwendungsfehler, der nicht in eine eingebaute Kategorie passt |
| `ValueError` | Wenn der Wert inhaltlich falsch ist -- passt semantisch |
| `TypeError` | Wenn der Typ falsch ist |
| `RuntimeError` | Wenn ein Fehler zur Laufzeit auftritt, der schwer kategorisierbar ist |
| Eigene Basisklasse | Wenn man eine Gruppe verwandter Fehler zusammenfassen will |

Der Vorteil von spezifischem Erben: Aufrufer die schon `except ValueError` haben, fangen die eigene Exception automatisch auch ab -- auch ohne den eigenen Typ zu kennen.

```python
class UngueltigeHPFehler(ValueError):
    """Erbt von ValueError -- HP-Wert ist inhaltlich ungueltig."""
    pass

def setze_hp(wert):
    if not isinstance(wert, int):
        raise TypeError(f'HP muss int sein')
    if wert < 0 or wert > 9999:
        raise UngueltigeHPFehler(f'HP muss zwischen 0 und 9999 liegen, nicht {wert}')
    return wert

# Spezifisch fangen
try:
    setze_hp(-5)
except UngueltigeHPFehler as e:
    print(f'HP-Fehler: {e}')

# Auch allgemein durch ValueError fangbar
try:
    setze_hp(-5)
except ValueError as e:
    print(f'ValueError: {e}  (Typ: {type(e).__name__})')
```

6. Pruefungsrelevante Muster -- Schritt fuer Schritt analysieren
</span>
</div>

PCAP-Pruefungen enthalten oft Aufgaben bei denen man Code mit eigenen Exceptions lesen und die Ausgabe vorhersagen muss. Das erfordert genau zu verstehen was `__init__`, `super().__init__()`, `self.args` und `str(e)` zusammen ergeben.

Die Vorgehensweise: Jeden Schritt von `__init__` in der richtigen Reihenfolge verfolgen und dabei beobachten wie sich `self.args` veraendert.

```python
# Muster 1: __init__ ruft super().__init__ mit veraendertem Argument auf
class VerdoppelteFehler(Exception):
    def __init__(self, msg):
        super().__init__(msg + msg)  # 'ab' -> args = ('abab',)

try:
    raise VerdoppelteFehler('ab')
except VerdoppelteFehler as e:
    print(e.args)   # ('abab',)
    print(e)        # abab
```

```python
# Muster 2: super().__init__ mit veraendertem Argument, dann self.args ueberschreiben
class UeberschriebenFehler(Exception):
    def __init__(self, msg):
        super().__init__(msg + msg)  # Schritt 1: args = ('abab',)
        self.args = (msg,)           # Schritt 2: args = ('ab',)  -- ueberschrieben!

try:
    raise UeberschriebenFehler('ab')
except UeberschriebenFehler as e:
    print(e.args)   # ('ab',)  -- Schritt 2 gewinnt
    print(e)        # ab       -- nicht abab!
```

```python
# Muster 3: Exception.__init__ statt super().__init__ -- Unterschied im except-Block?
class FehlerMitDirektemAufruf(Exception):
    def __init__(self, msg):
        Exception.__init__(self, msg + msg)  # direkt statt super()
        self.args = (msg,)                   # ueberschreiben

try:
    raise FehlerMitDirektemAufruf('ex')
except FehlerMitDirektemAufruf as e:
    print(e)        # ex  -- nicht exex
except Exception as e:
    print('Exception')  # dieser Block wird nicht erreicht

# Warum? FehlerMitDirektemAufruf ist spezifischer als Exception
# Python nimmt den ersten passenden except-Block -- das ist FehlerMitDirektemAufruf
```

```python
# Muster 4: Welcher except-Block greift?
class MeinFehler(Exception):
    pass

class SpeziellerFehler(MeinFehler):
    pass

try:
    raise SpeziellerFehler('Problem')
except MeinFehler as e:            # greift! SpeziellerFehler erbt von MeinFehler
    print(f'MeinFehler: {e}')
except SpeziellerFehler as e:      # wird nie erreicht -- MeinFehler war zuerst
    print('Speziell')

print()

# RICHTIGE Reihenfolge: spezifisch zuerst
try:
    raise SpeziellerFehler('Problem')
except SpeziellerFehler as e:      # spezifisch zuerst -- greift
    print(f'SpeziellerFehler: {e}')
except MeinFehler as e:
    print('MeinFehler')            # wird nicht erreicht
```

**Zusammenfassung**

| Konzept | Kernaussage |
|---------|-------------|
| Eigene Exception | Klasse die von `Exception` oder Unterklasse erbt |
| Minimale Form | `class X(Exception): pass` -- alles von Elternklasse geerbt |
| `super().__init__(msg)` | Setzt `self.args = (msg,)` -- empfohlener Weg |
| `Exception.__init__(self, msg)` | Identisch in einfacher Vererbung -- alternative Schreibweise |
| `self.args` | Tupel -- bestimmt was `str(e)` und `print(e)` anzeigen |
| `self.args` ueberschreiben | Moeglich nach dem Eltern-`__init__`-Aufruf -- letzter Wert gilt |
| Ein Element in args | `str(e)` gibt das Element direkt aus |
| Mehrere Elemente in args | `str(e)` gibt das gesamte Tupel aus |
| Eigene Hierarchien | Kindklassen eigener Exceptions sind sinnvoll fuer Gruppen von Fehlern |

**Checklist fuer Pruefungsaufgaben mit eigenen Exceptions:**
1. Was wird an `Exception.__init__` / `super().__init__` uebergeben? --> das setzt `self.args` initial
2. Wird `self.args` danach noch manuell gesetzt? --> dann gilt der neue Wert
3. Wie viele Elemente hat `self.args` am Ende? --> ein Element: direkte Ausgabe; mehrere: Tupel
4. Welcher `except`-Block greift? --> spezifischste Klasse zuerst, von oben nach unten
