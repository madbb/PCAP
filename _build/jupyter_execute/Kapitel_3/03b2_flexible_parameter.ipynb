1. Das Problem: Wie viele Argumente bekommt diese Funktion?
</span>
</div>

Normale Funktionen haben eine feste Anzahl Parameter:

```python
def addiere(a, b):
    return a + b
```

`addiere` erwartet genau zwei Argumente -- nicht eins, nicht drei. Das ist meistens gut so. Aber manchmal weiss man beim Schreiben der Funktion noch nicht, wie viele Argumente ankommen werden.

Stell dir vor, du willst eine Funktion schreiben die beliebig viele Zahlen addiert -- zwei, fuenf, zehn, egal:

````python
# Erster Versuch: feste Parameter
def addiere_zwei(a, b):          return a + b
def addiere_drei(a, b, c):       return a + b + c
def addiere_vier(a, b, c, d):    return a + b + c + d

# Das skaliert nicht -- und was wenn man 10 Zahlen addieren will?

# Zweiter Versuch: Liste als Parameter
def addiere_liste(zahlen):
    return sum(zahlen)

print(addiere_liste([1, 2, 3, 4, 5]))   # funktioniert
# Aber der Aufruf ist umstaendlich -- man muss jedes Mal eine Liste bauen
````

Es waere schoen, die Funktion so aufrufen zu koennen:

```python
addiere(1, 2, 3, 4, 5)   # ohne Liste, einfach kommasepariert
```

Genau dafuer gibt es `*args`.

2. *args -- beliebig viele Positionalargumente
</span>
</div>

Ein `*` vor einem Parameternamen in der Funktionsdefinition bedeutet: Alle weiteren Positionalargumente werden in diesem Parameter als **Tupel** gesammelt. Der Name `args` ist dabei nur Konvention -- `*zahlen` oder `*werte` funktioniert genauso. Das `*` ist das Entscheidende.

````python
def addiere(*args):
    print(f'args ist ein Tupel: {args}')
    print(f'Typ: {type(args)}')
    return sum(args)

print(addiere(1, 2))           # args = (1, 2)
print(addiere(1, 2, 3, 4, 5))  # args = (1, 2, 3, 4, 5)
print(addiere())               # args = ()  -- leeres Tupel, auch erlaubt
````

`args` ist ein normales Tupel -- man kann es iterieren, mit `len()` seine Laenge abfragen, mit Indexzugriff einzelne Elemente lesen. Der einzige Unterschied: Es wird automatisch aus beliebig vielen Argumenten zusammengebaut.

`*args` kann auch mit normalen Parametern kombiniert werden -- aber `*args` muss nach den normalen Parametern kommen:

````python
def held_aktionen(held_name, *aktionen):
    print(f'{held_name} fuehrt aus:')
    for aktion in aktionen:
        print(f'  - {aktion}')

held_aktionen('Aldric', 'Angriff', 'Verteidigung')
print()
held_aktionen('Lyria', 'Feuerball', 'Heilung', 'Teleport', 'Unsichtbarkeit')
print()
held_aktionen('Thorn')  # keine Aktionen -- *aktionen ist leer
````

<span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">
* beim Aufruf -- eine Liste auspacken
</span>
</div>

Das `*` funktioniert auch beim **Aufruf** einer Funktion -- umgekehrt. Es packt eine Liste oder ein Tupel in einzelne Argumente aus:

````python
def addiere(a, b, c):
    return a + b + c

werte = [1, 2, 3]

# Ohne *: Fehler -- eine Liste ist nicht drei Argumente
try:
    addiere(werte)
except TypeError as e:
    print(f'Fehler: {e}')

# Mit *: Liste wird in drei einzelne Argumente aufgepackt
print(addiere(*werte))   # identisch mit addiere(1, 2, 3)
````

3. **kwargs -- beliebig viele Keyword-Argumente
</span>
</div>

Was `*args` fuer Positionalargumente macht, macht `**kwargs` fuer **Keyword-Argumente** -- also Argumente die mit einem Namen uebergeben werden.

`**kwargs` sammelt alle benannten Argumente die keinem Parameter zugeordnet sind als **Dictionary**. Schluessel ist der Argumentname, Wert ist der Argumentwert.

````python
def erstelle_held(**kwargs):
    print(f'kwargs ist ein Dict: {kwargs}')
    print(f'Typ: {type(kwargs)}')
    for schluessel, wert in kwargs.items():
        print(f'  {schluessel}: {wert}')

erstelle_held(name='Aldric', klasse='Krieger', hp=100, staerke=15)
````

`kwargs` ist ein normales Dictionary. Man kann darauf zugreifen wie auf jedes andere Dict -- `.get()`, Indexzugriff, `.items()`, `.keys()`. Der einzige Unterschied: Es wird aus den uebergebenen Keyword-Argumenten zusammengebaut.

Ein praktischeres Beispiel -- eine Funktion die Helden-Attribute flexibel setzt:

````python
def helden_status(name, **attribute):
    print(f'--- {name} ---')
    # Bekannte Attribute mit Standardwerten
    hp       = attribute.get('hp',      100)
    staerke  = attribute.get('staerke', 10)
    klasse   = attribute.get('klasse',  'Unbekannt')
    print(f'Klasse:  {klasse}')
    print(f'HP:      {hp}')
    print(f'Staerke: {staerke}')
    # Unbekannte Attribute einfach ausgeben
    for key, val in attribute.items():
        if key not in ('hp', 'staerke', 'klasse'):
            print(f'{key}: {val}')

helden_status('Aldric', klasse='Krieger', hp=120, staerke=18, ruestung='Platte')
print()
helden_status('Lyria', klasse='Magierin', mana=80, zauberstaerke=25)
````

<span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">
** beim Aufruf -- ein Dictionary auspacken
</span>
</div>

Genau wie `*` eine Liste auspackt, packt `**` ein Dictionary in Keyword-Argumente aus:

````python
def erstelle_held(name, klasse, hp):
    print(f'{name} der {klasse} mit {hp} HP')

held_daten = {'name': 'Thorn', 'klasse': 'Waldlaeufer', 'hp': 90}

# Ohne **: jedes Argument einzeln uebergeben -- umstaendlich
erstelle_held(held_daten['name'], held_daten['klasse'], held_daten['hp'])

# Mit **: Dict wird in Keyword-Argumente aufgepackt
erstelle_held(**held_daten)   # identisch mit erstelle_held(name='Thorn', ...)
````

4. *args und **kwargs kombinieren
</span>
</div>

Beide koennen zusammen verwendet werden. Die Reihenfolge in der Signatur ist dabei fest:

```
def funktion(normal, *args, **kwargs)
              ↑        ↑       ↑
         normal    beliebig  beliebig
         Parameter positional keyword
```

````python
def kampflog(aktion, *beteiligte, **details):
    print(f'Aktion: {aktion}')
    print(f'Beteiligte: {beteiligte}')
    print(f'Details: {details}')
    print()

kampflog('Angriff', 'Aldric', 'Drache',
         schaden=45, kritisch=True, runde=3)

kampflog('Heilung', 'Lyria',
         menge=30, ziel='Aldric')

kampflog('Flucht', 'Thorn')   # nur Pflichtparameter + ein Beteiligter
````

5. Argumente transparent weitergeben -- der wichtigste Anwendungsfall
</span>
</div>

Der haeufigste Einsatz von `*args` und `**kwargs` ist das **transparente Weiterreichen** von Argumenten -- wenn eine Funktion eine andere aufruft und deren Argumente nicht kennt oder kennen soll.

Zwei konkrete Stellen wo das im Kurs auftaucht:

````python
# Anwendungsfall 1: Decorators
# Ein Decorator verhuellt eine Funktion -- er weiss nicht welche Parameter sie hat.
# Mit *args und **kwargs leitet er alle Argumente einfach weiter.

def logge_aufruf(funktion):
    def wrapper(*args, **kwargs):         # nimmt ALLES entgegen
        print(f'{funktion.__name__} aufgerufen')
        ergebnis = funktion(*args, **kwargs)  # reicht ALLES weiter
        print(f'{funktion.__name__} fertig')
        return ergebnis
    return wrapper

@logge_aufruf
def angreifen(held, ziel, schaden):
    print(f'  {held} greift {ziel} an ({schaden} Schaden)')
    return schaden

@logge_aufruf
def heilen(held, menge=20):
    print(f'  {held} heilt sich um {menge}')

angreifen('Aldric', 'Drache', 45)
print()
heilen('Lyria', menge=30)
````

Ohne `*args, **kwargs` muesste `wrapper` dieselbe Signatur haben wie die dekorierte Funktion -- das wuerde jeden Decorator auf genau eine Funktionssignatur beschraenken. Mit `*args, **kwargs` funktioniert ein Decorator fuer **jede beliebige Funktion**.

````python
# Anwendungsfall 2: Vererbungsketten
# In einer Klassenhierarchie muss super().__init__() Parameter an die
# naechste Klasse in der MRO weiterreichen -- auch wenn man nicht weiss
# welche Klasse das ist.

class Basis:
    def __init__(self, x, **kwargs):   # x nehmen, Rest weiterreichen
        super().__init__(**kwargs)
        self.x = x

class Erweiterung(Basis):
    def __init__(self, y, **kwargs):   # y nehmen, Rest weiterreichen
        super().__init__(**kwargs)
        self.y = y

class Kombination(Erweiterung, Basis):
    def __init__(self, x, y):
        super().__init__(x=x, y=y)    # alle als kwargs uebergeben

obj = Kombination(x=10, y=20)
print(obj.x, obj.y)   # 10  20 -- beide Attribute gesetzt
````

6. Keyword-only Parameter
</span>
</div>

Es gibt noch einen verwandten Mechanismus: Parameter die **nur** als Keyword-Argument uebergeben werden duerfen -- nicht als Positionalargument. Das erzwingt man, indem man den Parameter **nach** einem `*` in der Signatur platziert.

````python
# Normaler Parameter: positional oder keyword -- beides erlaubt
def angriff_normal(held, ziel, schaden):
    print(f'{held} -> {ziel}: {schaden}')

angriff_normal('Aldric', 'Drache', 45)          # positional
angriff_normal('Aldric', ziel='Drache', schaden=45)  # keyword

print()

# Keyword-only: Parameter nach * duerfen NUR als Keyword uebergeben werden
def angriff_sicher(held, ziel, *, schaden, kritisch=False):
    # schaden und kritisch sind keyword-only
    treffer = schaden * 2 if kritisch else schaden
    print(f'{held} -> {ziel}: {treffer} Schaden')

angriff_sicher('Aldric', 'Drache', schaden=45)               # ok
angriff_sicher('Aldric', 'Drache', schaden=45, kritisch=True) # ok

try:
    angriff_sicher('Aldric', 'Drache', 45)  # positional fuer schaden -- Fehler!
except TypeError as e:
    print(f'TypeError: {e}')
````

Keyword-only Parameter sind nuetzlich wenn ein Argument so wichtig oder fehlertraechtig ist, dass man sicherstellen will, dass es explizit benannt wird -- man sieht beim Aufruf sofort was gemeint ist.

---

**Zusammenfassung**

| Syntax | Wo | Bedeutung |
|--------|----|-----------|
| `*args` in Def | Parameter | Sammelt beliebig viele Positionalargumente als **Tupel** |
| `**kwargs` in Def | Parameter | Sammelt beliebig viele Keyword-Argumente als **Dictionary** |
| `*liste` beim Aufruf | Argument | Packt Liste/Tupel in einzelne Positionalargumente aus |
| `**dict` beim Aufruf | Argument | Packt Dictionary in einzelne Keyword-Argumente aus |
| `*, param` in Def | Parameter | Macht `param` zu einem Keyword-only Parameter |

**Reihenfolge in der Signatur:**
```python
def f(normal, *args, keyword_only, **kwargs):
```

**Warum das wichtig ist:**
- **Decorators** benutzen `*args, **kwargs` um jede beliebige Funktion huellen zu koennen
- **Mehrfachvererbung** benutzt `**kwargs` um Parameter entlang der MRO weiterzureichen ohne die Signatur aller Klassen kennen zu muessen
