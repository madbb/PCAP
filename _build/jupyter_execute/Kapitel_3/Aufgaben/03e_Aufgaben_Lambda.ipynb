**7.2**
```python
fn = lambda x: x ** 2 if x > 0 else 0
werte = [-2, -1, 0, 1, 2, 3]
print([fn(x) for x in werte])
```

**7.3**
```python
daten = [('Alice', 30), ('Bob', 25), ('Carol', 35)]
print(sorted(daten, key=lambda x: x[1]))
print(min(daten, key=lambda x: x[1]))
```

**7.4**
```python
fn = lambda x=1, y=2: x + y
print(fn())
print(fn(10))
print(fn(10, 20))
```
