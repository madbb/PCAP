1. Schreibe eine Funktion `erstelle_formatierer(prefix, suffix, grossschreiben=False)`.
   Sie gibt eine Funktion zurueck, die einen Text mit `prefix` und `suffix` umschliesst
   und optional in Grossbuchstaben umwandelt.
2. Erstelle und teste:
```python
html_fett  = erstelle_formatierer('<b>', '</b>')
eckig      = erstelle_formatierer('[', ']')
laut       = erstelle_formatierer('!!! ', ' !!!', grossschreiben=True)
```
3. Speichere alle drei in einer Liste.
   Wende jeden Formatierer auf `'Python'` an und gib die Ergebnisse aus.
4. Schreibe eine Funktion `wende_formatierer_an(text, formatierer_liste)`,
   die alle Formatierer nacheinander auf `text` anwendet.
   Teste sie mit allen drei Formatierern auf `'hallo'`.


````python
def erstelle_formatierer(prefix, suffix, grossschreiben = False):
    def formatierer(text):
        if grossschreiben:
            text = text.upper()
        return f"{prefix}{text}{suffix}"
    return formatierer

````

````python
html_fett   = erstelle_formatierer('<b>','</b>')
eckig      = erstelle_formatierer('[', ']')
laut       = erstelle_formatierer('!!! ', ' !!!', grossschreiben=True)

formatierer_liste = [html_fett,eckig,laut]

for fn in formatierer_liste:
    print(fn('python'))
````

````python
def wende_formatierer_an(text, formatierer_liste):
    ergebnis = text
    for fn in formatierer_liste:
        ergebnis = fn(ergebnis)
    return ergebnis
````

````python
print(wende_formatierer_an('hallo',formatierer_liste))
````
