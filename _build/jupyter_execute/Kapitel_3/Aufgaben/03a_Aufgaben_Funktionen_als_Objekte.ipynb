Gegeben:
```python

    {'id': 1, 'kunde': 'Alice', 'betrag': 59.90, 'status': 'offen'},
    {'id': 2, 'kunde': 'Bob',   'betrag': 12.50, 'status': 'bezahlt'},
    {'id': 3, 'kunde': 'Carol', 'betrag': 89.00, 'status': 'offen'},
    {'id': 4, 'kunde': 'Dave',  'betrag': 12.50, 'status': 'storniert'},
]
```
1. Schreibe eine Funktion `verarbeite(daten, transformation)`,
   die `transformation` auf jedes Element anwendet und eine neue Liste zurueckgibt.
2. Nutze sie um eine Liste aller Kundennamen zu erstellen.
3. Nutze sie um alle Betraege mit 1.19 zu multiplizieren (Bruttopreise).
4. Schreibe eine Funktion `filtere(daten, bedingung)`,
   die nur Elemente zurueckgibt, fuer die `bedingung` `True` ergibt.
5. Filtere alle offenen Bestellungen heraus.
6. Filtere alle Bestellungen mit einem Betrag ueber 20 Euro.


````python
def verarbeite(daten, transformation):
    ergebnis = []
    for element in daten:
        ergebnis.append(transformation(element))
    return ergebnis

def filtere(daten, bedingung):
    ergebnis = []
    for element in daten:
        if bedingung(element):
            ergebnis.append(element)
    return ergebnis

def ist_offen(bestellung):
    return bestellung['status'] == 'offen'

def betrag_uber_20(bestellung):
    return bestellung['betrag'] > 20

def hole_kunden_name(bestellung):
    return bestellung['kunde']

def berechne_brutto(bestellung):
    return round(bestellung['betrag'] * 1.19, 2)

````

````python
bestellungen = [
    {'id': 1, 'kunde': 'Alice', 'betrag': 59.90, 'status': 'offen'},
    {'id': 2, 'kunde': 'Bob',   'betrag': 12.50, 'status': 'bezahlt'},
    {'id': 3, 'kunde': 'Carol', 'betrag': 89.00, 'status': 'offen'},
    {'id': 4, 'kunde': 'Dave',  'betrag': 12.50, 'status': 'storniert'},
]
````

````python
def hole_kunden_name(bestellung):
    return bestellung['kunde']
````

````python
kundennamen = verarbeite(bestellungen,hole_kunden_name)
print(kundennamen)
````

````python
bruttopreise = verarbeite(bestellungen,berechne_brutto)
for num in bruttopreise:
    print(f"bruttopreis: {num}")
````

````python
offene = filtere(bestellungen, ist_offen)
for i in offene:
    print(i)
````

````python
über20 = filtere(bestellungen, betrag_uber_20)
for i in über20:
    print(i)
````
