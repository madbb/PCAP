**6.1**
```python
def mein_decorator(fn):
    def wrapper(*args, **kwargs):
        print('vorher')
        ergebnis = fn(*args, **kwargs)
        print('nachher')
        return ergebnis
    return wrapper

@mein_decorator
def sag_hallo():
    print('hallo')

sag_hallo()
```

**6.2**
```python
def mein_decorator(fn):
    def wrapper(*args, **kwargs):
        return fn(*args, **kwargs) * 2
    return wrapper

@mein_decorator
def addiere(a, b):
    return a + b

print(addiere(3, 4))
print(addiere.__name__)
```

**6.3**
```python
from functools import wraps

def mein_decorator(fn):
    @wraps(fn)
    def wrapper(*args, **kwargs):
        return fn(*args, **kwargs) * 2
    return wrapper

@mein_decorator
def addiere(a, b):
    '''Addiert zwei Zahlen.'''
    return a + b

print(addiere(3, 4))
print(addiere.__name__)
print(addiere.__doc__)
```
