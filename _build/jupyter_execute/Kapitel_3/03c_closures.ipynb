2. Was der Enclosing Scope ist
</span>
</div>

Aus Notebook 01b kennst du die LEGB-Regel: Python sucht Variablen in dieser Reihenfolge:

- **L**ocal -- innerhalb der aktuellen Funktion
- **E**nclosing -- in aeusseren Funktionen (von innen nach aussen)
- **G**lobal -- auf Modulebene
- **B**uilt-in -- Pythons eingebaute Namen

Das **E** ist der Schluessel. Wenn eine innere Funktion eine Variable benutzt, die sie selbst nicht definiert, schaut Python in der aeusseren Funktion nach:

````python
def aussen():
    nachricht = 'Hallo aus aussen'  # Variable im Enclosing Scope

    def innen():
        # innen hat keine eigene 'nachricht' -- Python sucht im Enclosing Scope
        print(nachricht)

    innen()

aussen()  # Hallo aus aussen
````

Das funktioniert, weil `innen` und `aussen` gleichzeitig aktiv sind -- `aussen` ist noch nicht beendet, wenn `innen` aufgerufen wird.

Aber was passiert, wenn `aussen` beendet ist und wir `innen` danach noch aufrufen wollen? Das ist genau der Fall beim `erstelle_bereichspruefer`-Beispiel.

3. Was eine Closure ist
</span>
</div>

Eine **Closure** ist eine innere Funktion, die sich Variablen aus dem Enclosing Scope "merkt" -- auch nachdem die aeussere Funktion beendet ist.

Man sagt: Die innere Funktion "schliesst" die Variablen aus ihrem Entstehungskontext ein. Daher der Name Closure (englisch: Abschluss, Einschliessung).

Schauen wir uns das Schritt fuer Schritt an:

````python
def erstelle_addierer(n):
    # n existiert im Enclosing Scope von 'addierer'
    def addierer(x):
        return x + n  # n wird aus dem Enclosing Scope geholt
    return addierer   # wir geben die innere Funktion zurueck

# erstelle_addierer(5) wird aufgerufen -- n = 5
# Die Funktion erstelle_addierer endet danach
# Aber 'addierer' mit n=5 wird zurueckgegeben und in plus5 gespeichert
plus5 = erstelle_addierer(5)

# erstelle_addierer ist laengst beendet -- aber n=5 lebt noch!
print(plus5(10))  # 15
print(plus5(3))   # 8

# Zweite, unabhaengige Closure mit n=100
plus100 = erstelle_addierer(100)
print(plus100(10))  # 110  -- eigenes n, unabhaengig von plus5
````

Python beweist uns, dass die Werte wirklich gespeichert sind -- mit `__closure__`:

````python
# __closure__ zeigt die eingeschlossenen Variablen
print(plus5.__closure__)                        # ein Tupel von 'cell'-Objekten
print(plus5.__closure__[0].cell_contents)       # 5
print(plus100.__closure__[0].cell_contents)     # 100

# Die freien Variablennamen zeigt __code__.co_freevars
print(plus5.__code__.co_freevars)  # ('n',)
````

4. Veraenderlicher Zustand mit nonlocal
</span>
</div>

Bisher haben die inneren Funktionen die Variablen nur **gelesen**. Was wenn wir sie auch **veraendern** wollen?

Das Problem: Wenn Python innerhalb einer Funktion eine Zuweisung sieht (`n = n + 1`), behandelt es `n` als **lokale** Variable dieser Funktion. Es wird nicht im Enclosing Scope gesucht.

Das Schluesselwort `nonlocal` loest das. Es sagt Python: *Diese Variable ist nicht lokal -- suche sie im Enclosing Scope.*

````python
# Erst das Problem zeigen:
def erstelle_zaehler_falsch():
    n = 0
    def erhoehe():
        n = n + 1   # Python sieht Zuweisung: n wird lokal behandelt
        return n    # aber lokal existiert n noch nicht -> Fehler
    return erhoehe

zaehler = erstelle_zaehler_falsch()
try:
    zaehler()
except UnboundLocalError as e:
    print(f'Fehler: {e}')
    print('Python sieht die Zuweisung und behandelt n als lokal -- aber lokal gibt es kein n')
````

````python
# Die Loesung: nonlocal
def erstelle_zaehler():
    n = 0
    def erhoehe():
        nonlocal n  # n ist NICHT lokal -- suche im Enclosing Scope
        n = n + 1
        return n
    return erhoehe

zaehler = erstelle_zaehler()
print(zaehler())  # 1
print(zaehler())  # 2
print(zaehler())  # 3

# Jeder Aufruf von erstelle_zaehler() erzeugt einen eigenen, unabhaengigen Zaehler
anderer_zaehler = erstelle_zaehler()
print(anderer_zaehler())  # 1  -- faengt bei 0 an, unabhaengig vom ersten
````

Dieses Muster -- Closure mit veraenderlichem Zustand -- ist ein leichtgewichtiger Ersatz fuer eine einfache Klasse. Der Zustand (`n`) ist in der Closure verborgen und nur ueber die Funktion zugaenglich.

5. Der klassische Closure-Fallstrick
</span>
</div>

Es gibt einen Fallstrick, der in der PCAP-Pruefung auftaucht. Er entsteht, wenn man in einer Schleife Closures erstellt.

Das Szenario: Wir wollen drei Funktionen erstellen -- eine gibt `0` zurueck, eine `1`, eine `2`.

````python
# Erster Versuch -- erscheint logisch, ist aber falsch
funktionen = []
for i in range(3):
    def f():
        return i   # i aus dem Enclosing Scope
    funktionen.append(f)

# Was erwarten wir?
# funktionen[0]() -> 0
# funktionen[1]() -> 1
# funktionen[2]() -> 2

# Was bekommen wir?
print([fn() for fn in funktionen])
````

Alle drei Funktionen geben `2` zurueck -- nicht `0`, `1`, `2`.

**Warum?** Alle drei Funktionen referenzieren dieselbe Variable `i` -- nicht deren **Wert** zum Zeitpunkt der Definition. Nach der Schleife hat `i` den Wert `2`. Alle drei Funktionen schauen in denselben Enclosing Scope und finden dort `i = 2`.

Die Closure speichert die **Referenz auf die Variable**, nicht eine Kopie des Werts.

````python
# Loesung 1: Default-Argument -- der Wert wird zum Zeitpunkt der Definition eingefroren
funktionen = []
for i in range(3):
    def f(i=i):   # i=i: der aktuelle Wert von i wird als Default gespeichert
        return i
    funktionen.append(f)

print([fn() for fn in funktionen])  # [0, 1, 2]
````

````python
# Loesung 2: Factory-Funktion -- jede Iteration bekommt ihren eigenen Enclosing Scope
def erstelle_funktion(wert):
    def f():
        return wert  # wert ist jetzt im eigenen Enclosing Scope von f
    return f

funktionen = [erstelle_funktion(i) for i in range(3)]
print([fn() for fn in funktionen])  # [0, 1, 2]
````

Loesung 2 ist sauberer und klarer verstaendlich: Jeder Aufruf von `erstelle_funktion(i)` erzeugt einen eigenen Enclosing Scope mit einem eigenen `wert`. Keine gemeinsame Variable, kein Problem.

**Zusammenfassung**

| Konzept | Erklaerung |
|---------|------------|
| Closure | Innere Funktion + eingeschlossene Variablen aus dem Enclosing Scope |
| Enclosing Scope | Der Namespace der aeusseren Funktion -- das E in LEGB |
| `nonlocal` | Erlaubt der inneren Funktion, Enclosing-Variablen zu veraendern |
| Closures speichern Referenzen | Nicht Werte -- Aenderungen nach der Definition sind sichtbar |
| Schleifenfalle | Alle Closures teilen dieselbe Schleifenvariable -- Loesung: Default-Argument oder Factory |

**Wann Closures einsetzen?** Wenn eine Funktion Zustand benoetigt, der zwischen Aufrufen erhalten bleibt -- aber eine ganze Klasse zu aufwendig waere. Fuer komplexeren Zustand oder mehrere Methoden ist eine Klasse die bessere Wahl.
