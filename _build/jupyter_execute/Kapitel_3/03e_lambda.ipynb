1. Das Problem: Funktionen fuer den Einmalgebrauch
</span>

In Notebook 03b haben wir hoeherwertige Funktionen kennengelernt -- Funktionen die andere Funktionen als Argument erwarten. Ein haeufiger Anwendungsfall ist `sorted()` mit einem `key`:

```python
personen = [{'name': 'Alice', 'alter': 30}, {'name': 'Bob', 'alter': 25}]
sorted(personen, key=nach_alter)
```

Dafuer mussten wir `nach_alter` vorher definieren:

```python
def nach_alter(person):
    return person['alter']
```

Diese Funktion hat einen ganzen Namen, eine eigene Zeile -- dabei tut sie nur eine einzige, winzige Sache. Und nach dem `sorted()`-Aufruf brauchen wir sie nie wieder.

Python hat dafuer eine kompaktere Loesung: **Lambda-Funktionen**.

2. Was ein Lambda ist
</span>
</div>

Ein Lambda ist eine **anonyme Funktion** -- eine Funktion ohne Namen, die in einer einzigen Zeile definiert wird.

**Syntax:**
```
lambda parameter: ausdruck
```

Schauen wir denselben Gedanken erst als normale Funktion, dann als Lambda:

````python
# Normale Funktion
def quadrat(x):
    return x ** 2

# Identisch als Lambda
quadrat_lambda = lambda x: x ** 2

# Beide verhalten sich gleich
print(quadrat(5))        # 25
print(quadrat_lambda(5)) # 25

# Ein Lambda ist ein Funktionsobjekt -- genau wie eine normale Funktion
print(type(quadrat_lambda))  # <class 'function'>
````

Das Lambda `lambda x: x ** 2` liest sich so: *"Nimm einen Parameter `x` und gib `x ** 2` zurueck."* Das `return` ist implizit -- der Ausdruck nach dem Doppelpunkt ist automatisch der Rueckgabewert.

Lambda-Funktionen koennen auch mehrere Parameter haben:

````python
# Mehrere Parameter
addiere = lambda a, b: a + b
print(addiere(3, 4))   # 7

# Kein Parameter
gruss = lambda: 'Hallo!'
print(gruss())         # Hallo!

# Mit Default-Wert
potenz = lambda x, n=2: x ** n
print(potenz(3))    # 9  -- Standard: hoch 2
print(potenz(3, 3)) # 27 -- hoch 3
````

3. Lambda als Argument -- der Hauptanwendungsfall
</span>
</div>

Der eigentliche Nutzen von Lambda liegt darin, direkt an der Stelle wo eine Funktion gebraucht wird eine zu schreiben -- ohne sie vorher benennen zu muessen. Das macht den Code oft lesbarer, weil der Gedanke nicht aufgesplittet wird:

````python
personen = [
    {'name': 'Alice', 'alter': 30},
    {'name': 'Bob',   'alter': 25},
    {'name': 'Carol', 'alter': 35},
]

# Ohne Lambda -- Funktion extra definieren
def nach_alter(p): return p['alter']
sortiert = sorted(personen, key=nach_alter)

# Mit Lambda -- direkt an Ort und Stelle
sortiert = sorted(personen, key=lambda p: p['alter'])

# Beide Varianten liefern dasselbe Ergebnis
for p in sortiert:
    print(p['name'], p['alter'])
````

````python
# Lambda funktioniert auch mit min(), max() und allen anderen
# hoeherwertigen Funktionen die ein key-Argument kennen

woerter = ['Banane', 'Apfel', 'Kiwi', 'Ananas']

# Nach Laenge sortieren
print(sorted(woerter, key=lambda w: len(w)))

# Kuerzestes Wort
print(min(woerter, key=lambda w: len(w)))

# Nach letztem Buchstaben sortieren
print(sorted(woerter, key=lambda w: w[-1]))

# Nach zwei Kriterien: erst Laenge, dann alphabetisch
print(sorted(woerter, key=lambda w: (len(w), w)))
````

4. Die Grenzen von Lambda
</span>
</div>

Lambda ist bewusst eingeschraenkt. Es kann nur einen **einzigen Ausdruck** enthalten -- keine Schleifen, keine `if`-Bloecke, kein `return`, keine Mehrfachzuweisungen.

Das ist keine Einschraenkung die Python vergessen hat -- es ist eine Designentscheidung: Lambda soll fuer einfache Einzeiler sein. Sobald die Logik komplexer wird, ist eine normale Funktion mit `def` die bessere Wahl.

````python
# Was Lambda kann: ein einziger Ausdruck
absolut = lambda x: x if x >= 0 else -x   # ternary-Ausdruck ist ok
print(absolut(-5))   # 5
print(absolut(3))    # 3

# Was Lambda NICHT kann -- mehrzeilige Logik
# Das hier ist KEIN gueltiges Lambda:
# falsch = lambda x:
#     if x > 0:
#         return x * 2
#     return 0
# --> SyntaxError

# Fuer solche Faelle: normale Funktion
def verarbeite(x):
    if x > 0:
        return x * 2
    return 0

print(verarbeite(5))   # 10
print(verarbeite(-3))  # 0
````

Wann Lambda, wann def?
</span>

| Lambda | def |
|--------|-----|
| Einzeiler, wird nur einmal benoetigt | Mehrere Zeilen oder wiederverwendet |
| Direkt als Argument uebergeben | Benoetigt einen eigenen Namen |
| Einfacher Ausdruck ohne Seiteneffekte | Komplexere Logik mit mehreren Schritten |

Eine Lambda-Funktion die so komplex ist, dass man sie kaum noch lesen kann, ist ein Zeichen: lieber `def` benutzen.

---

**Zusammenfassung**

Ein Lambda ist eine anonyme Einzeiler-Funktion. Syntax: `lambda parameter: ausdruck`.

- Das `return` ist implizit -- der Ausdruck ist der Rueckgabewert
- Nur ein einziger Ausdruck moeglich -- keine Schleifen, kein mehrzeiliger Code
- Hauptanwendungsfall: als Argument an hoeherwertige Funktionen wie `sorted()`, `min()`, `max()`
- Bei komplexerer Logik: lieber `def` verwenden
