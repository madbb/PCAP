1. Warum APIs? Das Problem mit isolierten Programmen
</span>
</div>

Stell dir vor, du baust eine Wetterapp. Du koenntest selbst Wetterstationen betreiben und Messdaten sammeln -- aber das ist aufwendig und teuer. Stattdessen gibt es Wetterdienste die diese Daten haben und sie anderen Programmen zur Verfuegung stellen.

Oder du baust eine App die Waehrungskurse anzeigt, GitHub-Issues verwaltet, Twitter-Posts liest, Google Maps-Daten nutzt. All das sind Daten und Funktionen die andere Unternehmen bereits haben -- und ueber eine **API** bereitstellen.

**API** steht fuer *Application Programming Interface* -- eine definierte Schnittstelle ueber die zwei Programme miteinander kommunizieren koennen. Dein Python-Programm fragt an, der Server antwortet.

Das Protokoll das dabei im Web verwendet wird heisst **HTTP** -- und das verstehen wir zuerst.

2. HTTP -- wie Browser und Server kommunizieren
</span>
</div>

Wenn du eine URL in den Browser eingibst, passiert folgendes:

1. Dein Browser schickt einen **Request** (Anfrage) an den Server
2. Der Server verarbeitet die Anfrage
3. Der Server schickt eine **Response** (Antwort) zurueck

Jeder Request besteht aus:
- einer **Methode** -- was will ich tun?
- einer **URL** -- mit wem rede ich und woruebers?
- optionalen **Headern** -- Metadaten (z.B. Sprache, Authentifizierung)
- optionalem **Body** -- Daten die mitgeschickt werden (z.B. Formularinhalt)

**Die vier wichtigsten HTTP-Methoden:**

| Methode | Bedeutung | Analogie |
|---------|-----------|----------|
| `GET` | Daten abrufen | Brief an die Bibliothek: *Schick mir Buch X* |
| `POST` | Neuen Datensatz erstellen | Formular einschicken |
| `PUT` | Datensatz ersetzen | Datensatz komplett ueberschreiben |
| `DELETE` | Datensatz loeschen | Datensatz entfernen |

Jede Response hat einen **Statuscode** -- eine dreistellige Zahl die sagt ob alles geklappt hat:

| Code | Bedeutung |
|------|-----------|
| `200` | OK -- alles gut |
| `201` | Created -- Datensatz wurde erstellt |
| `400` | Bad Request -- deine Anfrage ist fehlerhaft |
| `401` | Unauthorized -- du musst dich authentifizieren |
| `403` | Forbidden -- du hast keine Berechtigung |
| `404` | Not Found -- die Ressource existiert nicht |
| `500` | Internal Server Error -- Fehler auf dem Server |

3. REST -- eine Konvention fuer API-Design
</span>
</div>

**REST** (Representational State Transfer) ist kein Protokoll sondern eine **Konvention** -- eine Sammlung von Regeln wie eine gute API aufgebaut sein sollte.

Die Kernidee: Alles ist eine **Ressource** -- ein Nutzer, ein Produkt, ein Beitrag. Jede Ressource hat eine eigene URL. Was mit ihr gemacht wird, bestimmt die HTTP-Methode.

Ein Beispiel fuer eine gut designte REST-API fuer ein Helden-System:

```
GET    /helden         -- alle Helden abrufen
GET    /helden/42      -- Held mit ID 42 abrufen
POST   /helden         -- neuen Helden erstellen
PUT    /helden/42      -- Held 42 ersetzen
DELETE /helden/42      -- Held 42 loeschen
```

Daten werden fast immer als **JSON** uebertragen -- das kennst du bereits.

6. Fehlerbehandlung -- nicht jeder Request klappt
</span>
</div>

Netzwerke sind unzuverlaessig. Server sind manchmal nicht erreichbar. Die URL koennte falsch sein. Die API koennte sich geaendert haben. Robuster Code behandelt diese Faelle.

Es gibt zwei Arten von Fehlern:
1. **Netzwerkfehler** -- der Server ist nicht erreichbar, Timeout. `requests` wirft eine Exception.
2. **HTTP-Fehler** -- der Server antwortet, aber mit einem Fehlercode (4xx, 5xx). `requests` wirft **keine** Exception -- der Statuscode muss manuell geprueft werden.

````python
import requests

# HTTP-Fehler -- Server antwortet mit 404
response = requests.get('https://jsonplaceholder.typicode.com/posts/99999')
print(f'Statuscode: {response.status_code}')  # 404

# requests wirft KEINE Exception bei 4xx/5xx!
# Manuell pruefen:
if response.status_code == 200:
    print('Erfolg:', response.json())
elif response.status_code == 404:
    print('Nicht gefunden')
else:
    print(f'Fehler: {response.status_code}')

# Eleganter: raise_for_status() wirft bei 4xx/5xx eine Exception
try:
    response.raise_for_status()
except requests.HTTPError as e:
    print(f'HTTP-Fehler: {e}')
````

````python
import requests

# Robuste Hilfsfunktion fuer GET-Requests
def hole_daten(url, params=None, timeout=10):
    """
    Schickt einen GET-Request und gibt die JSON-Daten zurueck.
    Gibt None zurueck wenn etwas schieflaeuft.
    """
    try:
        response = requests.get(url, params=params, timeout=timeout)
        response.raise_for_status()   # Exception bei 4xx/5xx
        return response.json()
    except requests.ConnectionError:
        print('Fehler: Keine Verbindung zum Server')
    except requests.Timeout:
        print(f'Fehler: Timeout nach {timeout}s')
    except requests.HTTPError as e:
        print(f'HTTP-Fehler: {e}')
    return None


# Anwendung
post = hole_daten('https://jsonplaceholder.typicode.com/posts/1')
if post:
    print(post['title'])

# Nicht vorhandene Ressource
result = hole_daten('https://jsonplaceholder.typicode.com/posts/99999')
````

7. POST -- Daten senden
</span>
</div>

Mit `POST` schickt man Daten an den Server -- zum Beispiel um einen neuen Datensatz zu erstellen. Die Daten werden als JSON im Request-Body mitgeschickt.

````python
import requests

# Neuen Post erstellen
neuer_post = {
    'title':  'Mein erster API-Post',
    'body':   'Das ist der Inhalt des Posts.',
    'userId': 1
}

response = requests.post(
    'https://jsonplaceholder.typicode.com/posts',
    json=neuer_post   # json= serialisiert automatisch und setzt Content-Type Header
)

print(f'Statuscode: {response.status_code}')  # 201 Created
print(response.json())                         # Server antwortet mit dem erstellten Objekt
````

````python
import requests

BASE_URL = 'https://jsonplaceholder.typicode.com'

def alle_user_holen():
    response = requests.get(f'{BASE_URL}/users')
    response.raise_for_status()
    return response.json()

def posts_von_user(user_id):
    response = requests.get(f'{BASE_URL}/posts', params={'userId': user_id})
    response.raise_for_status()
    return response.json()

def kommentare_zu_post(post_id):
    response = requests.get(f'{BASE_URL}/posts/{post_id}/comments')
    response.raise_for_status()
    return response.json()


# Alle User laden und den aktivsten finden
user = alle_user_holen()
print(f'{len(user)} User geladen\n')

# Fuer die ersten 3 User: Posts zaehlen
for u in user[:3]:
    posts = posts_von_user(u['id'])
    print(f'{u["name"]}: {len(posts)} Posts')

    # Kommentare zum ersten Post
    if posts:
        kommentare = kommentare_zu_post(posts[0]['id'])
        print(f'  Erster Post hat {len(kommentare)} Kommentare')
````

**Zusammenfassung**

| Konzept | Erklaerung |
|---------|------------|
| HTTP | Protokoll fuer Kommunikation im Web |
| GET / POST / PUT / DELETE | HTTP-Methoden -- was will ich tun? |
| Statuscode | 200=OK, 201=Created, 404=Not Found, 500=Server Error |
| REST | Konvention: Ressourcen haben URLs, Methode bestimmt Aktion |
| `requests.get(url)` | GET-Request -- Daten abrufen |
| `requests.post(url, json=daten)` | POST-Request -- Daten senden |
| `response.status_code` | Statuscode der Antwort |
| `response.json()` | Antwort als Dictionary/Liste |
| `response.raise_for_status()` | Exception bei 4xx/5xx |
| `params={}` | Query-Parameter werden an URL angehaengt |
| `headers={}` | HTTP-Header mitschicken (z.B. API-Key) |
| API-Key in Umgebungsvariable | Niemals direkt im Code speichern |
