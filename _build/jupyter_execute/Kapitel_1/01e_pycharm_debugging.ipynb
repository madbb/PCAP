Was ist Debugging?</span></div>

Debugging bedeutet, einen Fehler im laufenden Programm zu finden und zu beheben. Es gibt zwei Ansaetze:

**Print-Debugging:** `print()`-Anweisungen einstreuen, um Zwischenwerte sichtbar zu machen.

```python
def berechne(werte):
    print('DEBUG: werte =', werte)  # <-- manuell eingebaut
    summe = sum(werte)
    print('DEBUG: summe =', summe)
    return summe / len(werte)
```

Das funktioniert -- ist aber muehsam, unvollstaendig und muss hinterher wieder entfernt werden.

**Debugger:** Das Programm haelt an einem Breakpoint an. Du kannst dann alle Variablen inspizieren, den Call Stack sehen und den Code Schritt fuer Schritt ausfuehren -- ohne eine einzige Zeile zu veraendern.

Sobald ein Problem komplexer als "welchen Wert hat x?" ist, ist der Debugger die bessere Wahl.

<div style="background:#0b1929;border-left:4px solid #4fc3f7;border-radius:4px;padding:0.75em 1.2em;margin:2rem 0 1.2rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(1rem,2.2vw,1.25rem);font-weight:600;color:#e8eaf6;letter-spacing:-0.01em;"><span style="font-size:0.65rem;letter-spacing:0.1em;text-transform:uppercase;color:#4fc3f7;background:rgba(79,195,247,0.12);border:1px solid rgba(79,195,247,0.3);border-radius:3px;padding:0.2em 0.55em;margin-right:0.75em;vertical-align:middle;font-family:'Segoe UI',sans-serif;white-space:nowrap;">Kursinhalt</span>4. Fehlertypen in Python</span></div>

Bevor wir debuggen, muessen wir Fehler lesen koennen. Python unterscheidet drei Kategorien:

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">4.1 SyntaxError</span></div>

Tritt auf, bevor das Programm laeuft -- beim Parsen. Der Code verletzt die Python-Grammatik.

````python
# SyntaxError -- diesen Code nicht ausfuehren, nur lesen
# def greet(name)
#     return 'Hallo ' + name
#
# Fehler:   File "skript.py", line 1
#             def greet(name)
#                            ^
#           SyntaxError: expected ':'

# Python zeigt genau die Stelle an. Hier fehlt der Doppelpunkt.
# PyCharm markiert SyntaxErrors bereits beim Tippen rot -- noch vor dem Ausfuehren.

print('SyntaxErrors entstehen vor der Ausfuehrung -- beim Parsen.')
print('PyCharm zeigt sie sofort an.')
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">4.2 RuntimeError (Exceptions)</span></div>

Tritt waehrend der Ausfuehrung auf. Die Syntax ist korrekt, aber etwas geht schief -- z.B. Division durch Null, fehlender Schluessel in einem Dictionary, falscher Typ.

````python
# Traceback lesen lernen
def berechne_durchschnitt(werte):
    return sum(werte) / len(werte)

def verarbeite(datensatz):
    gefiltert = [x for x in datensatz if x > 0]
    return berechne_durchschnitt(gefiltert)

try:
    ergebnis = verarbeite([-1, -2, -3])  # alle negativ -> gefiltert ist leer
except ZeroDivisionError as e:
    print('Fehler abgefangen:', e)
    print()
    print('Was der Traceback ohne try/except zeigen wuerde:')
    print('  Traceback (most recent call last):')
    print('    File "skript.py", line 10, in <module>')
    print('      ergebnis = verarbeite([-1, -2, -3])')
    print('    File "skript.py", line 7, in verarbeite')
    print('      return berechne_durchschnitt(gefiltert)')
    print('    File "skript.py", line 3, in berechne_durchschnitt')
    print('      return sum(werte) / len(werte)')
    print('  ZeroDivisionError: division by zero')
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Den Traceback lesen</span></div>

Der Traceback ist von **unten nach oben** zu lesen:

1. **Unterste Zeile:** Der eigentliche Fehler (`ZeroDivisionError: division by zero`)
2. **Zeile darueber:** Die Stelle im Code, wo der Fehler ausgeloest wurde
3. **Weiter hoch:** Der Aufrufpfad -- welche Funktion hat welche aufgerufen
4. **Oberste Zeile:** Der urspruengliche Einstiegspunkt

Der Fehler passiert unten -- die Ursache liegt oft weiter oben im Aufrufpfad.

````python
# Die haeufigsten Fehlertypen

fehler = [
    ('NameError',        lambda: print(nicht_definiert)),
    ('TypeError',        lambda: 'Text' + 42),
    ('IndexError',       lambda: [1, 2, 3][10]),
    ('KeyError',         lambda: {'a': 1}['b']),
    ('AttributeError',   lambda: (42).upper()),
    ('ValueError',       lambda: int('kein_int')),
    ('ZeroDivisionError',lambda: 1 / 0),
]

for name, fn in fehler:
    try:
        fn()
    except Exception as e:
        print(f'{name:22} --> {e}')
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">4.3 Logikfehler</span></div>

Der gefaehrlichste Fehlertyp: Das Programm laeuft ohne Exception -- gibt aber falsche Ergebnisse.

````python
# Logikfehler: kein Absturz, aber falsches Ergebnis
def celsius_zu_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

def fahrenheit_zu_celsius(fahrenheit):
    # Fehler: Klammern vergessen
    return fahrenheit - 32 * 5 / 9   # falsch: 32 * 5/9 = 17.78 wird subtrahiert
    # richtig waere: return (fahrenheit - 32) * 5 / 9

# Kein Fehler, keine Exception
grad_c = 100
grad_f = celsius_zu_fahrenheit(grad_c)
zurueck = fahrenheit_zu_celsius(grad_f)

print(f'{grad_c} Celsius -> {grad_f} Fahrenheit -> {zurueck:.2f} Celsius')
print(f'Erwartet: {grad_c}, bekommen: {zurueck:.2f}')
print(f'Differenz: {abs(zurueck - grad_c):.2f}')
````

Ein Breakpoint markiert eine Zeile, bei der das Programm pausiert. PyCharm: Klick auf die Zeilennummer links, oder `F9` auf der Zeile.

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Debugger-Steuerung</span></div>

Wenn das Programm an einem Breakpoint haelt, stehen folgende Aktionen zur Verfuegung:

| Aktion | Shortcut | Was passiert |
|--------|----------|--------------|
| **Step Over** | `F8` | Naechste Zeile ausfuehren; Funktionsaufrufe werden nicht betreten |
| **Step Into** | `F7` | In den Funktionsaufruf der naechsten Zeile hineinspringen |
| **Step Into My Code** | `Alt+Shift+F7` | Nur in eigenen Code, nicht in Bibliotheken |
| **Step Out** | `Shift+F8` | Aktuelle Funktion zu Ende fuehren, in aufrufende Funktion zurueck |
| **Run to Cursor** | `Alt+F9` | Bis zur Zeile mit dem Cursor weiterlaufen |
| **Resume** | `F9` | Bis zum naechsten Breakpoint weiterlaufen |
| **Stop** | `Ctrl+F2` | Programm abbrechen |

Der Call Stack</span></div>

Der Call Stack (Aufrufstapel) zeigt, welche Funktionen gerade aktiv sind -- und in welcher Reihenfolge sie aufgerufen wurden.

Stell dir den Stack als einen Stapel Platten vor: jede Funktion legt eine neue Platte oben drauf. Wenn sie endet, wird die Platte wieder entfernt.

````python
import traceback

def funktion_c():
    # Wir geben den aktuellen Call Stack aus, ohne einen Fehler zu werfen
    print('Aktueller Call Stack (innerste Funktion zuerst):')
    for zeile in traceback.format_stack():
        print(zeile.strip())

def funktion_b():
    funktion_c()

def funktion_a():
    funktion_b()

funktion_a()
````

Im PyCharm-Debugger zeigt das **Frames**-Panel denselben Stack -- aber interaktiv. Du kannst auf jeden Frame klicken und siehst dann die Variablen dieser Funktion zum Zeitpunkt des Breakpoints.

Das ist besonders nuetzlich, wenn eine Funktion tief verschachtelt ist und du verstehen willst, mit welchen Werten sie aufgerufen wurde.

<div style="background:#0b1929;border-left:4px solid #4fc3f7;border-radius:4px;padding:0.75em 1.2em;margin:2rem 0 1.2rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(1rem,2.2vw,1.25rem);font-weight:600;color:#e8eaf6;letter-spacing:-0.01em;"><span style="font-size:0.65rem;letter-spacing:0.1em;text-transform:uppercase;color:#4fc3f7;background:rgba(79,195,247,0.12);border:1px solid rgba(79,195,247,0.3);border-radius:3px;padding:0.2em 0.55em;margin-right:0.75em;vertical-align:middle;font-family:'Segoe UI',sans-serif;white-space:nowrap;">Kursinhalt</span>7. Debugging-Workflow</span></div>

Ein systematisches Vorgehen spart Zeit:

````python
# Beispiel: Diese Funktion hat einen Fehler -- finde ihn

def berechne_rabatt(preis, prozent):
    """Berechnet den Preis nach Abzug des Rabatts."""
    rabatt = preis / 100 * prozent
    return preis - rabatt

def berechne_gesamtpreis(artikel):
    """Berechnet den Gesamtpreis einer Liste von (preis, rabatt_prozent)-Tupeln."""
    gesamt = 0
    for preis, rabatt in artikel:
        endpreis = berechne_rabatt(preis, rabatt)
        gesamt += endpreis
    return gesamt

# Testfall: 100 Euro ohne Rabatt + 200 Euro mit 10% Rabatt = 100 + 180 = 280
artikel = [(100, 0), (200, 10)]
ergebnis = berechne_gesamtpreis(artikel)
erwartet = 280

print(f'Ergebnis:  {ergebnis}')
print(f'Erwartet:  {erwartet}')
print(f'Korrekt:   {ergebnis == erwartet}')
````

````python
# Debugging-Schritt 1: Eingrenzen -- wo liegt das Problem?
print('=== Einzelne Schritte pruefen ===')

# berechne_rabatt direkt testen
print('100 Euro, 0% Rabatt:  ', berechne_rabatt(100, 0))   # erwartet: 100
print('200 Euro, 10% Rabatt: ', berechne_rabatt(200, 10))  # erwartet: 180
print('100 Euro, 100% Rabatt:', berechne_rabatt(100, 100)) # erwartet: 0
print('100 Euro, 50% Rabatt: ', berechne_rabatt(100, 50))  # erwartet: 50

# Sieht korrekt aus -- also liegt das Problem in berechne_gesamtpreis?
print()
print('Gesamtpreis fuer [(100,0)]:       ', berechne_gesamtpreis([(100, 0)]))
print('Gesamtpreis fuer [(200,10)]:      ', berechne_gesamtpreis([(200, 10)]))
print('Gesamtpreis fuer [(100,0),(200,10)]:', berechne_gesamtpreis([(100, 0), (200, 10)]))
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Der systematische Debugging-Prozess</span></div>

1. **Fehler reproduzieren:** Einen minimalen Testfall finden, der den Fehler zuverlaessig ausloest
2. **Hypothese formulieren:** Was koennte falsch sein?
3. **Eingrenzen:** Wo genau liegt das Problem? (Binaere Suche: Mitte testen, eine Haelfte ausschliessen)
4. **Bestaetigen:** Breakpoint setzen, Variablen pruefen, Hypothese verifizieren
5. **Beheben:** Aenderung vornehmen
6. **Verifizieren:** Testfall erneut ausfuehren, pruefen ob weitere Tests noch funktionieren

> **Haeufiger Fehler:** Direkt Code aendern, ohne den Fehler vollstaendig verstanden zu haben. Das fuehrt oft dazu, dass man das Symptom behebt, nicht die Ursache.

Ein Breakpoint haelt nur an, wenn eine Bedingung wahr ist. Nuetzlich bei Schleifen mit vielen Iterationen.

In PyCharm: Rechtsklick auf den roten Breakpoint-Punkt > `Condition` eingeben, z.B. `i == 42` oder `wert < 0`.

Im **Variables**-Panel koennen Ausdruecke als "Watch" hinzugefuegt werden. Sie werden bei jedem Schritt neu ausgewertet und zeigen immer den aktuellen Wert.

`Run > View Breakpoints > Python Exception Breakpoints`: Haelt automatisch an, wenn eine bestimmte Exception (oder jede Exception) ausgeloest wird -- auch wenn sie spaeter abgefangen wird.

````python
# Alternative zu print-Debugging: das logging-Modul
import logging

# Logging konfigurieren
logging.basicConfig(
    level=logging.DEBUG,
    format='%(levelname)s | %(funcName)s | %(message)s'
)

def berechne_mit_logging(werte):
    logging.debug(f'Eingabe: {werte}')
    if not werte:
        logging.warning('Leere Liste uebergeben')
        return None
    ergebnis = sum(werte) / len(werte)
    logging.debug(f'Ergebnis: {ergebnis}')
    return ergebnis

print(berechne_mit_logging([10, 20, 30]))
print(berechne_mit_logging([]))
````

Das `logging`-Modul ist die professionelle Alternative zu `print`-Debugging:

| Level | Wann verwenden |
|-------|----------------|
| `DEBUG` | Detaillierte Diagnoseinformationen |
| `INFO` | Normaler Programmablauf, wichtige Ereignisse |
| `WARNING` | Etwas Unerwartetes, Programm laeuft weiter |
| `ERROR` | Ernstes Problem, Funktion kann nicht ausgefuehrt werden |
| `CRITICAL` | Schwerer Fehler, Programm kann nicht weiterlaufen |

In der Produktion stellt man das Level auf `WARNING` -- dann verschwinden alle `DEBUG`- und `INFO`-Meldungen automatisch, ohne eine Zeile Code zu aendern.

````python
def ist_palindrom(text):
    """Gibt True zurueck, wenn text ein Palindrom ist (vorwaerts = rueckwaerts)."""
    text = text.lower()
    for i in range(len(text)):
        if text[i] != text[len(text) - i]:   # Fehler hier
            return False
    return True

# Testfaelle
print(ist_palindrom('racecar'))   # erwartet: True
print(ist_palindrom('hello'))     # erwartet: False
print(ist_palindrom('Anna'))      # erwartet: True
print(ist_palindrom('a'))         # erwartet: True

# Fehleranalyse:
# ...
````

Ersetze in der folgenden Funktion alle `print`-Anweisungen durch geeignete `logging`-Aufrufe mit dem richtigen Level. Das Programm soll in der Produktion (`level=logging.WARNING`) keine Ausgaben erzeugen, in der Entwicklung (`level=logging.DEBUG`) alle.

````python
def lade_konfiguration(pfad):
    print(f'Lade Konfiguration aus {pfad}')          # Debug-Info
    try:
        with open(pfad) as f:
            inhalt = f.read()
        print(f'Konfiguration geladen: {len(inhalt)} Zeichen')  # Debug-Info
        return inhalt
    except FileNotFoundError:
        print(f'Konfigurationsdatei nicht gefunden: {pfad}')    # Warnung
        print('Verwende Standardkonfiguration')                  # Info
        return ''
    except PermissionError:
        print(f'Keine Leseberechtigung fuer {pfad}')            # Fehler
        return None

# Deine ueberarbeitete Version hier:
````

Die folgende Funktion liefert fuer manche Eingaben ein falsches Ergebnis. Ohne sie zu veraendern: Beschreibe Schritt fuer Schritt, wie du mit dem PyCharm-Debugger vorgehst, um den Fehler zu finden. Welchen Breakpoint setzt du zuerst? Was schaust du im Variables-Panel an?

````python
def finde_maximum(zahlen):
    """Gibt den groessten Wert in der Liste zurueck."""
    maximum = 0
    for zahl in zahlen:
        if zahl > maximum:
            maximum = zahl
    return maximum

print(finde_maximum([3, 7, 2, 9, 4]))    # erwartet: 9
print(finde_maximum([-3, -7, -1, -4]))   # erwartet: -1
print(finde_maximum([]))                  # erwartet: ?

# Deine Debugging-Strategie:
# ...
````
