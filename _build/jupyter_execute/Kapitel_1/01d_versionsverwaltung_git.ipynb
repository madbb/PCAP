Warum Versionsverwaltung?</span></div>

Stell dir vor, du arbeitest an einem Python-Projekt. Du aenderst eine Funktion -- und drei Stunden spaeter merkst du, dass die alte Version besser war. Oder ein Kollege hat gleichzeitig dieselbe Datei bearbeitet.

Ohne Versionsverwaltung loesen die meisten das so:

```
skript.py
skript_neu.py
skript_neu2.py
skript_final.py
skript_final_wirklich.py
skript_final_wirklich_v2.py
```

Git loest das sauber: Es speichert die vollstaendige Geschichte jeder Datei, wer was wann geaendert hat -- und erlaubt es, jederzeit auf jeden frueheren Zustand zurueckzugehen.

**Git ist heute der Standard.** Kein professionelles Softwareprojekt arbeitet ohne es.

- **Working Directory:** Die Dateien, die du gerade bearbeitest
- **Staging Area:** Aenderungen, die du fuer den naechsten Commit vorgemerkt hast
- **Repository:** Die vollstaendige Geschichte aller Commits (liegt im `.git`-Ordner)

Der Zwischenschritt ueber die Staging Area ist ein Feature, kein Bug: Er erlaubt dir, aus vielen gleichzeitigen Aenderungen gezielt nur bestimmte in einen Commit zu packen.

Git einrichten</span></div>

Vor dem ersten Einsatz muss Git wissen, wer du bist. Diese Angaben landen in jedem Commit.

````python
# Git-Konfiguration pruefen
import subprocess

def git(befehl, cwd=None):
    """Hilfsfunktion: Git-Befehl ausfuehren und Ausgabe zurueckgeben."""
    ergebnis = subprocess.run(
        ['git'] + befehl.split(),
        capture_output=True, text=True, cwd=cwd
    )
    ausgabe = ergebnis.stdout.strip() or ergebnis.stderr.strip()
    return ausgabe

print('Name:  ', git('config user.name'))
print('Email: ', git('config user.email'))
````

````python
# Falls noch nicht konfiguriert -- einmalig ausfuehren:
# Diese Befehle im Terminal ausfuehren, nicht hier im Notebook

print('Im Terminal ausfuehren:')
print('  git config --global user.name "Dein Name"')
print('  git config --global user.email "deine@email.de"')
print('  git config --global init.defaultBranch main')
print()
print('Aktuelle globale Konfiguration:')
print(git('config --global --list'))
````

Erstes Repository</span></div>

Alle folgenden Beispiele fuehren wir in einem temporaeren Verzeichnis durch, damit nichts am echten System veraendert wird.

````python
import os
import shutil

# Arbeitsverzeichnis fuer die Demo
REPO = '/tmp/git_demo'
if os.path.exists(REPO):
    shutil.rmtree(REPO)
os.makedirs(REPO)

def g(befehl):
    """Git im Demo-Repo ausfuehren."""
    return git(befehl, cwd=REPO)

def schreibe(dateiname, inhalt):
    """Datei im Demo-Repo anlegen oder ueberschreiben."""
    with open(os.path.join(REPO, dateiname), 'w') as f:
        f.write(inhalt)

# Repository initialisieren
print(g('init'))
print(g('config user.name "Demo User"'))
print(g('config user.email "demo@example.com"'))
````

````python
# Erste Datei anlegen
schreibe('rechner.py', '''def addiere(a, b):
    return a + b

def subtrahiere(a, b):
    return a - b
''')

# Status pruefen: Was sieht Git?
print('=== git status ===')
print(g('status'))
````

````python
# Datei zur Staging Area hinzufuegen
print(g('add rechner.py'))

print('=== git status nach add ===')
print(g('status'))
````

````python
# Commit erstellen
print(g('commit -m "Grundfunktionen: addiere und subtrahiere"'))

print()
print('=== git log ===')
print(g('log --oneline'))
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Was macht eine gute Commit-Nachricht aus?</span></div>

Eine Commit-Nachricht beantwortet die Frage: **Was und warum -- nicht wie.**

| Schlecht | Gut |
|----------|-----|
| `fix` | `Divisionsfehler bei Eingabe 0 behoben` |
| `changes` | `CSV-Export: Sonderzeichen in Spaltennamen escaped` |
| `asdf` | `Login-Timeout von 30 auf 60 Sekunden erhoeht (#42)` |
| `update` | `Veraltete requests-Version auf 2.31 aktualisiert` |

Konvention: Erste Zeile max. 72 Zeichen, Imperativ (`Behebt`, `Fuegt hinzu`, `Entfernt`), kein Punkt am Ende.

<div style="background:#0b1929;border-left:4px solid #4fc3f7;border-radius:4px;padding:0.75em 1.2em;margin:2rem 0 1.2rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(1rem,2.2vw,1.25rem);font-weight:600;color:#e8eaf6;letter-spacing:-0.01em;"><span style="font-size:0.65rem;letter-spacing:0.1em;text-transform:uppercase;color:#4fc3f7;background:rgba(79,195,247,0.12);border:1px solid rgba(79,195,247,0.3);border-radius:3px;padding:0.2em 0.55em;margin-right:0.75em;vertical-align:middle;font-family:'Segoe UI',sans-serif;white-space:nowrap;">Kursinhalt</span>5. .gitignore -- Was nicht ins Repository gehoert</span></div>

Nicht alles in einem Projektordner soll versioniert werden. Die `.gitignore`-Datei sagt Git, welche Dateien und Verzeichnisse ignoriert werden sollen.

Fuer Python-Projekte gibt es eine Standardliste:

````python
gitignore_inhalt = '''# Bytecode und Cache
__pycache__/
*.py[cod]
*.pyo

# Virtuelle Umgebungen
.venv/
venv/
env/
.env

# Distribution / Packaging
dist/
build/
*.egg-info/

# Tests und Coverage
.pytest_cache/
.coverage
htmlcov/

# IDE-Konfigurationen
.idea/
.vscode/
*.iml

# Betriebssystem
.DS_Store
Thumbs.db

# Geheimnisse -- niemals ins Repository!
.env
secrets.py
config_local.py
'''

schreibe('.gitignore', gitignore_inhalt)
print(g('add .gitignore'))
print(g('commit -m "gitignore fuer Python-Projekte hinzugefuegt"'))
print(g('log --oneline'))
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Warum .venv nicht ins Repository?</span></div>

Eine virtuelle Umgebung enthaelt Hunderte von Dateien, ist betriebssystemabhaengig (Windows-Pfade funktionieren nicht auf Linux) und ist jederzeit aus einer `requirements.txt` reproduzierbar.

Der korrekte Workflow:

```bash
# Abhaengigkeiten sichern
pip freeze > requirements.txt
git add requirements.txt
git commit -m "Abhaengigkeiten dokumentiert"

# Auf einem anderen Rechner wiederherstellen
python -m venv .venv
.venv\Scripts\activate        # Windows
source .venv/bin/activate     # Linux/Mac
pip install -r requirements.txt
```

Aenderungen rueckgaengig machen</span></div>

Das ist eine der wichtigsten Faehigkeiten -- und der Grund, warum man ueberhaupt Git benutzt.

````python
# Szenario: Datei versehentlich beschaedigt
schreibe('rechner.py', '# Ups, alles geloescht')
print('Inhalt nach dem Unfall:')
print(open(os.path.join(REPO, 'rechner.py')).read())

# Noch nicht gestaget -- Datei auf letzten Commit zuruecksetzen
print()
print(g('restore rechner.py'))

print('Inhalt nach git restore:')
print(open(os.path.join(REPO, 'rechner.py')).read())
````

````python
# Szenario: Datei versehentlich gestaget
schreibe('versehen.py', 'diese Datei sollte nicht committet werden')
print(g('add versehen.py'))
print('Nach add:', g('status'))

print()
# Aus der Staging Area entfernen (Datei bleibt unveraendert)
print(g('restore --staged versehen.py'))
print('Nach restore --staged:', g('status'))
````

````python
# Szenario: Letzten Commit rueckgaengig machen (sicher: neuer Revert-Commit)
print('Commit rueckgaengig machen (sicher, Geschichte bleibt erhalten):')
print('  git revert HEAD            -- letzten Commit umkehren')
print('  git revert HEAD~2          -- vorletzten Commit umkehren')
print()
print('Zu einem alten Stand springen (fuer Inspektion, nicht fuer Aenderungen):')
print('  git checkout <commit-hash> -- Repository auf alten Stand setzen')
print('  git switch -              -- zurueck zum letzten Branch')
print()
print('ACHTUNG -- diese Befehle aendern die Geschichte (nur im lokalen Repo):')
print('  git reset HEAD~1 --soft   -- letzten Commit aufloesen, Aenderungen bleiben gestaget')
print('  git reset HEAD~1 --mixed  -- letzten Commit aufloesen, Aenderungen im Working Dir')
print('  git reset HEAD~1 --hard   -- letzten Commit und alle Aenderungen vernichten')
````

Branches</span></div>

Ein Branch ist ein unabhaengiger Entwicklungsstrang. Er erlaubt es, an einem Feature oder Bugfix zu arbeiten, ohne den Hauptzweig zu beeinflussen.

```
main:    A --- B --- C ------- F
                      \       /
feature:              D --- E
```

Commits D und E passieren auf dem Feature-Branch. Wenn das Feature fertig ist, wird es in `main` gemergt (F).

````python
# Aktuellen Branch anzeigen
print('Aktueller Branch:', g('branch'))

# Neuen Branch erstellen und wechseln
print(g('switch -c feature/division'))
print('Nach switch:', g('branch'))
````

````python
# Auf dem Feature-Branch arbeiten
schreibe('rechner.py', '''def addiere(a, b):
    return a + b

def subtrahiere(a, b):
    return a - b

def multipliziere(a, b):
    return a * b

def dividiere(a, b):
    if b == 0:
        raise ValueError("Division durch Null ist nicht erlaubt")
    return a / b
''')

print(g('add rechner.py'))
print(g('commit -m "dividiere mit Nullpruefung hinzugefuegt"'))
print()
print(g('log --oneline --graph --all'))
````

````python
# Zurueck zu main und Feature mergen
print(g('switch main'))
print()

# Vor dem Merge: main hat die dividiere-Funktion noch nicht
print('Inhalt rechner.py auf main (vor merge):')
inhalt = open(os.path.join(REPO, 'rechner.py')).read()
print('dividiere vorhanden:', 'dividiere' in inhalt)

print()
print(g('merge feature/division -m "Feature: dividiere integriert"'))
print()
print('dividiere vorhanden nach merge:', 'dividiere' in open(os.path.join(REPO, 'rechner.py')).read())
print()
print(g('log --oneline --graph --all'))
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Merge-Konflikte</span></div>

Ein Konflikt entsteht, wenn dieselbe Stelle einer Datei auf zwei Branches unterschiedlich veraendert wurde. Git kann nicht automatisch entscheiden, welche Version richtig ist.

````python
# Konflikt simulieren
print(g('switch -c branch_a'))
schreibe('info.txt', 'Version von Branch A\n')
print(g('add info.txt'))
print(g('commit -m "info.txt auf branch_a erstellt"'))

print(g('switch main'))
schreibe('info.txt', 'Version von main\n')
print(g('add info.txt'))
print(g('commit -m "info.txt auf main erstellt"'))

# Jetzt mergen -- das gibt einen Konflikt
ergebnis = subprocess.run(
    ['git', 'merge', 'branch_a'],
    capture_output=True, text=True, cwd=REPO
)
print(ergebnis.stdout or ergebnis.stderr)
````

````python
# So sieht eine Konfliktdatei aus
print('Inhalt von info.txt mit Konfliktmarkierungen:')
print(open(os.path.join(REPO, 'info.txt')).read())

print()
print('Erklaerung:')
print('  <<<<<<< HEAD         -- Beginn der eigenen Version (main)')
print('  =======              -- Trenner zwischen den Versionen')
print('  >>>>>>> branch_a     -- Ende der anderen Version (branch_a)')
print()
print('Loesung: Datei manuell bearbeiten, Markierungen entfernen, dann:')
print('  git add info.txt')
print('  git commit')
````

````python
# Konflikt aufloesen
schreibe('info.txt', 'Zusammengefuehrte Version: main und branch_a\n')
print(g('add info.txt'))
print(g('commit -m "Konflikt in info.txt aufgeloest"'))
print()
print(g('log --oneline --graph --all'))
````

Remote-Repository (GitHub / GitLab)</span></div>

Ein Remote-Repository ist eine Kopie des Repositorys auf einem Server. Es dient als zentraler Austauschpunkt fuer Teams und als Backup.

```
Lokales Repo         Remote Repo (GitHub)
    |                       |
    |  git push  ---------->|
    |                       |
    |  git pull  <----------|
    |                       |
    |  git fetch <----------| (nur herunterladen, nicht mergen)
```

````python
print('Typischer Workflow mit Remote-Repository:')
print()
print('# Neues Repo auf GitHub anlegen, dann lokal verknuepfen:')
print('  git remote add origin https://github.com/user/repo.git')
print('  git push -u origin main')
print()
print('# Vorhandenes Repo klonen:')
print('  git clone https://github.com/user/repo.git')
print()
print('# Aenderungen hochladen:')
print('  git push')
print()
print('# Aenderungen holen und mergen:')
print('  git pull')
print()
print('# Nur holen, noch nicht mergen:')
print('  git fetch')
print('  git diff main origin/main   # Unterschiede anzeigen')
print('  git merge origin/main       # dann mergen')
````

Branch-Strategie fuer Python-Projekte</span></div>

In professionellen Projekten gibt es Konventionen, welche Branches wofuer genutzt werden. Ein einfaches Modell:

| Branch | Zweck |
|--------|-------|
| `main` | Stabiler, produktionsreifer Code |
| `develop` | Integrationsbranch fuer Features |
| `feature/name` | Einzelnes Feature in Entwicklung |
| `bugfix/name` | Fehlerbehebung |
| `hotfix/name` | Kritischer Fix direkt auf main |

Fuer kleine Projekte und Einzelpersonen reicht oft: `main` fuer stabilen Code, Feature-Branches fuer jede Aenderung.

Git in PyCharm</span></div>

PyCharm hat eine vollstaendige Git-Integration -- du musst die meisten Befehle nie in der Kommandozeile tippen.

**Wichtige Stellen in PyCharm:**

- `Git`-Menue in der Menueleiste: alle Befehle
- `Commit`-Tab (links): Staging Area und Commit-Nachricht
- `Git`-Tab (unten): Log, Branches, Remote
- Dateifarben im Projektbaum:
  - **Gruen:** neue, noch nicht commitete Datei
  - **Blau:** geaenderte, committete Datei
  - **Grau:** ignorierte Datei (in .gitignore)
  - **Rot:** Konflikt

**Konflikte loesen:** PyCharm hat einen eingebauten Merge-Editor mit drei Spalten (lokal / Ergebnis / remote) -- deutlich komfortabler als die Textmarkierungen.

<div style="background:#0b1929;border-left:4px solid #4fc3f7;border-radius:4px;padding:0.75em 1.2em;margin:2rem 0 1.2rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(1rem,2.2vw,1.25rem);font-weight:600;color:#e8eaf6;letter-spacing:-0.01em;">Zusammenfassung</span></div>

| Befehl | Was er tut |
|--------|------------|
| `git init` | Neues Repository anlegen |
| `git status` | Zustand des Working Directory anzeigen |
| `git add <datei>` | Datei zur Staging Area hinzufuegen |
| `git add .` | Alle Aenderungen stagen |
| `git commit -m "..."` | Commit erstellen |
| `git log --oneline` | Geschichte anzeigen |
| `git diff` | Aenderungen seit letztem Commit anzeigen |
| `git restore <datei>` | Datei auf letzten Commit zuruecksetzen |
| `git restore --staged <datei>` | Datei aus Staging Area entfernen |
| `git switch -c <n>` | Neuen Branch erstellen und wechseln |
| `git switch <n>` | Branch wechseln |
| `git merge <n>` | Branch mergen |
| `git push` | Commits zum Remote senden |
| `git pull` | Commits vom Remote holen und mergen |
| `git clone <url>` | Repository klonen |

**Merksatz:** Commit oft, commit klein, commit mit aussagekraeftiger Nachricht. Ein Commit pro logischer Aenderung -- nicht pro Arbeitsstunde.
