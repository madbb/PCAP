1. Was ist ein Callback?
</span>
</div>

In 11a haben wir Fenster, Widgets und Layout gebaut -- aber die Widgets tun noch nichts. Ein Button der nicht reagiert ist sinnlos. Um Widgets zum Leben zu erwecken, braucht man **Callbacks**.

Ein **Callback** ist eine Funktion die nicht sofort aufgerufen wird, sondern zu einem späteren Zeitpunkt -- nämlich dann wenn ein bestimmtes Ereignis eintritt. Man übergibt die Funktion als Argument und sagt damit: "Wenn der Button geklickt wird, ruf diese Funktion auf."

Das Schlüsselprinzip: Man übergibt die **Referenz** auf die Funktion -- nicht das Ergebnis des Aufrufs.

```python
command=meine_funktion     # RICHTIG: Referenz -- tkinter ruft sie später auf
command=meine_funktion()   # FALSCH:  sofortiger Aufruf -- gibt None zurück
```

Das ist ein Fehler der sehr häufig passiert. Wenn man `meine_funktion()` mit Klammern schreibt, wird die Funktion sofort beim Erstellen des Buttons ausgeführt -- und der Rückgabewert (meistens `None`) wird als `command` gesetzt. Der Button tut dann beim Klicken gar nichts. Keine Fehlermeldung, einfach Stille.

````python
# fenster_08_callback.py
import tkinter as tk

# Diese Funktion wird als Callback registriert.
# Sie wird NICHT sofort aufgerufen -- tkinter ruft sie auf wenn der Nutzer klickt.
def bei_klick():
    print('Button wurde geklickt!')

root = tk.Tk()
root.title('Erster Callback')
root.geometry('250x120')

# command=bei_klick: Referenz auf die Funktion, KEINE Klammern
# tkinter speichert die Referenz und ruft sie bei jedem Klick auf
tk.Button(root, text='Klick mich', command=bei_klick).pack(pady=30)

# Zum Vergleich -- was passiert mit Klammern:
# command=bei_klick()  <-- bei_klick() wird JETZT ausgefuehrt, gibt None zurueck
#                          Der Button bekommt None als command -> tut nichts

root.mainloop()
````

Die Ausgabe `'Button wurde geklickt!'` erscheint im Terminal -- nicht im Fenster. Das ist für den Anfang in Ordnung. Später schreiben wir die Ausgabe in ein Label im Fenster selbst.

2. Eingaben auslesen -- entry.get(), entry.delete(), entry.insert()
</span>
</div>

Mit `entry.get()` liest man den aktuellen Inhalt eines Eingabefelds aus. Das klingt einfach -- aber es gibt drei Details die man kennen muss:

**`get()` gibt immer einen String zurück -- auch wenn der Nutzer eine Zahl eingibt.** `'42'` ist nicht dasselbe wie `42`. Wenn man mit der Eingabe rechnen will, muss man sie zuerst umwandeln: `int(entry.get())` oder `float(entry.get())`. Diese Umwandlung kann scheitern -- sie gehört immer in einen `try/except`-Block.

**Leeres Feld:** Wenn das Eingabefeld leer ist, gibt `get()` einen leeren String `''` zurück -- nicht `None`. Das ist wichtig für die Fehlerbehandlung: `if not entry.get()` prüft ob das Feld leer ist.

**`entry.delete(start, end)`** -- löscht Zeichen aus dem Eingabefeld. Die beiden Parameter sind Zeichenpositionen: `0` ist das erste Zeichen, `tk.END` ist eine tkinter-Konstante die "bis zum Ende" bedeutet. `entry.delete(0, tk.END)` löscht also den gesamten Inhalt. Man kann auch nur Teile löschen: `entry.delete(0, 3)` löscht die ersten drei Zeichen.

**Was ist `tk.END`?** Eine von tkinter definierte Konstante -- ihr Wert ist der String `'end'`. tkinter benutzt solche Konstanten an vielen Stellen. Man könnte `entry.delete(0, 'end')` schreiben und es wäre dasselbe -- aber `tk.END` ist lesbarer und weniger fehleranfällig.

**`entry.insert(position, text)`** -- fügt Text an einer Position ein. `entry.insert(0, 'Hallo')` fügt am Anfang ein. `entry.insert(tk.END, ' Welt')` hängt am Ende an. Wenn man das Feld komplett neu befüllen will, löscht man zuerst mit `delete(0, tk.END)` und fügt dann neu ein.

````python
# fenster_09_entry.py
import tkinter as tk

def auslesen():
    inhalt = eingabe.get()           # gibt immer einen String zurueck!
    print(f'Inhalt: {repr(inhalt)}') # repr() macht leer vs. Text sichtbar
    print(f'Typ: {type(inhalt)}')    # immer <class 'str'>

def leeren():
    # delete(start, end): loescht von Position 0 bis zum Ende
    # tk.END ist eine Konstante -- ihr Wert ist der String 'end'
    eingabe.delete(0, tk.END)

def neu_befuellen():
    eingabe.delete(0, tk.END)        # erst loeschen -- sonst wird Text nur angehaengt
    eingabe.insert(0, 'Hallo Welt')  # dann neu einsetzen -- 0 = am Anfang

def anhaengen():
    # insert(tk.END, text): haengt Text ans Ende des bestehenden Inhalts an
    eingabe.insert(tk.END, '!')

root = tk.Tk()
root.title('Entry-Methoden')
root.geometry('320x200')

tk.Label(root, text='Eingabefeld:').pack(pady=(12, 2))

# width=30: Breite in ZEICHEN -- nicht Pixel (wie bei allen Entry- und Button-Widgets)
eingabe = tk.Entry(root, width=30)
eingabe.pack(pady=4)

frame = tk.Frame(root)
frame.pack(pady=8)

tk.Button(frame, text='Auslesen',     command=auslesen).pack(    side='left', padx=4)
tk.Button(frame, text='Leeren',       command=leeren).pack(      side='left', padx=4)
tk.Button(frame, text='Befuellen',    command=neu_befuellen).pack(side='left', padx=4)
tk.Button(frame, text='! anhaengen',  command=anhaengen).pack(   side='left', padx=4)

root.mainloop()
````

Was passiert wenn man `int(entry.get())` aufruft und das Feld leer ist oder Text enthält? Ein `ValueError`. In einem normalen Python-Skript würde der Fehler im Terminal erscheinen und das Programm beenden. In tkinter ist das anders.

**tkinter schluckt Fehler in Callbacks manchmal stillschweigend.** Das Fenster bleibt offen, der Button tut nichts, und man sieht keine Fehlermeldung. Das ist gefährlicher als ein Programmabsturz -- man weiß nicht warum etwas nicht funktioniert.

Deshalb gilt in tkinter: **Callbacks immer mit try/except absichern.** Fehler müssen explizit behandelt werden. Die beste Strategie: Fehlermeldungen direkt im Fenster anzeigen, zum Beispiel indem man ein Label rot einfärbt.

````python
# fenster_10_fehler.py
import tkinter as tk

def berechne():
    try:
        # float() statt int() -- akzeptiert auch Dezimalzahlen
        wert = float(eingabe.get())      # kann ValueError werfen wenn kein gueltiger Wert
        ergebnis = wert * 2
        # Ergebnis anzeigen -- Textfarbe auf schwarz zuruecksetzen
        anzeige.config(text=f'Ergebnis: {ergebnis}', fg='black')
    except ValueError:
        # Fehlermeldung IM FENSTER anzeigen -- nicht auf der Konsole
        # fg='red': Textfarbe auf rot setzen damit der Fehler auffaellt
        anzeige.config(text='Bitte eine gueltige Zahl eingeben!', fg='red')

root = tk.Tk()
root.title('Fehlerbehandlung')
root.geometry('300x140')

eingabe = tk.Entry(root, width=22)
eingabe.pack(pady=10)

tk.Button(root, text='Berechnen (Wert * 2)', command=berechne).pack()

# Dieses Label zeigt entweder das Ergebnis oder eine Fehlermeldung
# Am Anfang ist es leer
anzeige = tk.Label(root, text='')
anzeige.pack(pady=10)

root.mainloop()
````

3. Widgets dynamisch ändern -- .config() und die Dictionary-Syntax
</span>
</div>

Aus 11a kennen wir `.config()` -- damit kann man Parameter eines Widgets nachträglich ändern. Das ist der Standardweg um die Oberfläche dynamisch zu aktualisieren.

Man kann alle Parameter die man beim Erstellen angegeben hat auch nachträglich ändern: `text`, `bg`, `fg`, `font`, `width`, `state` und mehr.

**`.cget(parametername)`** -- liest den aktuellen Wert eines Parameters aus:

```python
aktueller_text = label.cget('text')   # gibt den aktuellen Text zurück
```

**Alternative Schreibweise mit Dictionary-Syntax:** In fremdem Code sieht man häufig eine andere Schreibweise:

```python
label.config(text='Neuer Text')   # Variante A -- am gebräuchlichsten
label['text'] = 'Neuer Text'      # Variante B -- Dictionary-Syntax
```

Beide tun exakt dasselbe. Variante B sieht aus wie ein Dictionary-Zugriff -- und das ist kein Zufall. tkinter speichert Widget-Optionen intern tatsächlich ähnlich wie ein Dictionary. Man liest auch so:

```python
print(label['text'])    # wie label.cget('text')
```

Variante A bevorzugen weil lesbarer -- aber Variante B in fremdem Code erkennen können.

````python
# fenster_11_config.py
import tkinter as tk

def rot():
    # .config() aendert mehrere Parameter auf einmal
    anzeige.config(text='Rot!', bg='red', fg='white')

def gruen():
    anzeige.config(text='Gruen!', bg='lightgreen', fg='black')

def zurueck():
    # Variante B: Dictionary-Syntax -- tut dasselbe wie .config()
    anzeige['text'] = 'Normal'
    anzeige['bg']   = 'lightyellow'
    anzeige['fg']   = 'black'

def aktuellen_text_ausgeben():
    # .cget('parametername') liest den aktuellen Wert eines Parameters
    print(f'Aktueller Text: {anzeige.cget("text")}')
    print(f'Gleich mit []: {anzeige["text"]}')

root = tk.Tk()
root.title('.config() und Dictionary-Syntax')
root.geometry('320x180')

# Das Label das veraendert wird
# width=20: Breite in Zeichen -- verhindert dass das Fenster bei kurzem Text schrumpft
anzeige = tk.Label(root, text='Normal', bg='lightyellow', width=20, font=('Arial', 13))
anzeige.pack(pady=15)

# Buttons in einer Reihe
frame = tk.Frame(root)
frame.pack()

tk.Button(frame, text='Rot',           command=rot).pack(                  side='left', padx=4)
tk.Button(frame, text='Gruen',         command=gruen).pack(                side='left', padx=4)
tk.Button(frame, text='Normal',        command=zurueck).pack(              side='left', padx=4)
tk.Button(frame, text='Text ausgeben', command=aktuellen_text_ausgeben).pack(side='left', padx=4)

root.mainloop()
````

4. StringVar, IntVar, BooleanVar -- tkinter-Variablen
</span>
</div>

Es gibt eine elegantere Alternative zu `entry.get()` und `label.config()` -- tkinters eigene **Variable-Klassen**.

Was sind das? Spezielle Objekte die einen Wert speichern -- und alle Widgets die mit dieser Variable verbunden sind, aktualisieren sich automatisch wenn der Wert sich ändert. Man muss nicht manuell mit `.config()` nachziehen.

tkinter kennt vier Variable-Klassen:

| Klasse | Speichert | Verwendung |
|--------|-----------|------------|
| `tk.StringVar()` | Text (`str`) | Entry, Label, Optionmenus |
| `tk.IntVar()` | Ganzzahl (`int`) | Checkbutton, Radiobutton |
| `tk.DoubleVar()` | Dezimalzahl (`float`) | Scale-Widget |
| `tk.BooleanVar()` | Wahrheitswert (`bool`) | Checkbutton |

Man verbindet ein Widget mit einer Variablen über den `textvariable=`-Parameter (bei Entry und Label) oder den `variable=`-Parameter (bei Checkbutton und Radiobutton):

```python
mein_text = tk.StringVar()                          # Variable erstellen
entry = tk.Entry(root, textvariable=mein_text)      # Entry verbinden
label = tk.Label(root, textvariable=mein_text)      # Label verbinden
```

Jetzt sind Entry und Label **mit derselben Variable** verbunden. Was der Nutzer ins Entry-Feld tippt, erscheint sofort im Label. Man braucht keinen Button und keinen Callback.

**`.get()` und `.set()`** sind die Methoden um den Wert zu lesen und zu setzen:

```python
mein_text.set('Hallo')    # Wert setzen -- alle verbundenen Widgets aktualisieren sich
wert = mein_text.get()    # Wert lesen
```

**Wann `entry.get()`, wann `StringVar`?**
- `entry.get()` ist einfacher wenn man den Wert nur einmal beim Button-Klick braucht
- `StringVar` ist eleganter wenn mehrere Widgets denselben Wert teilen sollen, oder wenn man auf jede Änderung sofort reagieren will (mit `trace` -- dazu gleich mehr)

````python
# fenster_12_stringvar.py
import tkinter as tk

root = tk.Tk()
root.title('StringVar')
root.geometry('320x200')

# StringVar erstellen -- noch mit keinem Widget verbunden
eingabe_var = tk.StringVar()

tk.Label(root, text='Eingabe:').pack(pady=(12, 2))

# textvariable=: verbindet das Entry mit der StringVar
# Jede Aenderung im Feld aktualisiert automatisch eingabe_var
tk.Entry(root, textvariable=eingabe_var, width=28).pack(pady=4)

tk.Label(root, text='Vorschau (live):').pack(pady=(10, 2))

# Dieses Label zeigt automatisch was im Entry steht --
# ohne Button, ohne Callback, einfach weil es dieselbe Variable teilt
tk.Label(root, textvariable=eingabe_var, bg='lightyellow', width=28).pack(pady=4)

def vorbefuellen():
    # .set() setzt den Wert der Variable --
    # BEIDE verbundenen Widgets (Entry und Label) aktualisieren sich sofort
    eingabe_var.set('Hallo Welt')

def wert_lesen():
    # .get() liest den aktuellen Wert -- egal ob Entry oder Variable
    print(f'Aktueller Wert: {repr(eingabe_var.get())}')

frame = tk.Frame(root)
frame.pack(pady=8)
tk.Button(frame, text='Vorbefuellen', command=vorbefuellen).pack(side='left', padx=5)
tk.Button(frame, text='Wert ausgeben', command=wert_lesen).pack(side='left', padx=5)

root.mainloop()
````

<span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">trace_add -- bei jeder Änderung reagieren</span>

Mit `textvariable=` aktualisieren sich Widgets automatisch -- aber man kann noch einen Schritt weitergehen: Eine eigene Funktion aufrufen die bei **jeder Änderung** ausgeführt wird. Das nennt sich **Trace**.

```python
var.trace_add('write', callback)
```

`'write'` bedeutet: reagiere wenn der Wert der Variable geschrieben (verändert) wird. Es gibt auch `'read'` (beim Lesen) und `'unset'` (beim Löschen) -- aber `'write'` ist bei weitem das Häufigste.

**Der Callback bekommt drei Argumente** die tkinter intern braucht -- man nennt sie per Konvention `name`, `index`, `mode`. Man braucht sie fast nie, muss sie aber in der Signatur stehen haben:

```python
def bei_aenderung(name, index, mode):
    aktueller_wert = var.get()
    # ...
```

**Typischer Einsatz:** Suchfeld das die Liste sofort filtert während der Nutzer tippt -- ohne dass der Nutzer einen Button drücken muss. Oder ein Passwortfeld das die Stärke live anzeigt.

````python
# fenster_13_trace.py -- Live-Zeichenzaehler
import tkinter as tk

def bei_eingabe(name, index, mode):
    # Diese Funktion wird bei JEDER Aenderung der Variable aufgerufen --
    # also bei jedem Tastendruck, nicht erst beim Klick auf einen Button.
    # name, index, mode: interne tkinter-Argumente -- hier nicht benoetigt,
    # muessen aber in der Signatur stehen
    aktuell = eingabe_var.get()
    anzahl  = len(aktuell)

    # Zeichenzaehler aktualisieren
    zaehler_var.set(f'{anzahl} / 50 Zeichen')

    # Farbe wechseln wenn nahe am Limit
    if anzahl > 50:
        zaehler_label.config(fg='red')
    elif anzahl > 40:
        zaehler_label.config(fg='orange')
    else:
        zaehler_label.config(fg='gray')

    # Vorschau aktualisieren (auch ohne textvariable moeglich)
    vorschau.config(text=aktuell if aktuell else '(leer)')

root = tk.Tk()
root.title('trace_add -- Live-Reaktion')
root.geometry('350x180')

eingabe_var = tk.StringVar()
zaehler_var = tk.StringVar(value='0 / 50 Zeichen')

# 'write': Callback ausfuehren wenn Variable VERAENDERT wird
# bei_eingabe wird bei jedem Tastendruck aufgerufen
eingabe_var.trace_add('write', bei_eingabe)

tk.Label(root, text='Nachricht:').pack(pady=(12, 2))
tk.Entry(root, textvariable=eingabe_var, width=35).pack()

# Zeichenzaehler -- textvariable zeigt den Zaehlertext live
zaehler_label = tk.Label(root, textvariable=zaehler_var, fg='gray', font=('Arial', 9))
zaehler_label.pack()

tk.Label(root, text='Vorschau:').pack(pady=(8, 2))
vorschau = tk.Label(root, text='(leer)', fg='blue')
vorschau.pack()

root.mainloop()
````

5. Checkbutton und Radiobutton
</span>
</div>

Bisher haben wir nur Buttons und Entry-Felder gesehen. Für Ja/Nein-Entscheidungen und Auswahlgruppen gibt es zwei weitere wichtige Widgets.

**Checkbutton** -- eine Checkbox die an- und abgehakt werden kann. Jede Checkbox ist unabhängig von den anderen. Man verbindet sie mit einer `BooleanVar` (oder `IntVar`) über den `variable=`-Parameter:

```python
opt_a = tk.BooleanVar()   # speichert True/False
tk.Checkbutton(root, text='Option A', variable=opt_a)
```

`opt_a.get()` gibt `True` zurück wenn die Box angehakt ist, `False` wenn nicht. Man kann auch direkt setzen: `opt_a.set(True)` setzt die Checkbox.

**Radiobutton** -- mehrere Optionen von denen immer genau eine ausgewählt ist (wie bei alten Radiogeräten). Alle Radiobuttons einer Gruppe teilen dieselbe `StringVar` oder `IntVar`. Jeder Button hat einen eigenen `value` der in die Variable geschrieben wird wenn er ausgewählt wird:

```python
auswahl = tk.StringVar(value='a')   # Startwert
tk.Radiobutton(root, text='Option A', variable=auswahl, value='a')
tk.Radiobutton(root, text='Option B', variable=auswahl, value='b')
tk.Radiobutton(root, text='Option C', variable=auswahl, value='c')
```

Wenn der Nutzer "Option B" anklickt, setzt tkinter automatisch `auswahl.set('b')`. Die anderen Buttons werden dabei automatisch abgewählt. `auswahl.get()` gibt dann `'b'` zurück.

**Warum `BooleanVar` für Checkbutton und nicht `StringVar`?** Weil `BooleanVar` direkt `True`/`False` zurückgibt statt `'1'`/`'0'` -- das macht die Abfrage einfacher.

````python
# fenster_14_check_radio.py
import tkinter as tk

def auswerten():
    zeilen = []

    # Checkbuttons: .get() gibt True oder False zurueck
    zeilen.append(f'Tomaten:  {"Ja" if tomaten_var.get() else "Nein"}')
    zeilen.append(f'Kaese:    {"Ja" if kaese_var.get()   else "Nein"}')
    zeilen.append(f'Oliven:   {"Ja" if oliven_var.get()  else "Nein"}')

    # Radiobutton: .get() gibt den value des ausgewaehlten Buttons zurueck
    zeilen.append(f'Groesse:  {groesse_var.get()}')

    ergebnis.config(text='\n'.join(zeilen))

root = tk.Tk()
root.title('Checkbutton und Radiobutton')
root.geometry('280x320')

# --- Checkbuttons ---
tk.Label(root, text='Zutaten:', font=('Arial', 10, 'bold')).pack(pady=(10, 2), anchor='w', padx=15)

# BooleanVar: speichert True/False je nach Hakenstatus
# value=False: am Anfang nicht angehakt
tomaten_var = tk.BooleanVar(value=False)
kaese_var   = tk.BooleanVar(value=True)   # am Anfang angehakt
oliven_var  = tk.BooleanVar(value=False)

# variable=: verbindet den Checkbutton mit der BooleanVar
# Haken setzen -> var.get() gibt True zurueck
# Haken loesen -> var.get() gibt False zurueck
tk.Checkbutton(root, text='Tomaten', variable=tomaten_var).pack(anchor='w', padx=25)
tk.Checkbutton(root, text='Kaese',   variable=kaese_var).pack(  anchor='w', padx=25)
tk.Checkbutton(root, text='Oliven',  variable=oliven_var).pack( anchor='w', padx=25)

# --- Radiobuttons ---
tk.Label(root, text='Groesse:', font=('Arial', 10, 'bold')).pack(pady=(10, 2), anchor='w', padx=15)

# StringVar: alle drei Radiobuttons teilen DIESELBE Variable
# value='M': Startwert -- 'M' ist am Anfang ausgewaehlt
groesse_var = tk.StringVar(value='M')

# variable= und value=: alle drei zeigen auf dieselbe Variable,
# haben aber verschiedene value= -- das unterscheidet sie
# Wenn der Nutzer 'S' waehlt: groesse_var.get() gibt 'S' zurueck
tk.Radiobutton(root, text='Klein (S)', variable=groesse_var, value='S').pack(anchor='w', padx=25)
tk.Radiobutton(root, text='Mittel (M)', variable=groesse_var, value='M').pack(anchor='w', padx=25)
tk.Radiobutton(root, text='Gross (L)',  variable=groesse_var, value='L').pack(anchor='w', padx=25)

tk.Button(root, text='Bestellung anzeigen', command=auswerten).pack(pady=8)

# justify='left': mehrzeiliger Text linksbuendig ausrichten
ergebnis = tk.Label(root, text='', justify='left', font=('Courier', 9))
ergebnis.pack()

root.mainloop()
````

6. Lambda bei command -- Argumente an Callbacks übergeben
</span>
</div>

Was wenn man einer Callback-Funktion ein Argument mitgeben will? `command=` erwartet eine Funktion ohne Argumente -- man kann nicht einfach `command=meine_funktion('hallo')` schreiben, weil das die Funktion sofort aufrufen würde (Klammernproblem von Abschnitt 1).

Die Lösung: **lambda**. Man verpackt den Aufruf in eine parameterlose Lambda-Funktion:

```python
command=lambda: meine_funktion('hallo')
```

Das erzeugt eine neue, parameterlose Funktion. tkinter speichert die Referenz auf diese Lambda-Funktion. Beim Klick ruft tkinter die Lambda auf -- und die Lambda ruft dann `meine_funktion('hallo')` auf.

Das ist besonders nützlich wenn man mehrere Buttons hat die dieselbe Funktion aufrufen, aber mit verschiedenen Argumenten.

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;">
<span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Die Schleifenfalle -- warum lambda f=farbe statt lambda: farbe</span>
</div>

Wenn man Buttons in einer Schleife erzeugt, gibt es eine gefährliche Falle:

```python
farben = ['rot', 'gruen', 'blau']
for farbe in farben:
    tk.Button(root, command=lambda: print(farbe)).pack()
```

Man erwartet drei Buttons die jeweils ihre Farbe ausgeben. Tatsächlich geben alle drei `'blau'` aus -- den letzten Wert der Schleifenvariable.

**Warum?** Weil Lambda keine eigene Kopie von `farbe` erstellt -- es speichert nur die Referenz auf die Variable. Nach der Schleife hat `farbe` den Wert `'blau'`. Alle drei Lambdas zeigen auf dieselbe Variable und lesen beim Klick `'blau'`.

**Die Lösung:** Den Wert zum Zeitpunkt der Lambda-Erstellung einfrieren mit einem Default-Argument:

```python
command=lambda f=farbe: print(f)   # f=farbe: Wert jetzt festhalten
```

`f=farbe` ist ein Default-Parameter der Lambda. Er wird sofort beim Erstellen ausgewertet -- nicht erst beim Aufruf. Jede Lambda bekommt so ihren eigenen eingefrorenen Wert. Das ist dasselbe Prinzip das wir aus den Closures in Kapitel 3 kennen.

````python
# fenster_15_lambda.py
import tkinter as tk

def setze_hintergrund(farbe):
    # Diese Funktion erwartet ein Argument -- command= kann sie nicht direkt benutzen
    hintergrund.config(bg=farbe)

root = tk.Tk()
root.title('Lambda bei command')
root.geometry('350x220')

hintergrund = tk.Frame(root, bg='white')
hintergrund.pack(fill='both', expand=True)

button_frame = tk.Frame(hintergrund)
button_frame.pack(pady=10)

# Alle vier Buttons rufen dieselbe Funktion auf, aber mit verschiedenem Argument.
# lambda loest das Problem -- ohne lambda muesste man vier separate Funktionen schreiben.
farben = [
    ('Rot',    'lightcoral'),
    ('Gruen',  'lightgreen'),
    ('Blau',   'lightblue'),
    ('Gelb',   'lightyellow'),
]

for name, farbe in farben:
    tk.Button(
        button_frame,
        text=name,
        # f=farbe: Default-Argument friert den Wert zum Zeitpunkt der Lambda-Erstellung ein.
        # Ohne f=farbe wuerden alle Buttons denselben Wert benutzen (den letzten aus der Schleife).
        command=lambda f=farbe: setze_hintergrund(f)
    ).pack(side='left', padx=6)

# Erklaerung der Schleifenfalle als Kommentar:
# FALSCH: command=lambda: setze_hintergrund(farbe)
#         Nach der Schleife hat farbe den Wert 'lightyellow' -- alle Buttons machen dasselbe
# RICHTIG: command=lambda f=farbe: setze_hintergrund(f)
#          f=farbe wird sofort ausgewertet -- jede Lambda hat ihre eigene Kopie

tk.Label(hintergrund, text='Buttons anklicken um Farbe zu aendern',
         bg='white', fg='gray').pack(pady=5)

root.mainloop()
````

<span style="font-size:clamp(1rem,2.2vw,1.25rem);font-weight:600;color:#e8eaf6;letter-spacing:-0.01em;">
<span style="font-size:0.65rem;letter-spacing:0.1em;text-transform:uppercase;color:#4fc3f7;background:rgba(79,195,247,0.12);border:1px solid rgba(79,195,247,0.3);border-radius:3px;padding:0.2em 0.55em;margin-right:0.75em;vertical-align:middle;white-space:nowrap;">Kursinhalt</span>
7. Widgets aktivieren und deaktivieren -- state=
</span>
</div>

Widgets können in verschiedenen Zuständen sein. Der Zustand steuert ob das Widget auf Eingaben reagiert oder nicht -- und wie es aussieht.

**`state='normal'`** -- das Widget funktioniert normal. Das ist der Standardzustand.

**`state='disabled'`** -- das Widget ist deaktiviert. Es wird ausgegraut dargestellt, Klicks werden ignoriert, Tastatureingaben werden nicht akzeptiert. Ein deaktivierter Button kann nicht geklickt werden, ein deaktiviertes Entry-Feld kann nicht bearbeitet werden.

**`state='readonly'`** -- **nur bei Entry** verfügbar. Das Feld zeigt den Text an, der Nutzer kann ihn aber nicht ändern. Man kann den Inhalt aber trotzdem programmatisch mit `.delete()` und `.insert()` verändern. Das ist nützlich für Felder die Ergebnisse anzeigen -- der Nutzer soll lesen aber nicht tippen.

**Warum `disabled` und nicht einfach das Widget verstecken?** `disabled` ist besser als das Widget zu entfernen, weil der Nutzer sehen kann dass die Option existiert -- sie ist nur gerade nicht verfügbar. Das ist ein wichtiges GUI-Designprinzip: Dem Nutzer zeigen was möglich wäre, nicht alles verstecken.

Den Zustand ändert man mit `.config(state=...)`:

```python
button.config(state='disabled')   # deaktivieren
button.config(state='normal')     # wieder aktivieren
```

````python
# fenster_16_state.py
import tkinter as tk

def aktivieren():
    ziel_button.config(state='normal')          # Widget wieder aktivieren
    readonly_entry.config(state='normal')       # readonly Entry wieder bearbeitbar
    status.config(text='Alles aktiv', fg='green')

def deaktivieren():
    ziel_button.config(state='disabled')        # Button ausgegraut, kein Klick moeglich
    readonly_entry.config(state='readonly')     # Entry anzeigen aber nicht bearbeitbar
    status.config(text='Button deaktiviert, Entry nur lesbar', fg='gray')

def ziel_aktion():
    status.config(text='Ziel-Button wurde geklickt!', fg='blue')

root = tk.Tk()
root.title('Widget-Zustaende')
root.geometry('340x230')

tk.Label(root, text='Dieser Button kann aktiviert/deaktiviert werden:',
         anchor='w').pack(fill='x', padx=15, pady=(12, 2))

# Der Ziel-Button der veraendert wird
ziel_button = tk.Button(root, text='Ich bin der Ziel-Button', command=ziel_aktion)
ziel_button.pack(pady=4)

tk.Label(root, text='Dieses Entry-Feld kann readonly geschaltet werden:',
         anchor='w').pack(fill='x', padx=15, pady=(8, 2))

# Entry mit state='readonly': Nutzer kann lesen aber nicht tippen
# Der Inhalt kann aber programmatisch gesetzt werden
readonly_entry = tk.Entry(root, width=30)
readonly_entry.insert(0, 'Diesen Text sieht man, kann ihn aber nicht aendern')
readonly_entry.pack(pady=4)

# Steuerbuttons
frame = tk.Frame(root)
frame.pack(pady=8)
tk.Button(frame, text='Aktivieren',   command=aktivieren).pack(  side='left', padx=6)
tk.Button(frame, text='Deaktivieren', command=deaktivieren).pack(side='left', padx=6)

status = tk.Label(root, text='')
status.pack(pady=4)

# Beim Start deaktivieren -- zeigt den Ausgangszustand
deaktivieren()

root.mainloop()
````

<span style="font-size:clamp(1rem,2.2vw,1.25rem);font-weight:600;color:#e8eaf6;letter-spacing:-0.01em;">
<span style="font-size:0.65rem;letter-spacing:0.1em;text-transform:uppercase;color:#4fc3f7;background:rgba(79,195,247,0.12);border:1px solid rgba(79,195,247,0.3);border-radius:3px;padding:0.2em 0.55em;margin-right:0.75em;vertical-align:middle;white-space:nowrap;">Kursinhalt</span>
8. bind() -- auf Tastatur und Maus reagieren
</span>
</div>

`command=` funktioniert nur bei Buttons -- und nur für einfache Mausklicks. Für alles andere gibt es `bind()`. Damit kann man auf jedes Ereignis reagieren: Tastaturdrücke, Mausklicks auf beliebige Widgets, Mausbewegungen, Fensterereignisse.

```python
widget.bind('<Ereignisname>', callback_funktion)
```

Der Ereignisname steht in spitzen Klammern und folgt einem festen Schema. Hier die wichtigsten:

| Ereignis | Beschreibung |
|----------|--------------|
| `'<Return>'` | Enter-Taste gedrückt |
| `'<Escape>'` | Escape-Taste gedrückt |
| `'<space>'` | Leertaste gedrückt |
| `'<KeyPress>'` | Irgendeine Taste gedrückt |
| `'<KeyRelease>'` | Taste losgelassen |
| `'<Button-1>'` | Linke Maustaste geklickt |
| `'<Button-3>'` | Rechte Maustaste geklickt |
| `'<Double-Button-1>'` | Doppelklick mit linker Taste |
| `'<Motion>'` | Maus bewegt sich über Widget |
| `'<Enter>'` | Maus fährt über Widget (nicht Enter-Taste!) |
| `'<Leave>'` | Maus verlässt Widget |
| `'<FocusIn>'` | Widget bekommt Tastaturfokus |
| `'<FocusOut>'` | Widget verliert Tastaturfokus |

**Achtung bei `<Enter>` und `<Return>`:** `<Return>` ist die Enter-Taste. `<Enter>` ist das Ereignis wenn die Maus ein Widget betritt (hover). Das ist verwirrend -- gut einprägen.

**Das Event-Objekt:** Callbacks die mit `bind()` registriert werden, bekommen automatisch ein **Event-Objekt** als erstes Argument übergeben. Das ist kein optionaler Parameter -- tkinter ruft den Callback immer mit diesem Argument auf. Wenn man es in der Signatur weglässt, wirft Python beim Klick einen `TypeError`.

Das Event-Objekt enthält Informationen über das Ereignis:

```python
def bei_taste(event):
    print(event.keysym)    # Name der Taste: 'Return', 'a', 'space', ...
    print(event.char)      # Das Zeichen falls es eines ist: 'a', '3', ...
    print(event.x)         # Mausposition X relativ zum Widget
    print(event.y)         # Mausposition Y relativ zum Widget
    print(event.widget)    # Das Widget auf dem das Ereignis auftrat
```

Wenn man das Event-Objekt nicht braucht, benennt man den Parameter trotzdem -- per Konvention `event` oder kurz `_`.

````python
# fenster_17_bind.py -- bind() mit verschiedenen Ereignissen
import tkinter as tk

def bei_enter_taste(event):
    # event wird hier nicht benoetigt -- muss aber in der Signatur stehen
    # tkinter uebergibt es immer, Python wuerde TypeError werfen wenn es fehlt
    berechne()

def bei_escape(event):
    eingabe.delete(0, tk.END)           # Feld leeren
    ergebnis.config(text='', fg='black')

def bei_tastendruck(event):
    # event.keysym: Name der Taste als String
    # event.char: das Zeichen falls es druckbar ist
    info.config(text=f'Taste: {event.keysym!r}  |  Zeichen: {event.char!r}')

def bei_maus_ein(event):
    # <Enter>: Maus faehrt UEBER das Widget -- nicht die Enter-Taste!
    berechne_button.config(bg='lightblue')

def bei_maus_aus(event):
    # <Leave>: Maus verlasst das Widget
    berechne_button.config(bg='SystemButtonFace')  # Standard-Buttonfarbe

def berechne():
    try:
        zahl = float(eingabe.get())
        ergebnis.config(text=f'{zahl} ^ 2 = {zahl ** 2}', fg='black')
    except ValueError:
        ergebnis.config(text='Keine gueltige Zahl!', fg='red')

root = tk.Tk()
root.title('bind() -- verschiedene Ereignisse')
root.geometry('360x210')

tk.Label(root, text='Zahl eingeben:').pack(pady=(12, 2))

eingabe = tk.Entry(root, width=22)
eingabe.pack(pady=4)

# bind() auf dem Entry-Feld:
# <Return>: Enter-Taste loest Berechnung aus
eingabe.bind('<Return>', bei_enter_taste)
# <Escape>: Feld leeren
eingabe.bind('<Escape>', bei_escape)
# <KeyPress>: Bei JEDER Taste info aktualisieren -- zeigt event.keysym
eingabe.bind('<KeyPress>', bei_tastendruck)

berechne_button = tk.Button(root, text='Quadrat berechnen (oder Enter)', command=berechne)
berechne_button.pack(pady=5)

# bind() auf dem Button:
# <Enter>: Maus faehrt ueber den Button -> Farbe aendern (Hover-Effekt)
berechne_button.bind('<Enter>', bei_maus_ein)
berechne_button.bind('<Leave>', bei_maus_aus)

ergebnis = tk.Label(root, text='', font=('Arial', 11))
ergebnis.pack(pady=4)

# Zeigt welche Taste gerade gedrueckt wurde
info = tk.Label(root, text='', fg='gray', font=('Courier', 9))
info.pack()

# Fokus direkt ins Eingabefeld -- Nutzer kann sofort tippen
eingabe.focus()

root.mainloop()
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;">
<span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">Fokusmanagement -- wo landet der erste Tastendruck?</span>
</div>

Wenn ein Fenster öffnet, hat zunächst kein Widget den Tastaturfokus. Der Nutzer müsste erst ins Eingabefeld klicken bevor er tippen kann -- das ist schlechte Bedienbarkeit.

**`widget.focus()`** -- setzt den Tastaturfokus sofort auf dieses Widget. Man ruft es einmal auf nachdem alle Widgets erstellt wurden, kurz vor `root.mainloop()`:

```python
eingabe.focus()   # Fokus direkt ins Eingabefeld -- Nutzer kann sofort tippen
root.mainloop()
```

**Tabulator-Reihenfolge:** Der Nutzer kann mit der Tab-Taste von Widget zu Widget springen. Die Reihenfolge richtet sich nach der Reihenfolge in der die Widgets erstellt wurden. Man kann die Reihenfolge mit `widget.lift()` und `widget.lower()` beeinflussen -- aber meistens ist die natürliche Reihenfolge die richtige.

**`FocusIn` und `FocusOut`:** Mit `bind('<FocusIn>', ...)` kann man reagieren wenn ein Widget den Fokus bekommt -- zum Beispiel um Platzhaltentexte zu löschen wenn der Nutzer in ein Eingabefeld klickt.

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;">
<span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">def funktion(event=None) -- ein Callback für beides</span>
</div>

Oft soll dieselbe Funktion sowohl als `command=` beim Button als auch als `bind()`-Callback funktionieren. Das Problem: `command=` ruft den Callback ohne Argumente auf, `bind()` übergibt immer ein Event-Objekt.

Die Lösung: **`event=None` als Default-Parameter**.

```python
def umrechnen(event=None):
    # Logik hier
    pass

tk.Button(root, command=umrechnen)         # ruft umrechnen() auf -- event bekommt None
eingabe.bind('<Return>', umrechnen)        # ruft umrechnen(event) auf -- event ist das Event-Objekt
```

Mit `event=None` als Default-Wert kann die Funktion auf beide Arten aufgerufen werden:
- Als Button-Callback: `umrechnen()` -- `event` bekommt den Wert `None`
- Als bind-Callback: `umrechnen(event_objekt)` -- `event` bekommt das Event-Objekt

Da man das Event-Objekt in der Funktion meistens sowieso nicht braucht, stört es nicht dass es manchmal `None` ist.

<div style="background:#0b1929;border-left:4px solid #4fc3f7;border-radius:4px;padding:0.75em 1.2em;margin:2rem 0 1.2rem 0;font-family:'Segoe UI',sans-serif;">
<span style="font-size:clamp(1rem,2.2vw,1.25rem);font-weight:600;color:#e8eaf6;letter-spacing:-0.01em;">
<span style="font-size:0.65rem;letter-spacing:0.1em;text-transform:uppercase;color:#4fc3f7;background:rgba(79,195,247,0.12);border:1px solid rgba(79,195,247,0.3);border-radius:3px;padding:0.2em 0.55em;margin-right:0.75em;vertical-align:middle;white-space:nowrap;">Kursinhalt</span>
9. Abschlussbeispiel -- der Celsius-Umrechner komplett
</span>
</div>

Alle Konzepte aus diesem Notebook in einem vollständigen Beispiel: Callback, `entry.get()`, Fehlerbehandlung, Label dynamisch aktualisieren, Enter-Taste via `bind()`, `event=None`-Trick, Fokusmanagement, `state=` für den Button und zwei Richtungen der Umrechnung über dieselbe Funktion mit Lambda.

````python
# grad_umrechner_fertig.py -- Abschlussbeispiel
import tkinter as tk

# event=None: diese Funktion kann sowohl als command= als auch als bind()-Callback benutzt werden.
# Als command= ruft tkinter sie ohne Argument auf -> event bekommt None.
# Als bind()-Callback uebergibt tkinter das Event-Objekt -> event ist ein Event.
# Da wir event nicht brauchen, ist das egal.
def umrechnen(richtung, event=None):
    # richtung='ct' = Celsius -> Fahrenheit
    # richtung='fc' = Fahrenheit -> Celsius
    try:
        if richtung == 'cf':
            wert = float(von_eingabe.get())
            ergebnis = wert * 9 / 5 + 32
            ergebnis_label.config(
                text=f'{wert:.1f} °C  =  {ergebnis:.1f} °F',
                fg='black'
            )
        else:
            wert = float(von_eingabe.get())
            ergebnis = (wert - 32) * 5 / 9
            ergebnis_label.config(
                text=f'{wert:.1f} °F  =  {ergebnis:.1f} °C',
                fg='black'
            )
    except ValueError:
        # Fehler im Fenster anzeigen -- nicht auf der Konsole
        ergebnis_label.config(text='Bitte eine Zahl eingeben!', fg='red')

def leeren():
    von_eingabe.delete(0, tk.END)         # Entry-Feld leeren
    ergebnis_label.config(text='', fg='black')
    von_eingabe.focus()                   # Fokus wieder ins Eingabefeld

def richtung_geaendert(*args):
    # Wird aufgerufen wenn der Nutzer zwischen C->F und F->C wechselt
    leeren()
    if richtung_var.get() == 'cf':
        eingabe_label.config(text='Celsius:')
    else:
        eingabe_label.config(text='Fahrenheit:')

root = tk.Tk()
root.title('Grad-Umrechner')
root.geometry('340x220')
root.resizable(False, False)   # Fenstergrösse fixieren

# StringVar fuer die Richtung -- trace_add reagiert auf Aenderungen
richtung_var = tk.StringVar(value='cf')
richtung_var.trace_add('write', richtung_geaendert)

# --- Richtungsauswahl ---
radio_frame = tk.Frame(root)
radio_frame.pack(pady=(12, 4))
tk.Radiobutton(radio_frame, text='Celsius -> Fahrenheit',  variable=richtung_var, value='cf').pack(side='left', padx=8)
tk.Radiobutton(radio_frame, text='Fahrenheit -> Celsius',  variable=richtung_var, value='fc').pack(side='left', padx=8)

# --- Eingabe ---
eingabe_frame = tk.Frame(root)
eingabe_frame.pack(padx=20, pady=4, fill='x')

# Label wird von richtung_geaendert() aktualisiert
eingabe_label = tk.Label(eingabe_frame, text='Celsius:', width=12, anchor='w')
eingabe_label.grid(row=0, column=0, sticky='w')

von_eingabe = tk.Entry(eingabe_frame, width=14)
von_eingabe.grid(row=0, column=1, padx=8)

# Enter-Taste im Eingabefeld: lambda weil richtung_var.get() uebergeben werden muss
# lambda event=...: damit event=None-Trick und Argument gleichzeitig funktionieren
von_eingabe.bind('<Return>', lambda e: umrechnen(richtung_var.get()))
von_eingabe.bind('<Escape>', lambda e: leeren())

# --- Ergebnis ---
ergebnis_label = tk.Label(root, text='', font=('Arial', 12))
ergebnis_label.pack(pady=8)

# --- Buttons ---
button_frame = tk.Frame(root)
button_frame.pack(pady=4)

# lambda: richtung_var.get() liefert den aktuellen Wert ('cf' oder 'fc')
# ohne lambda koennte man richtung_var.get() nicht als Argument uebergeben
tk.Button(button_frame, text='Umrechnen', width=12,
          command=lambda: umrechnen(richtung_var.get())).pack(side='left', padx=6)
tk.Button(button_frame, text='Leeren',    width=12,
          command=leeren).pack(side='left', padx=6)

# Fokus direkt ins Eingabefeld -- Nutzer kann sofort tippen
von_eingabe.focus()

root.mainloop()
````

**Zusammenfassung**

| Konzept | Erklärung |
|---------|------------|
| Callback | Funktion die später aufgerufen wird -- Referenz übergeben, nicht aufrufen |
| `command=funktion` | Richtig -- Referenz auf die Funktion |
| `command=funktion()` | Falsch -- sofortiger Aufruf, gibt `None` zurück |
| `entry.get()` | Gibt immer einen **String** zurück -- auch bei Zahlen |
| Leeres Feld | `get()` gibt `''` zurück, nicht `None` |
| `tk.END` | tkinter-Konstante für "bis zum Ende", Wert ist der String `'end'` |
| `entry.delete(0, tk.END)` | Gesamten Inhalt löschen |
| `entry.insert(0, text)` | Text am Anfang einsetzen |
| `entry.insert(tk.END, text)` | Text ans Ende anhängen |
| try/except in Callbacks | Pflicht -- tkinter schluckt Fehler manchmal stillschweigend |
| `widget.config(param=wert)` | Parameter nachträglich ändern |
| `widget.cget('param')` | Aktuellen Wert eines Parameters lesen |
| `widget['param'] = wert` | Alternative Schreibweise zu `.config()` |
| `tk.StringVar()` | tkinter-Variable für Text |
| `tk.IntVar()` | tkinter-Variable für Ganzzahlen |
| `tk.BooleanVar()` | tkinter-Variable für True/False |
| `textvariable=var` | Widget mit Variable verbinden (Entry, Label) |
| `variable=var` | Widget mit Variable verbinden (Checkbutton, Radiobutton) |
| `var.get()` / `var.set()` | Wert lesen und setzen |
| `var.trace_add('write', cb)` | Callback bei jeder Änderung der Variable aufrufen |
| Trace-Callback-Signatur | `def cb(name, index, mode):` -- drei Argumente, meist ignoriert |
| `Checkbutton` + `BooleanVar` | Checkbox -- `var.get()` gibt `True`/`False` |
| `Radiobutton` + `StringVar` + `value=` | Gruppe mit Einzelauswahl -- alle teilen dieselbe Variable |
| `command=lambda: f(arg)` | Argumente an Callbacks übergeben |
| `lambda f=val: ...` | Wert in Schleife einfrieren -- Schleifenfalle vermeiden |
| `state='normal'` | Widget aktiv |
| `state='disabled'` | Widget deaktiviert -- ausgegraut, keine Eingaben |
| `state='readonly'` | Nur Entry: anzeigen aber nicht bearbeiten |
| `widget.bind('<Return>', cb)` | Auf Enter-Taste reagieren |
| `widget.bind('<Escape>', cb)` | Auf Escape-Taste reagieren |
| `widget.bind('<Button-1>', cb)` | Auf Mausklick reagieren |
| `widget.bind('<Enter>', cb)` | Maus fährt über Widget -- nicht die Enter-Taste! |
| Event-Objekt | bind()-Callbacks bekommen immer ein `event`-Argument |
| `event.keysym` | Name der gedrückten Taste |
| `event.char` | Das gedrückte Zeichen |
| `event.x`, `event.y` | Mausposition relativ zum Widget |
| `def f(event=None)` | Funktion die sowohl als `command=` als auch als `bind()`-Callback funktioniert |
| `widget.focus()` | Tastaturfokus auf dieses Widget setzen |
