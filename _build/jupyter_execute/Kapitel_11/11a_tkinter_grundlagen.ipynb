1. Was ist tkinter?
</span>
</div>

Bisher liefen alle Programme in der Konsole. Das reicht fuer Entwickler -- aber nicht fuer normale Nutzer. Eine **grafische Benutzeroberflaeche** (GUI) macht ein Programm zugaenglich: Buttons, Eingabefelder, Listen, Menus.

**tkinter** ist Pythons eingebautes GUI-Framework. Es ist in der Standardbibliothek, braucht keine Installation und laeuft auf Windows, macOS und Linux. Es ist nicht das modernste Framework -- aber es ist das einfachste fuer den Einstieg und reicht fuer alle Projekte in diesem Kurs.

Drei Konzepte tragen die gesamte tkinter-Logik:

1. Es gibt ein **Hauptfenster** -- das Wurzelelement jeder Anwendung
2. In das Fenster kommen **Widgets** -- sichtbare Elemente wie Buttons, Labels, Eingabefelder
3. Die Anwendung laeuft in einer **Ereignisschleife** -- sie wartet auf Klicks und Tastatureingaben und reagiert darauf

2. Der Import -- warum immer `import tkinter as tk`?
</span>
</div>

```python
import tkinter as tk
```

Das `as tk` ist technisch nicht noetig -- man koennte auch `import tkinter` schreiben und dann `tkinter.Label(...)` benutzen. Aber `tk` als Abkuerzung ist der universelle Standard in der gesamten tkinter-Welt. Jede Dokumentation, jedes Tutorial, jede Stack-Overflow-Antwort benutzt `tk`. Deshalb halten wir uns daran.

Manchmal sieht man auch:
```python
from tkinter import *
```
Das importiert alles direkt in den Namespace -- dann schreibt man `Label(...)` statt `tk.Label(...)`. Wir vermeiden das, weil es unuebersichtlich wird und Namen aus dem Modul mit eigenen Variablennamen kollidieren koennen.

3. Das Grundgeruest -- drei Zeilen die immer da sind
</span>
</div>

````python
# fenster_01.py
import tkinter as tk

root = tk.Tk()                    # Hauptfenster erstellen
root.title('Mein erstes Fenster') # Text in der Titelleiste
root.geometry('400x300')          # Startgroesse: 400 Pixel breit, 300 hoch

# Hier kommen spaeter Widgets rein

root.mainloop()                   # Ereignisschleife -- blockiert bis Fenster zu
````

**Zeile fuer Zeile:**

`root = tk.Tk()` -- erstellt das Hauptfenster als Objekt. Der Name `root` ist Konvention -- koennte auch `fenster` oder `app` heissen. `Tk()` ist die Klasse, `root` ist die Instanz.

`root.title('...')` -- setzt den Text der in der Titelleiste des Fensters erscheint.

`root.geometry('400x300')` -- setzt die Startgroesse. Das Format ist immer `'BreitexHoehe'` -- ein kleines x als Trennzeichen, kein Leerzeichen. Ohne diesen Aufruf waehlt tkinter eine Standardgroesse.

`root.mainloop()` -- das ist der entscheidende Aufruf. Er uebergibt die Kontrolle vollstaendig an tkinter. Ab hier laeuft intern eine Schleife die staendig prueft: Hat der Nutzer geklickt? Eine Taste gedrueckt? Das Fenster verschoben? Diese Schleife laeuft bis das Fenster geschlossen wird -- erst dann kehrt `mainloop()` zurueck und das Programm endet. Ohne diesen Aufruf wuerde das Fenster kurz aufflackern und sofort verschwinden.

4. Widget-Parameter -- was ist bg, fg, text, width?
</span>
</div>

Wenn man ein Widget erstellt, uebergibt man Konfigurationsoptionen als Keyword-Argumente:

```python
label = tk.Label(root, text='Hallo', bg='lightblue', fg='darkblue')
```

Diese Parameter sind **keine** normalen `__init__`-Parameter und **keine** Instanzvariablen die man mit `label.bg` abrufen koennte. tkinter verwaltet sie intern in einer eigenen Konfigurationstabelle. Deshalb kann man sie auch nachtraeglich aendern -- mit `.config()`:

```python
label.config(text='Welt', bg='lightyellow')   # Parameter aendern
print(label.cget('text'))                      # Parameter auslesen -> 'Welt'
```

**Die haeufigsten Parameter und ihre Abkuerzungen:**

| Kurzform | Langform | Bedeutung |
|----------|----------|-----------|
| `bg` | `background` | Hintergrundfarbe des Widgets |
| `fg` | `foreground` | Textfarbe ("Vordergrund") |
| `text` | -- | Der angezeigte Text |
| `width` | -- | Breite -- bei Entry und Button in **Zeichen**, bei Frame in Pixeln |
| `height` | -- | Hoehe -- bei Frame in Pixeln |
| `font` | -- | Schriftart, z.B. `('Arial', 12, 'bold')` |
| `padx` | -- | Innenabstand links/rechts innerhalb des Widgets |
| `pady` | -- | Innenabstand oben/unten innerhalb des Widgets |

Beide Schreibweisen funktionieren -- `bg='lightblue'` und `background='lightblue'` sind identisch. Die Kurzformen sind in der Praxis Standard.

**Wie weiss man welche Parameter ein Widget hat?** `help(tk.Label)`  Mit der Zeit kennt man die gaengigsten auswendig.

Jedes Widget wird in zwei Schritten verwendet -- erstellen und platzieren. Ohne den zweiten Schritt ist das Widget unsichtbar.

Man kann ein Widget in einer Variable speichern -- oder den Aufruf direkt verketten. Beides ist gueltig:


````python
# Variante A: Widget in Variable speichern
# Benutzen wenn man das Widget spaeter nochmal ansprechen will
label = tk.Label(root, text='Hallo')
label.pack()
# ... spaeter ...
label.config(text='Welt')   # moeglich weil Variable existiert
````

````python
# Variante B: direkt verketten
# Benutzen wenn man das Widget nach dem Platzieren nie mehr anfasst
tk.Label(root, text='Hallo').pack()
````



In Beispiel-Code sieht man oft Variante B fuer einfache Labels die sich nie aendern.

````python
# fenster_02.py
import tkinter as tk

root = tk.Tk()
root.title('Widgets')
root.geometry('320x200')

# Label: text= was angezeigt wird, bg= Hintergrund, fg= Textfarbe
label = tk.Label(root, text='Ich bin ein Label', bg='lightyellow', fg='darkblue')
label.pack(pady=5)

# Entry: einzeiliges Eingabefeld
# width= Breite in ZEICHEN (nicht Pixel!)
eingabe = tk.Entry(root, width=30)
eingabe.pack(pady=5)

# Entry mit show='*': jedes Zeichen wird als * angezeigt -- fuer Passwoerter
passwort = tk.Entry(root, width=30, show='*')
passwort.pack(pady=5)

# Button: text= Beschriftung, width= Breite in Zeichen
# command= wird in 11b erklaert
button = tk.Button(root, text='Klick mich', width=15)
button.pack(pady=5)

# Parameter nachtraeglich aendern mit .config()
label.config(text='Geaenderter Text', bg='lightcoral')

root.mainloop()
````

padx und pady tauchen in tkinter an zwei verschiedenen Stellen auf -- mit unterschiedlicher Bedeutung:

**Als Parameter beim Erstellen eines Widgets** -- steuert den Innenabstand zwischen dem Inhalt und dem Rand des Widgets selbst:

```python
tk.Button(root, text='OK', padx=20, pady=10)
# Der Button wird innen groesser -- mehr Luft um den Text
```

**Als Parameter in `.pack()`** -- steuert den Aussenabstand zwischen dem Widget und seinen Nachbarn:

```python
label.pack(pady=5)
# 5 Pixel Abstand oben und unten zum naechsten Widget
```

Der einfachste Merker: Widget-Parameter = Innenabstand. pack-Parameter = Aussenabstand.

````python
# pad_vergleich.py
import tkinter as tk

root = tk.Tk()
root.title('padx pady Vergleich')
root.geometry('250x250')

# Kein pad -- Widgets kleben aneinander
tk.Label(root, text='Kein pad', bg='lightblue').pack()
tk.Label(root, text='Kein pad', bg='lightgreen').pack()

tk.Label(root, text='----------').pack()

# pady in pack() -- Aussenabstand zum Nachbar
tk.Label(root, text='pady=10 in pack', bg='lightblue').pack(pady=10)
tk.Label(root, text='pady=10 in pack', bg='lightgreen').pack(pady=10)

tk.Label(root, text='----------').pack()

# padx/pady als Widget-Parameter -- Innenabstand
tk.Label(root, text='padx=20 pady=10 am Widget', bg='lightblue', padx=20, pady=10).pack()
tk.Label(root, text='padx=20 pady=10 am Widget', bg='lightgreen', padx=20, pady=10).pack()

root.mainloop()
````

7. pack -- Layout Schritt fuer Schritt
</span>
</div>

`pack()` ist der einfachste Layout-Manager. Er stapelt Widgets der Reihe nach in einen Container. Standardmaessig von oben nach unten.

Wir gehen jeden Parameter einzeln durch -- mit einem eigenen Beispiel.

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;">
<span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">side -- an welche Wand wird das Widget gedrueckt?</span>
</div>

`side` bestimmt an welche Seite des verfuegbaren Platzes das Widget gedrueckt wird:

- `side='top'` -- an die Oberkante (Standardwert wenn man nichts angibt)
- `side='bottom'` -- an die Unterkante
- `side='left'` -- an die linke Seite
- `side='right'` -- an die rechte Seite

**Wichtig zu verstehen:** Jedes Widget belegt nach dem Platzieren einen Bereich. Das naechste Widget bekommt den Rest. Deshalb ist die Reihenfolge der `.pack()`-Aufrufe entscheidend -- und `side='bottom'` muss vor `side='top'` (fuer den Inhaltsbereich) aufgerufen werden wenn man Toolbar + Inhalt + Statusleiste bauen will.

````python
# fenster_03_side.py -- side ausprobieren
import tkinter as tk

root = tk.Tk()
root.title('side')
root.geometry('300x200')

# Reihenfolge bestimmt das Ergebnis!
# top belegt einen Streifen oben vom verbleibenden Platz
tk.Label(root, text='top', bg='lightblue').pack(side='top')
# bottom belegt einen Streifen unten vom verbleibenden Platz
tk.Label(root, text='bottom', bg='lightgreen').pack(side='bottom')
# left belegt einen Streifen links vom verbleibenden Platz
tk.Label(root, text='left', bg='lightyellow').pack(side='left')
# right belegt einen Streifen rechts vom verbleibenden Platz
tk.Label(root, text='right', bg='lightcoral').pack(side='right')
# Was uebrig bleibt: die Mitte -- kein Widget dort in diesem Beispiel

root.mainloop()
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;">
<span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">fill -- den zugewiesenen Bereich ausfuellen</span>
</div>

Ohne `fill` ist ein Widget genau so gross wie sein Inhalt -- also so breit wie der Text. Mit `fill` dehnt es sich innerhalb des ihm zugewiesenen Bereichs aus:

- `fill='x'` -- horizontal ausdehnen (die volle Breite des Bereichs belegen)
- `fill='y'` -- vertikal ausdehnen (die volle Hoehe des Bereichs belegen)
- `fill='both'` -- in beide Richtungen ausdehnen

**Was ist der "zugewiesene Bereich"?** Ein Widget mit `side='top'` bekommt einen horizontalen Streifen zugewiesen -- so breit wie der Container, aber nur so hoch wie der Inhalt. `fill='x'` dehnt das Widget so aus dass es diesen Streifen in der Breite ausfuellt. `fill='y'` haette keinen Effekt weil der Streifen bereits die Hoehe des Inhalts hat -- es gibt keinen zusaetzlichen vertikalen Spielraum.

````python
# fenster_04_fill.py -- fill vergleichen
import tkinter as tk

root = tk.Tk()
root.title('fill')
root.geometry('300x200')

# Ohne fill: Label ist nur so breit wie der Text
tk.Label(root, text='kein fill', bg='lightblue').pack(side='top', pady=5)

# fill='x': Label dehnt sich auf volle Fensterbreite aus
tk.Label(root, text='fill=x', bg='lightgreen').pack(side='top', fill='x', pady=5)

# fill='x' hat Wirkung -- fill='y' haette hier keine, weil der
# Bereich (horizontaler Streifen) keinen extra vertikalen Platz hat
tk.Label(root, text='fill=x (gleich)', bg='lightyellow').pack(side='top', fill='x', pady=5)

root.mainloop()
````

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;">
<span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">expand -- extra Platz beim Vergroessern des Fensters beanspruchen</span>
</div>

Der Unterschied zwischen `fill` und `expand` ist subtil aber wichtig:

- `fill` sagt: **nutze** den Platz der dir bereits zugewiesen wurde
- `expand=True` sagt: **fordere** zusaetzlichen Platz an wenn das Fenster groesser wird

Ohne `expand=True` behalten alle Widgets ihre Groesse wenn das Fenster vergroessert wird. Der leere Platz sammelt sich unten. Mit `expand=True` bekommt das Widget diesen extra Platz -- und mit `fill='both'` benutzt es ihn auch.

Fuer einen Inhaltsbereich der mit dem Fenster mitwachsen soll: **immer beides zusammen** -- `fill='both', expand=True`.

````python
# fenster_05_expand.py -- Fenster mit der Maus vergroessern!
import tkinter as tk

root = tk.Tk()
root.title('expand -- Fenster jetzt vergroessern!')
root.geometry('300x200')

# Feste Toolbar: fill='x' aber kein expand
# -> beim Vergroessern nur breiter, nicht hoeher
tk.Label(root, text='Toolbar', bg='#d0d0d0').pack(side='top', fill='x')

# Inhaltsbereich: fill='both' UND expand=True
# -> bekommt allen extra Platz, dehnt sich in beide Richtungen aus
tk.Label(root, text='Inhalt waechst mit', bg='white').pack(side='top', fill='both', expand=True)

# Feste Statusleiste: fill='x' aber kein expand
tk.Label(root, text='Statusleiste', bg='#d0d0d0').pack(side='bottom', fill='x')

root.mainloop()
````

<span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;">padx und pady -- Aussenabstand</span>
</div>

`padx` und `pady` in `.pack()` fuer den **Aussenabstand** um das Widget -- also der Abstand zwischen dem Widget und seinen Nachbarn:

- `padx=10` -- 10 Pixel Abstand links und rechts des Widgets
- `pady=5` -- 5 Pixel Abstand oben und unten des Widgets
- `padx=(10, 20)` -- asymmetrisch: 10 links, 20 rechts

**Nicht zu verwechseln** mit `padx`/`pady` als Widget-Parameter -- die steuern den **Innenabstand** (zwischen Inhalt und Rahmen des Widgets selbst):

```python
# Innenabstand: Button wird innen groesser
tk.Button(root, text='OK', padx=20, pady=10)

# Aussenabstand: Abstand zum Nachbar-Widget
button.pack(padx=10, pady=5)
```

8. grid -- praezise im Raster
</span>
</div>

`grid()` platziert Widgets in einem unsichtbaren Tabellenraster. Man gibt Zeile (`row`) und Spalte (`column`) an -- tkinter berechnet die tatsaechliche Position selbst. Zeilen und Spalten entstehen automatisch, man muss sie nicht deklarieren.

**sticky** -- Ausrichtung innerhalb der Zelle. Funktioniert wie Kompassrichtungen:
- `sticky='w'` -- West = linksbuendig in der Zelle
- `sticky='e'` -- East = rechtsbuendig
- `sticky='ew'` -- Ost und West = Widget streckt sich horizontal
- `sticky='nsew'` -- alle vier Richtungen = Zelle komplett ausfuellen

**columnspan** -- Widget ueber mehrere Spalten erstrecken (wie "Zellen zusammenfuehren" in Excel).

**columnconfigure(n, weight=1)** -- legt fest welche Spalte extra Platz bekommt wenn das Fenster breiter wird. Ohne das bleiben alle Spalten fest -- die Eingabefelder wachsen nicht mit.

````python
# fenster_06_grid.py
import tkinter as tk

root = tk.Tk()
root.title('grid')
root.geometry('320x180')

# Spalte 1 (die Eingabefelder) soll sich ausdehnen wenn Fenster breiter wird
# weight=1 bedeutet: diese Spalte bekommt den extra Platz
root.columnconfigure(1, weight=1)

# row=0: erste Zeile. column=0: erste Spalte
# sticky='w': Widget in der Zelle linksbuendig ausrichten
tk.Label(root, text='Name:').grid(     row=0, column=0, sticky='w', padx=10, pady=8)
tk.Label(root, text='E-Mail:').grid(   row=1, column=0, sticky='w', padx=10, pady=8)
tk.Label(root, text='Passwort:').grid( row=2, column=0, sticky='w', padx=10, pady=8)

# sticky='ew': Eingabefeld fuellt die gesamte Zellenbreite aus
tk.Entry(root).grid(           row=0, column=1, sticky='ew', padx=10, pady=8)
tk.Entry(root).grid(           row=1, column=1, sticky='ew', padx=10, pady=8)
tk.Entry(root, show='*').grid( row=2, column=1, sticky='ew', padx=10, pady=8)

# columnspan=2: Button ueber beide Spalten
tk.Button(root, text='Anmelden').grid(row=3, column=0, columnspan=2, pady=10)

root.mainloop()
````

9. Frame -- Bereiche trennen und pack + grid kombinieren

**Regel: Innerhalb desselben Containers darf man pack und grid nicht mischen.** tkinter wirft sonst einen Fehler -- die beiden Layout-Manager "streiten" sich um die Kontrolle.

Die Loesung ist `Frame`. Ein Frame ist ein unsichtbarer Container der Widgets gruppiert. Jeder Frame hat seinen eigenen Layout-Manager -- unabhaengig vom Rest der Anwendung. Damit kann man gleichzeitig pack (im Hauptfenster fuer die Bereiche) und grid (im Inhaltsbereich fuer ein Formular) benutzen, weil sie in verschiedenen Containern sind.

**pack_propagate(False)** -- ein Frame passt sich standardmaessig an die Groesse seines Inhalts an. Das bedeutet: wenn man `height=40` setzt aber der Inhalt kleiner ist, schrumpft der Frame auf den Inhalt. `pack_propagate(False)` schaltet dieses Verhalten ab -- der Frame behaelt seine angegebene Groesse.

````python
# fenster_07_frames.py
import tkinter as tk

root = tk.Tk()
root.title('Frames')
root.geometry('400x300')

# -------------------------------------------------------
# Hauptfenster: pack() fuer die drei Bereiche
# -------------------------------------------------------

# Toolbar oben
toolbar = tk.Frame(root, bg='#e0e0e0', height=40)
toolbar.pack(side='top', fill='x')
# pack_propagate(False): Frame soll height=40 behalten
# auch wenn die Buttons kleiner sind
toolbar.pack_propagate(False)

# Statusleiste unten -- MUSS vor dem Inhaltsbereich kommen
statusleiste = tk.Frame(root, bg='#e0e0e0', height=24)
statusleiste.pack(side='bottom', fill='x')
statusleiste.pack_propagate(False)

# Inhaltsbereich Mitte -- bekommt den Rest
inhalt = tk.Frame(root, bg='white')
inhalt.pack(side='top', fill='both', expand=True)

# -------------------------------------------------------
# Innerhalb der Toolbar: pack() mit side='left'
# -------------------------------------------------------
tk.Button(toolbar, text='Neu').pack(     side='left', padx=5, pady=5)
tk.Button(toolbar, text='Oeffnen').pack( side='left', padx=5, pady=5)
tk.Button(toolbar, text='Speichern').pack(side='left', padx=5, pady=5)

# -------------------------------------------------------
# Innerhalb der Statusleiste: pack()
# -------------------------------------------------------
tk.Label(statusleiste, text='Bereit', bg='#e0e0e0').pack(side='left', padx=8)

# -------------------------------------------------------
# Innerhalb des Inhaltsbereichs: grid() fuer ein Formular
# Das ist kein Mischen! grid laeuft in inhalt, pack in root/toolbar/status
# -------------------------------------------------------
inhalt.columnconfigure(1, weight=1)
tk.Label(inhalt, text='Titel:',    bg='white').grid(row=0, column=0, sticky='w', padx=15, pady=12)
tk.Entry(inhalt).grid(                          row=0, column=1, sticky='ew',   padx=15, pady=12)
tk.Label(inhalt, text='Interpret:',bg='white').grid(row=1, column=0, sticky='w', padx=15, pady=12)
tk.Entry(inhalt).grid(                          row=1, column=1, sticky='ew',   padx=15, pady=12)

root.mainloop()
````

**Zusammenfassung**

| Konzept | Erklaerung |
|---------|------------|
| `import tkinter as tk` | Standard-Importname -- `tk` ist universelle Konvention |
| `tk.Tk()` | Hauptfenster erstellen |
| `root.geometry('BxH')` | Startgroesse: kleines x, kein Leerzeichen |
| `root.mainloop()` | Ereignisschleife -- blockiert bis Fenster geschlossen |
| Widget-Parameter | Keyword-Argumente des Konstruktors -- intern als Konfiguration gespeichert |
| `.config(param=wert)` | Parameter nachtraeglich aendern |
| `.cget('param')` | Parameter auslesen |
| `bg` / `background` | Hintergrundfarbe -- beide Schreibweisen funktionieren |
| `fg` / `foreground` | Textfarbe |
| `width` bei Entry/Button | Breite in **Zeichen** (nicht Pixeln!) |
| Hex-Farben | `'#RRGGBB'` -- praeziser als benannte Farben |
| Variable vs. verkettung | Variable wenn Widget spaeter benutzt wird; `.pack()` direkt wenn nicht |
| `pack(side=)` | Wand an die gedrueckt wird: `'top'`, `'bottom'`, `'left'`, `'right'` |
| `pack(fill=)` | Bereich ausfuellen: `'x'`, `'y'`, `'both'` |
| `pack(expand=True)` | Extra Platz beanspruchen wenn Fenster waechst |
| `padx`/`pady` bei pack | Aussenabstand um das Widget |
| `padx`/`pady` als Widget-Parameter | Innenabstand zwischen Inhalt und Widget-Rahmen |
| `grid(row=, column=)` | Rasterpositionierung |
| `grid(sticky=)` | Ausrichtung in der Zelle: `'w'`=links, `'e'`=rechts, `'nsew'`=alles |
| `grid(columnspan=)` | Widget ueber mehrere Spalten |
| `columnconfigure(n, weight=1)` | Spalte darf beim Vergroessern wachsen |
| pack + grid nicht mischen | Nur innerhalb desselben Containers -- Frames loesen das |
| `Frame` | Unsichtbarer Container mit eigenem Layout-Manager |
| `pack_propagate(False)` | Frame behaelt seine angegebene Groesse statt auf Inhalt zu schrumpfen |
| `root.resizable(w, h)` | Fenstergroesse fixieren |
| `root.minsize() / maxsize()` | Groessenbereich einschraenken |
