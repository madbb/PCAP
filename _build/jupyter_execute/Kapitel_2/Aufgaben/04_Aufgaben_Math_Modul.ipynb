**Schritt 1:** Erstelle eine neue Datei `mathe_tools.py` im selben Verzeichnis wie dieses Notebook.
Inhalt der Datei:
```python
import math

MODUL_INFO = "Mathe Tools v1.0"

def kreis_flaeche(radius):
    '''Berechnet die Flaeche eines Kreises.'''
    return math.pi * radius ** 2

def ist_perfekte_quadratzahl(n):
    '''Gibt True zurueck, wenn n eine perfekte Quadratzahl ist.'''
    if n < 0:
        return False
    wurzel = math.sqrt(n)
    return wurzel == math.floor(wurzel)
```
