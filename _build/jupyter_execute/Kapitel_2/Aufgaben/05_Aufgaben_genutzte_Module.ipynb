```python
# Lösung  Aufgabe 15:

# 1. Importe
import platform
from random import seed, randint, uniform
from math import sqrt, pow, log

# 2. Systeminformationen als Seed-Basis
node_name_laenge = len(platform.node())
python_major_version = int(platform.python_version_tuple()[0])

seed_basis = node_name_laenge + python_major_version
if seed_basis == 0:
    seed_basis = 1 # Sicherstellen, dass Seed nicht 0 ist, falls beide Werte 0 wären

# 3. Zufallsgenerator initialisieren
seed(seed_basis)

# 4. Zufällige Operationen
# Stelle sicher, dass randint einen gültigen Bereich hat, falls seed_basis klein ist.
# Für randint(1, N) muss N >= 1 sein. Da seed_basis >= 1 ist, ist das hier sicher.
a = randint(1, seed_basis)
b = uniform(0.1, 5.0)

ergebnis1 = sqrt(a) + pow(b, 2)

if a > 0: # log(0) ist nicht definiert, log negativer Zahlen auch nicht für reelle Zahlen
    ergebnis2 = log(a) * b
else: # Sollte durch randint(1, seed_basis) nicht eintreten, aber defensiv programmiert
    ergebnis2 = 0.0

# 5. Ausgabe
print("Verwendete Seed-Basis:", seed_basis)
print(f"Generierte Zufallszahl a (Ganzzahl): {a}")
print(f"Generierte Zufallszahl b (Fließkomma): {b:.4f}")
print(f"Ergebnis 1 (sqrt(a) + pow(b, 2)): {ergebnis1:.4f}")
print(f"Ergebnis 2 (log(a) * b oder 0): {ergebnis2:.4f}")
```
