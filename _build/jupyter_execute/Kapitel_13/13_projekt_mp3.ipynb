Du baust ein Programm das eine Sammlung von MP3-Dateien verwaltet. Jeder der viele Musikdateien hat kennt das Problem: Man weiss nicht mehr was man hat, Duplikate schleichen sich ein, Dateinamen sind chaotisch, und die eingebetteten Metadaten -- Interpret, Album, Jahr -- stimmen nicht immer mit dem Dateinamen ueberein.

Dein Programm soll diesem Chaos Herr werden. Es durchsucht ein Verzeichnis nach MP3-Dateien, liest ihre Metadaten, zeigt sie uebersichtlich an und erlaubt dem Nutzer seine Sammlung zu durchsuchen, zu filtern und zu verwalten. Am Ende soll der Nutzer seine Bibliothek im Griff haben.

Wie das Programm aussieht -- ob es eine Kommandozeilenanwendung ist, ein einfaches Textmenue, oder eine grafische Oberflaeche -- das entscheidest du. Auch wie weit du die Features ausbaust liegt bei dir. Das Mindestziel ist ein Programm das nutzbar ist und echte Arbeit abnimmt.

Dieses Projekt bringt viele Themen aus dem Kurs zusammen -- nicht als Uebung, sondern als echtes Problem das geloest werden muss:

**Dateisystem und I/O** -- du musst Verzeichnisse durchsuchen, Dateien finden, Pfade zusammenbauen und Daten persistent speichern. Was du in Kapitel 7 gelernt hast, wird hier zum Werkzeug.

**OOP** -- eine Musikbibliothek besteht aus Objekten: Songs, Alben, Interpreten. Wie du diese modellierst, welche Klassen du baust und wie sie zusammenhaengen, ist eine der ersten Entwurfsentscheidungen die du treffen musst.

**Fehlerbehandlung** -- nicht jede Datei hat vollstaendige Metadaten. Nicht jedes Verzeichnis ist lesbar. Dein Programm muss mit unvollstaendigen oder fehlenden Daten umgehen, ohne abzusturzen.

**Strings und Suche** -- der Nutzer gibt einen Suchbegriff ein. Dein Programm muss ihn mit Titeln, Interpreten und Alben abgleichen. Gross- und Kleinschreibung, Teilstrings, Sonderzeichen -- alles Faelle die man bedenken muss.

**Persistenz** -- die eingelesene Bibliothek soll gespeichert werden damit man sie nicht bei jedem Start neu scannen muss. Wie macht man das? In welchem Format? Was passiert wenn sich die Dateien in der Zwischenzeit geaendert haben?

**Fehlende oder fehlerhafte Metadaten**
Nicht jede MP3-Datei hat vollstaendige Tags. Manche haben gar keine. Manche haben Tags in einem alten Format das mutagen anders liest. Teste dein Programm mit echten Dateien -- nicht nur mit perfekt getaggten Beispieldateien.

**Zeichenkodierung**
ID3-Tags koennen in verschiedenen Kodierungen gespeichert sein -- ASCII, Latin-1, UTF-8, UTF-16. mutagen abstrahiert das meistens -- aber bei aelteren Dateien kann es zu Darstellungsproblemen kommen. Behalte das im Hinterkopf wenn Umlaute oder Sonderzeichen falsch angezeigt werden.

**Pfade auf verschiedenen Betriebssystemen**
Windows benutzt Backslashes, Linux und macOS Slashes. Wenn du Pfade als Strings zusammenbaust, kann das zu Problemen fuehren. `pathlib` loest das -- es weiss welches Betriebssystem aktiv ist und baut den Pfad richtig zusammen.

**Performance bei grossen Sammlungen**
Wenn jemand 10.000 MP3-Dateien hat, kann das initiale Einlesen eine Weile dauern. Ueberlege ob du den Nutzer ueber den Fortschritt informierst -- eine einfache Ausgabe wie `Lade Datei 3 von 247...` macht einen grossen Unterschied in der Nutzererfahrung.

**Die gespeicherte Bibliothek veraltet**
Was passiert wenn der Nutzer MP3-Dateien loescht oder verschiebt, nachdem die Bibliothek gespeichert wurde? Beim naechsten Start zeigt dein Programm Eintraege fuer Dateien die nicht mehr existieren. Wie erkennst du das? Was tust du damit?

**Schritt 4 -- Persistenz:** Die Bibliothek als JSON speichern und wieder laden. Pruefen ob die geladenen Daten identisch mit den gescannten sind.

**Schritt 6 -- Ausbau:** Eigene Ideen umsetzen. Statistiken, GUI, Playlisten -- was immer dich interessiert.

Du brauchst MP3-Dateien zum Testen. Ein paar Moeglichkeiten:

- **Eigene Sammlung:** Wenn du MP3-Dateien hast, nutze sie -- am besten erst mit einem kleinen Testordner mit 5-10 Dateien, nicht gleich mit der gesamten Sammlung.

- **Freie Musik:** Auf `freemusicarchive.org` gibt es lizenzfreie MP3-Dateien mit vollstaendigen Metadaten.

- **Testdateien erstellen:** mutagen kann auch Metadaten schreiben. Du koenntest dir selbst ein paar leere MP3-Dateien mit verschiedenen Tags erstellen -- inklusive absichtlich fehlenden Tags um deine Fehlerbehandlung zu testen.

Teste bewusst mit Edge Cases: eine Datei ohne Titel, eine ohne Interpret, eine mit Sonderzeichen im Dateinamen, eine in einem tief verschachtelten Unterordner.
