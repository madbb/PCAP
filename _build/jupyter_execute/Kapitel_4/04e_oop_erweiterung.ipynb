1. Komposition -- hat-ein statt ist-ein
</span>
</div>

Vererbung drueckt eine **ist-ein**-Beziehung aus: Ein Krieger **ist ein** Held. Aber nicht alles laesst sich so beschreiben.

Ein Held **hat eine** Waffe. Ein Held **hat ein** Inventar. Das ist eine **hat-ein**-Beziehung -- und dafuer ist **Komposition** der richtige Ansatz: Das Objekt einer Klasse wird als Attribut in einer anderen Klasse gespeichert.

Warum ist das besser als Vererbung in solchen Faellen? Eine Waffe ist kein Held. Wenn `Held` von `Waffe` erben wuerde, bekaeeme er alle Waffen-Methoden -- das ergibt keinen Sinn.

```python
class Waffe:
    def __init__(self, name, schaden, typ):
        self.name   = name
        self.schaden = schaden
        self.typ    = typ

    def __str__(self):
        return f'{self.name} ({self.typ}, {self.schaden} Schaden)'


class Ruestung:
    def __init__(self, name, schutz):
        self.name   = name
        self.schutz = schutz

    def __str__(self):
        return f'{self.name} ({self.schutz} Schutz)'


class Held:
    def __init__(self, name, hp):
        self.name    = name
        self.hp      = hp
        self.waffe   = None   # Komposition: Waffe ist ein Attribut
        self.ruestung = None

    def ausruesten(self, gegenstand):
        if isinstance(gegenstand, Waffe):
            self.waffe = gegenstand
            print(f'{self.name} traegt jetzt {gegenstand.name}')
        elif isinstance(gegenstand, Ruestung):
            self.ruestung = gegenstand
            print(f'{self.name} traegt jetzt {gegenstand.name}')

    def angreifen(self):
        if self.waffe:
            return self.waffe.schaden
        return 5  # Faustkampf

    def __str__(self):
        w = str(self.waffe) if self.waffe else 'keine'
        r = str(self.ruestung) if self.ruestung else 'keine'
        return f'{self.name} | HP: {self.hp} | Waffe: {w} | Ruestung: {r}'


aldric = Held('Aldric', 100)
print(aldric)

schwert    = Waffe('Eisenschwert', 25, 'Nahkampf')
kettenhemd = Ruestung('Kettenhemd', 10)

aldric.ausruesten(schwert)
aldric.ausruesten(kettenhemd)
print(aldric)
print(f'Angriff: {aldric.angreifen()} Schaden')
```

Der Vorteil der Komposition: Waffe und Ruestung sind eigenstaendige Klassen. Sie koennen unabhaengig entwickelt, getauscht und wiederverwendet werden. Aldric kann jederzeit eine bessere Waffe anlegen -- ohne dass sich seine eigene Klasse aendert.

2. Dunder-Methoden -- eigene Operatoren definieren
</span>
</div>

`__str__` kennen wir bereits -- Python ruft es bei `print()` auf. Das Prinzip gilt fuer viele weitere Operationen: Python hat fuer fast jeden Operator eine entsprechende Dunder-Methode. Wenn wir `+` auf zwei Objekte anwenden, ruft Python intern `__add__` auf.

Das erlaubt uns, eigene Klassen so zu gestalten, dass sie sich wie eingebaute Typen verhalten.

```python
class Inventar:
    """Das Inventar eines Helden -- eine Liste von Gegenstaenden."""

    def __init__(self, gegenstaende=None):
        self._items = list(gegenstaende) if gegenstaende else []

    def __len__(self):
        """Wird aufgerufen bei len(inventar)"""
        return len(self._items)

    def __contains__(self, gegenstand):
        """Wird aufgerufen bei 'Schwert' in inventar"""
        return gegenstand in self._items

    def __add__(self, other):
        """Wird aufgerufen bei inventar1 + inventar2 -- kombiniert Inventare"""
        return Inventar(self._items + other._items)

    def __iter__(self):
        """Wird aufgerufen bei 'for x in inventar'"""
        return iter(self._items)

    def __str__(self):
        if not self._items:
            return 'Inventar: (leer)'
        return 'Inventar: ' + ', '.join(self._items)


rucksack = Inventar(['Schwert', 'Schild', 'Heiltrank'])
truhe    = Inventar(['Goldmuenzen', 'Schluessel'])

print(len(rucksack))              # __len__: 3
print('Schwert' in rucksack)      # __contains__: True
print('Zauberstab' in rucksack)   # __contains__: False

alles = rucksack + truhe          # __add__
print(alles)                      # __str__

print('Alle Gegenstaende:')
for item in rucksack:             # __iter__
    print(f'  - {item}')
```

**Uebersicht wichtiger Dunder-Methoden:**

| Methode | Wird aufgerufen bei |
|---------|--------------------|
| `__init__` | `Klasse()` -- Objekterstellung |
| `__str__` | `print(obj)`, `str(obj)` |
| `__repr__` | `repr(obj)`, in der REPL |
| `__len__` | `len(obj)` |
| `__contains__` | `x in obj` |
| `__add__` | `obj1 + obj2` |
| `__eq__` | `obj1 == obj2` |
| `__lt__` | `obj1 < obj2` |
| `__iter__` | `for x in obj` |

3. Design Patterns -- bewaehrte Loesungsschablonen
</span>
</div>

Design Patterns sind bewaehrte Loesungen fuer wiederkehrende Probleme. Drei davon sind besonders verbreitet und tauchen im Berufsalltag regelmaessig auf.

```python
# SINGLETON -- nur eine Instanz existiert je
# Anwendungsfall: Die Spielwelt soll nur einmal existieren

class Spielwelt:
    _instanz = None

    def __new__(cls):
        if cls._instanz is None:
            cls._instanz = super().__new__(cls)
            cls._instanz.name = 'Aetherion'
            cls._instanz.stufe = 1
        return cls._instanz

welt1 = Spielwelt()
welt2 = Spielwelt()

print(welt1 is welt2)    # True -- immer dieselbe Instanz
print(welt1.name)        # Aetherion

welt1.stufe = 5
print(welt2.stufe)       # 5 -- dieselbe Instanz, dieselbe Aenderung
```

```python
# FACTORY -- Objekterstellung kapseln
# Anwendungsfall: Held basierend auf Klassennamen erstellen

class Held:
    def __init__(self, name, hp, staerke):
        self.name    = name
        self.hp      = hp
        self.staerke = staerke
    def __str__(self):
        return f'{self.name} ({self.__class__.__name__}) HP:{self.hp} STR:{self.staerke}'

class Krieger(Held):
    def __init__(self, name): super().__init__(name, 120, 18)

class Magier(Held):
    def __init__(self, name): super().__init__(name, 70, 8)

class Waldlaeufer(Held):
    def __init__(self, name): super().__init__(name, 90, 14)


def erstelle_held(klasse: str, name: str) -> Held:
    """Factory: erstellt den richtigen Helden basierend auf dem Klassennamen."""
    klassen = {
        'krieger':     Krieger,
        'magier':      Magier,
        'waldlaeufer': Waldlaeufer,
    }
    if klasse.lower() not in klassen:
        raise ValueError(f'Unbekannte Klasse: {klasse}')
    return klassen[klasse.lower()](name)


# Helden per Name erstellen -- die Logik steckt in der Factory
party = [
    erstelle_held('krieger',     'Aldric'),
    erstelle_held('magier',      'Lyria'),
    erstelle_held('waldlaeufer', 'Thorn'),
]

for held in party:
    print(held)
```

```python
# OBSERVER -- Ereignisse verteilen
# Anwendungsfall: Wenn ein Held stirbt, sollen mehrere Systeme reagieren

class Ereignissystem:
    def __init__(self):
        self._handler = {}   # Ereignisname -> Liste von Funktionen

    def anmelden(self, ereignis, funktion):
        self._handler.setdefault(ereignis, []).append(funktion)

    def ausloesen(self, ereignis, **daten):
        for fn in self._handler.get(ereignis, []):
            fn(**daten)


ereignisse = Ereignissystem()

# Verschiedene Systeme reagieren auf dasselbe Ereignis
ereignisse.anmelden('held_gestorben',
    lambda name, **_: print(f'LOG: {name} ist gestorben'))
ereignisse.anmelden('held_gestorben',
    lambda name, **_: print(f'UI: Zeige Game-Over fuer {name}'))
ereignisse.anmelden('held_gestorben',
    lambda name, **_: print(f'AUDIO: Spiele Sterbesequenz'))

# Ereignis ausloesen -- alle Handler werden benachrichtigt
ereignisse.ausloesen('held_gestorben', name='Aldric')
```

**Zusammenfassung**

| Konzept | Wann einsetzen |
|---------|---------------|
| **Vererbung** | Wenn eine `ist-ein`-Beziehung besteht: Krieger ist ein Held |
| **Komposition** | Wenn eine `hat-ein`-Beziehung besteht: Held hat eine Waffe |
| **Dunder-Methoden** | Wenn eigene Klassen wie eingebaute Typen funktionieren sollen |
| **Singleton** | Wenn nur eine Instanz existieren soll: Spielwelt, Konfiguration |
| **Factory** | Wenn die Objekterstellung komplex oder variabel ist |
| **Observer** | Wenn bei einem Ereignis mehrere Komponenten reagieren sollen |
