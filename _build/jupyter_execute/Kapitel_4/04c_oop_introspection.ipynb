1. Was ist Introspection und wozu braucht man sie?
</span>

**Introspection** bedeutet: Ein Programm kann sich zur Laufzeit selbst untersuchen. Es kann fragen: Welche Attribute hat dieses Objekt? Von welcher Klasse erbt es? Hat es diese bestimmte Methode?

Stell dir vor, du baust ein Kampfsystem. Verschiedene Helden haben verschiedene Faehigkeiten -- nicht jeder kann zaubern, nicht jeder kann heilen. Bevor du eine Faehigkeit aufrufst, willst du pruefen ob der Held sie ueberhaupt hat:

```python
class Krieger:
    def __init__(self, name):
        self.name   = name
        self.waffe  = 'Schwert'

    def angreifen(self):
        return f'{self.name} schlaegt mit {self.waffe} zu!'

class Magierin:
    def __init__(self, name):
        self.name   = name
        self.mana   = 80

    def zaubern(self):
        return f'{self.name} wirkt einen Zauber!'

    def heilen(self):
        return f'{self.name} heilt einen Verbündeten!'


aldric = Krieger('Aldric')
lyria  = Magierin('Lyria')

helden = [aldric, lyria]

# Wir wollen zaubern() aufrufen -- aber nicht alle koennen das
for held in helden:
    # Wie pruefen wir ob der Held zaubern() hat, ohne einen Fehler zu riskieren?
    pass
```

Ohne Introspection muessten wir `isinstance()` oder try/except verwenden. Mit `hasattr()` geht es direkt und sauber.

2. hasattr(), getattr(), setattr()
</span>
</div>

`hasattr(objekt, 'name')` prueft ob ein Attribut oder eine Methode existiert -- `True` oder `False`. Der Name wird als String uebergeben, weil wir zur Laufzeit pruefen -- nicht zur Compile-Zeit.

```python
# Das Kampfsystem mit hasattr
for held in helden:
    print(f'--- {held.name} ---')
    if hasattr(held, 'angreifen'):
        print(held.angreifen())
    if hasattr(held, 'zaubern'):
        print(held.zaubern())
    if hasattr(held, 'heilen'):
        print(held.heilen())
```

```python
# getattr() -- Attributwert dynamisch lesen
aldric = Krieger('Aldric')

# Normal: aldric.name
# Mit getattr: dynamisch, der Attributname kann eine Variable sein
attribut = 'waffe'
print(getattr(aldric, attribut))            # Schwert

# Fallback-Wert wenn Attribut nicht existiert
print(getattr(aldric, 'mana', 'kein Mana')) # kein Mana -- Fallback

# setattr() -- Attribut dynamisch setzen
setattr(aldric, 'schild', 'Turmschild')
print(aldric.schild)                        # Turmschild

# Praktisch: alle Attribute eines Helden auf einmal setzen
ausruestung = {'waffe': 'Axt', 'ruestung': 'Kettenhemd', 'helm': 'Eisenhelm'}
for attr, wert in ausruestung.items():
    setattr(aldric, attr, wert)

print(aldric.waffe, aldric.ruestung, aldric.helm)
```

3. __name__, __module__ und __bases__
</span>
</div>

Klassen tragen eingebaute Attribute mit Informationen ueber ihre eigene Struktur. Diese Attribute existieren an der **Klasse** selbst, nicht an den Instanzen.

```python
class Charakter:
    pass

class Held(Charakter):
    pass

class Krieger(Held):
    pass

aldric = Krieger()

# __name__: Der Name der Klasse als String
print(Krieger.__name__)    # 'Krieger'
print(Held.__name__)       # 'Held'

# __module__: Das Modul in dem die Klasse definiert ist
print(Krieger.__module__)  # '__main__' (in einem Notebook/Skript)

# __bases__: Ein Tupel der direkten Elternklassen
print(Krieger.__bases__)   # (<class '__main__.Held'>,)
print(Held.__bases__)      # (<class '__main__.Charakter'>,)
print(Charakter.__bases__) # (<class 'object'>,)  -- alles erbt von object!
```

__name__ an Klasse vs. Instanz
</span>

```python
class Held:
    pass

aldric = Held()

# __name__ existiert an der KLASSE
print(Held.__name__)        # 'Held'

# Nicht an der INSTANZ
try:
    print(aldric.__name__)  # AttributeError!
except AttributeError as e:
    print(f'Fehler: {e}')

# Umweg: Klasse des Objekts abfragen, dann __name__
print(type(aldric).__name__)           # 'Held'
print(aldric.__class__.__name__)       # 'Held' -- aequivalent
```

4. isinstance() und issubclass()
</span>
</div>

`type()` gibt die exakte Klasse eines Objekts zurueck. `isinstance()` prueft dagegen die gesamte Vererbungskette -- es gibt `True` zurueck, wenn das Objekt eine Instanz der Klasse oder einer ihrer Unterklassen ist.

Das ist besonders wichtig wenn man mit Vererbung arbeitet -- und ein zentrales Werkzeug fuer die PCAP-Pruefung.

```python
class Charakter:
    pass

class Held(Charakter):
    pass

class Krieger(Held):
    pass

aldric = Krieger()

# type() -- nur die direkte Klasse
print(type(aldric))              # <class '__main__.Krieger'>
print(type(aldric) == Krieger)   # True
print(type(aldric) == Held)      # False -- aldric ist Krieger, nicht Held

print()

# isinstance() -- prueft die gesamte Vererbungskette
print(isinstance(aldric, Krieger))    # True  -- ist ein Krieger
print(isinstance(aldric, Held))       # True  -- Krieger erbt von Held
print(isinstance(aldric, Charakter))  # True  -- Held erbt von Charakter
print(isinstance(aldric, str))        # False -- keine Verwandtschaft
```

```python
# issubclass() -- prueft Beziehung zwischen zwei Klassen (nicht Instanzen!)
print(issubclass(Krieger, Held))       # True
print(issubclass(Krieger, Charakter))  # True  -- auch ueber mehrere Ebenen
print(issubclass(Held, Krieger))       # False -- andersrum gilt nicht
print(issubclass(Krieger, Krieger))    # True  -- eine Klasse ist Unterklasse von sich selbst

# Kombination: Alle Helden in einer gemischten Liste identifizieren
class Gegner:
    pass

truppe = [Krieger(), Held(), Gegner(), Krieger()]

helden = [x for x in truppe if isinstance(x, Held)]
print(f'Helden in der Truppe: {len(helden)}')   # 3 -- Held und beide Krieger
```

---

**Zusammenfassung**

| Funktion / Attribut | Was es tut |
|---------------------|------------|
| `hasattr(obj, 'x')` | Prueft ob Attribut oder Methode `x` existiert |
| `getattr(obj, 'x')` | Liest Attribut `x` dynamisch (auch mit Fallback-Wert) |
| `setattr(obj, 'x', wert)` | Setzt Attribut `x` dynamisch |
| `Klasse.__name__` | Name der Klasse als String -- nur an der Klasse, nicht Instanz |
| `Klasse.__module__` | Modul in dem die Klasse definiert ist |
| `Klasse.__bases__` | Tupel der direkten Elternklassen |
| `isinstance(obj, Klasse)` | Prueft Vererbungskette -- auch Elternklassen geben True |
| `issubclass(Kind, Eltern)` | Prueft Klassen-Beziehung, nicht Instanzen |
