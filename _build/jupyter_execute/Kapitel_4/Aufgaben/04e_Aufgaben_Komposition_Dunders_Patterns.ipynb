**5.1**
```python
class Held:
    def __init__(self, name, hp):
        self.name = name
        self.hp   = hp

    def __str__(self):
        return f'{self.name}: {self.hp} HP'

    def __repr__(self):
        return f"Held('{self.name}', {self.hp})"

a = Held('Aldric', 100)
print(a)
print(repr(a))
print([a])
```

**5.2**
```python
class Trupp:
    def __init__(self, helden):
        self._helden = helden

    def __len__(self):
        return len(self._helden)

    def __add__(self, other):
        return Trupp(self._helden + other._helden)

a = Trupp(['Aldric', 'Lyria'])
b = Trupp(['Thorn'])
c = a + b
print(len(a))
print(len(c))
```

**5.3**
```python
class Held:
    def __init__(self, name, hp):
        self.name = name
        self.hp   = hp

    def __eq__(self, other):
        return self.hp == other.hp

a = Held('Aldric', 100)
b = Held('Lyria', 100)
c = Held('Thorn', 80)

print(a == b)
print(a == c)
print(a is b)
```
