**3.1**
```python
class Held:
    def __init__(self, name, hp):
        self.name = name
        self.hp   = hp

    def status(self):
        return f'{self.name}: {self.hp} HP'

class Krieger(Held):
    def __init__(self, name, hp, staerke):
        super().__init__(name, hp)
        self.staerke = staerke

aldric = Krieger('Aldric', 100, 15)
print(aldric.status())
print(aldric.staerke)
print(isinstance(aldric, Held))
```

**3.2**
```python
class Held:
    def angreifen(self):
        return 10

class Krieger(Held):
    def angreifen(self):
        return super().angreifen() + 5

class Paladin(Krieger):
    def angreifen(self):
        return super().angreifen() * 2

print(Held().angreifen())
print(Krieger().angreifen())
print(Paladin().angreifen())
```

**7.1**
```python
class A:
    def hallo(self):
        return 'A'

class B(A):
    def hallo(self):
        return 'B'

class C(A):
    def hallo(self):
        return 'C'

class D(B, C):
    pass

print(D().hallo())
print([k.__name__ for k in D.__mro__])
```

**7.2**
```python
class Held:
    def __str__(self):
        return 'Held'

class Krieger(Held):
    pass

class Paladin(Krieger):
    def __str__(self):
        return 'Paladin, ' + super().__str__()

print(str(Paladin()))
```
