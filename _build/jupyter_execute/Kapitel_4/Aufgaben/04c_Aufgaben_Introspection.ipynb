**4.1**
```python
class Charakter: pass
class Held(Charakter): pass
class Krieger(Held): pass

aldric = Krieger()

print(isinstance(aldric, Krieger))
print(isinstance(aldric, Held))
print(isinstance(aldric, Charakter))
print(isinstance(aldric, object))
print(isinstance(aldric, str))
```

**4.2**
```python
print(type(aldric) == Krieger)
print(type(aldric) == Held)
print(type(aldric) is Krieger)
```

**4.3**
```python
print(issubclass(Krieger, Held))
print(issubclass(Krieger, Charakter))
print(issubclass(Held, Krieger))
print(issubclass(Krieger, Krieger))
print(issubclass(Krieger, object))
```

**4.4**
```python
class Gegner: pass

truppe = [Krieger(), Held(), Gegner(), Krieger(), Charakter()]
helden = [x for x in truppe if isinstance(x, Held)]
print(len(helden))
```
