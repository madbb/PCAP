**4.1**
```python
class Held:
    def __init__(self, name, hp):
        self.name = name
        self.hp   = hp

    def status(self):
        return f'{self.name}: {self.hp} HP'

a = Held('Aldric', 100)
b = Held('Lyria', 80)
print(a.status())
print(b.status())
print(a.hp == b.hp)
```

**4.2**
```python
class Held:
    def __init__(self, name, hp):
        self.name = name
        self.hp   = hp

    def schaden(self, wert):
        self.hp -= wert

a = Held('Aldric', 100)
b = a
b.schaden(30)
print(a.hp)
print(b.hp)
print(a is b)
```
