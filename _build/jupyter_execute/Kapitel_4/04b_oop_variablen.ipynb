1. Eine neue Frage: Was gehoert zum Helden, was zur Gilde?
</span>
</div>

In 04a haben wir Attribute kennengelernt, die wir in `__init__` mit `self.name = name` gesetzt haben. Diese Attribute gehoeren zu einem bestimmten Objekt -- jeder Held hat seinen eigenen Namen, seine eigenen HP.

Aber jetzt kommt eine neue Anforderung: Unsere Helden gehoeren alle derselben Gilde an. Die Gilde hat einen Namen und eine maximale Mitgliederzahl. Wie speichert man das?

Erster Gedanke: einfach als weiteres Attribut im Konstruktor:

```python
class Held:
    def __init__(self, name, klasse, hp):
        self.name        = name
        self.klasse      = klasse
        self.hp          = hp
        self.gildenname  = 'Die Silberne Klinge'   # jeder Held bekommt dasselbe
        self.max_mitglieder = 20                   # jeder Held bekommt dasselbe

aldric = Held('Aldric', 'Krieger', 100)
lyria  = Held('Lyria', 'Magierin', 80)

print(aldric.gildenname)   # Die Silberne Klinge
print(lyria.gildenname)    # Die Silberne Klinge
```

Das funktioniert -- aber es hat ein Problem: Der Gildenname wird fuer jedes Objekt einzeln gespeichert. Bei hundert Helden liegt derselbe String hundertmal im Speicher. Und wenn die Gilde umbenannt wird, muss man jeden einzelnen Helden aendern.

Es gibt einen besseren Weg: **Klassenvariablen**.

2. Klassenvariablen -- geteilt von allen Instanzen
</span>
</div>

Eine **Klassenvariable** wird direkt im Klassenkoerper definiert -- nicht in `__init__`, nicht mit `self`. Sie gehoert zur Klasse selbst, nicht zu einer einzelnen Instanz. Alle Objekte dieser Klasse teilen sich diese eine Variable.

```python
class Held:
    # Klassenvariablen -- gehoeren zur Klasse, geteilt von allen Instanzen
    gildenname      = 'Die Silberne Klinge'
    max_mitglieder  = 20

    def __init__(self, name, klasse, hp):
        # Instanzvariablen -- gehoeren zum einzelnen Objekt
        self.name   = name
        self.klasse = klasse
        self.hp     = hp

aldric = Held('Aldric', 'Krieger', 100)
lyria  = Held('Lyria',  'Magierin', 80)

# Zugriff ueber das Objekt -- Python findet die Klassenvariable
print(aldric.gildenname)   # Die Silberne Klinge
print(lyria.gildenname)    # Die Silberne Klinge

# Zugriff direkt ueber die Klasse -- der sauberere Weg
print(Held.gildenname)     # Die Silberne Klinge
print(Held.max_mitglieder) # 20
```

Der grosse Vorteil: Wenn die Gilde umbenannt wird, aendern wir `Held.gildenname` an einer einzigen Stelle -- und alle Helden sehen sofort den neuen Namen.

```python
# Umbenennung der Gilde -- nur eine Aenderung noetig
Held.gildenname = 'Die Goldene Flamme'

print(aldric.gildenname)   # Die Goldene Flamme -- automatisch aktualisiert
print(lyria.gildenname)    # Die Goldene Flamme
print(Held.gildenname)     # Die Goldene Flamme
```

3. Die PCAP-Falle: Klassenvariable per Instanz ueberschreiben
</span>
</div>

Hier lauert eine der wichtigsten Fallen fuer die PCAP-Pruefung. Was passiert, wenn wir eine Klassenvariable ueber ein Objekt setzen?

```python
class Held:
    gildenname = 'Die Silberne Klinge'

    def __init__(self, name):
        self.name = name

aldric = Held('Aldric')
lyria  = Held('Lyria')

# Was passiert wenn wir gildenname ueber aldric setzen?
aldric.gildenname = 'Freie Klingen'  # Achtung!

print(aldric.gildenname)  # Freie Klingen
print(lyria.gildenname)   # Die Silberne Klinge  -- lyria unveraendert!
print(Held.gildenname)    # Die Silberne Klinge  -- Klasse unveraendert!
```

Was ist passiert? `aldric.gildenname = 'Freie Klingen'` hat **keine** Klassenvariable veraendert -- es hat eine **neue Instanzvariable** an `aldric` erstellt, die die Klassenvariable fuer dieses eine Objekt **verdeckt**.

Die Klasse und alle anderen Instanzen sind unveraendert. Nur `aldric` hat jetzt seine eigene `gildenname`-Variable im Vordergrund.

Um die Klassenvariable wirklich zu aendern, muss man ueber die Klasse selbst gehen:

```python
class Held:
    gildenname = 'Die Silberne Klinge'
    def __init__(self, name): self.name = name

aldric = Held('Aldric')
lyria  = Held('Lyria')

# RICHTIG: Klassenvariable aendern -- alle sehen es
Held.gildenname = 'Die Goldene Flamme'
print(aldric.gildenname)  # Die Goldene Flamme
print(lyria.gildenname)   # Die Goldene Flamme

# FALLE: per Instanz setzen -- erzeugt versteckte Instanzvariable
aldric.gildenname = 'Freie Klingen'
print(aldric.gildenname)  # Freie Klingen  -- Instanzvariable verdeckt Klassenvariable
print(lyria.gildenname)   # Die Goldene Flamme  -- unveraendert
print(Held.gildenname)    # Die Goldene Flamme  -- unveraendert
```

Ein klassisches, sinnvolles Beispiel fuer Klassenvariablen: mitzaehlen wie viele Instanzen erschaffen wurden.

```python
class Held:
    anzahl_helden = 0   # Klassenvariable -- zaehlt alle Helden

    def __init__(self, name, klasse):
        self.name   = name
        self.klasse = klasse
        Held.anzahl_helden += 1   # bei jeder Erschaffung erhoehen

    def __str__(self):
        return f'{self.name} ({self.klasse})'


print(f'Helden in der Gilde: {Held.anzahl_helden}')  # 0

aldric = Held('Aldric', 'Krieger')
lyria  = Held('Lyria',  'Magierin')
thorn  = Held('Thorn',  'Waldlaeufer')

print(f'Helden in der Gilde: {Held.anzahl_helden}')  # 3
print(f'Auch erreichbar: {aldric.anzahl_helden}')    # 3 -- ueber Instanz lesbar
```

4. __dict__ -- ein Blick hinter die Kulissen
</span>
</div>

Jedes Objekt und jede Klasse traegt intern ein Dictionary mit sich -- `__dict__`. Es zeigt genau welche Attribute an welcher Stelle gespeichert sind. Das ist besonders nuetzlich um die Instanz/Klassen-Trennung sichtbar zu machen.

```python
class Held:
    gildenname = 'Die Silberne Klinge'  # Klassenvariable

    def __init__(self, name, klasse, hp):
        self.name   = name    # Instanzvariablen
        self.klasse = klasse
        self.hp     = hp

aldric = Held('Aldric', 'Krieger', 100)

# __dict__ des Objekts: nur Instanzvariablen
print('Objekt __dict__:', aldric.__dict__)

# __dict__ der Klasse: Klassenvariablen und Methoden
print('Klassen __dict__ (Auszug):')
for key, value in Held.__dict__.items():
    if not key.startswith('__'):   # Dunder-Eintraege ueberspringen
        print(f'  {key}: {value}')
```

Man sieht klar: `gildenname` steht im Dictionary der **Klasse**, nicht im Dictionary des Objekts. Wenn `aldric.gildenname` aufgerufen wird, findet Python es nicht im Objekt-Dict und schaut dann in der Klasse nach -- das ist die normale Suchkette.

Sobald wir `aldric.gildenname = 'Freie Klingen'` setzen, landet das im Objekt-Dict von `aldric` -- und ab dann wird die Klassenvariable fuer dieses Objekt nicht mehr gefunden.

5. Private Attribute und Name Mangling
</span>
</div>

In 04a haben wir den einfachen Unterstrich `_hp` kennengelernt -- eine Konvention die sagt: *Bitte nicht direkt anfassen.* Python erzwingt das aber nicht.

Mit **zwei** Unterstrichen (`__hp`) geht Python einen Schritt weiter und benennt das Attribut intern um -- das nennt sich **Name Mangling**. Das Attribut heisst dann nicht mehr `__hp` sondern `_Klassenname__hp`.

Warum ist das nuetzlich? Um in Vererbungshierarchien zu verhindern, dass eine Kindklasse versehentlich ein gleichnamiges Attribut der Elternklasse ueberschreibt.

```python
class Held:
    def __init__(self, name, geheimcode):
        self.name        = name          # oeffentlich
        self._fraktion   = 'Gilde'       # geschuetzt (Konvention)
        self.__geheimcode = geheimcode   # privat (Name Mangling)

    def code_korrekt(self, eingabe):
        return eingabe == self.__geheimcode


aldric = Held('Aldric', 'schild42')

print(aldric.name)       # Aldric    -- oeffentlich, kein Problem
print(aldric._fraktion)  # Gilde     -- geht, aber Konvention sagt: lass es

# Direkter Zugriff auf __geheimcode funktioniert nicht mehr
try:
    print(aldric.__geheimcode)
except AttributeError as e:
    print(f'AttributeError: {e}')

# Python hat den Namen intern veraendert -- so kommt man dran (aber nie tun!)
print(aldric._Held__geheimcode)   # schild42 -- so heisst es wirklich

# Der saubere Weg: ueber die vorgesehene Methode
print(aldric.code_korrekt('schild42'))  # True
print(aldric.code_korrekt('falsch'))    # False
```

Und jetzt der eigentliche Nutzen -- Name Mangling schuetzt in Vererbungshierarchien:

```python
class Held:
    def __init__(self, name):
        self.name  = name
        self.__rang = 'Rekrut'   # wird zu _Held__rang

    def rang(self):
        return self.__rang       # greift auf _Held__rang zu

class EliteHeld(Held):
    def __init__(self, name):
        super().__init__(name)
        self.__rang = 'Elite'    # wird zu _EliteHeld__rang -- ANDERER Name!

aldric = EliteHeld('Aldric')

# rang() in der Elternklasse greift auf _Held__rang zu -- unveraendert!
print(aldric.rang())             # Rekrut -- nicht 'Elite', kein versehentliches Ueberschreiben

# Beide Werte existieren unabhaengig
print(aldric._Held__rang)        # Rekrut
print(aldric._EliteHeld__rang)   # Elite
```

---

**Zusammenfassung**

| Konzept | Wo definiert | Gehoert zu | Zugriff |
|---------|-------------|-----------|--------|
| Instanzvariable | `__init__` mit `self.x = ...` | Dem einzelnen Objekt | `objekt.x` |
| Klassenvariable | Im Klassenkoerper ohne `self` | Der Klasse, geteilt von allen | `Klasse.x` oder `objekt.x` |
| `_x` | Ueberall | Konvention: intern | Technisch ueberall, aber bitte nicht |
| `__x` | Ueberall | Wirklich intern | Wird zu `_Klassenname__x` umbenannt |
| `__dict__` | -- | Jedes Objekt und jede Klasse | Zeigt alle gespeicherten Attribute |
