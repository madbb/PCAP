1. Warum ueberhaupt OOP?
</span>
</div>

Bisher hast du Programme geschrieben, die aus Variablen und Funktionen bestehen. Das reicht fuer einfache Aufgaben gut aus.

Stell dir jetzt vor, du entwickelst ein Fantasy-Rollenspiel. Du brauchst Helden -- und jeder Held hat einen Namen, eine Klasse, Lebenspunkte und eine Staerke. Du fangst so an:

````python
# Erster Versuch: einfach Variablen
name    = 'Aldric'
klasse  = 'Krieger'
hp      = 100
staerke = 15

print(f'{name} der {klasse} hat {hp} HP und {staerke} Staerke.')
````

Funktioniert. Jetzt soll Aldric angreifen koennen. Du schreibst eine Funktion:

````python
def angriff(name, staerke):
    schaden = staerke * 2
    print(f'{name} greift an und macht {schaden} Schaden!')

angriff(name, staerke)
````

Gut. Jetzt kommt ein zweiter Held hinzu. Und ein dritter. Und jeder soll angreifen, Schaden nehmen und geheilt werden koennen.

````python
# Das skaliert nicht
held1_name = 'Aldric';  held1_hp = 100; held1_staerke = 15
held2_name = 'Lyria';   held2_hp = 80;  held2_staerke = 12
held3_name = 'Thorn';   held3_hp = 90;  held3_staerke = 14

def angriff(name, staerke):
    print(f'{name} macht {staerke * 2} Schaden!')

def schaden_nehmen(name, hp, schaden):
    hp -= schaden
    print(f'{name} hat noch {hp} HP.')
    return hp  # wir muessen den neuen Wert zurueckgeben und speichern!

held1_hp = schaden_nehmen(held1_name, held1_hp, 30)
angriff(held2_name, held2_staerke)
````

Siehst du die Probleme?

- Die Daten eines Helden (`held1_name`, `held1_hp`, ...) und die Funktionen die darauf arbeiten gehoeren zusammen -- aber sie sind getrennt. Kein gemeinsamer Behaelter.
- Jede Funktion muss alle benoetigen Werte als Parameter bekommen und veraenderte Werte wieder zurueckgeben.
- Bei zehn Helden hat man dreissig Variablen und kaum noch Ueberblick.

**Objektorientierte Programmierung (OOP)** ist der Ausweg. Die Idee: Daten und die Funktionen die auf ihnen arbeiten gehoeren in einen gemeinsamen Behaelter -- ein **Objekt**. Den Bauplan fuer diesen Behaelter nennt man **Klasse**.

2. Klasse und Objekt
</span>
</div>

Eine **Klasse** ist ein Bauplan. Sie beschreibt, welche Eigenschaften und Faehigkeiten ein Held haben soll -- aber sie ist selbst noch kein konkreter Held. Sie ist die Vorlage.

Ein **Objekt** (oder **Instanz**) ist das Konkrete, das nach dem Bauplan erschaffen wird. Aldric ist ein Objekt, Lyria ist ein Objekt -- beide wurden nach demselben Bauplan `Held` erschaffen, sind aber eigenstaendige Entitaeten mit eigenen Werten.

In Python definiert man eine Klasse mit dem Schluesselwort `class`:

````python
class Held:
    pass   # vorerst leer -- der Bauplan existiert, hat aber noch keinen Inhalt

# Objekte aus dem Bauplan erschaffen
aldric = Held()
lyria  = Held()

# Beide sind vom Typ Held
print(type(aldric))              # <class '__main__.Held'>
print(isinstance(aldric, Held))  # True

# Aber es sind zwei separate Objekte im Speicher
print(aldric is lyria)           # False -- verschiedene Identitaeten
````

Wir koennen Objekte erschaffen -- aber sie haben noch keine Eigenschaften. Wie bekommt Aldric seinen Namen und seine Lebenspunkte? Dafuer gibt es den Konstruktor.

3. Der Konstruktor __init__
</span>
</div>

Wenn `Held('Aldric', 'Krieger', 100, 15)` aufgerufen wird, braucht Python eine Moeglichkeit, die uebergebenen Werte am neuen Objekt zu speichern. Das erledigt der **Konstruktor** -- eine Methode mit dem festen Namen `__init__`.

Python ruft `__init__` automatisch auf, sobald ein neues Objekt erschaffen wird. Der erste Parameter heisst immer `self` -- das ist die Referenz auf das gerade erschaffene Objekt selbst. Ueber `self.name = name` speichern wir Werte direkt am Objekt, als sogenannte **Attribute**.

````python
class Held:
    def __init__(self, name, klasse, hp, staerke):
        # self ist das neue Objekt -- wir haengen die Werte daran
        self.name    = name
        self.klasse  = klasse
        self.hp      = hp
        self.staerke = staerke

# Python ruft intern auf: Held.__init__(aldric, 'Aldric', 'Krieger', 100, 15)
aldric = Held('Aldric', 'Krieger', 100, 15)
lyria  = Held('Lyria',  'Magierin', 80, 12)
thorn  = Held('Thorn',  'Waldlaeufer', 90, 14)

# Punktnotation fuer Zugriff auf Attribute
print(aldric.name)    # Aldric
print(lyria.klasse)   # Magierin
print(thorn.hp)       # 90

# Jedes Objekt hat seine eigenen, unabhaengigen Werte
print(aldric.staerke, lyria.staerke, thorn.staerke)  # 15  12  14
````

Das ist schon ein grosser Schritt: `aldric.hp` gehoert zu Aldric, `lyria.hp` gehoert zu Lyria. Sie koennen sich nicht mehr gegenseitig in die Quere kommen. Und wenn ein vierter Held hinzukommt, schreiben wir nur `Held('Mira', 'Schurkin', 75, 13)` -- die Klasse bleibt unveraendert.

4. Methoden -- das Verhalten des Helden
</span>
</div>

Attribute beschreiben was ein Held **ist**. Methoden beschreiben was ein Held **kann**.

Eine Methode ist eine Funktion innerhalb einer Klasse. Auch sie bekommt `self` als ersten Parameter -- und damit automatisch Zugriff auf alle Attribute des Objekts. Das ist der entscheidende Unterschied zu einer normalen Funktion: Sie muss ihre Daten nicht als Parameter bekommen, sie hat sie bereits.

````python
class Held:
    def __init__(self, name, klasse, hp, staerke):
        self.name    = name
        self.klasse  = klasse
        self.hp      = hp
        self.staerke = staerke

    def angreifen(self):
        # self.staerke -- kein Parameter noetig, der Held kennt seine eigene Staerke
        schaden = self.staerke * 2
        print(f'{self.name} greift an und macht {schaden} Schaden!')

    def erleidet_schaden(self, schaden):
        self.hp -= schaden
        if self.hp <= 0:
            self.hp = 0
            print(f'{self.name} wurde besiegt!')
        else:
            print(f'{self.name} hat noch {self.hp} HP.')

    def heilen(self, menge):
        self.hp += menge
        print(f'{self.name} wird geheilt. HP: {self.hp}')


aldric = Held('Aldric', 'Krieger', 100, 15)
lyria  = Held('Lyria', 'Magierin', 80, 12)

aldric.angreifen()          # kein Argument noetig
lyria.erleidet_schaden(30)  # nur der Schadenswert -- nicht Name und HP
lyria.heilen(15)
````

Vergleiche das mit dem Ansatz von vorhin:

```python
# Vorher: Funktion braucht alle Daten als Parameter
held1_hp = schaden_nehmen(held1_name, held1_hp, 30)

# Jetzt: Methode hat die Daten bereits
lyria.erleidet_schaden(30)
```

Der Aufruf `aldric.angreifen()` sieht so aus als ob wir `self` nicht angeben -- aber Python uebergibt `aldric` automatisch als `self`. Intern ist das identisch mit `Held.angreifen(aldric)`.

self vergessen -- TypeError mit seltsamer Meldung
</span>
</div>

````python
class Held:
    def __init__(self, name):
        self.name = name

    def angreifen(self):   # self ist Pflicht
        print(f'{self.name} greift an!')

    def info():            # self VERGESSEN -- haeufiger Anfaengerfehler
        print('Ich bin ein Held')

aldric = Held('Aldric')
aldric.angreifen()   # funktioniert

try:
    aldric.info()    # TypeError -- Python uebergibt aldric als erstes Argument
                     # aber info() erwartet kein Argument
except TypeError as e:
    print(f'Fehler: {e}')

# Technisch: self ist kein Schluesselwort, nur Konvention
# In der PCAP-Pruefung kann der Parameter auch anders heissen
class Test:
    def __init__(this, wert):   # 'this' statt 'self' -- funktioniert, aber nie machen
        this.wert = wert

t = Test(42)
print(t.wert)  # 42
````

5. __str__ -- der Charakter-Status
</span>
</div>

Wenn wir `print(aldric)` aufrufen, ohne etwas zu definieren, zeigt Python eine wenig hilfreiche Speicheradresse. Mit der Methode `__str__` koennen wir bestimmen, was angezeigt wird -- quasi den Status-Screen des Helden.

`__str__` muss einen String zurueckgeben. Python ruft sie automatisch auf, wenn ein Objekt mit `print()` ausgegeben oder mit `str()` in einen String umgewandelt wird.

````python
class Held:
    def __init__(self, name, klasse, hp, staerke):
        self.name    = name
        self.klasse  = klasse
        self.hp      = hp
        self.staerke = staerke

    def erleidet_schaden(self, schaden):
        self.hp = max(0, self.hp - schaden)

    def ist_am_leben(self):
        return self.hp > 0

    def __str__(self):
        status = '✓ lebt' if self.ist_am_leben() else '✗ besiegt'
        return (f'--- {self.name} ({self.klasse}) ---\n'
                f'  HP: {self.hp}  |  STR: {self.staerke}  |  {status}')


aldric = Held('Aldric', 'Krieger', 100, 15)
print(aldric)       # ruft __str__ auf

print()
aldric.erleidet_schaden(120)
print(aldric)       # Status hat sich veraendert
````

6. Kapselung -- Schutz vor unsinnigen Werten
</span>
</div>

Momentan kann jeder die Attribute direkt setzen -- auch auf Werte die im Spielkontext keinen Sinn ergeben:

````python
aldric = Held('Aldric', 'Krieger', 100, 15)

# Von aussen direkt veraenderbar -- das kann zu Problemen fuehren
aldric.hp = -999     # negative HP -- bricht die Spiellogik
aldric.staerke = 0   # Staerke 0 -- Angriffe machen keinen Schaden
print(aldric)
````

**Kapselung** bedeutet: Interne Daten werden nicht einfach von aussen veraendert -- der Zugriff laeuft ueber Methoden, die sicherstellen dass alles konsistent bleibt.

Python hat keine echten privaten Attribute, aber eine Konvention: Ein fuehrender Unterstrich `_hp` signalisiert: *"Bitte nicht direkt von aussen anfassen."* Das ist kein erzwungener Schutz, aber ein klares Signal an andere Entwickler.

````python
class Held:
    def __init__(self, name, klasse, hp, staerke):
        self.name     = name
        self.klasse   = klasse
        self._hp      = hp       # _ = intern, bitte nicht direkt setzen
        self._staerke = max(1, staerke)  # Staerke mindestens 1

    def erleidet_schaden(self, schaden):
        # Die Methode sorgt dafuer, dass HP nie unter 0 faellt
        self._hp = max(0, self._hp - schaden)

    def heilen(self, menge):
        # Nur positive Heilmenge wird akzeptiert
        if menge > 0:
            self._hp += menge

    def angreifen(self):
        return self._staerke * 2

    def __str__(self):
        status = '✓ lebt' if self._hp > 0 else '✗ besiegt'
        return f'[{self.klasse}] {self.name} | HP: {self._hp} | {status}'


aldric = Held('Aldric', 'Krieger', 100, 15)
print(aldric)

aldric.erleidet_schaden(999)  # Egal wie viel Schaden -- HP bleibt bei 0
print(aldric)

aldric.heilen(50)
print(aldric)
````

Der Held kontrolliert jetzt selbst, was mit seinen HP passiert. Von aussen kann man ihn nur ueber definierte Methoden beeinflussen -- `erleidet_schaden()` und `heilen()`. Das ist Kapselung in der Praxis.

**Zusammenfassung**

| Begriff | Bedeutung | Fantasy-Analogie |
|---------|-----------|------------------|
| Klasse | Bauplan / Vorlage | Das Charakter-Erstellungsformular |
| Objekt / Instanz | Konkretes Exemplar | Aldric, Lyria, Thorn |
| Attribut | Eigenschaft des Objekts | Name, HP, Staerke |
| `__init__` | Konstruktor -- beim Erschaffen aufgerufen | Charakter betritt die Welt |
| `self` | Referenz auf das Objekt selbst | Der Held selbst |
| Methode | Funktion der Klasse -- kennt ihre eigenen Daten | angreifen(), heilen() |
| `__str__` | Textdarstellung des Objekts | Status-Screen des Helden |
| Kapselung | Daten und Methoden gebuendelt, Zugriff kontrolliert | HP kann nicht auf -999 gesetzt werden |
