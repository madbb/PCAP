1. Das Problem: Wiederholung in aehnlichen Klassen
</span>
</div>

Unser Spiel bekommt verschiedene Heldenklassen: Krieger, Magier, Waldlaeufer. Jede hat eigene Faehigkeiten -- aber auch vieles gemeinsam: Name, HP, Staerke, `angreifen()`, `erleidet_schaden()`, `__str__`.

Ohne Vererbung wuerde man jede Klasse einzeln schreiben:

````python
class Krieger:
    def __init__(self, name, hp, staerke):
        self.name    = name
        self.hp      = hp
        self.staerke = staerke

    def angreifen(self):           # identisch in jeder Klasse
        return self.staerke * 2

    def erleidet_schaden(self, s): # identisch in jeder Klasse
        self.hp = max(0, self.hp - s)

    def spezial(self):
        return f'{self.name} laeuft in den Kampf!'

class Magier:
    def __init__(self, name, hp, staerke):  # dieselben drei Zeilen nochmal
        self.name    = name
        self.hp      = hp
        self.staerke = staerke

    def angreifen(self):           # Code-Duplikat!
        return self.staerke * 2

    def erleidet_schaden(self, s): # Code-Duplikat!
        self.hp = max(0, self.hp - s)

    def spezial(self):
        return f'{self.name} wirkt einen Feuerball!'

# Und Waldlaeufer waere das dritte Mal dasselbe...
````

Das verletzt das DRY-Prinzip massiv. Und wenn wir `erleidet_schaden()` verbessern wollen, muessen wir jede Klasse einzeln anfassen.

**Vererbung** loest das: Gemeinsamen Code einmal in einer Basisklasse schreiben. Kindklassen erben alles davon und fuegen nur ihr Eigenes hinzu.

2. Einfache Vererbung und super()
</span>
</div>

Die Basisklasse `Held` enthaelt alles Gemeinsame. Die Kindklassen erben von ihr -- in Python schreibt man das als `class Krieger(Held)`. Die Kindklasse bekommt damit automatisch alle Methoden und Attribute der Elternklasse.

`super()` gibt den Zugriff auf die Elternklasse. In `__init__` der Kindklasse nutzt man es, um den Konstruktor der Elternklasse aufzurufen -- sonst muesste man alle gemeinsamen Attribute nochmal manuell setzen.

````python
class Held:
    """Basisklasse -- enthaelt alles was alle Helden gemeinsam haben."""

    def __init__(self, name, hp, staerke):
        self.name    = name
        self.hp      = hp
        self.staerke = staerke

    def angreifen(self):
        return self.staerke * 2

    def erleidet_schaden(self, schaden):
        self.hp = max(0, self.hp - schaden)

    def ist_am_leben(self):
        return self.hp > 0

    def __str__(self):
        return f'{self.name} | HP: {self.hp} | STR: {self.staerke}'


class Krieger(Held):              # erbt von Held
    def __init__(self, name, hp, staerke, ruestung):
        super().__init__(name, hp, staerke)  # Elternkonstruktor aufrufen
        self.ruestung = ruestung             # eigenes Attribut hinzufuegen

    def spezial(self):            # eigene Methode
        return f'{self.name} setzt Schlachtruf ein! Alle erhalten +5 STR.'


class Magier(Held):
    def __init__(self, name, hp, staerke, mana):
        super().__init__(name, hp, staerke)
        self.mana = mana

    def spezial(self):
        if self.mana >= 20:
            self.mana -= 20
            return f'{self.name} wirkt Feuerball! 40 Schaden!'
        return f'{self.name} hat zu wenig Mana!'


class Waldlaeufer(Held):
    def __init__(self, name, hp, staerke, pfeile):
        super().__init__(name, hp, staerke)
        self.pfeile = pfeile

    def spezial(self):
        if self.pfeile > 0:
            self.pfeile -= 1
            return f'{self.name} schiesst einen Pfeil! Noch {self.pfeile} Pfeile.'
        return f'{self.name} hat keine Pfeile mehr!'


# Alle Klassen in Aktion
aldric = Krieger('Aldric', 100, 15, 'Plattenpanzer')
lyria  = Magier('Lyria', 80, 10, 60)
thorn  = Waldlaeufer('Thorn', 90, 13, 20)

for held in [aldric, lyria, thorn]:
    print(held)             # __str__ aus Held
    print(held.spezial())   # eigene spezial()-Methode
    print()
````

3. Overriding und Polymorphismus
</span>
</div>

Was wenn eine Kindklasse eine geerbte Methode anders verhalten soll? Ein Krieger greift anders an als ein Magier -- der Basisangriff passt nicht fuer alle.

**Overriding** bedeutet: Die Kindklasse definiert eine Methode mit demselben Namen neu. Die eigene Version hat Vorrang vor der geerbten. Das ist kein Fehler -- es ist gewollt.

````python
class Held:
    def __init__(self, name, hp, staerke):
        self.name    = name
        self.hp      = hp
        self.staerke = staerke

    def angreifen(self):
        return self.staerke * 2   # Standard-Angriff

    def __str__(self):
        return f'{self.name} (HP: {self.hp})'


class Krieger(Held):
    def __init__(self, name, hp, staerke, ruestung):
        super().__init__(name, hp, staerke)
        self.ruestung = ruestung

    def angreifen(self):          # Override: eigene Version
        return self.staerke * 3   # Krieger schlaegt haerter zu

    def erleidet_schaden(self, schaden):  # Override: Ruestung reduziert Schaden
        reduziert = max(0, schaden - self.ruestung)
        self.hp = max(0, self.hp - reduziert)
        return reduziert


class Magier(Held):
    def __init__(self, name, hp, staerke, mana):
        super().__init__(name, hp, staerke)
        self.mana = mana

    def angreifen(self):          # Override: Magier nutzt Mana
        if self.mana >= 10:
            self.mana -= 10
            return self.staerke * 4   # Zauberer trifft magisch
        return self.staerke * 1       # ohne Mana: schwacher Angriff


aldric = Krieger('Aldric', 100, 15, 5)
lyria  = Magier('Lyria', 80, 10, 60)
basis  = Held('Bauer', 50, 8)      # kein Override, Standard-Angriff

for kaempfer in [aldric, lyria, basis]:
    print(f'{kaempfer}: Angriff = {kaempfer.angreifen()}')
````

Das ist **Polymorphismus** -- "viele Formen". Alle drei Objekte haben eine Methode `angreifen()`. Der aufrufende Code muss nicht wissen ob er einen Krieger, Magier oder Bauern vor sich hat -- er ruft einfach `kaempfer.angreifen()` auf, und Python weiss welche Version gemeint ist.

Die Schleife `for kaempfer in [aldric, lyria, basis]` behandelt alle gleich -- obwohl sie unterschiedliche Klassen sind.

4. Mehrfachvererbung und MRO
</span>
</div>

Python erlaubt es, von mehreren Klassen gleichzeitig zu erben. Stell dir vor, ein Paladin ist sowohl Krieger als auch Heiler -- er hat Faehigkeiten von beiden.

````python
class Kaempfer:
    def angreifen(self):
        return f'{self.name} schlaegt zu!'

class Heiler:
    def heilen(self):
        return f'{self.name} heilt einen Verbündeten!'

# Paladin erbt von beiden
class Paladin(Kaempfer, Heiler):
    def __init__(self, name):
        self.name = name

    def heiliger_schlag(self):
        return f'{self.name} trifft und heilt gleichzeitig!'


valeria = Paladin('Valeria')
print(valeria.angreifen())      # von Kaempfer geerbt
print(valeria.heilen())         # von Heiler geerbt
print(valeria.heiliger_schlag()) # eigene Methode

# Vererbungskette pruefen
print(isinstance(valeria, Kaempfer))  # True
print(isinstance(valeria, Heiler))    # True
````

Aber was passiert wenn beide Elternklassen eine Methode mit demselben Namen haben? Python folgt der **MRO (Method Resolution Order)** -- einer festen Suchreihenfolge, die mit `__mro__` abfragbar ist.

````python
class A:
    def faehigkeit(self):
        return 'Faehigkeit von A'

class B(A):
    def faehigkeit(self):
        return 'Faehigkeit von B'

class C(A):
    def faehigkeit(self):
        return 'Faehigkeit von C'

class D(B, C):   # erbt von B und C
    pass

d = D()
print(d.faehigkeit())   # Welche Version gewinnt?

# MRO zeigt die Suchreihenfolge
print(D.__mro__)
# D -> B -> C -> A -> object
# Python sucht von links nach rechts -- B kommt zuerst, also gewinnt B
````

5. Das Diamant-Problem
</span>
</div>

Das **Diamant-Problem** tritt auf, wenn eine Klasse ueber zwei verschiedene Wege von derselben Basisklasse erbt. Die Vererbungsstruktur sieht aus wie eine Raute:

```
       Wesen
      /     \
 Kaempfer  Heiler
      \     /
       Paladin
```

Die Frage ist: Was passiert wenn `Paladin.__init__` aufgerufen wird und jede Klasse `super().__init__()` aufruft -- wer ruft am Ende `Wesen.__init__` auf, und wie oft?

Schauen wir uns zuerst an was ohne besondere Massnahmen passiert:

````python
# ERSTE VERSION -- funktioniert NICHT
# Beachte: jede Klasse hat ihre eigenen Parameter in __init__

class Held:
    def __init__(self, name):
        self.name = name

class Kaempfer(Held):
    def __init__(self, name, staerke):
        super().__init__(name)   # <-- was ruft super() hier auf?
        self.staerke = staerke

class Heiler(Held):
    def __init__(self, name, heilkraft):
        super().__init__(name)
        self.heilkraft = heilkraft

class Paladin(Kaempfer, Heiler):
    def __init__(self, name, staerke, heilkraft):
        super().__init__(name, staerke)   # ruft Kaempfer auf
        self.heilkraft = heilkraft

try:
    valeria = Paladin('Valeria', 12, 18)
except TypeError as e:
    print(f'TypeError: {e}')
````

Der Fehler lautet: `Heiler.__init__() missing 1 required positional argument: 'heilkraft'`

Das ist verwirrend -- wir rufen `Paladin('Valeria', 12, 18)` auf, nicht `Heiler` direkt. Was ist passiert?

**Der Schluessel: `super()` folgt nicht der Klassenhierarchie, sondern der MRO.**

Die MRO von `Paladin` ist:

````python
# MRO anzeigen
print([k.__name__ for k in Paladin.__mro__])
# ['Paladin', 'Kaempfer', 'Heiler', 'Held', 'object']
````

Python ruft `super().__init__()` nicht nach der Vererbungshierarchie auf, sondern nach der MRO -- linear von links nach rechts:

1. `Paladin.__init__` ruft `super().__init__(name, staerke)`
   --> laut MRO: naechste Klasse ist **Kaempfer**

2. `Kaempfer.__init__` bekommt `name='Valeria'`, `staerke=12` -- ok
   ruft dann `super().__init__(name)` auf
   --> laut MRO: naechste Klasse ist **Heiler** -- nicht Wesen!

3. `Heiler.__init__` erwartet `(name, heilkraft)` -- bekommt aber nur `(name)`    --> **TypeError**

Das ist das Diamant-Problem: `super()` in `Kaempfer` landet bei `Heiler`, nicht bei `Wesen`. Die Klassen muessen so gebaut sein, dass sie miteinander funktionieren -- auch wenn sie in der MRO aufeinandertreffen.

Die Loesung: Jede Klasse muss unbekannte Parameter mit `**kwargs` weiterreichen, damit die MRO-Kette nicht bricht.

````python
# KORREKTE VERSION -- mit **kwargs

class Held:
    def __init__(self, name, **kwargs):    # **kwargs: nimmt alles uebrige entgegen
        super().__init__(**kwargs)          # und reicht es weiter (fuer object)
        print(f'Held.__init__: {name}')
        self.name = name

class Kaempfer(Held):
    def __init__(self, staerke, **kwargs): # nimmt staerke heraus
        super().__init__(**kwargs)          # reicht den Rest weiter
        print(f'Kaempfer.__init__: staerke={staerke}')
        self.staerke = staerke

class Heiler(Held):
    def __init__(self, heilkraft, **kwargs): # nimmt heilkraft heraus
        super().__init__(**kwargs)            # reicht den Rest weiter
        print(f'Heiler.__init__: heilkraft={heilkraft}')
        self.heilkraft = heilkraft

class Paladin(Kaempfer, Heiler):
    def __init__(self, name, staerke, heilkraft):
        # Alle Parameter als Keyword-Argumente -- jede Klasse nimmt sich was sie kennt
        super().__init__(name=name, staerke=staerke, heilkraft=heilkraft)

    def __str__(self):
        return f'Paladin {self.name} | STR: {self.staerke} | HLK: {self.heilkraft}'


print('MRO:', [k.__name__ for k in Paladin.__mro__])
print()
valeria = Paladin('Valeria', 12, 18)
print()
print(valeria)
````

Jetzt laeuft die Kette sauber durch:

1. `Paladin.__init__` ruft `super().__init__(name=..., staerke=..., heilkraft=...)`
2. **Kaempfer** nimmt `staerke` heraus, reicht `name` und `heilkraft` weiter
3. **Heiler** nimmt `heilkraft` heraus, reicht `name` weiter
4. **Held** nimmt `name` heraus, reicht nichts weiter

Jede Klasse bedient sich nur an dem was sie kennt -- der Rest wandert durch. `Wesen.__init__` wird genau einmal aufgerufen, obwohl zwei Pfade zu ihm fuehren. Das ist kooperative Mehrfachvererbung in Python.

6. is und is not -- Identitaet pruefen
</span>
</div>

`==` prueft ob zwei Objekte **gleich sind** (gleicher Wert). `is` prueft ob es **dasselbe Objekt** ist (dieselbe Speicheradresse). Der Unterschied ist subtil aber wichtig -- besonders in der PCAP-Pruefung.

````python
class Held:
    def __init__(self, name):
        self.name = name

aldric1 = Held('Aldric')
aldric2 = Held('Aldric')  # gleiches Aussehen, anderes Objekt
aldric3 = aldric1          # dieselbe Referenz

# == prueft Gleichheit (ruft __eq__ auf -- hier nicht definiert, also Identitaet)
print(aldric1 == aldric2)   # False -- verschiedene Objekte, kein __eq__ definiert
print(aldric1 == aldric3)   # True  -- selbes Objekt

# is prueft Identitaet (selbe Speicheradresse)
print(aldric1 is aldric2)   # False -- verschiedene Objekte
print(aldric1 is aldric3)   # True  -- aldric3 zeigt auf aldric1
print(aldric1 is not aldric2)  # True

# Praktisch: None-Prüfung immer mit is, nicht mit ==
held = None
print(held is None)         # True  -- empfohlene Form
print(held is not None)     # False
````

**Zusammenfassung**

| Konzept | Erklaerung |
|---------|------------|
| Vererbung | `class Kind(Eltern)` -- Kind bekommt alle Methoden und Attribute |
| `super()` | Zugriff auf Elternklasse -- in `__init__` zum Weiterreichen der Argumente |
| Overriding | Kindklasse definiert geerbte Methode neu -- eigene Version hat Vorrang |
| Polymorphismus | Verschiedene Klassen, gleicher Methodenname -- Python ruft die richtige Version auf |
| Mehrfachvererbung | `class Kind(A, B)` -- erbt von beiden |
| MRO | Suchreihenfolge fuer Methoden -- von links nach rechts, dann aufwaerts |
| Diamant-Problem | Gemeinsame Basisklasse ueber zwei Wege -- Python loest mit `super()` und MRO |
| `is` / `is not` | Identitaetspruefung -- selbes Objekt, nicht gleicher Wert |
