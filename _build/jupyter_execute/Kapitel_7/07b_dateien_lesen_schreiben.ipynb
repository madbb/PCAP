1. open() und close() -- und warum with besser ist
</span>
</div>

Eine Datei oeffnen heisst: Das Betriebssystem reserviert Ressourcen. Diese Ressourcen muessen wieder freigegeben werden -- mit `close()`. Vergisst man `close()`, kann es zu Datenverlust oder Speicherlecks kommen.

Das Problem: Wenn zwischen `open()` und `close()` ein Fehler auftritt, wird `close()` nie aufgerufen.

```python
# Manuell -- fehleranfaellig
f = open('test.txt', 'w', encoding='utf-8')
f.write('Zeile 1\n')
f.write('Zeile 2\n')
f.close()   # muss explizit aufgerufen werden

print('Geschlossen:', f.closed)   # True
```

```python
# Das Problem ohne with: close() wird bei Fehler uebersprungen
f = open('test.txt', 'w', encoding='utf-8')
try:
    f.write('Zeile 1\n')
    raise RuntimeError('Simulierter Fehler')
    f.write('Zeile 2\n')   # wird nie erreicht
except RuntimeError:
    pass
# f.close() vergessen -- Datei bleibt offen!
print('Geschlossen:', f.closed)  # False!
f.close()   # manuelles Nachbessern noetig

# Die Loesung: with-Statement -- schliesst IMMER, auch bei Fehlern
with open('test.txt', 'w', encoding='utf-8') as f:
    f.write('Zeile 1\n')
    f.write('Zeile 2\n')
    f.write('Zeile 3\n')
# Hier ist f automatisch geschlossen
print('Geschlossen:', f.closed)  # True
```

2. Lesen: read(), readline() und readlines()
</span>
</div>

Es gibt drei Wege um Dateiinhalt zu lesen -- je nachdem wie viel man auf einmal einlesen will:

- `read()` -- alles auf einmal als ein String
- `readline()` -- eine Zeile auf einmal
- `readlines()` -- alle Zeilen als Liste

```python
# Testdatei erstellen
with open('helden.txt', 'w', encoding='utf-8') as f:
    f.write('Aldric,Krieger,100\n')
    f.write('Lyria,Magierin,80\n')
    f.write('Thorn,Waldlaeufer,90\n')

# read() -- gesamter Inhalt als ein String
with open('helden.txt', 'r', encoding='utf-8') as f:
    alles = f.read()
    print(repr(alles))

# read(n) -- maximal n Zeichen
with open('helden.txt', 'r', encoding='utf-8') as f:
    erste_10 = f.read(10)
    print(repr(erste_10))
```

```python
# readline() -- eine Zeile (inkl. \n)
with open('helden.txt', 'r', encoding='utf-8') as f:
    zeile1 = f.readline()
    zeile2 = f.readline()
    print(repr(zeile1))   # 'Aldric,Krieger,100\n'
    print(repr(zeile2))   # 'Lyria,Magierin,80\n'

# Schleife mit readline() -- leerer String signalisiert Dateiende
with open('helden.txt', 'r', encoding='utf-8') as f:
    while True:
        zeile = f.readline()
        if zeile == '':    # Dateiende
            break
        print(zeile.strip())
```

```python
# readlines() -- alle Zeilen als Liste
with open('helden.txt', 'r', encoding='utf-8') as f:
    zeilen = f.readlines()
    print(zeilen)    # ['Aldric,Krieger,100\n', 'Lyria,...]
    print(type(zeilen))  # list

# CSV-Daten verarbeiten
with open('helden.txt', 'r', encoding='utf-8') as f:
    for zeile in f.readlines():
        teile = zeile.strip().split(',')
        print(f'{teile[0]} ist ein {teile[1]} mit {teile[2]} HP')
```

Der Cursor -- zweimaliges Lesen liefert leeren String
</span>
</div>

Beim Lesen merkt sich Python eine **Cursorposition** -- wo im Stream man gerade ist. Nach `read()` steht der Cursor am Ende. Ein zweites `read()` gibt dann einen leeren String zurueck, weil nichts mehr kommt.

```python
with open('helden.txt', 'r', encoding='utf-8') as f:
    erstes  = f.read()
    zweites = f.read()   # Cursor ist am Ende!
    print(f'Erstes Lesen:  {len(erstes)} Zeichen')
    print(f'Zweites Lesen: {len(zweites)} Zeichen')  # 0!

# Loesung: seek(0) setzt den Cursor zurueck
with open('helden.txt', 'r', encoding='utf-8') as f:
    erstes = f.read()
    f.seek(0)            # zurueck zum Anfang
    zweites = f.read()
    print(f'Nach seek(0): {len(zweites)} Zeichen')   # wieder alle
```

3. Schreiben: write()
</span>
</div>

```python
# write() -- String schreiben, gibt Anzahl geschriebener Zeichen zurueck
with open('ausgabe.txt', 'w', encoding='utf-8') as f:
    n = f.write('Erste Zeile\n')
    print(f'Geschrieben: {n} Zeichen')
    f.write('Zweite Zeile\n')

# Modus 'w' ueberschreibt -- Modus 'a' haengt an
with open('ausgabe.txt', 'a', encoding='utf-8') as f:
    f.write('Dritte Zeile\n')   # wird angehaengt, nicht ueberschrieben

with open('ausgabe.txt', 'r', encoding='utf-8') as f:
    print(f.read())
```

4. errno -- Fehlercodes bei I/O-Problemen
</span>
</div>

Wenn eine Datei nicht gefunden wird oder die Berechtigungen fehlen, wirft Python eine `OSError`. Diese Exception hat ein `.errno`-Attribut -- eine numerische Kennung die genauer beschreibt was schief gelaufen ist. Das `errno`-Modul stellt benannte Konstanten fuer diese Codes bereit.

```python
import errno

# Wichtige Fehlercodes
print('ENOENT (nicht gefunden):', errno.ENOENT)  # 2
print('EACCES (kein Zugriff):  ', errno.EACCES)  # 13
print('EEXIST (existiert schon):', errno.EEXIST) # 17

# errno in der Praxis -- gezielt auf bestimmte Fehler reagieren
def lade_heldendaten(pfad):
    try:
        with open(pfad, 'r', encoding='utf-8') as f:
            return f.read()
    except OSError as e:
        if e.errno == errno.ENOENT:
            print(f'Datei nicht gefunden: {pfad}')
        elif e.errno == errno.EACCES:
            print(f'Kein Lesezugriff auf: {pfad}')
        else:
            print(f'I/O-Fehler {e.errno}: {e.strerror}')
        return None

lade_heldendaten('nicht_vorhanden.txt')
```

5. bytearray -- veraenderlicher Byte-Puffer
</span>
</div>

Im Binaermodus arbeitet man mit `bytes` -- aber `bytes` ist unveraenderlich, genau wie `str`. Wenn man Binaerdaten lesen und veraendern will, braucht man `bytearray` -- die veraenderliche Variante.

Der Unterschied: `bytes` kann man nicht an Ort und Stelle veraendern. `bytearray` schon -- man kann einzelne Bytes direkt setzen, ohne einen neuen Puffer zu erstellen.

```python
# bytes: unveraenderlich
b = b'Hallo'
try:
    b[0] = 104   # TypeError!
except TypeError as e:
    print(f'TypeError: {e}')

# bytearray: veraenderlich
ba = bytearray(b'Hallo')
print(ba)          # bytearray(b'Hallo')
ba[0] = 104        # h statt H (ASCII 72 -> 104)
print(ba)          # bytearray(b'hallo')

# Indexzugriff liefert int -- nicht bytes!
print(b[0])        # 72  -- nicht b'H'
print(ba[0])       # 104
print(b[0:1])      # b'H' -- Slicing gibt bytes zurueck
```

```python
# bytearray als Puffer beim Binaerlesen
# Binaerdatei erstellen
with open('binaer.bin', 'wb') as f:
    f.write(bytes(range(16)))   # Bytes 0-15

# In bytearray-Puffer lesen und bearbeiten
puffer = bytearray(8)   # Puffer fuer 8 Bytes

with open('binaer.bin', 'rb') as f:
    gelesene_bytes = f.readinto(puffer)  # direkt in Puffer lesen
    print(f'Gelesen: {gelesene_bytes} Bytes')
    print(f'Puffer: {list(puffer)}')

# Einzelne Bytes veraendern -- ohne neues Objekt zu erstellen
puffer[0] = 255
print(f'Veraendert: {list(puffer)}')

# Konvertierung zwischen bytes und bytearray
als_bytes     = bytes(puffer)      # bytearray -> bytes
als_bytearray = bytearray(als_bytes)  # bytes -> bytearray
print(type(als_bytes), type(als_bytearray))
```
