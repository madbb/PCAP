**2.1** Was gibt `type(inhalt)` aus?
```python
with open('test.txt', 'w') as f:
    f.write('Hallo')

with open('test.txt', 'r') as f:
    inhalt = f.read()
print(type(inhalt))
```

**2.2** Was gibt `type(inhalt)` aus?
```python
with open('test.txt', 'rb') as f:
    inhalt = f.read()
print(type(inhalt))
```

**2.3** Was steht danach in der Datei?
```python
with open('test.txt', 'w') as f:
    f.write('Erste Version')

with open('test.txt', 'w') as f:
    f.write('Zweite Version')

with open('test.txt', 'r') as f:
    print(f.read())
```

**2.4** Was steht danach in der Datei?
```python
with open('test.txt', 'w') as f:
    f.write('Erste Version')

with open('test.txt', 'a') as f:
    f.write(' Zweite Version')

with open('test.txt', 'r') as f:
    print(f.read())
```
