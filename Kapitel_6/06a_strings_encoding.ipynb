1. Das Grundproblem: Computer kennen keine Buchstaben
</span>
</div>

Ein Computer speichert alles als Zahlen -- genauer: als Folgen von Nullen und Einsen. Zahlen, Bilder, Musik -- alles ist letztlich eine Zahl.

Aber was ist mit Text? Das Zeichen `'A'` ist keine Zahl. Damit ein Computer Text speichern kann, braucht man eine **Vereinbarung**: Jedes Zeichen bekommt eine Nummer. Das nennt man **Encoding** oder **Zeichenkodierung**.

Die erste weitverbreitete Vereinbarung hiess **ASCII** (American Standard Code for Information Interchange), entwickelt in den 1960ern. Sie legte fest: `'A'` hat die Nummer 65, `'B'` hat 66, `'a'` hat 97, und so weiter.

```python
# Python zeigt uns die Zahlen hinter den Zeichen -- mit ord()
print(ord('A'))   # 65
print(ord('B'))   # 66
print(ord('a'))   # 97  -- Kleinbuchstaben haben andere Nummern!
print(ord('0'))   # 48  -- die Ziffer 0, nicht die Zahl null
print(ord(' '))   # 32  -- sogar ein Leerzeichen hat eine Nummer

# Und chr() macht das Umgekehrte: Nummer -> Zeichen
print(chr(65))    # A
print(chr(66))    # B
print(chr(97))    # a
```

ASCII definiert 128 Zeichen -- die Nummern 0 bis 127. Das reicht fuer englische Buchstaben, Ziffern und Sonderzeichen. Aber fuer Umlaute wie `ä`, `ö`, `ü` oder gar chinesische Zeichen ist kein Platz. Das war fuer die ersten Computer ausreichend -- aber nicht fuer eine globale Welt.

```python
# Die druckbaren ASCII-Zeichen (32-126)
print('Code | Zeichen')
print('-' * 15)
for i in range(32, 127):
    if i % 16 == 0:
        print()
    print(f'{i:3}: {chr(i)}  ', end='')
print()
```

2. Unicode -- ein Standard fuer alle Sprachen der Welt
</span>
</div>

In den 1980ern gab es ein Chaos: Jedes Land entwickelte eigene Erweiterungen von ASCII. Deutsche Texte nutzten einen anderen Code als japanische, franzoesische anders als arabische. Dateien die auf einem System korrekt aussahen, zeigten auf einem anderen nur Kauderwelsch.

Die Loesung: **Unicode**. Ein einziges, universelles System das jedem Zeichen aller Sprachen der Welt eine eindeutige Nummer zuweist -- den **Codepoint**. Heute hat Unicode ueber 140.000 Zeichen -- von Lateinisch ueber Arabisch und Chinesisch bis zu Emojis.

Die ersten 128 Unicode-Codepoints sind identisch mit ASCII -- das war bewusst, um Kompatibilitaet zu gewaehrleisten.

```python
# Unicode geht weit ueber ASCII hinaus
print(ord('ä'))    # 228  -- jenseits von ASCII (0-127)
print(ord('€'))    # 8364
print(ord('中'))   # 20013 -- chinesisches Zeichen

# Umgekehrt: aus Codepoint ein Zeichen
print(chr(228))    # ä
print(chr(8364))   # €
print(chr(20013))  # 中

# Python 3 Strings sind intern immer Unicode
text = 'Hallo Welt auf Chinesisch: 你好世界'
print(text)
print(f'Laenge: {len(text)} Zeichen')
```

3. UTF-8 -- wie Unicode gespeichert wird
</span>
</div>

Unicode definiert die Nummern. Aber wie werden diese Nummern tatsaechlich als Bytes gespeichert? Dafuer gibt es verschiedene **Encodings**. Das verbreiteste ist **UTF-8**.

UTF-8 ist clever: Haeufige Zeichen (ASCII, lateinische Buchstaben) brauchen nur 1 Byte. Seltenere Zeichen brauchen 2, 3 oder 4 Bytes. Das macht UTF-8 platzsparend und gleichzeitig kompatibel mit altem ASCII-Code.

In Python ist der Unterschied zwischen `str` (Unicode-Text) und `bytes` (rohe Bytes) wichtig:

```python
text = 'Hallo'

# str -> bytes: encode()
als_utf8  = text.encode('utf-8')
als_ascii = text.encode('ascii')
print(als_utf8)    # b'Hallo'  -- 5 Bytes, identisch mit ASCII
print(als_ascii)   # b'Hallo'

# bytes -> str: decode()
zurueck = als_utf8.decode('utf-8')
print(zurueck)     # Hallo
print(type(text))      # <class 'str'>
print(type(als_utf8))  # <class 'bytes'>
```

```python
# Zeichen ausserhalb von ASCII brauchen mehr Bytes in UTF-8
umlaut = 'ä'
print(f'Unicode-Codepoint: {ord(umlaut)}')
print(f'UTF-8 Bytes:       {umlaut.encode("utf-8")}')   # 2 Bytes
print(f'Anzahl Bytes:      {len(umlaut.encode("utf-8"))}')

print()
chinesisch = '中'
print(f'Unicode-Codepoint: {ord(chinesisch)}')
print(f'UTF-8 Bytes:       {chinesisch.encode("utf-8")}')  # 3 Bytes
print(f'Anzahl Bytes:      {len(chinesisch.encode("utf-8"))}')
```

```python
# ASCII kann keine Umlaute -- UnicodeEncodeError
try:
    'Straße'.encode('ascii')
except UnicodeEncodeError as e:
    print(f'Fehler: {e}')

# Mit UTF-8 kein Problem
print('Straße'.encode('utf-8'))   # b'Stra\xc3\x9fe'

# Falsche Decodierung ergibt Kauderwelsch
bytes_daten = 'Straße'.encode('utf-8')
try:
    falsch = bytes_daten.decode('ascii')
except UnicodeDecodeError as e:
    print(f'Falsche Decodierung: {e}')
```

4. ord() und chr() -- zwischen Zeichen und Zahlen wechseln
</span>
</div>

`ord()` und `chr()` sind die direkten Werkzeuge um zwischen Zeichen und Unicode-Codepoints zu wechseln. Sie werden in der PCAP-Pruefung haeufig eingesetzt -- oft in Kombination mit String-Operationen oder Vergleichen.

```python
# ord() und chr() sind Umkehrfunktionen
print(ord('A'))         # 65
print(chr(ord('A')))    # 'A'  -- hin und zurueck

# Warum ist das nuetzlich? Weil Zeichenmanipulation oft ueber Zahlen geht

# Beispiel: Naechsten Buchstaben im Alphabet
def naechster_buchstabe(zeichen):
    return chr(ord(zeichen) + 1)

print(naechster_buchstabe('A'))   # B
print(naechster_buchstabe('Z'))   # [  -- Vorsicht: nach Z kommt kein A mehr!

# Beispiel: Gross- in Kleinbuchstaben umwandeln (manuell)
def zu_kleinbuchstabe(zeichen):
    if 'A' <= zeichen <= 'Z':
        return chr(ord(zeichen) + 32)   # Abstand zwischen A(65) und a(97) ist 32
    return zeichen

print(zu_kleinbuchstabe('H'))   # h
print(zu_kleinbuchstabe('a'))   # a  -- unveraendert
```

```python
# PCAP-relevantes Wissen: Codepoint-Reihenfolge bestimmt String-Vergleiche
print('A' < 'B')    # True  -- 65 < 66
print('Z' < 'a')    # True  -- 90 < 97 -- Grossbuchstaben kommen vor Kleinbuchstaben!
print('z' < 'A')    # False

# Das erklaert warum sorted() ohne key Grossbuchstaben zuerst sortiert
woerter = ['banane', 'Apfel', 'kirsche', 'Dattel']
print(sorted(woerter))            # ['Apfel', 'Dattel', 'banane', 'kirsche']
print(sorted(woerter, key=str.lower))  # ['Apfel', 'banane', 'Dattel', 'kirsche']
```

5. Indexing, Slicing und Iteration
</span>
</div>

Ein String ist in Python eine **geordnete, unveraenderliche Folge von Zeichen**. Geordnet bedeutet: Jedes Zeichen hat eine Position -- den Index. Unveraenderlich (immutable) bedeutet: Man kann einen String nicht an Ort und Stelle veraendern -- jede Manipulation erzeugt einen neuen String.

```python
text = 'Python'
#       0 1 2 3 4 5   positiver Index
#      -6-5-4-3-2-1   negativer Index

# Einzelne Zeichen -- Indexzugriff
print(text[0])    # P  -- erstes Zeichen
print(text[5])    # n  -- letztes Zeichen
print(text[-1])   # n  -- letztes Zeichen (von hinten)
print(text[-2])   # o  -- vorletztes Zeichen

# Slicing: text[start:stop:schritt]
print(text[0:3])   # Pyt  -- Index 0, 1, 2 (stop ist exklusiv!)
print(text[2:])    # thon -- von Index 2 bis Ende
print(text[:4])    # Pyth -- von Anfang bis Index 3
print(text[::2])   # Pto  -- jedes zweite Zeichen
print(text[::-1])  # nohtyP -- rueckwaerts
```

```python
# Strings sind immutable -- direktes Aendern ist nicht moeglich
text = 'Python'
try:
    text[0] = 'J'   # TypeError!
except TypeError as e:
    print(f'TypeError: {e}')

# Stattdessen: neuen String erzeugen
neuer_text = 'J' + text[1:]
print(neuer_text)   # Jython
print(text)         # Python -- Original unveraendert
```

6. Operatoren: in, not in, +, *, Vergleiche
</span>
</div>

```python
text = 'Python ist toll'

# in / not in -- Teilstring pruefen
print('Python' in text)       # True
print('Java' not in text)     # True
print('python' in text)       # False -- Gross-/Kleinschreibung beachtet

# + -- Verkettung (Konkatenation)
gruss  = 'Hallo' + ', ' + 'Welt' + '!'
print(gruss)

# * -- Wiederholung
print('Ha' * 3)    # HaHaHa
print('-' * 30)    # 30 Bindestriche

# Vergleiche -- basieren auf Codepoints
print('abc' == 'abc')   # True
print('abc' < 'abd')    # True -- 'c' (99) < 'd' (100)
print('abc' < 'abcd')   # True -- kuerzer kommt zuerst bei gleichem Anfang
print('B' < 'a')        # True -- Grossbuchstaben (66) < Kleinbuchstaben (97)
```

```python
# Iteration -- Zeichen fuer Zeichen durchlaufen
for zeichen in 'Python':
    print(zeichen, end=' ')   # P y t h o n
print()

# Haeufigkeiten zaehlen
text = 'Mississippi'
haeufigkeit = {}
for zeichen in text:
    haeufigkeit[zeichen] = haeufigkeit.get(zeichen, 0) + 1
print(haeufigkeit)   # {'M': 1, 'i': 4, 's': 4, 'p': 2}
```

7. Escape-Sequenzen
</span>
</div>

Manche Zeichen koennen nicht direkt in einen String geschrieben werden -- zum Beispiel ein Zeilenumbruch. Dafuer gibt es **Escape-Sequenzen**: Ein Backslash `\` gefolgt von einem Zeichen bekommt eine besondere Bedeutung.

```python
# Haeufige Escape-Sequenzen
print('Zeile 1\nZeile 2')       # \n = Zeilenumbruch
print('Spalte1\tSpalte2')        # \t = Tab
print('Er sagte: \'Hallo\'')    # \' = einfaches Anfuehrungszeichen
print('Pfad: C:\\Users\\Aldric') # \\ = einfacher Backslash

# Unicode-Zeichen per Codepoint einbetten
print('\u00e4')   # ä  -- Unicode-Escape mit 4 Hex-Ziffern
print('\u20ac')   # €
print('\U0001F600')  # 😀 -- 8 Hex-Ziffern fuer Zeichen ausserhalb der Basic Multilingual Plane

# Raw-Strings: Backslash wird nicht als Escape interpretiert
normal = 'C:\\Users\\neu'
raw    = r'C:\Users\neu'    # r vor dem String
print(normal == raw)        # True -- identisch
```
