**Checkliste -- das muss sitzen:**

- [ ] `ord()` und `chr()` -- Richtung merken
- [ ] ASCII 0-127, Grossbuchstaben (65-90) < Kleinbuchstaben (97-122)
- [ ] Vergleiche basieren auf Codepoints -- `'Z' < 'a'` ist True
- [ ] Slicing: `stop` ist exklusiv, negative Indizes, `[::-1]` dreht um
- [ ] Strings sind immutable -- `text[0] = 'X'` gibt TypeError
- [ ] `isdigit()`: `-5` und `3.14` geben False
- [ ] `join()`: auf dem Trennzeichen aufrufen, nicht auf der Liste
- [ ] `split()` mit maxsplit, leere Strings bei doppeltem Trenner
- [ ] `sort()` gibt `None` zurueck -- `sorted()` gibt neue Liste
- [ ] `find()` gibt `-1`, `index()` gibt `ValueError`
- [ ] `rfind()` -- letztes Vorkommen

```python
# Codepoints und Vergleiche
print(ord('A'), ord('Z'))   # 65, 90
print(ord('a'), ord('z'))   # 97, 122
print(ord('0'), ord('9'))   # 48, 57

# Vergleiche -- haeufige PCAP-Falle
print('A' < 'a')    # True  -- 65 < 97
print('Z' < 'a')    # True  -- 90 < 97  (alle Grossbuchstaben < Kleinbuchstaben)
print('z' < 'A')    # False
print('abc' < 'abd')  # True  -- 'c'(99) < 'd'(100)
print('abc' < 'abcd') # True  -- kuerzer kommt zuerst bei gleichem Anfang
```

```python
text = 'Python'
#       0 1 2 3 4 5
#      -6-5-4-3-2-1

print(text[0])     # P
print(text[-1])    # n  -- letztes Zeichen
print(text[1:4])   # yth  -- stop=4 ist exklusiv!
print(text[::-1])  # nohtyP
print(text[::2])   # Pto

# Immutabilitaet
try:
    text[0] = 'J'
except TypeError as e:
    print(f'TypeError: {e}')
```

```python
# isdigit() -- Fallen
print('123'.isdigit())    # True
print('-5'.isdigit())     # False -- Minus ist keine Ziffer
print('3.14'.isdigit())   # False -- Punkt ist keine Ziffer
print(''.isdigit())       # False -- leerer String
print('²'.isdigit())      # True  -- Unicode-Hochzahl gilt als Ziffer
```

```python
# split() -- Fallen
print('a,,b'.split(','))       # ['a', '', 'b']  -- leerer String!
print('a:b:c'.split(':', 1))   # ['a', 'b:c']    -- maxsplit=1

# join() -- auf dem Trennzeichen aufrufen
print('-'.join(['a', 'b', 'c']))  # a-b-c
# ['a','b','c'].join('-')  --> AttributeError!

# sort() gibt None zurueck
woerter = ['c', 'a', 'b']
ergebnis = woerter.sort()
print(ergebnis)   # None -- nicht die sortierte Liste!
print(woerter)    # ['a', 'b', 'c']  -- die Liste selbst ist sortiert
```

```python
text = 'Python ist toll und Python macht Spass'

print(text.find('Python'))    # 0   -- erstes Vorkommen
print(text.rfind('Python'))   # 19  -- letztes Vorkommen
print(text.find('Java'))      # -1  -- nicht gefunden

try:
    text.index('Java')         # ValueError!
except ValueError as e:
    print(f'ValueError: {e}')
```
