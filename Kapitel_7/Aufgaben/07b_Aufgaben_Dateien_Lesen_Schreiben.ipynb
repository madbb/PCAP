Gegeben ist eine Datei `data.txt` mit folgendem Inhalt (drei Zeilen):
```
Zeile1
Zeile2
Zeile3
```
Den Code nicht ausfuehren. Ergebnis als Kommentar schreiben, dann pruefen.

**4.1**
```python
with open('data.txt', 'r') as f:
    print(repr(f.readline()))
    print(repr(f.readline()))
```

**4.2**
```python
with open('data.txt', 'r') as f:
    zeilen = f.readlines()
print(len(zeilen))
print(repr(zeilen[0]))
print(repr(zeilen[-1]))
```

**4.3**
```python
with open('data.txt', 'r') as f:
    f.read()
    rest = f.read()
print(repr(rest))
print(len(rest))
```

**4.4**
```python
with open('data.txt', 'r') as f:
    print(repr(f.read(6)))
    print(repr(f.read(6)))
```
