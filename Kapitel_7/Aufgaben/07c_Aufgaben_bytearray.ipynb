**4.1**
```python
ba = bytearray(b'ABC')
print(ba[0])
print(type(ba[0]))
print(ba[0:2])
print(type(ba[0:2]))
```

**4.2**
```python
ba = bytearray([65, 66, 67])
print(ba.decode('utf-8'))
ba[1] = 90
print(ba.decode('utf-8'))
```

**4.3**
```python
ba = bytearray(b'Hallo')
b  = bytes(ba)
print(ba == b)
print(ba is b)
print(type(ba) == type(b))
```

**4.4**
```python
ba = bytearray(3)
print(list(ba))
ba[0] = 255
ba[1] = 0
ba[2] = 128
print(list(ba))
```
