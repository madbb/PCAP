Das Modul `platform` ermoeglicht den Zugriff auf Daten der zugrunde liegenden Plattform, d. h. Hardware-, Betriebssystem- und Interpreter-Versionsinformationen.

<div style="background:#0d1f2d;border-left:3px solid #1e4d6b;border-radius:3px;padding:0.55em 1em;margin:1.5rem 0 0.9rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(0.9rem,1.8vw,1.05rem);font-weight:600;color:#c5cae9;letter-spacing:-0.005em;"><span style="font-size:0.6rem;letter-spacing:0.1em;text-transform:uppercase;color:#80cbc4;background:rgba(128,203,196,0.1);border:1px solid rgba(128,203,196,0.25);border-radius:3px;padding:0.15em 0.5em;margin-right:0.65em;vertical-align:middle;white-space:nowrap;">PCAP 1.4</span>Die Funktion platform()</span></div>

Es gibt eine Funktion, die alle darunterliegenden Schichten auf einen Blick anzeigen kann – sie heißt ebenfalls `platform`. Sie gibt einen String zurueck, der die Umgebung beschreibt – daher richtet sich ihre Ausgabe eher an Menschen als an automatisierte Verarbeitung.

```python
platform(aliased=False, terse=False)
```

- `aliased` → Wenn auf `True` gesetzt, kann die Funktion alternative Namen der zugrunde liegenden Schichten anzeigen, anstelle der ueblichen Bezeichnungen.
- `terse` → Wenn auf `True` gesetzt, kann die Funktion versuchen, eine kuerzere Form der Ausgabe zu liefern (sofern moeglich).

```python
from platform import platform

print(platform())
print(platform(1))
print(platform(0, 1))
```

````python
#Teste es selber
from platform import platform

print(platform())
print(platform(1))
print(platform(0, 1))
````

Wenn du wissen moechtest, welche Version von Python deinen Code ausfuehrt, kannst du das mit speziellen Funktionen ueberpruefen:

- `python_implementation()` → gibt einen String zurueck, der die Python-Implementierung angibt (in der Regel `CPython`, es sei denn, du nutzt eine nicht-kanonische Python-Variante)

- `python_version_tuple()` → gibt ein Tupel mit drei Elementen zurueck:
    - dem Hauptversionsanteil von Python
    - dem Nebenversionsteil
    - der Patch-Level-Nummer

```python
from platform import python_implementation, python_version_tuple

print(python_implementation())

for atr in python_version_tuple():
    print(atr)
```

````python
# Teste es selber
from platform import python_implementation, python_version_tuple

print(python_implementation())

for atr in python_version_tuple():
    print(atr)
````

````python
# Alle Informationen auf einen Blick
import platform

print("System:", platform.system())
print("Plattform:", platform.platform())
print("Maschine:", platform.machine())
print("Prozessor:", platform.processor())
print("OS-Version:", platform.version())
print("Python-Implementierung:", platform.python_implementation())
print("Python-Version:", platform.python_version_tuple())
````
