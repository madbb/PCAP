```python
import random

MODUL_NAME = "Mini Random Tools"

def wuerfeln(anzahl_wuerfe=1):
    '''Simuliert das Würfeln mit einem Standard-Würfel.'''
    ergebnisse = []
    for _ in range(anzahl_wuerfe):
        wurf = random.randint(1, 6)
        ergebnisse.append(wurf)
    return ergebnisse

def zufalls_element_aus_liste(input_liste):
    '''Wählt ein zufälliges Element aus der gegebenen Liste.'''
    if not input_liste: # Überprüft, ob die Liste leer ist
        return None
    return random.choice(input_liste)

if __name__ == "__main__":
    print(f"--- {MODUL_NAME} wird direkt ausgeführt ---")
    
    print("Test Würfeln (3x):", wuerfeln(3))
    print("Test Würfeln (1x):", wuerfeln()) # Standardwert nutzen

    test_liste = ["rot", "grün", "blau"]
    print(f"Test Zufallselement aus {test_liste}:", zufalls_element_aus_liste(test_liste))
    
    leere_liste = []
    print(f"Test Zufallselement aus {leere_liste}:", zufalls_element_aus_liste(leere_liste))
    print(f"--- Ende Selbsttest {MODUL_NAME} ---")
```

````python
# Antowrt 6:
import math
from random import randrange

basis = randrange(2, 5) # Mögliche Werte: 2, 3, 4
exponent = randrange(2, 4) # Mögliche Werte: 2, 3

potenz = math.pow(basis, exponent)

print(f"Basis: {basis}, Exponent: {exponent}")
print("Potenz:", potenz)
````
