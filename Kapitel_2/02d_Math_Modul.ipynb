Die Funktion `math.sqrt(x)` berechnet die **Quadratwurzel** von `x`.

**Mathematische Definition:** Die Zahl `r` fuer die gilt: `r × r = x`

$$\sqrt{x} = r \quad \text{so dass} \quad r^2 = x$$

Beispiele: `sqrt(9) = 3`, `sqrt(16) = 4`, `sqrt(2) ≈ 1.414`

Wichtig: `math.sqrt()` gibt immer einen `float` zurueck. Bei negativen Zahlen: `ValueError`.

````python
print(math.pow(3, 4))
````

**Rechnet:** $x^y$

**Beispiel:** `math.pow(3, 4)` ergibt `81`, weil `3 × 3 × 3 × 3 = 81`

Beachte: `math.pow()` liefert immer eine Float-Zahl. Alternativ kannst du den `**`-Operator verwenden.

`math.trunc(x)` schneidet die Nachkommastellen einfach ab -- es wird immer **Richtung Null** gerundet.

Das ist der Unterschied zu `floor()`: Bei negativen Zahlen verhält sich `trunc()` anders!

- `trunc(5.9) = 5`  (nach unten -- wie floor)
- `trunc(-5.2) = -5`  (nach oben Richtung Null -- anders als floor(-5.2) = -6!)

Merkhilfe: `trunc()` entfernt einfach alles nach dem Komma. `floor()` rundet immer zur kleineren Zahl.

````python
math.trunc(5.9)
````

Die Funktion `math.factorial(n)` berechnet die **Fakultaet** einer nicht-negativen ganzen Zahl.

**Mathematische Definition:**

$$n! = n \times (n-1) \times (n-2) \times \ldots \times 1$$

Beispiele:
- `5! = 5 × 4 × 3 × 2 × 1 = 120`
- `3! = 3 × 2 × 1 = 6`
- `0! = 1` (per Definition)

Wichtig: Nur fuer ganze, nicht-negative Zahlen. Bei Kommazahlen oder negativen Werten: `ValueError`.

Die Funktion `math.hypot(a, b)` berechnet die **Hypotenuse** eines rechtwinkligen Dreiecks -- also die laengste Seite gegenueber dem rechten Winkel.

**Mathematische Grundlage -- Satz des Pythagoras:**

$$c = \sqrt{a^2 + b^2}$$

Wobei `a` und `b` die Katheten (die beiden kuerzeren Seiten) und `c` die Hypotenuse ist.

Beispiel: Bei einem Dreieck mit den Katheten `3` und `4` ist die Hypotenuse `5` (das bekannte 3-4-5-Dreieck).

````python
import math

# Klassisches 3-4-5-Dreieck
print(math.hypot(3, 4))    # 5.0

# Praxisbeispiel: Diagonale eines Monitors berechnen
# Ein 16:9 Monitor mit 40 cm Breite und 22.5 cm Hoehe
breite = 40.0
hoehe  = 22.5
diagonale = math.hypot(breite, hoehe)
print(f'Monitordiagonale: {diagonale:.1f} cm')  # ca. 45.8 cm

# Entfernung zwischen zwei Punkten in einem Koordinatensystem
x1, y1 = 1, 2
x2, y2 = 4, 6
entfernung = math.hypot(x2 - x1, y2 - y1)
print(f'Entfernung: {entfernung:.2f}')  # 5.0
````
