Die Funktion seed()</span></div>

Der Zufallsgenerator startet immer mit einem sogenannten **Startwert** (Seed). Normalerweise nutzt Python dafuer die aktuelle Systemzeit – das Ergebnis ist also bei jedem Aufruf verschieden.

Mit `seed()` kannst du diesen Startwert selbst festlegen. Das macht die Ausgabe **reproduzierbar** – nuetzlich fuer Tests, Debugging oder Demonstrationen.

```python
random.seed(a=None, version=2)
```

- `a` → Der Startwert. Bei `None` (Standard) wird die Systemzeit verwendet.
- `version` → Interner Parameter, normalerweise nicht aendern.

Beide Durchlaeufe liefern dieselben Zahlen, weil der Seed identisch ist.

````python
# Teste es selber
import random

random.seed(42)
print(random.randint(1, 100))
print(random.randint(1, 100))

random.seed(42)
print(random.randint(1, 100))
print(random.randint(1, 100))
````

Die grundlegendste Funktion des Moduls heißt genauso wie das Modul selbst: `random()`.

Sie gibt einen zufaelligen **Gleitkommawert** zurueck – immer im Bereich:

```
0.0 <= x < 1.0
```

Die obere Grenze `1.0` ist **ausgeschlossen**.

```python
from random import random

print(random())
print(random())
```

Moegliche Ausgabe:
```
0.6394267984578837
0.025010755222666936
0.27502931836911926
```

````python
# Teste es selber
from random import random

print(random())
print(random())
print(random())
````

Die Funktion uniform()</span></div>

Wenn du einen Gleitkommawert in einem **eigenen Bereich** brauchst, verwende `uniform()`.

```python
random.uniform(a, b)
```

- Gibt einen Wert `x` zurueck, sodass `a <= x <= b`.
- **Beide** Grenzen koennen im Ergebnis vorkommen (im Gegensatz zu `random()`).

````python
# Teste es selber
import random

print(random.uniform(1.5, 4.5))
print(random.uniform(0, 100))
````

Fuer ganze Zufallszahlen in einem bestimmten Bereich gibt es `randint()`.

- Gibt eine ganzzahlige Zufallszahl zurueck: `a <= x <= b`.

Die Funktion randrange()</span></div>

`randrange()` ist flexibler als `randint()`: Du kannst zusaetzlich eine Schrittweite angeben.

```python
random.randrange(start, stop, step=1)
```

- `start` → Startwert (inklusive)
- `stop` → Endwert (**exklusive** – wie bei `range()`)
- `step` → Schrittweite (Standard: 1)

````python
# Teste es selber
import random

print(random.randrange(0, 10))
print(random.randrange(0, 10, 2))
print(random.randrange(10, 100, 10))
````

Die Funktion sample()</span></div>

Wenn du **mehrere** Elemente ohne Wiederholung auswaehlen willst, ist `sample()` die richtige Wahl.

```python
random.sample(population, k)
```

- `population` → Die Sequenz, aus der gewaehlt wird
- `k` → Anzahl der Elemente, die ausgewaehlt werden sollen
- Jedes Element kann **nur einmal** vorkommen (keine Wiederholung)
- Wenn `k` groesser als die Laenge der Sequenz: `ValueError`

````python
# Teste es selber
import random

zahlen = list(range(1, 50))
print(random.sample(zahlen, 6))
````

Die Funktion shuffle()</span></div>

`shuffle()` mischt eine Liste **in-place** – also direkt in der Originalliste, ohne eine neue zu erstellen.

```python
random.shuffle(x)
```

- `x` → Die Liste, die gemischt werden soll
- Gibt **`None`** zurueck – der Rueckgabewert ist nicht die gemischte Liste!
- Funktioniert nur mit **veraenderbaren** Sequenzen (Listen, nicht Tupel)

````python
# Teste es selber
import random

karten = ['Ass', 'Koenig', 'Dame', 'Bube', '10']
print("Vorher:", karten)

random.shuffle(karten)
print("Nachher:", karten)
````

**Haeufiger Fehler:** `shuffle()` gibt `None` zurueck – diesen Code **nicht** schreiben:

````python
import random

karten = ['Ass', 'Koenig', 'Dame']

# FALSCH -- ergebnis ist None!
ergebnis = random.shuffle(karten)
print(ergebnis)  # None

# RICHTIG -- die originale Liste wurde veraendert
print(karten)
````

Die Funktion choices()</span></div>

`choices()` waehlt **mehrere Elemente mit Wiederholung** aus – im Gegensatz zu `sample()`.

```python
random.choices(population, weights=None, k=1)
```

- `population` → Die Sequenz, aus der gewaehlt wird
- `weights` → Optionale Gewichtung (haeufigere Elemente wahrscheinlicher)
- `k` → Anzahl der Ergebnisse
- Ein Element **kann mehrfach** vorkommen

````python
# Teste es selber
import random

print(random.choices(['Kopf', 'Zahl'], k=5))

print(random.choices(['rot', 'blau'], weights=[3, 1], k=10))
````

<div style="background:#0b1929;border-left:4px solid #4fc3f7;border-radius:4px;padding:0.75em 1.2em;margin:2rem 0 1.2rem 0;font-family:'Segoe UI',sans-serif;"><span style="font-size:clamp(1rem,2.2vw,1.25rem);font-weight:600;color:#e8eaf6;letter-spacing:-0.01em;"><span style="font-size:0.65rem;letter-spacing:0.1em;text-transform:uppercase;color:#4fc3f7;background:rgba(79,195,247,0.12);border:1px solid rgba(79,195,247,0.3);border-radius:3px;padding:0.2em 0.55em;margin-right:0.75em;vertical-align:middle;font-family:'Segoe UI',sans-serif;white-space:nowrap;">Kursinhalt</span>11. Alle Funktionen auf einen Blick</span></div>

| Funktion | Rueckgabe | Wiederholung | Anmerkung |
|---|---|---|---|
| `random()` | Float 0.0–1.0 | - | Basiswert |
| `uniform(a, b)` | Float in [a, b] | - | Beide Grenzen inklusive |
| `randint(a, b)` | Int in [a, b] | - | Beide Grenzen inklusive |
| `randrange(start, stop, step)` | Int in [start, stop) | - | stop ist exklusive |
| `choice(seq)` | 1 Element | - | Sequenz muss nicht leer sein |
| `choices(seq, k=n)` | Liste mit n Elementen | Ja | Gewichtung moeglich |
| `sample(seq, k)` | Liste mit k Elementen | Nein | k <= len(seq) |
| `shuffle(x)` | None (in-place) | - | Veraendert Originalliste |

````python
# Alle Funktionen in der Uebersicht
import random

random.seed(0)

print("random():           ", random.random())
print("uniform(1, 10):     ", random.uniform(1, 10))
print("randint(1, 100):    ", random.randint(1, 100))
print("randrange(0,10,2):  ", random.randrange(0, 10, 2))

seq = ['a', 'b', 'c', 'd', 'e']
print("choice(seq):        ", random.choice(seq))
print("choices(seq, k=3):  ", random.choices(seq, k=3))
print("sample(seq, k=3):   ", random.sample(seq, k=3))

liste = [1, 2, 3, 4, 5]
random.shuffle(liste)
print("shuffle([1..5]):     ", liste)
````
