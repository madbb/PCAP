Python kennt fuenf Wege, etwas zu importieren:

````python
import math                        # Zugriff ueber math.sqrt()
import math as m                   # Alias: m.sqrt()
from math import sqrt              # Direkt: sqrt()
from math import sqrt as wurzel    # Direkt mit Alias: wurzel()
from math import *                 # Alles -- in Produktionscode vermeiden

print(math.sqrt(16))   # 4.0
print(m.sqrt(16))      # 4.0
print(sqrt(16))        # 4.0
print(wurzel(16))      # 4.0
````

`import math` traegt nur den Namen `math` in den Namespace ein. `from math import sqrt` traegt nur `sqrt` ein -- `math` selbst ist dann nicht verfuegbar.

`from math import *` importiert alles, was nicht mit `_` beginnt. In der Pruefung auftauchende Falle: danach kann man `math.sqrt()` **nicht** nutzen, nur `sqrt()` direkt.

````python
import sys

# sys.path: Liste der Verzeichnisse, in denen Python nach Modulen sucht
for pfad in sys.path:
    print(pfad if pfad else '(aktuelles Verzeichnis)')
````

````python
import math

# dir() zeigt alle Namen eines Moduls
print(dir(math))

# Nur die oeffentlichen (ohne fuehrendem Unterstrich)
print([n for n in dir(math) if not n.startswith('_')])
````

````python
# __name__ ist '__main__' wenn die Datei direkt ausgefuehrt wird
# und der Modulname wenn sie importiert wird

print(__name__)   # In diesem Notebook: '__main__'

import math
print(math.__name__)   # 'math'
````

````python
# Das ist der Grund fuer dieses Muster in .py-Dateien:

def berechne(x):
    return x * 2

if __name__ == '__main__':
    # Dieser Block laeuft nur bei direktem Aufruf, nicht beim Import
    print(berechne(5))
````

`__pycache__` ist ein Verzeichnis, das Python automatisch anlegt. Es enthaelt den kompilierten Bytecode (`.pyc`-Dateien) der importierten Module. Python prueft beim naechsten Import, ob die `.pyc`-Datei aktuell ist -- wenn ja, wird sie direkt geladen statt neu kompiliert. Gehoert nicht ins Git-Repository.

Die Pruefung testet drei Module konkret: `math`, `random`, `platform`.

````python
import math

print(math.ceil(4.1))       # 5  -- aufrunden
print(math.floor(4.9))      # 4  -- abrunden
print(math.trunc(4.9))      # 4  -- Nachkommastellen abschneiden
print(math.factorial(5))    # 120
print(math.sqrt(25))        # 5.0
print(math.hypot(3, 4))     # 5.0 -- Hypotenuse: sqrt(3^2 + 4^2)
````

````python
import random

random.seed(42)                        # Reproduzierbarer Zufallswert
print(random.random())                 # Float zwischen 0.0 und 1.0
print(random.randint(1, 6))            # Ganzzahl 1-6 (inklusive beider Grenzen)
print(random.choice(['a', 'b', 'c'])) # Zufaelliges Element
print(random.sample([1,2,3,4,5], 3))  # 3 Elemente ohne Wiederholung
````

````python
import platform

print(platform.platform())              # z.B. 'Windows-10-...' oder 'Linux-...'
print(platform.machine())              # z.B. 'AMD64'
print(platform.processor())            # Prozessor-Info
print(platform.system())               # 'Windows', 'Linux', 'Darwin'
print(platform.python_version_tuple()) # ('3', '11', '0')
print(platform.python_implementation())# 'CPython'
````

Ein Paket ist ein Verzeichnis mit einer `__init__.py`-Datei. Es fasst mehrere Module zusammen.

````python
import os

# Paketstruktur erstellen (Demo)
os.makedirs('mein_paket', exist_ok=True)

with open('mein_paket/__init__.py', 'w') as f:
    f.write('# Paket-Initialisierung\n')
    f.write('VERSION = "1.0"\n')

with open('mein_paket/rechner.py', 'w') as f:
    f.write('def addiere(a, b):\n    return a + b\n')

# Paket importieren
import mein_paket
print(mein_paket.VERSION)

from mein_paket import rechner
print(rechner.addiere(3, 4))

from mein_paket.rechner import addiere
print(addiere(10, 5))
````

Die `__init__.py` kann leer sein oder Code enthalten. Sie macht das Verzeichnis zu einem Paket. Ohne sie behandelt Python das Verzeichnis nicht als importierbares Paket.

Oeffentliche Variablen in einem Modul sind alle Namen ohne fuehrenden Unterstrich. Mit `__all__` kann ein Modul explizit steuern, was bei `from modul import *` exportiert wird.
