Welche der folgenden Aussagen gibt `True` aus?

```python
class Tier: pass
class Saeugetier(Tier): pass
class Hund(Saeugetier): pass

h = Hund()
```

A) `print(isinstance(h, Tier) and not isinstance(Saeugetier(), Hund))`  
B) `print(type(h) is Tier)`
