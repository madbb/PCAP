**Frage 20 von 20** | Syllabus: 5.4 / 5.5 – I/O-Modi Uebersicht

Welche Kombination aus Modus und Verhalten ist korrekt? (Wähle zwei)

A) `'r'` – Lesen; Datei muss existieren, sonst `FileNotFoundError`  
B) `'w'` – Schreiben; haengt an vorhandene Datei an  
C) `'a'` – Schreiben; haengt an vorhandene Datei an, erstellt sie falls nicht vorhanden  
D) `'x'` – Schreiben; oeffnet auch vorhandene Dateien ohne Fehler


````python
import os

# 'r' - Datei muss existieren
try:
    open(tmppath("nope2.txt"), "r")
except FileNotFoundError:
    print("'r' auf fehlende Datei: FileNotFoundError")

# 'w' - ueberschreibt
p = tmppath("modus_w.txt")
with open(p, "w") as f: f.write("Alt")
with open(p, "w") as f: f.write("Neu")
with open(p, "r") as f: print("'w' Inhalt:", repr(f.read()))

# 'a' - haengt an
p2 = tmppath("modus_a.txt")
with open(p2, "w") as f: f.write("Alt")
with open(p2, "a") as f: f.write("Neu")
with open(p2, "r") as f: print("'a' Inhalt:", repr(f.read()))

# 'x' - schlaegt fehl wenn Datei existiert
try:
    open(p, "x")
except FileExistsError:
    print("'x' auf vorhandene Datei: FileExistsError")
````

---

## Loesungen

1. **B) `FileNotFoundError`** - Modus `'r'` setzt voraus, dass die Datei existiert. `FileNotFoundError` ist eine Unterklasse von `OSError`.

2. **B) `"Zweiter"`** - Modus `'w'` oeffnet die Datei zum Schreiben und leert sie dabei vollstaendig. Der vorherige Inhalt geht verloren.

3. **B) `"AB"`** - Modus `'a'` (append) haengt neuen Inhalt ans Ende der Datei. Die Datei wird erstellt, falls sie noch nicht existiert.

4. **C) `5`** - `write()` gibt die Anzahl der geschriebenen Zeichen als `int` zurueck. `"Hallo"` hat 5 Zeichen.

5. **C) `"AB"`** - `write()` fuegt keinen automatischen Zeilenumbruch ein. Mehrere `write()`-Aufrufe haengen den Text nahtlos aneinander.

6. **B) `'Hallo'`** - `read(n)` liest maximal `n` Zeichen. `read(5)` liest die ersten 5 Zeichen: `'Hallo'`.

7. **B) `'Zeile1\n'`** - `readline()` liest eine Zeile einschliesslich des abschliessenden Zeilenumbruchs `\n`.

8. **C) `<class 'list'>` gefolgt von `3`** - `readlines()` gibt eine Liste aller Zeilen zurueck. Jede Zeile ist ein String mit `\n` am Ende.

9. **B) `True`** - Das `with`-Statement ruft `__exit__` auf, das `close()` ausfuehrt. Nach dem Block ist `f.closed` immer `True`.

10. **C) `ValueError`** - Eine geschlossene Datei kann nicht mehr gelesen werden. Python wirft `ValueError: I/O operation on closed file.`

11. **C) `<class 'str'>` gefolgt von `<class 'bytes'>`** - Im Textmodus gibt `read()` einen `str` zurueck. Im Binaermodus (`'rb'`) gibt `read()` ein `bytes`-Objekt zurueck.

12. **B) `b'ABC'`** - Im Binaermodus wird ein `bytes`-Objekt zurueckgegeben. `\x41`, `\x42`, `\x43` sind die ASCII-Codes fuer `A`, `B`, `C`.

13. **B) `bytearray(b'ABC')`** - `readinto()` schreibt die Bytes direkt in den uebergebenen `bytearray`-Puffer. Der Rueckgabetyp ist weiterhin `bytearray`.

14. **C) `FileExistsError`** - Modus `'x'` (exclusive creation) erstellt eine neue Datei und schlaegt mit `FileExistsError` fehl, wenn die Datei bereits existiert.

15. **C) `2`** - `errno.ENOENT` hat den Wert `2` und steht fuer "No such file or directory". `errno.EACCES` (`13`) steht fuer Zugriffsverweigerung.

16. **B) `True`** - `OSError` (und alle Unterklassen wie `FileNotFoundError`) haben ein `.errno`-Attribut mit dem numerischen Fehlercode. Bei fehlender Datei entspricht dieser `errno.ENOENT` (2).

17. **B)** - `sys.stdin`, `sys.stdout` und `sys.stderr` sind dateiaehnliche Objekte (`TextIOWrapper`). Sie koennen umgeleitet werden, z.B. beim Pipen in der Shell oder per `sys.stdout = open(...)`.

18. **A) und C)** - Textmodus gibt `str` zurueck, Binaermodus gibt `bytes` zurueck.

19. **B) `'Zeile1\n'` gefolgt von `'Zeile1'`** - `readlines()` behaelt die Zeilenumbrueche in den Strings. `strip()` entfernt fuehrendes und nachfolgendes Whitespace einschliesslich `\n`.

20. **A) und C)** - `'r'` erfordert eine vorhandene Datei. `'a'` haengt an und erstellt die Datei bei Bedarf. `'w'` ueberschreibt (nicht anhängen). `'x'` schlaegt bei vorhandener Datei fehl.
