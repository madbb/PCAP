**Frage 13 von 20** | Syllabus: 4.5 – Polymorphismus

Was gibt der folgende Code aus?

```python
class Form:
    def flaeche(self):
        return 0

class Rechteck(Form):
    def __init__(self, b, h):
        self.b = b
        self.h = h
    def flaeche(self):
        return self.b * self.h

class Quadrat(Rechteck):
    def __init__(self, s):
        super().__init__(s, s)

formen = [Rechteck(3, 4), Quadrat(5)]
for f in formen:
    print(f.flaeche())
```

A) `0` gefolgt von `0`  
B) `12` gefolgt von `5`  
C) `12` gefolgt von `25`  
D) `TypeError`


````python
class Form:
    def flaeche(self):
        return 0

class Rechteck(Form):
    def __init__(self, b, h):
        self.b = b
        self.h = h
    def flaeche(self):
        return self.b * self.h

class Quadrat(Rechteck):
    def __init__(self, s):
        super().__init__(s, s)

formen = [Rechteck(3, 4), Quadrat(5)]
for f in formen:
    print(f.flaeche())
````

**Frage 20 von 20** | Syllabus: 4.1 – OOP-Konzepte: Encapsulation

Welche Aussage beschreibt Encapsulation (Kapselung) am genauesten?

A) Eine Klasse kann von mehreren Elternklassen erben.

B) Daten und die Methoden, die auf diese Daten zugreifen, werden in einer Klasse zusammengefasst und der direkte externe Zugriff wird eingeschraenkt.

C) Eine Methode verhaelt sich je nach Objekttyp unterschiedlich.

D) Eine Klasse uebernimmt Attribute und Methoden einer anderen Klasse.


---

## Loesungen

1. **C) `2`** - `zaehler` ist ein Klassenattribut. Jede Instanziierung erhoegt ihn ueber `Geraet.zaehler += 1`.

2. **B) `[42]`** - `inhalt` ist ein mutable Klassenattribut. Alle Instanzen teilen dasselbe Listenobjekt, da kein neues erzeugt wird.

3. **B) `10`** - `del c.wert` loescht das Instanzattribut. Der Attributzugriff faellt danach auf das Klassenattribut `Config.wert = 10` zurueck.

4. **A) und C)** - Klassenattribute stehen in `Klasse.__dict__`, Instanzattribute in `instanz.__dict__`. `art` gehoert der Klasse, `hoehe` der Instanz.

5. **C) `AttributeError`** - Name Mangling benennt `__betrag` intern zu `_Konto__betrag` um. Der Zugriff ueber `k.__betrag` schlaegt fehl.

6. **C) `k._Konto__betrag`** - Das Muster ist `_Klassenname__attributname`.

7. **C) `rot`** - `Lampe.beschreibe(l)` ist eine ungebundene Form des Methodenaufrufs. `l` wird explizit als `self` uebergeben.

8. **B) `Buch` gefolgt von `0.0`** - Der Konstruktor hat einen Standardwert `preis=0.0`. Wird `preis` nicht uebergeben, gilt der Default.

9. **B) `True`, `False`, `False`** - `leistung` ist ein Instanzattribut. Es liegt in der Instanz `m`, nicht in der Klasse `Motor` selbst.

10. **B) `Hund` gefolgt von `(<class '__main__.Tier'>,)`** - `__name__` gibt den Klassennamen als String zurueck. `__bases__` ist ein Tupel der direkten Elternklassen.

11. **C) `True`, `False`, `True`** - `isinstance` prueft die gesamte Vererbungskette. `Hund` erbt von `Tier` (True), nicht von `Katze` (False), aber jede Klasse erbt von `object` (True).

12. **C) `Basis` gefolgt von `Kind`** - `Kind` ueberschreibt `__str__`. Jedes Objekt nutzt seine eigene Implementierung.

13. **C) `12` gefolgt von `25`** - `Quadrat` erbt `flaeche()` von `Rechteck`. `super().__init__(5, 5)` setzt `b=5, h=5`, also `5*5=25`.

14. **B) `10` gefolgt von `20`** - `super().__init__(wert)` ruft den Elternkonstruktor auf. Beide Attribute werden korrekt gesetzt.

15. **B) `False` gefolgt von `True`** - Ohne `__eq__` vergleicht `==` auf Identitaet. `a` und `b` sind verschiedene Objekte (False). `c = a` zeigt auf dasselbe Objekt (True).

16. **B) `Basis`** - `Kind` ueberschreibt `__str__` nicht. Die Methode wird von `Basis` geerbt.

17. **C) `3` gefolgt von `4`** - Die Zuweisung `auto.raeder = 3` erzeugt ein neues Instanzattribut. Das Klassenattribut bleibt unveraendert.

18. **B) `True` gefolgt von `False`** - `getattr` mit Standardwert wirft keine Ausnahme bei fehlendem Attribut, sondern gibt den Standardwert zurueck.

19. **B) und C)** - Name Mangling erzeugt `_Zaehler__wert`. Zugriff ist ueber Instanz und Klasse moeglich.

20. **B)** - Kapselung buendelt Daten und Methoden und schuetzt den internen Zustand. A = Mehrfachvererbung, C = Polymorphismus, D = Vererbung.
